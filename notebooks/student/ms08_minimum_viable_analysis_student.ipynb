## 🎯 Definition of Done

You are done with this sprint when your submission carries all five parts,
applied to **your own project**. You are not producing new evidence this week.
You are producing the sentences your evidence licenses.

1. **The research note.** One to two pages for a stranger who has only the
   page, opening with a dated version line and a reason a reader could use to
   reconstruct your thinking. Five moves in order: the **question** on a
   problem-gap-question spine, as your Research Contract declares it; the
   **method in one plain-language paragraph**; the **results with uncertainty
   in the same eye-span** as every number; the **bounded headline claim**
   carrying its kind and its reach; and the **limitations then the
   references**, led by the limitation your M7 audit could not fix. One small
   card rides beside the method paragraph: question kind, identification
   status, result, claim boundary.
2. **The claim-to-evidence table.** Every claim in the note gets a row with
   four columns: claim, evidence, verification, boundary. This is your **claim
   ledger**, the record of what your research asserts, and it is a separate
   instrument from the AI Research Ledger. The rule it enforces: a claim with
   no row does not appear in the note. Spot-check one row before you submit.
3. **The AI disclosure statement.** Drafted **from your ledger, never from
   memory**: which tool, for which task, and how you verified what came back,
   plus what you *sent* to tools and what you did not verify.
4. **AI Research Ledger rows.** One row for every AI-assisted step, each with
   a named verification method.
5. **The version line and the dossier update line.** The artifact opens with
   **Book Milestone 9, version 1**, dated, with its reason. It closes with the
   line naming where the note, the table, and the disclosure now live.

> **A question that often comes up here:** *"The Expo is the deadline that
> counts. Shouldn't I build the poster first and write the note after?"* No,
> and the reason is protective. The poster is next week's job, and it will be
> an adaptation of what you write now. A claim that has never been written in
> full sentences has never been tested for size, and the fastest way to
> manufacture an overclaim is to shrink a result that was never bounded in
> prose.

### 🤝 AI Research Partner

AI is your arm and your research assistant in this sprint, not your brain.
Task it with the mechanical work, keep every judgment for yourself.

**Good to delegate here:**
- Red-teaming a headline sentence you wrote yourself, and generating hostile
  questions a stranger could ask about the note.
- Assembling a disclosure draft from ledger rows you paste in.
- Tightening prose you already wrote, rechecked against the table afterward.

**Never delegate (these are yours to decide):**
- The headline's verb and its reach. A verb one size too strong turns an
  honest finding into an **overclaim**, a sentence that asserts more than the
  design established.
- Which claims get a table row and which get cut. A claim you cannot fit into
  a row is a claim to cut or a check to run, and both send you back before the
  note ships.
- Whether a causal question keeps its causal label. It does. A causal question
  your design cannot identify is **causal, currently unidentified**, never
  relabeled descriptive.

Whatever you delegate, verify what comes back and log it in your **AI
Research Ledger** below. See `ai_resources/human_responsibility_checklist.md`
for the full never-delegate list.

```python
# Table-and-headline checker — edit the EDIT ME blocks to match YOUR note.
# Part 1 reads your claim-to-evidence table: every claim needs a row whose
# verification names a real check and whose boundary says where the claim
# stops. Part 2 reads your headline on TWO INDEPENDENT dimensions, kind and
# reach, each carrying its own warrant. Part 3 checks the eye-span rule.

# EDIT ME — one row per claim that will appear in the note.
claim_table = [
    {"claim": "EDIT ME (a sentence from your note)",
     "evidence": "EDIT ME (the result or source behind it, with its uncertainty)",
     "verification": "EDIT ME (the cell, the clean-restart trace, the pre-listed check)",
     "boundary": "EDIT ME (how far it reaches, and no further)"},
]

# EDIT ME — your headline and the two dimensions, declared separately.
headline       = "EDIT ME (one sentence carrying its kind and its reach)"
question_kind  = "descriptive"     # "descriptive" or "causal" (the question's own words)
identification = "not applicable"  # causal only: "identified" or "currently unidentified"
reach          = "data at hand"    # "data at hand", "population", or "unseen cases"
kind_warrant   = "EDIT ME (what licenses the verb: assignment, or an argued identification)"
reach_warrant  = "EDIT ME (what licenses the reach: a sampling strategy, or a held-out check)"

VAGUE = {"verified", "checked", "yes", "done", "i checked it"}
print("— Part 1: claim-to-evidence table —")
for i, row in enumerate(claim_table, 1):
    blanks = [k for k, v in row.items() if "EDIT ME" in v or not v.strip()]
    if blanks:
        print(f"  row {i}: ⚠️ still a placeholder in {', '.join(blanks)}")
    elif row["verification"].strip().lower() in VAGUE:
        print(f"  row {i}: ⚠️ vague verification. Name the check precisely "
              "enough to open.")
    else:
        print(f"  row {i}: ✓ complete")
print("  The rule: a claim with no row does not appear in the note.")

print("\n— Part 2: kind and reach, warranted separately —")
CAUSAL_VERBS = ["causes", "drives", "improves", "boosts", "raises", "reduces",
                "leads to", "results in", "makes"]
WIDE_WORDS = ["in general", "everyone", "nationwide", "the population",
              "all people", "generally"]
low = headline.lower()
verb_hits = [v for v in CAUSAL_VERBS if v in low]
reach_hits = [w for w in WIDE_WORDS if w in low]

if "EDIT ME" in headline:
    print("  ⚠️ headline still a placeholder. Paste your real sentence and "
          "rerun.")
if question_kind == "causal" and identification == "currently unidentified":
    print("  kind: causal target, currently unidentified. Report the "
          "association your data produced,")
    print("        and keep the causal label on the question.")
    if verb_hits:
        print(f"  ⚠️ {verb_hits} in the headline. Your design has not earned "
              "the causal verb yet.")
elif question_kind == "descriptive" and verb_hits:
    print(f"  ⚠️ causal verb(s) {verb_hits} in a descriptive headline. Use "
          "'was associated with'.")
else:
    print("  kind: no classic verb upgrade found. Read it once more against "
          "your route.")
if reach != "population" and reach_hits:
    print(f"  ⚠️ {reach_hits} reaches past '{reach}'. Population reach is a "
          "second claim needing a second argument.")
for label, w in (("kind", kind_warrant), ("reach", reach_warrant)):
    print(f"  {label} warrant: " + ("⚠️ still a placeholder" if "EDIT ME" in w
                                    else "✓ named"))

print("\n— Part 3: uncertainty in the same eye-span —")
BOUND_CUES = ["to ", "±", "give or take", "interval", "plus or minus", "range"]
sentences = [s.strip() for s in headline.replace("!", ".").split(".") if s.strip()]
with_number = [s for s in sentences if any(ch.isdigit() for ch in s)]
if not with_number:
    print("  ⚠️ no number in the headline. A bounded claim carries its range.")
else:
    for s in with_number:
        ok = any(c in s.lower() for c in BOUND_CUES)
        print(f"  {'✓' if ok else '⚠️'} \"{s[:60]}...\": bound "
              f"{'rides with the number' if ok else 'is missing from this sentence'}")
```

```python
# Spot-check one row before you submit. Reopen the traced output and read the
# number again: a table you have not spot-checked is a list of hopes.
# The block below stands in for your own cell so the check runs today. Swap
# the stand-in for YOUR computation, keeping the compare at the bottom.

n_per   = 300                              # EDIT ME — your group sizes
arm_hi  = rng.normal(3.55, 0.90, n_per)    # EDIT ME — replace with your data
arm_lo  = rng.normal(3.18, 0.90, n_per)
recomputed = arm_hi.mean() - arm_lo.mean()
se = np.sqrt(arm_hi.var(ddof=1) / n_per + arm_lo.var(ddof=1) / n_per)
lo, hi = recomputed - 1.96 * se, recomputed + 1.96 * se

# Deliberately set to a value carried from an older draft, so the default run
# shows you the warning path. Replace it with the number YOUR note prints.
NOTE_SAYS = 0.37   # EDIT ME

print(f"cell recomputes: {recomputed:.2f}  (95% interval {lo:.2f} to {hi:.2f})")
print(f"the note prints: {NOTE_SAYS:.2f}")
matched = abs(NOTE_SAYS - recomputed) < 0.02
print("verdict:", "✓ matched, and the interval travels with it" if matched
      else "⚠️ does not match. Fix the note or rerun the cell before you ship.")
print("Record the row you spot-checked and this result in your verification "
      "column.")
```

### 🗡️ Red-Team Exchange

The attack runs at the Friday studio, where your note meets its first
strangers. Two rounds, and the second is the drill this milestone is built
around.

**Round 1: a peer reads the note the way a stranger would.**
They have only the page. They hunt the two silent upgrades: a verb stronger
than the design established, and a reach the design never bought. They also
read for the eye-span rule, since a number in one line with its interval three
lines later is, for most readers, a number with no interval. Your AI assistant
joins as a second hostile reader using the headline prompt above.

**Round 2: the trace-or-cut drill.**
Go sentence by sentence. Every claim gets matched to a row in your
claim-to-evidence table. A sentence with a row stays. A sentence with no row is
traced back to the output that supports it, or cut before you submit. There is
no third option, and there is no "I will add the row later." Finish the round
by spot-checking one row at random with the cell above.

**What to log.** The strongest hit from each round and what you changed. A
sentence you cut is a result of the drill, so write down which one and why.

> **A question that often comes up here:** *"A sentence I like has no row, but
> I am sure it is true. Can it stay if I soften it?"* No. Softening turns an
> unsupported claim into a vague unsupported claim, and it still travels with
> your name on it. Run the check that earns the row, or cut the sentence and
> keep the note honest.
