## 🎯 Definition of Done

You are done with this sprint when your submission carries all seven parts of
the audit, applied to **your own verified result**:

1. **The pre-listed checks.** Three robustness or placebo checks committed in
   writing *before* you look at any of their results, each named by what it
   varies (the sample, the measurement, the specification, or the metric) or
   named as a **negative test** or a leave-one-out influence check, and each
   tied to the reviewer attack it answers. You report all three, whatever
   they show.
2. **What survived.** Your headline claim after the checks ran, reworded to
   carry its **range** and its compass boundary. Two guard-rails by name: a
   same-sign specification spread is a direction plus a range, never an
   uncertainty interval; and a complete-case contrast compares whoever
   happened to stay measurable in each arm, so it is not automatically an
   effect for any shared group of stayers: name the question it actually
   answers.
3. **The verified AI-review trail.** The **Robustness & Sensitivity
   Reviewer** run (the required GenAI Studio reviewer for M8), every flag
   marked confirmed or refuted by a data check, and the single most confident
   **wrong** flag you caught, named for what it is.
4. **The remaining limitation and the adjudication.** The one weakness no
   check could fix, stated as expertise, then the close in three lines: what
   survived, what did not, and what you still cannot rule out.
5. **The audit notebook.** Seeded (`SEED = 464`), rerunning top to bottom,
   with your three checks as runnable cells and the negative test's
   prediction written in markdown *above* its cell, so the order of
   commitment is on the record.
6. **AI Research Ledger rows.** One row for every AI-assisted step, each with
   a named verification method against your own data.
7. **The version line and the dossier update line.** The artifact opens with
   **Book Milestone 8, version 1**, dated, with its reason; it ends with the
   dossier line naming where the robustness and diagnostic record now lives.

> **A question that often comes up here:** *"What happens if a check breaks
> my result?"* You are graded on the audit, not on survival. A finding that
> fails its own pre-listed checks sends you back to your route or your
> pipeline; that is the honest path, not softer wording. A broken result
> found now costs a revision. A broken result found by a reader later costs
> the claim.

```python
# Pre-listing scaffold + null-spread demo — edit the EDIT ME blocks.
# Part 1 puts your three checks on the record BEFORE you run anything.
# Part 2 shows how a negative test is judged: against the null reference
# spread, never against exact zero.

# EDIT ME — your three pre-listed checks. Fill these in BEFORE computing any
# check result; the audit reports all three, whatever they show.
pre_listed = [
    {"name": "EDIT ME (e.g., drop the largest hall)", "varies": "sample",
     "answers": "EDIT ME (which reviewer attack does this answer?)"},
    {"name": "EDIT ME (e.g., alternative outcome coding)", "varies": "measurement",
     "answers": "EDIT ME"},
    {"name": "EDIT ME (e.g., placebo outcome the design says is untouched)",
     "varies": "negative test", "answers": "EDIT ME"},
]

print("— Pre-listed checks (committed before looking) —")
for i, c in enumerate(pre_listed, 1):
    flag = "" if "EDIT ME" not in (c["name"] + c["answers"]) else "  ⚠️ still a placeholder"
    print(f"  {i}. [{c['varies']}] {c['name']} → answers: {c['answers']}{flag}")

# --- Part 2: the null reference spread, demonstrated -----------------------
# A placebo estimate is judged against the range the test produces when
# nothing is really there. We simulate that range by shuffling group labels:
# every shuffled contrast is a draw from "nothing is going on."
n = 120
group = np.repeat([0, 1], n // 2)
outcome = rng.normal(3.2, 0.9, size=n)          # no group effect by construction
null_draws = []
for _ in range(2000):
    shuffled = rng.permutation(group)
    null_draws.append(outcome[shuffled == 1].mean() - outcome[shuffled == 0].mean())
null_draws = np.array(null_draws)
lo, hi = np.percentile(null_draws, [2.5, 97.5])

placebo_estimate = 0.03   # EDIT ME — your negative test's actual estimate

fig, ax = plt.subplots(figsize=(9, 4.2))
ax.hist(null_draws, bins=40, color="#e8eef6", edgecolor="#33475b")
ax.axvline(placebo_estimate, color="#b3261e", linewidth=2,
           label=f"placebo estimate {placebo_estimate:.2f}")
ax.axvline(lo, color="#33475b", linestyle="dashed")
ax.axvline(hi, color="#33475b", linestyle="dashed",
           label=f"null reference spread {lo:.2f} to {hi:.2f}")
ax.set_title("Judge the negative test against the null spread, never exact zero",
             fontsize=11)
ax.set_xlabel("contrast when nothing is really there")
ax.legend()
plt.tight_layout()
plt.show()

verdict = "PASS (the null spread covers it)" if lo <= placebo_estimate <= hi \
          else "FAIL (a value the spread cannot explain)"
print(f"✓ Negative-test verdict at {placebo_estimate:.2f}: {verdict}")
```
