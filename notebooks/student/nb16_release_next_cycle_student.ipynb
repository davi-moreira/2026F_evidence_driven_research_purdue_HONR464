## 1. Wire the Review Team You Can Defend

**Guiding question:** *how do you split a closing review into jobs you can
grade, and put them in an order that stops an early mistake from flowing
downstream?*

> *"Bring me a draft, not a pile of AI output. I want to know which part each
> tool touched, what you told it to do, and how you checked it. If you cannot
> tell me that, you did not write it."*
> — your **thesis advisor**, reading the draft you are about to defend

First, be exact about what you are running. Your review team is a set of
**reviewer roles you open, task, and sequence yourself**. You choose the order,
you carry each output to the next role, and you decide at every hand-off.
Nothing in this course plans or acts on its own, and a chat you sequenced by
hand is never described as autonomous. That sentence has to survive a
skeptical question, so make it true first and say it second.

Four terms turn a pile of roles into a workflow you can supervise.

- **Scoped role**: a job narrow enough to state in one sentence and check in
  one pass. Example: "flag every sentence in my results that claims more than
  my sample supports." A scoped role has a finish line you can see from
  outside. "Make it better" does not.
- **Worker-critic pair**: one role produces something, and a separate role
  attacks what it produced. Example: a worker drafts your limitations
  paragraph, then a critic hunts for the limitation the worker left out. A
  worker alone is confident about its own quality, which is the one blind spot
  it cannot see.
- **Dependency**: a required order between two jobs, where one needs the
  other's output before it can start. Example: a critic cannot attack a draft
  that does not exist yet.
- **Critical path**: the longest chain of must-follow steps in your workflow.
  Example: worker drafts, critic attacks, you integrate is three steps that
  cannot collapse into fewer. That chain sets the minimum number of rounds
  your review will take.

The book works this on a small health study: a post-lunch-walk write-up sent
to a clarity reviewer, a methodologist, and a citation-checker at the same
time, then integrated by the author, then tightened by an editor. Its critical
path is three arrows deep and runs three loops at its widest, and the
integrate node is the one node with no role attached to it. The cell below
builds that same shape as a **directed graph**, a picture of arrows in which
each arrow means "this feeds that."

### 🛠️ Run It Live: your review team as a graph

**🔴 Live in class: we run this one together.**

Run the cell below. It builds the parallel wiring, prints the node count, the
arrow count, the critical-path depth, and the widest layer, then rebuilds the
same team with the skeptic waiting on the reviewer. Read the numbers off the
printout before the next markdown cell, and check them against what you just

```python
# Build the closing-review team as a directed graph.
# Nodes: five AI roles (worker, three critics, editor) plus YOU, the integrator.
# (networkx is imported once in the setup cell.)
G = nx.DiGraph()
parallel_edges = [
    ("Worker\ndraft", "Reviewer"),
    ("Worker\ndraft", "Methodologist"),
    ("Worker\ndraft", "Skeptic"),
    ("Reviewer", "You\nintegrate"),
    ("Methodologist", "You\nintegrate"),
    ("Skeptic", "You\nintegrate"),
    ("You\nintegrate", "Editor"),
]
G.add_edges_from(parallel_edges)

generations = list(nx.topological_generations(G))
depth = nx.dag_longest_path_length(G)             # longest dependency chain, in arrows
width = max(len(layer) for layer in generations)  # most roles that can run at once

print("Closing-review team (three critics in PARALLEL):")
print(f"  roles + you (nodes)     : {G.number_of_nodes()}")
print(f"  dependencies (arrows)   : {G.number_of_edges()}")
print(f"  critical-path depth     : {depth} arrows  ({len(generations)} rounds)")
print(f"  maximum parallel width  : {width} roles at once")
print(f"  the critical path       : {' -> '.join(p.replace(chr(10), ' ') for p in nx.dag_longest_path(G))}")

# The SEQUENTIAL variant: the skeptic now waits for the reviewer's notes.
H = nx.DiGraph()
H.add_edges_from([
    ("Worker\ndraft", "Reviewer"),
    ("Worker\ndraft", "Methodologist"),
    ("Reviewer", "Skeptic"),               # skeptic now depends on the reviewer
    ("Skeptic", "You\nintegrate"),
    ("Methodologist", "You\nintegrate"),
    ("You\nintegrate", "Editor"),
])
gens_seq = list(nx.topological_generations(H))
print("\nIf the skeptic instead waits for the reviewer (one critic made SEQUENTIAL):")
print(f"  critical-path depth     : {nx.dag_longest_path_length(H)} arrows  ({len(gens_seq)} rounds)")
print(f"  maximum parallel width  : {max(len(l) for l in gens_seq)} roles at once")

# Draw the parallel workflow.
pos = {
    "Worker\ndraft":    (0.0, 1.0),
    "Reviewer":          (1.6, 1.9),
    "Methodologist":     (1.6, 1.0),
    "Skeptic":           (1.6, 0.1),
    "You\nintegrate":   (3.2, 1.0),
    "Editor":            (4.8, 1.0),
}
colors = {
    "Worker\ndraft": "#4C72B0", "Reviewer": "#8AA9D6", "Methodologist": "#8AA9D6",
    "Skeptic": "#8AA9D6", "You\nintegrate": "#55A868", "Editor": "#4C72B0",
}
fig, ax = plt.subplots(figsize=(10, 4.5))
nx.draw_networkx_nodes(G, pos, node_color=[colors[n] for n in G.nodes()],
                       edgecolors="#333333", linewidths=1.5, node_size=4600, ax=ax)
nx.draw_networkx_labels(G, pos, font_size=8.5, ax=ax)
nx.draw_networkx_edges(G, pos, arrowstyle="-|>", arrowsize=16, edge_color="#777777",
                       width=1.5, node_size=4600, ax=ax)
ax.set_title("A closing-review team: three critics in parallel, then you integrate")
ax.axis("off")
plt.tight_layout()
plt.show()

print("\n✓ Team built. The green node is YOU: the integrator no role can replace.")
```

```python
# Self-check: the parallel team has depth 3 and width 3; sequencing a critic costs a round.
assert G.number_of_nodes() == 6, "expected 5 AI roles + you"
assert G.number_of_edges() == 7, "expected 7 dependency arrows"
assert depth == 3, "parallel critical-path depth should be 3 arrows"
assert width == 3, "three critics should run in parallel (width 3)"
assert nx.dag_longest_path_length(H) == 4, "sequencing the skeptic should deepen the path to 4"
print(f"✓ Self-check passed: parallel depth {depth} arrows (width {width}); the sequential "
      f"variant deepens to {nx.dag_longest_path_length(H)}.\n"
      f"  Parallel buys speed and three lenses at once; sequencing buys the skeptic the "
      f"reviewer's context, at the cost of one more round.")
```

## 2. What Is Their Agreement Worth?

**Guiding question:** *when four roles agree, how many checks do you actually

### 🛠️ Run It Live: independent versus correlated reviewers

**🔴 Live in class: we run this one together.**

Run the cell below. It simulates 5,000 chapters, each carrying one hidden
flaw, reviewed twice: once by a team of four whose errors are independent, and
once by a team of four that shares a blind spot. It prints how often each team
misses a flaw unanimously and plots the two rates side by side. Read both
numbers before the next markdown cell.

```python
# Two four-reviewer teams on 5,000 flawed chapters. Fresh generator so this
# cell is reproducible on its own, whatever ran before it (seed 464).
rng_sim = np.random.default_rng(SEED)

n_chapters = 5000
k = 4        # reviewers per team
detect = 0.5 # each reviewer catches the hidden flaw about half the time, on its own
rho = 0.40   # the correlated team shares a blind spot on ~40% of chapters

# Independent team: each reviewer catches the flaw on its own. A MISS = all four fail.
caught_indep = rng_sim.random((n_chapters, k)) < detect
miss_indep = ~caught_indep.any(axis=1)
rate_indep = miss_indep.mean()

# Correlated team: on a shared-blind-spot chapter EVERY reviewer misses together;
# otherwise each catches on its own, exactly like the independent team.
blind = rng_sim.random(n_chapters) < rho
caught_corr = (rng_sim.random((n_chapters, k)) < detect) & (~blind[:, None])
miss_corr = ~caught_corr.any(axis=1)
rate_corr = miss_corr.mean()

print("A team MISS = all four reviewers say 'fine' while a real flaw is present.")
print(f"  Independent team miss rate : {rate_indep:.1%}   ({miss_indep.sum()} of {n_chapters})")
print(f"  Correlated team miss rate  : {rate_corr:.1%}   ({miss_corr.sum()} of {n_chapters})")
print(f"  The correlated team is unanimous-and-wrong about {rate_corr/rate_indep:.1f}x as often.")

fig, ax = plt.subplots(figsize=(7, 5))
# Independent gets the calm blue; correlated gets the warning orange.
ax.bar(["Independent\nreviewers", "Correlated\nreviewers"],
       [rate_indep, rate_corr], color=["#4C72B0", "#DD8452"], edgecolor="white")
ax.set_ylim(0, 0.55)
ax.set_ylabel("Share of flawed chapters the team MISSES")
ax.set_title("Same four reviewers: a shared blind spot manufactures correlated error")
for i, r in enumerate([rate_indep, rate_corr]):
    ax.text(i, r + 0.012, f"{r:.1%}", ha="center", fontsize=11)
plt.tight_layout()
plt.show()
```

```python
# Self-check: the correlated team misses real flaws several times more often.
assert 0.04 < rate_indep < 0.09, "independent miss rate drifted — is the seed 464?"
assert 0.40 < rate_corr < 0.47, "correlated miss rate drifted — is the seed 464?"
assert rate_corr > 5 * rate_indep, "the shared blind spot should dominate"
print(f"✓ Self-check passed: independent misses {rate_indep:.1%} of flawed chapters, "
      f"correlated misses {rate_corr:.1%}.\n  Four reviewers either way, about "
      f"{rate_corr/rate_indep:.0f}x the false-confidence rate. Agreement is worth only "
      f"the independence behind it.")
```

*Reading the output.* The independent team is unanimous-and-wrong on about
**6%** of flawed chapters. The correlated team, the same four reviewers, is
unanimous-and-wrong on about **43%**, roughly seven times as often. Nothing
changed except whether the reviewers could miss the flaw *together*. When they
share a blind spot, four "looks fine" verdicts are not four confirmations.
They are close to one.

> **A question that often comes up here:** *"Conflict feels like the problem.
> Is agreement not what I want?"* It is the reverse. Conflict announces itself
> and tells you where to look. Shared blind spots feel like confirmation while
> quietly agreeing on something false. When roles agree on something that
> matters, your job is to ask whether they were ever independent before you
> trust it.

Close every AI tool for this section. Here is a real draft sentence that a

**Move one: recompute the number.** Before you run the next cell, commit a

**The formalization.** Three diagnoses, and they call for opposite responses.

- **Real disagreement**: the roles read the evidence differently. Example: your
  reviewer calls a sentence clear and strong while your methodologist flags the
  same sentence as reaching past your data. This is the useful case, and it
  points straight at the claim to check.
- **Correlated error**: the roles agree, but on the same wrong assumption, so
  the agreement proves nothing. Example: four roles built in one tool from one
  briefing all miss the same overreaching verb. Their consensus is one signal,
  not four.
- **False consensus**: the roles agree only because your prompt framed the task
  so they had to. Example: you ask "confirm my poster is ready" and every role
  obliges. You led all the witnesses.

Three more terms carry the decision. An **independence check** is the test that
makes agreement worth more than a head count, and you just ran two of them.
**Escalation to human judgment** is the moment an AI output would *settle*
rather than *inform* a decision that is yours, so you stop and decide it
yourself. Example: a role proposes calling your result causal, which would
settle your claim boundary, so it escalates to you. A **human override** is
what you do there: you take the pen back, in your own words, backed by your own
check, even when every role agreed on something else. The five triggers that
force an escalation are listed in
[`ai_resources/escalation_protocol.md`](https://github.com/davi-moreira/2026F_evidence_driven_research_purdue_HONR464/blob/main/ai_resources/escalation_protocol.md),
and any one is enough.

Close the block on the rule that follows from the simulation: a decision that
counted the roles is not a decision. It is a vote, and votes among look-alike
models are meaningless.

> **A question that often comes up here:** *"If I keep overriding the tools,
> why use them at all?"* Because they are genuinely good at the parts you can
> delegate: drafting, surfacing leads, listing threats, tightening prose.
> Escalation is not distrust of everything. It keeps a human at the few
> decisions that carry your name. A semester-long ledger with zero escalations
> is a warning sign rather than a badge. It usually means AI was never used on
> anything that mattered, or a boundary was crossed and nobody noticed.

```python
# Sweep the team size. Fresh generator so the sweep is reproducible on its own (seed 464).
rng_sweep = np.random.default_rng(SEED)

team_sizes = [2, 4, 8, 16]
indep_rates, corr_rates = [], []
for kk in team_sizes:
    caught_i = rng_sweep.random((n_chapters, kk)) < detect
    indep_rates.append((~caught_i.any(axis=1)).mean())
    bl = rng_sweep.random(n_chapters) < rho
    caught_c = (rng_sweep.random((n_chapters, kk)) < detect) & (~bl[:, None])
    corr_rates.append((~caught_c.any(axis=1)).mean())

print("Miss rate as the team grows:")
print(f"  {'reviewers':>10} | {'independent':>12} | {'correlated':>11}")
for kk, ri, rc in zip(team_sizes, indep_rates, corr_rates):
    print(f"  {kk:>10} | {ri:>11.1%} | {rc:>10.1%}")
print(f"\n  The correlated floor sits near the shared-blind-spot rate ({rho:.0%}): "
      f"adding roles cannot push below it.")

fig, ax = plt.subplots()
ax.plot(team_sizes, indep_rates, marker="o", color="#4C72B0", label="independent reviewers")
ax.plot(team_sizes, corr_rates, marker="s", color="#DD8452", label="correlated reviewers")
ax.axhline(rho, color="#8C8C8C", linestyle=":", linewidth=1.3, label=f"shared blind-spot rate ({rho:.0%})")
ax.set_xlabel("Number of reviewers on the team")
ax.set_ylabel("Share of flawed chapters missed")
ax.set_title("More reviewers help only if their errors are independent")
ax.legend()
plt.tight_layout()
plt.show()
```

## 4. The Stale-Run Rule and the Seven-Row Release Audit

**Guiding question:** *when does a passed cold run stop being evidence about the
package you are about to release?*

> *"I download your package a year from now and press run. I was not in the
> room, I do not know what you meant, and I cannot ask. The only version of your
> work that exists for me is the one in the folder."*
> — a **replication reviewer**, the stranger your release is really for

Your package earned its credibility last week the hard way, through a **cold
run**: a run of the packaged files from a fresh start, on a machine that is not
yours, with no author in the room. This week adds the rule that protects that
credibility until release.

A **stale run** is a cold run that predates a change to what it verified. The
test is one question. Since the run finished, did anything change that a
stranger would download, run, or follow? A number, a claim, the data, the code,
the environment record, or an instruction in the README: any one of those makes
the run stale. Example: your run passed on day 0, you corrected one mistyped
data value on day 4, and the day-0 run is now a certificate for a package that
no longer exists.

The rule deliberately ignores the author's label for the change. "Just a
wording fix" is a judgment about your own work, made by the one person who
cannot see it from outside, at the exact moment you most want to be done. The
boundary does the judging instead: package changed, run stale. A stale run has
one honest remedy, which is to **rerun from the packaged files**, record the
date, and let the new run testify.

**🔴 Live in class: we run this one together.**

Run the cell below. It applies one question to each change: did it land after
the cold run *and* touch something a stranger would download, run, or follow?
Read the six verdicts and the release decision off the printout before the next

```python
# The stale-run test: one package, one cold run, six changes on record.
# Day 0 is the day last week's cold run finished clean.
cold_run_day = 0

changes = pd.DataFrame([
    {"day": 2, "where": "research note (packaged)",
     "change": "headline sentence reworded ('just a wording fix')", "in_package": True},
    {"day": 3, "where": "defense slide deck (not packaged)",
     "change": "closing figure added to the slides", "in_package": False},
    {"day": 4, "where": "data file (packaged)",
     "change": "one mistyped value corrected", "in_package": True},
    {"day": 5, "where": "analysis notebook (packaged)",
     "change": "seed switched during exploration and saved", "in_package": True},
    {"day": 6, "where": "README (packaged)",
     "change": "install instructions edited for clarity", "in_package": True},
    {"day": 6, "where": "AI Research Ledger (dossier, not packaged)",
     "change": "new row logging yesterday's verification", "in_package": False},
])

# The rule: after the run + inside the package = the run is stale.
changes["makes_run_stale"] = (changes["day"] > cold_run_day) & changes["in_package"]

print("Cold run finished clean on day 0. Changes on record since:")
for _, ev in changes.iterrows():
    verdict = "STALE-MAKING" if ev["makes_run_stale"] else "run stays fresh"
    print(f"  day {ev['day']}: {ev['where']:<42} -> {verdict}")

n_stale = int(changes["makes_run_stale"].sum())
print(f"\nStale-making changes: {n_stale} of {len(changes)}")
print("Release decision:", "RERUN REQUIRED before release."
      if n_stale else "The cold run still testifies.")

# Timeline: green line = the cold run; orange = package-touching change (stale-making).
fig, ax = plt.subplots(figsize=(10, 4))
event_colors = ["#DD8452" if s else "#4C72B0" for s in changes["makes_run_stale"]]
ax.scatter(changes["day"], range(len(changes)), c=event_colors, s=150, zorder=3,
           edgecolors="#333333", linewidths=1.0)
for i, (_, ev) in enumerate(changes.iterrows()):
    ax.text(ev["day"] + 0.18, i, ev["where"], va="center", fontsize=9)
ax.axvline(cold_run_day, color="#55A868", linewidth=2.5)
ax.text(cold_run_day + 0.08, len(changes) - 0.35, "cold run\nfinished clean",
        color="#55A868", fontsize=9, va="top")
ax.set_yticks([])
ax.set_xlim(-1.0, 10.5)
ax.set_xlabel("Days since the cold run")
ax.set_title("Every orange change touches the package — each one alone makes the run stale")
plt.tight_layout()
plt.show()
```

```python
# Self-check: four of the six changes touch the package, so the run is stale.
assert n_stale == 4, "expected 4 package-touching changes after the cold run"
assert bool(changes.loc[0, "makes_run_stale"]), "the 'wording fix' edits the packaged note — stale-making"
assert not bool(changes.loc[1, "makes_run_stale"]), "slides are not in the package"
assert not bool(changes.loc[5, "makes_run_stale"]), "a ledger row documents; a stranger reruns the package, not the ledger"
print(f"✓ Self-check passed: {n_stale} of 6 changes make the run stale, the 'wording fix' included.\n"
      "  The rule reads the package boundary, never the author's label for the change.")
```

**three-movement defense** takes about six minutes, two per movement, followed
by cross-examination.

- **Movement 1, the claim.** Your headline claim and the exact edge it stops
  at. Example: "in my sample I find X; I do not claim it holds for the
  population."
- **Movement 2, the choice.** The central choice you made and the road you did
  not take.
- **Movement 3, the verification.** How you checked, what survived, which
  sources you confirmed yourself. Verification leads; it is not an afterthought.

Two judgments decide how the cross-examination goes. **Defend or concede**: you
defend when the objection asks for something your evidence already licenses, and
you point at the row that licenses it. You concede when the objection asks for a
crossing your design never paid for, and then you name the smaller true claim on
the spot. Conceding the right point on time is scored as strength. And the
**forbidden moves** are three ways to lose the room: inventing a new claim under
pressure, un-conceding by insistence rather than evidence, and blaming the tool.
"The AI gave me that" is never a defense.

Every non-defender arrives with at least one ledger-grounded question, and the
defender either answers from a real ledger row or concedes and names the smaller
true claim.

> **A question that often comes up here:** *"What if I get a question I cannot
> answer?"* Answer from your ledger if the row exists. If it does not, concede
> and name the claim you can defend, plus how one would find out. A defense is
> not a performance of certainty. It is an honest account of what your evidence
> holds and where it stops.

After the rehearsal, your research lead states the room's finding and its
uncertainty in two sentences, and your instructor confirms the seeded flaw,
reads the corrected audit row aloud, and confirms the two calibration items.
Records get corrected there, before anything wrong enters a ledger.

A research cycle does not trail off. It ends with a hand-forward. The
**next-study agenda** is the last page of your dossier: the question this work
opened, tied to a limitation you could not resolve. Its one discipline is the
verb tense of honesty. The agenda is written as **directions, never findings**.

"The next study should randomize the reminder to test whether it causes the gap
to close" is a direction. "The gap likely persists" is a finding you never
showed, smuggled past your own audit in the future tense.

Stopping becomes respectable here. You are not abandoning the open question. You
are naming it, pointing the next cycle at it, and signing the rule that says why
you stop here.
