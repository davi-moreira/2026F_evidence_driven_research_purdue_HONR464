## 2. Your working rules and your delegation map

Your **working rules** are three lists with real entries drawn from the work you
can already see coming: **delegate freely** (work you could check trivially, or
would not sign for anyway), **delegate then verify** (work a tool does faster but
that you confirm before it enters your artifact), and **never delegate** (the
decisions from your responsibility statement).

Your **delegation map** applies those rules to the specific work ahead, including
this milestone's own declaration. Mark that work explicitly: candidate wordings
may travel to a tool, while the lead question, its kind, its reach, and its
boundary may not. A map that never mentions this week's actual work is decoration.

> **A question that often comes up here:** *"Does 'delegate freely' mean
> unchecked?"* No. It means you do not owe a separate verification step for that
> item. The moment a freely delegated task produces a fact, a number, or a citation
> that enters your artifact, it moves to the middle list and gets verified like
> everything else there.

```python
# Working rules and delegation map — edit the three lists and the task table.
# RULES is the single source of truth: the map below is checked against it, so a
# task you file under a rule that does not exist is caught here, not in grading.

RULES = {
    "delegate freely": [
        "EDIT ME: work you could check trivially, or would not sign for anyway",
    ],
    "delegate then verify": [
        "EDIT ME: work a tool does faster but you confirm before it enters the artifact",
    ],
    "never delegate": [
        "EDIT ME: a decision from your responsibility statement",
    ],
}

# The work ahead, each task filed under one rule. Include THIS milestone's own
# declaration work: candidate wordings travel, the lead question does not.
DELEGATION_MAP = [
    {"task": "Generate candidate wordings of my lead question", "rule": "delegate freely"},
    {"task": "Choose the lead question, its kind, its reach, and its boundary",
     "rule": "never delegate"},
    {"task": "EDIT ME: another task you can already see coming this term",
     "rule": "delegate then verify"},
]

print("Your working rules")
for rule, entries in RULES.items():
    print(f"  {rule.upper()}")
    for e in entries:
        print(f"     - {e}")
print()

print("Your delegation map")
unknown = []
for row in DELEGATION_MAP:
    ok = row["rule"] in RULES
    unknown += [] if ok else [row["rule"]]
    print(f"  [{row['rule']:<21}] {row['task']}")
print()

covers_declaration = any("question" in r["task"].lower() for r in DELEGATION_MAP)
unwritten = [e for entries in RULES.values() for e in entries if "edit me" in e.lower()]

if unknown:
    print(f"⚠️ Filed under a rule that is not in RULES: {', '.join(sorted(set(unknown)))}")
if unwritten:
    print(f"⚠️ {len(unwritten)} rule entry(ies) still unwritten. Real entries only.")
if covers_declaration and not unknown and not unwritten:
    print("✓ Rules written and the map covers this week's own declaration work.")
else:
    print("Finish the lists, then re-run. The map has to name the declaration work.")
```

## 6. Kind and reach, fixed by your question's own words

Place your question on the **inquiry compass**, the course's two-axis
classification of what a question asks. **Kind** is descriptive or causal: does
answering require a **counterfactual contrast**, a comparison against a world that
did not happen, or only the world as it is or was? **Reach** is the units the
answer must hold for: the data at hand, a population beyond your data, or cases
not yet seen.

Quote the words in your own sentence that fix each axis. "How much more" fixes
descriptive; "because of" or "would have been" fixes causal; "in these records"
fixes the data at hand; "undergraduates nationally" fixes a population; "next
term's cohort" fixes unseen cases. If nothing in your sentence fixes the reach,
the sentence is not finished yet.

**The kind rule, and it is load-bearing here.** Your question's kind is fixed by
its words, never by the data you happen to be able to get. A causal question your
data cannot identify stays **causal, currently unidentified**. Write that status
down and carry it: it is honest, it is gradable, and M5 exists precisely to give
such a question a design. Renaming it to fit weak evidence is the failure this
milestone is built to catch, and it caps the first rubric row at Beginning.

If your question is causal, record its **comparison world** as well: what is being
compared to what, in one sentence.

> **A question that often comes up here:** *"My question is causal and I have no
> experiment. Should I rewrite it?"* Changing your question is a human decision you
> may make, openly, as a recorded revision. Relabelling it quietly is a different
> act: it hides the gap between what you want to know and what your evidence can
> support. Declare it causal and currently unidentified, then report the
> association you can see plus the warrant you are missing.

## 7. The field card and the stranger test

Four fields, one line each: **objective** (what the answer is for), **units** (the
who or what each row of your evidence stands for), **outcome** (the observable
thing you will actually record), and **conditions** (the comparison or the window
your question sets).

Then run the **stranger test**. Hand the question sentence to someone who knows
nothing about your project and ask them three things: what one unit is, what you
would record, and who the answer is about. Write down where they stalled. That is
your next revision, and reporting it is part of the grade.

The cell below assembles the declaration and the field card, and checks what a
reader checks first: that the words you quote as fixing your kind and your reach
really do appear in your question sentence.

```python
# Declaration card — edit every value, then run. The checks below read only
# from these two dicts, so a rewrite of your question re-runs every check.

DECLARATION = {
    "lead_question": "EDIT ME: the single sentence your evidence will answer",
    "kind": "descriptive",              # "descriptive" or "causal" — fixed by the WORDS
    "kind_words": "EDIT ME: the words in your sentence that fix the kind",
    "reach": "data at hand",            # "data at hand", "population", or "unseen cases"
    "reach_words": "EDIT ME: the words in your sentence that fix the reach",
    # causal questions only: "identified" if a design warrants the contrast,
    # "currently unidentified" if it does not yet. Descriptive: leave as "n/a".
    "identification": "n/a",
    "comparison_world": "EDIT ME (causal only): what is compared to what",
}

FIELD_CARD = {
    "objective":  "EDIT ME: what the answer is for, in one sentence",
    "units":      "EDIT ME: the who or what each row of your evidence stands for",
    "outcome":    "EDIT ME: the observable, measurable thing you will record",
    "conditions": "EDIT ME: the comparison or the window your question sets",
}

KINDS = {"descriptive", "causal"}
REACHES = {"data at hand", "population", "unseen cases"}

q = DECLARATION["lead_question"]
print("Lead question:")
print(f"   {q}\n")
print(f"Kind : {DECLARATION['kind']}   fixed by: \"{DECLARATION['kind_words']}\"")
print(f"Reach: {DECLARATION['reach']}   fixed by: \"{DECLARATION['reach_words']}\"")
if DECLARATION["kind"] == "causal":
    print(f"Identification: {DECLARATION['identification']}")
    print(f"Comparison world: {DECLARATION['comparison_world']}")
print()
print("Field card")
for field, line in FIELD_CARD.items():
    print(f"   {field:<11}: {line}")
print()

flags = []
if DECLARATION["kind"] not in KINDS:
    flags.append(f"kind must be one of {sorted(KINDS)}")
if DECLARATION["reach"] not in REACHES:
    flags.append(f"reach must be one of {sorted(REACHES)}")
for axis in ("kind", "reach"):
    words = DECLARATION[f"{axis}_words"].lower().strip()
    if words and words not in q.lower():
        flags.append(f"the {axis} words you quoted do not appear in your question sentence")
if DECLARATION["kind"] == "causal" and DECLARATION["identification"] not in {
        "identified", "currently unidentified"}:
    flags.append("a causal question needs 'identified' or 'currently unidentified'")
if "edit me" in " ".join(list(DECLARATION.values()) + list(FIELD_CARD.values())).lower():
    flags.append("some fields are still placeholders")

if flags:
    print("⚠️ Not declarable yet:")
    for f in flags:
        print(f"   - {f}")
else:
    print("✓ Kind and reach are both fixed by words that really are in your sentence,")
    print("  and the field card is complete. Now run the stranger test on the sentence.")
```
