## 1. Same Numbers, Two Pictures

*(minutes 9–13)*

**Guiding question:** *can a picture built from honest numbers still make a
claim the numbers never made?*

> *"From four feet away I read one thing: does your tallest bar mean what it
> looks like it means? If your axis starts halfway up the scale, I have caught
> you before I read a word of your methods, and now I distrust the whole
> board."*
> — a **skeptical peer** walking a poster session

Your claim is already written and already bounded. This week it moves into a
format a stranger meets in seconds, and the move is where claims quietly grow.
Four terms name how.

- **Figure honesty** is the rule that a picture cannot say more than its data.
  Example: if the real gap is half a point on a 1-to-7 scale, the bars should
  look about half a point apart, not twice as tall.
- **Truncated axis** is an axis that starts above the bottom of its scale
  instead of at the floor of the measure. Example: an axis running 4.26 to 4.9
  on a 1-to-7 question turns a small difference into a cliff.
- **Uncertainty on the page** is the interval or caveat printed in the same
  glance as the claim, never in a footnote. Example: the two interval bars drawn
  right on top of the two means.
- **Redundant encoding** means carrying the same distinction in more than one
  channel, so no single channel is load-bearing. Example: label each bar with
  its value and give each a pattern, so color becomes a bonus rather than the
  only signal.

> **A question that often comes up here:** *"Accessibility sounds like a
> compliance checkbox. Is it really an evidence issue?"* It is the same issue as
> every other audit on this list. Your claim is only as strong as the readers it
> reaches, and a figure that encodes its result in color alone makes a claim a
> color-blind reviewer literally cannot receive. Roughly one man in twelve is
> red-green color-blind. Reaching every reader is not politeness. It is the
> claim doing its job.

### 🛠️ Run It Live: one note sentence, drawn two ways

Run the cell. It loads the teaching survey, splits respondents by how responsive
they rated the government, draws a seeded subsample of 150 from each side so the
intervals are the size a real project produces, and prints the two means with
their 95% intervals, the gap, and the exaggeration a floored axis creates. Then
it draws both panels: the described one on the left, and on the right the same
numbers on the full 1-to-7 scale with value labels, intervals, a color-blind
distinguishable pair, and a hatch so color is never the only channel.

One caveat travels with every number here. `lapop_brazil` is a 10,000-row
resample with replacement of real survey interviews, realistic in its patterns
and manufactured in its precision, so it serves practice and never a claim about
Brazil. That caveat is itself a limitation your artifact would have to carry.

Expect two means about half a point apart, two intervals that overlap, and an
exaggeration factor far above 3.

```python
# One bounded sentence, drawn two ways, on a file anyone can reopen.
lapop = load_course_data("lapop_brazil.csv")
assert lapop.shape == (10000, 10), "unexpected shape — flag this!"
print("✓ Loaded lapop_brazil.csv —", lapop.shape[0], "rows,", lapop.shape[1], "columns\n")

# Two groups, split on how responsive the respondent rated the government.
low = lapop.loc[lapop.govt_responsive <= 3, "trust_police"].to_numpy()
high = lapop.loc[lapop.govt_responsive >= 4, "trust_police"].to_numpy()
print(f"Group A, rated government 1-3 : {low.size:,} rows in the file")
print(f"Group B, rated government 4-7 : {high.size:,} rows in the file")

# A seeded subsample of 150 per group: project-sized, so the intervals are real.
a = low[rng.choice(low.size, 150, replace=False)]
b = high[rng.choice(high.size, 150, replace=False)]

mean_a, mean_b = a.mean(), b.mean()
ci_a = 1.96 * a.std(ddof=1) / np.sqrt(len(a))     # 95% interval half-widths
ci_b = 1.96 * b.std(ddof=1) / np.sqrt(len(b))
gap = mean_b - mean_a
overlap = (mean_a + ci_a) > (mean_b - ci_b)
share_hi = (b >= 5).mean()                        # one extra number, for later

# ONE source of truth for the truncation: a floor just under the shorter bar.
floor = min(mean_a, mean_b) - 0.05
exaggeration = (mean_b - floor) / (mean_a - floor)

print(f"\nGroup A mean = {mean_a:.2f} ± {ci_a:.2f}   "
      f"Group B mean = {mean_b:.2f} ± {ci_b:.2f}")
print(f"gap = {gap:.2f} on a 1-to-7 scale        intervals overlap: {overlap}")
print(f"floored at {floor:.2f}, the taller bar is drawn {exaggeration:.1f}x the shorter one")
print(f"share of group B reporting trust of 5 or higher: {share_hi:.0%}")

fig, axes = plt.subplots(1, 2, figsize=(12, 4.8))

# LEFT: the panel as described. Bring the audits to it yourself; these comments
# say only what the code does, never which audit each choice trips.
axL = axes[0]
axL.bar(["Low responsiveness", "High responsiveness"], [mean_a, mean_b],
        color=["#D62728", "#2CA02C"])              # told apart by red vs green only
axL.set_ylim(floor, mean_b + 0.05)                 # axis floored just under the shorter bar
axL.set_title("Responsive government BUILDS police trust")
axL.set_ylabel("Trust in the police")
# no interval drawn, no value labels, no full scale shown.

# RIGHT: as the note reads, and reachable by every reader.
axR = axes[1]
bars = axR.bar(["Low responsiveness", "High responsiveness"], [mean_a, mean_b],
               yerr=[ci_a, ci_b], capsize=6,
               color=["#4C72B0", "#DD8452"],        # color-blind distinguishable pair
               hatch=["", "//"], edgecolor="white")  # redundant encoding: pattern backs color
axR.set_ylim(1, 7)                                   # the honest full scale of the measure
for bar, m in zip(bars, [mean_a, mean_b]):
    axR.annotate(f"{m:.2f}", (bar.get_x() + bar.get_width() / 2, m + 0.18),
                 ha="center", fontsize=11)           # a value a screen reader can be given
axR.set_ylabel("Mean trust in the police (1 = none, 7 = a lot)")
axR.set_title("Trust in the police by rated responsiveness, with 95% intervals")

plt.tight_layout()
plt.show()

# Self-check: both the distortion and the honest reading are really present.
assert mean_b > mean_a, "group B should read a little higher in this subsample"
assert exaggeration > 3, "the truncated axis should visibly inflate the gap"
assert overlap, "the two intervals overlap at this subsample size"
assert axes[0].get_ylim()[0] > 1, "the left axis is truncated by design"
assert axes[1].get_ylim() == (1.0, 7.0), "the right axis shows the honest full scale"
print("\n✓ Same numbers, drawn two ways: the left panel changes the finding, "
      "the right one repeats the sentence.")
```

**Reading the output.** The file splits 5,481 rows into group A and 4,519 into
group B, and the seeded subsamples land at 4.31 ± 0.33 against 4.83 ± 0.32. The
gap is 0.51 on a 1-to-7 scale, and the two intervals overlap. Floored at 4.26,
the taller bar is drawn 11.3 times the shorter one. Nothing was rounded, hidden,
or invented to produce that 11.3: it is what an axis choice does to a real half
point.

Read the overlap carefully, because the room usually flips too far. Overlapping
intervals do not make the gap fake. They say a small, real difference in this
file is being drawn as a chasm, so the honest headline has to carry both the
half point and the overlap. And the file is a teaching resample, so no claim
about any population travels either way.

*In plain terms, the picture says the left panel lies with true numbers and the
right panel repeats the sentence with the same ones.*

## 2. The Five Audits, Run on Someone Else's Artifact

*(minutes 13–41)*

**Guiding question:** *what does a structured audit catch that a careful glance
never will?*

A **gallery-walk audit** is a structured pass over an artifact where you run a
fixed checklist instead of glancing and admiring. It turns reactions into
located, fixable findings. Five questions form it.

- **Claim boundary.** Does the headline stay inside what the evidence licenses?
  A **claim boundary** is the line between what your evidence supports and what
  it does not, and it lives in the verb. Example: an observational comparison
  may say "goes with", never "builds".
- **Figure honesty.** Does the picture's impression match the number?
- **Read path.** Can a stranger find the headline claim in about ninety seconds?
  The **read path** is the order a reader's eye actually travels. Example: the
  eye should reach the claim before the method box, not after.
- **Uncertainty.** Is the interval or caveat in the same eye-span as the claim,
  or buried?
- **Accessibility.** Can a color-blind reader and a screen-reader user receive
  the claim at all?

A **hit** is a specific, located finding with a fix. Example: "the axis starts
at 4.26, so a half point is drawn as 11.3 times; run the axis 1 to 7." An audit
with no hits did not spare the artifact. It failed it.

**Today's artifact.** Your colleague's Studio-10 package is open in front of
you: a short venue contract, a draft board, a claim-evidence trace, and one
accessibility line. The file is a simulated case issued by your instructor,

4. **Required elements.** What the venue insists appears.
5. **Accessibility needs.** Captions, contrast, value labels, alt text.
6. **Submission rule.** What is delivered, in what file, by when.
7. **Question channel.** Live questions, written review, or none at all.

## 4. Compression, Inflation, and the Boundary Sentence

*(minutes 7–30)*

**Guiding question:** *what may a spoken version cut, and what must survive
every cut?*

> *"I stop at maybe forty posters in an afternoon. What I remember is not the
> one with the most data. It is the one whose presenter told me, in one clean
> breath, exactly what they found and exactly where it stops. The ones who
> oversold lost me on the second sentence."*
> — a **skeptical peer** working the room at a research conference

Speaking is faster than reading, so you compress on the fly, and that is where
claims quietly grow. Four terms carry the discipline.

- **Compression** is a shorter version of a claim that keeps its boundary.
  Example: "in this teaching file, more responsive government goes with a little
  more trust in the police" is the sentence, compressed.
- **Inflation** is a version that quietly makes the claim bigger. Example:
  "responsive government builds trust" is shorter and it is a different claim.
- The **excited-mouth upgrade** is the moment your spoken verb outruns your
  written one and the boundary moves while nobody is watching the words. You
  wrote "goes with"; the adrenaline says "builds".
- Your **boundary sentence** is one sentence naming what the claim does NOT
  show. Example: "this is a descriptive comparison in a teaching file, so it
  cannot say a responsive government causes the trust." You write it first and
  rehearse it until it is reflex, because it is the beat pressure deletes.

Monday's ladder still applies, one line and move on: a verb states a kind, a
phrase states a reach, and neither upgrade is free. Speech is that same ladder
running at speaking speed.

Everything you say sits on one spine, the **pitch architecture**: hook, claim,
evidence, boundary, invitation. The thirty-second version is hook, claim,
boundary, invitation, with no evidence beats yet. The three-minute version is
the same spine with two or three evidence beats added. The longer version *adds*
blocks. It never states a bigger claim.

> **A question that often comes up here:** *"How is compressing different from
> inflating? Both change the sentence."* Compression drops detail while keeping
> the boundary. Inflation moves the boundary while keeping the length. In the
> compressed line on the board, "when the government responds" smuggles in a
> cause the comparison never bought, "people" quietly swaps a teaching file for
> everyone, and the half point and the overlap vanished entirely. One version
> lost a number. The other lost the truth.

### 🛠️ Run It Live: the delivery check

Run the cell. It does two jobs at once. It turns a word count into seconds at a
clear speaking pace and prints the word budget for the thirty-second and
three-minute layers. And it scans a delivery for **self-erasing words**, the
words that bury a real finding under apology, and checks whether the delivery
ends on a next step rather than on the flaw.

Three sample deliveries are built in, all about Monday's claim: A is an apology
spiral, B is hedge-heavy, and C runs the honest move in order. Expect A and B to
trip several flags each and neither to end on a next step, and expect C to trip
none and to end on one.

Then paste your own delivery into `my_delivery` and run it again. The cell
settles your prediction. It never settles your judgment.

```python
# The delivery check: a pacing budget plus a self-erasing-word count.
RATE_WPM = 130   # a clear, unrushed pace; under pressure people speed up, so budget slow

def words_that_fit(seconds, wpm=RATE_WPM):
    # Seconds times words-per-second gives the word budget for that slice of time.
    return int(seconds * wpm / 60)

def duration_seconds(script, wpm=RATE_WPM):
    # Words in the script divided by words-per-second gives seconds aloud.
    return len(script.split()) / (wpm / 60.0)

APOLOGY_MARKERS = ["sorry", "just", "probably", "i guess", "i'm no", "not sure",
                   "sort of", "kind of", "maybe", "possibly", "all wrong", "can't really"]
NEXT_STEP_MARKERS = ["next step", "the study that could", "the design that could",
                     "would settle", "how you'd find out", "how one would find out"]

def audit_delivery(text):
    low = text.lower()
    flags = sum(low.count(m) for m in APOLOGY_MARKERS)
    ends_on_next_step = any(m in low for m in NEXT_STEP_MARKERS)
    return len(text.split()), duration_seconds(text), flags, ends_on_next_step

budget_30, budget_3min = words_that_fit(30), words_that_fit(180)
print(f"At {RATE_WPM} words per minute, an honest budget is:")
print(f"  30-second layer : about {budget_30} words")
print(f"  3-minute layer  : about {budget_3min} words\n")

deliveries = {
    "A (apology spiral)": (
        "Honestly this is probably all wrong, it's just a teaching resample of 10,000 "
        "rows, and I'm no statistician, so I guess you can't really read anything "
        "into half a point."),
    "B (hedge-heavy)": (
        "It sort of seems like maybe there might be a kind of weak difference, "
        "possibly, but I'm not sure it really holds up."),
    "C (boundary, then the next step)": (
        "The gap is about half a point on a 1-to-7 scale and the two intervals "
        "overlap, so I report a small difference in this file, not a precise number. "
        "Nobody was assigned anything here, so it cannot show that a responsive "
        "government raises trust in the police. The study that could settle it "
        "follows the same people as their view of government changes."),
}

# Paste your own 30-second delivery here, then run this cell again.
my_delivery = ""

print(f"{'Delivery':<34}{'words':>7}{'seconds':>9}{'flags':>7}{'ends on next step?':>21}")
for name, text in deliveries.items():
    w, s, f, n = audit_delivery(text)
    print(f"{name:<34}{w:>7}{s:>9.0f}{f:>7}{('yes' if n else 'no'):>21}")
if my_delivery.strip():
    w, s, f, n = audit_delivery(my_delivery)
    print(f"{'MINE':<34}{w:>7}{s:>9.0f}{f:>7}{('yes' if n else 'no'):>21}")

# Self-check: the pacing math is an identity and the samples behave as described.
assert words_that_fit(60) == 130, "one minute must be 130 words at 130 wpm"
assert budget_30 == 65 and budget_3min == 390, "word budgets drifted — check RATE_WPM"
fa, fb, fc = (audit_delivery(t)[2] for t in deliveries.values())
na, nb_, nc = (audit_delivery(t)[3] for t in deliveries.values())
assert fa >= 5 and fb >= 5 and fc == 0, "sample flag counts drifted"
assert nc and not na and not nb_, "only the third delivery should end on a next step"
assert audit_delivery(deliveries["C (boundary, then the next step)"])[0] <= budget_30, \
    "the honest sample must fit its own 30-second budget"
print(f"\n✓ Self-check passed: A trips {fa} flags, B trips {fb}, C trips {fc}; "
      f"only C ends on a next step, inside the {budget_30}-word budget.")
```

## 5. Defend, Concede, or Say You Do Not Know

*(minutes 30–47)*

**Guiding question:** *when a stranger pushes, how do you decide on the spot
whether to hold your claim or narrow it?*

> *"When a presenter tells me, unprompted, exactly what their study cannot show,
> I trust everything else they said more. When they insist it is airtight, I
> start looking for the crack myself."*
> — a **policy stakeholder** who has heard a thousand pitches

Three lenses cover almost every hard question, and knowing the lens tells you
the answer. The **outsider** does not share your vocabulary and asks whether the
finding matters and how far it travels: the test is whether your claim survives
translation. The **specialist** knows your tools and asks about a design choice:
the test is whether your specific choices were defensible. The **skeptic**
grants your data and offers a rival explanation for the same pattern: the test
is your claim boundary, and it is the one you usually cannot study your way out
of at the artifact.

Under every answer sits one judgment. You **defend** when the objection asks for
something your evidence already licenses: you point at the claim-evidence row
and hold your ground with the number. You **concede** when the objection asks
for a crossing your design never paid for: you narrow the claim out loud, and
you say the narrower wording on the spot. The test is mechanical. Does my
evidence back this? If yes, defend. If no, concede and state the smaller claim
that survives. A graceful concession raises your credibility; an upgrade under
pressure destroys it fastest.

When you genuinely do not know, the **honest don't-know** keeps the concession
expert. It has three required parts: what you cannot say, the adjacent thing you
CAN stand on, and the study that would settle it. Example: "I cannot say whether
this holds outside the file; what I can stand on is that the gap is stable
across both halves of it; what would settle it is a sample drawn from a real
frame." A prepared don't-know reads as command of your evidence. Bluffing,
inventing an answer you do not have, is the one move that ends your credibility
on the spot.

And the move that speaks a limit as expertise is the **ULN move**: uncertainty,
then limitation, then next step, three sentences with a straight spine. The
**next step** is what turns a limitation from an apology into a research agenda.

> **A question that often comes up here:** *"Is saying 'I do not know' at my own
> artifact a disaster?"* Only as a bare shrug. In full form it is the opposite:
> it names exactly where your knowledge stops, which is the most credible thing
> you can say. The presenter who answers "sure, basically it causes it" to a
> friendly stranger has just committed the excited-mouth upgrade in public, and
> any careful visitor now trusts nothing else on the artifact.

**Optional depth: the spine, drawn.** The next cell draws the five beats and
shows how the two layers nest on them. The boundary beat is drawn with a heavy
dark outline rather than a color code, so it survives a grayscale print. Expect
one path from hook to invitation, with a shortcut arrow from claim straight to
boundary: that shortcut is the thirty-second layer.

```python
# OPTIONAL DEPTH — the pitch spine: five beats, two lengths.
G = nx.DiGraph()
pos = {
    "Hook":       (0.0, 0.0),
    "Claim":      (1.3, 0.0),
    "Evidence 1": (2.6, 0.6),
    "Evidence 2": (2.6, -0.6),
    "Boundary\nsentence": (3.9, 0.0),
    "Invitation": (5.2, 0.0),
}
G.add_edges_from([
    ("Hook", "Claim"),
    ("Claim", "Evidence 1"),
    ("Claim", "Evidence 2"),
    ("Evidence 1", "Boundary\nsentence"),
    ("Evidence 2", "Boundary\nsentence"),
    ("Claim", "Boundary\nsentence"),        # the 30-second shortcut
    ("Boundary\nsentence", "Invitation"),
])

nodes = list(pos)
outline = ["#4C72B0"] * len(nodes)
widths = [1.5] * len(nodes)
bi = nodes.index("Boundary\nsentence")
outline[bi], widths[bi] = "#222222", 3.5      # heavy outline, not a color code

fig, ax = plt.subplots(figsize=(11, 4.5))
nx.draw_networkx_nodes(G, pos, node_color=["#DCE6F1"] * len(nodes),
                       edgecolors=outline, linewidths=widths, node_size=4600, ax=ax)
nx.draw_networkx_labels(G, pos, font_size=9, ax=ax)
nx.draw_networkx_edges(G, pos, arrowstyle="-|>", arrowsize=16, edge_color="#666666",
                       width=1.4, node_size=4600, ax=ax)
ax.set_title("The spine: one skeleton, two lengths")
ax.axis("off")
ax.text(0.0, -1.4, "30-second layer: Hook to Claim to Boundary to Invitation (no evidence beats)",
        fontsize=9, color="#333333")
ax.text(0.0, -1.7, "3-minute layer: the same spine, with evidence beats added",
        fontsize=9, color="#333333")
plt.tight_layout()
plt.show()

print("✓ Spine drawn. The heavy-outlined boundary beat is the one you rehearse to reflex.")
print("  The longer layer ADDS evidence to this spine; it never enlarges the claim.")
```

**Reading the diagram.** Every version walks the same path left to right: hook,
claim, then either straight to the boundary or through the evidence beats first,
and out on the invitation. The heavy-outlined boundary beat is the one you
cannot skip at either length.

*In plain terms, the picture says the short version is the skeleton and the long
one hangs evidence on it without growing the head.*

**Optional depth: the same limitation, five ways.** The next cell draws a
1-to-5 scale of how a limitation can land, from an apology spiral at 1 to
expertise at 5. Score grows with marker size rather than color, so the scale
reads in grayscale. Expect the same limitation at every point on it.

```python
# OPTIONAL DEPTH — the strength scale: how the SAME limitation can land.
labels = ["1\nApology\nspiral", "2\nHedge-\nheavy", "3\nFlat", "4\nInformative",
          "5\nExpertise"]
sounds = ["'probably all\nwrong, sorry'", "finding buried\nunder qualifiers",
          "limitation stated,\nno next step", "limitation +\nnext step",
          "precise bound +\nlimitation + the\ndesign that resolves it"]
scores = [1, 2, 3, 4, 5]

fig, ax = plt.subplots(figsize=(11, 3.4))
ax.plot(scores, [0] * 5, color="#888888", linewidth=2, zorder=1)
ax.scatter(scores, [0] * 5, s=[140, 200, 260, 340, 440], color="#4C72B0",
           edgecolors="#222222", linewidths=1.5, zorder=3)   # size, not colour
for x, lab, snd in zip(scores, labels, sounds):
    ax.text(x, 0.16, lab, ha="center", va="bottom", fontsize=9)
    ax.text(x, -0.16, snd, ha="center", va="top", fontsize=8, color="#333333")
ax.axvspan(3.5, 5.5, color="#DCE6F1", alpha=0.6, zorder=0)
ax.text(4.5, 0.42, "aim here", ha="center", fontsize=9, color="#1a1a1a")
ax.set_xlim(0.5, 5.5)
ax.set_ylim(-0.5, 0.5)
ax.axis("off")
ax.set_title("The same limitation can sound like weakness or like expertise")
plt.tight_layout()
plt.show()

print("✓ Scale drawn. A 1 or a 2 misstates the evidence as badly as HIDING the limit would.")
print("  The growing marker is one limitation, delivered with more spine.")
```

**Reading the diagram.** The identical limitation can land anywhere on this
scale. At a 1 it erases the finding. At a 5 it names a precise bound, the thing
the design cannot establish, and the study that would resolve it. Nothing about
the finding changed between the two.

*In plain terms, the picture says your limitation is not the problem. The spine
you say it with decides whether a stranger trusts you more or less.*
