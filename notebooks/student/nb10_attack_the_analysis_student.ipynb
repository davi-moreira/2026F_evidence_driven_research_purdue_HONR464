## 1. From a Result to a Claim: the Verb Carries the License

**Guiding question:** *what sentence can your design and your stress tests
actually support?*

> *"Your analysis produced a number. Your note will produce a sentence. Between
> the two sits every mistake I am paid to catch: the verb that outruns the
> design, the reach the frame never covered, the certainty the interval never
> offered. Show me a sentence sized to its evidence and I will read the rest."*
> — a **journal reviewer**, on where reviews are won and lost

Two words name the gap this week closes. A **result** is what your analysis
printed: an estimate, an interval, a table. Example: a correlation of 0.46 with
its interval. A **claim** is the sentence you write about it, and the sentence
always says more than the number unless you bound it. Example: "trust in the
police and the military move together in this sample" versus "military trust
drives confidence in the police." Same result; two different claims; only one
is licensed.

The license lives in the **claim verb**, the verb that states what your
evidence establishes. "Is related to" claims a pattern. "Drives" claims a
cause. "Across Latin America" claims a reach. None of these upgrades is free,
and a p-value pays for none of them: **statistical significance** only says the
pattern would be surprising if there were nothing but noise. It speaks to
chance, not to kind, not to reach, and never to direction.

**The claim-license matrix.** There is no ladder of claims, and that matters.
A claim makes two separate declarations, and they are the compass's own two
dimensions. The **kind** says what the claim asserts: descriptive, a pattern or
a level, or causal, one thing changing another. The **reach** says who the
claim covers: the data at hand, a population, or unseen cases. Every claim
states BOTH, and each dimension holds its own license with its own warrant.

| Dimension | Position | Verb and phrase family | The warrant it demands |
|---|---|---|---|
| Kind | **Descriptive** | is, sits at, is related to, moves with | the computation itself, plus honest uncertainty |
| Kind | **Causal** | changes, reduces, increases, drives | assignment, or a defended identification argument |
| Reach | **Data at hand** | "in the rows my frame reaches" | honest bookkeeping of what those rows are |
| Reach | **Population** | "among [the population], the share/level is" | a sampling-and-response accounting |
| Reach | **Unseen cases** | will, is expected to | prediction-time honesty plus a held-out check |

The dimensions do not rank each other. A forecast is a descriptive-kind claim
at unseen reach, not a step below cause; a causal claim still names its own
reach, for these units or for a population; and no cell outranks a
well-warranted claim in another cell. Off the matrix entirely floats "always,
everywhere, for everyone": no design licenses a universal claim. Three rules
govern the matrix:

1. **The design picks the license on each dimension**, never the wish and never
   the p-value.
2. **Claiming less is free.** You may narrow the reach, or state the
   descriptive pattern while the causal question stands open. Every extension,
   on either dimension, must show its warrant in the same breath.
3. **The boundary rides in the same sentence.** "X and Y move together" is
   unfinished; "in this teaching sample, X and Y move together (r 0.46)" is a
   claim that states its kind and its reach. A boundary in a footnote is a
   boundary the reader never meets.

The puzzle's headline fails on both dimensions. "Significant" is asked to mean
big and causal, which it never means. "Drives" claims the causal kind with no
identification argument, and could not even pick a direction if it had one.
"Across Latin America" claims a population reach that a teaching resample
licenses for nobody. And one more rule from Week 5 still stands: if the
colleague truly wants the causal question, it stays causal, **currently
unidentified**, stated in the open. It never gets relabeled descriptive because
the data ran weak.

> **A question that often comes up here:** *"Isn't a statistically significant
> result at least a strong result?"* No. Significance is a statement about
> noise, and with 10,000 rows almost any real pattern clears it. It cannot tell
> you the pattern is large, or important, or causal, or general. The interval
> tells you the size; the design tells you the kind; the frame tells you the
> reach. A note that leans on "significant" as its adjective is a note that has
> not yet decided what it is claiming.

## 2. Uncertainty and Limits: Rank the Findings by Their Evidence

**Guiding question:** *which of your findings has earned the right to lead the
note, and how big are your limitations allowed to be?*

Not all findings are equal, and pretending they are is how honest work
overclaims. One anchor and three terms keep the front of a note honest. The
anchor is your **predeclared primary**: the single analysis your Research
Contract and your M8 pre-listed card named as the headline answer before any
result existed. Example: here, the whole-sample analysis is the declared
primary, and the 40-row slice is not.

- **Results hierarchy** is the order you present findings in: the predeclared
  primary leads, the rest follow by design validity and audit survival, and
  exploratory or fragile results ride last, clearly labeled. Interval width is
  one input to that judgment, never the ordering rule. Example: the declared
  whole-sample analysis leads; a surprising result from a tiny undeclared
  subgroup comes last, labeled exploratory.
- **Inflated contribution** is a claim written to sound like a bigger finding
  than the evidence supports. Example: leading with a fragile forty-person
  number because it is the most surprising thing you found.
- **Limitations sized to evidence** are limitations stated in proportion to
  what actually threatens the claim. Example: naming the teaching-resample
  caveat and the coverage gap that most bend your number, instead of a generic
  "more research is needed."

> **A question that often comes up here:** *"Isn't my most surprising result
> the one I should lead with?"* Surprise and evidence are different axes. Your
> most surprising result is often your least supported one, precisely because
> it came from a small or unusual slice where noise runs high. Lead with what
> you can defend; the surprising finding earns a careful, clearly labeled
> paragraph near the end. A reviewer who finds a fragile result leading the
> note stops trusting the whole ranking.

### 🛠️ Run the Study: the same relationship, two tiers

Run the next three cells in order. The first computes the correlation between
reported trust in the police and trust in the military twice, whole sample and
40-row slice, each with a seeded 95% bootstrap interval. The whole-sample
analysis plays the declared primary here; the 40-row slice is an undeclared
exploratory cut. One caveat travels
with every number: `lapop_brazil` is a 10,000-row resample with replacement of
real survey interviews, realistic in its patterns but manufactured in its
precision, so it serves planning and practice, never substantive claims about
Brazil. Expect two positive correlations and two very different interval
widths. The second cell draws both intervals on one axis, and the third checks
the numbers.

```python
# Load the teaching survey and confirm its shape before trusting anything.
lapop = load_course_data("lapop_brazil.csv")
assert lapop.shape == (10000, 10), "unexpected shape — flag this!"
print("✓ Loaded lapop_brazil.csv —", lapop.shape[0], "rows,", lapop.shape[1], "columns")

# The declared primary: are police trust and military trust related?
r_full = lapop["trust_police"].corr(lapop["trust_military"])

def boot_corr_ci(frame, a, b, gen, n_boot=2000):
    """95% bootstrap interval for the correlation of two columns (seeded gen)."""
    va, vb = frame[a].to_numpy(), frame[b].to_numpy()
    m = len(frame)
    out = np.empty(n_boot)
    for i in range(n_boot):
        pick = gen.integers(0, m, m)          # resample rows with replacement
        out[i] = np.corrcoef(va[pick], vb[pick])[0, 1]
    return np.nanpercentile(out, [2.5, 97.5])

rng_boot = np.random.default_rng(SEED)        # isolated, reproducible on its own
lo_full, hi_full = boot_corr_ci(lapop, "trust_police", "trust_military", rng_boot)

# An UNDECLARED exploratory cut: the SAME relationship in a tiny 40-row slice.
small = lapop.sample(n=40, random_state=SEED)
r_small = small["trust_police"].corr(small["trust_military"])
lo_small, hi_small = boot_corr_ci(small, "trust_police", "trust_military", rng_boot)

print("\nRESULTS HIERARCHY (the predeclared primary leads; exploratory rides last, labeled):")
print(f"  Tier 1 declared primary : r = {r_full:.2f}   95% CI [{lo_full:.2f}, {hi_full:.2f}]"
      f"   width {hi_full - lo_full:.2f}   (n = {len(lapop):,})")
print(f"  Tier 3 exploratory      : r = {r_small:.2f}   95% CI [{lo_small:.2f}, {hi_small:.2f}]"
      f"   width {hi_small - lo_small:.2f}   (n = 40)")
```

```python
# Figure: the two intervals on one axis. Each tier gets its OWN hue; neither
# reuses the other's, so the wide interval never blends into the tight one.
fig, ax = plt.subplots(figsize=(9, 3.6))
ax.plot([lo_full, hi_full], [1, 1], color="#4C72B0", linewidth=3, solid_capstyle="round")
ax.scatter([r_full], [1], color="#4C72B0", s=110, zorder=3)
ax.plot([lo_small, hi_small], [0, 0], color="#8172B3", linewidth=3, solid_capstyle="round")
ax.scatter([r_small], [0], color="#8172B3", s=110, zorder=3)
ax.axvline(0, color="gray", linestyle=":", linewidth=1.2, label="no relationship (r = 0)")
ax.set_yticks([0, 1])
ax.set_yticklabels(["Exploratory\n(n = 40)", "Declared primary\n(n = 10,000)"])
ax.set_ylim(-0.6, 1.6)
ax.set_xlim(-0.2, 1.0)
ax.set_xlabel("Correlation r (police trust, military trust), with 95% bootstrap interval")
ax.set_title("A results hierarchy: the declared primary leads, the undeclared slice waits at the back")
ax.legend(loc="lower right")
plt.tight_layout()
plt.show()

print("✓ Figure drawn. The declared primary's interval is narrow; the 40-row")
print("  interval is far wider, one more reason the undeclared slice cannot lead.")
```

```python
# Self-check: the whole sample gives a tight interval; the tiny slice a wide one.
assert 0.40 < r_full < 0.50, "headline correlation drifted — check the data"
assert (hi_full - lo_full) < 0.10, "headline interval should be tight at n = 10,000"
assert (hi_small - lo_small) > (hi_full - lo_full), "the n=40 interval should be wider"
ratio = (hi_small - lo_small) / (hi_full - lo_full)
print(f"✓ Self-check passed: headline r = {r_full:.2f} on a tight interval "
      f"[{lo_full:.2f}, {hi_full:.2f}];\n  the 40-row interval is about {ratio:.0f}x wider, "
      f"so it belongs at the BACK of the note, labeled exploratory.")
```

**Reading the output.** Both slices measured the same relationship, and both
point estimates are positive. The whole-sample estimate is r = 0.46 on an
interval of [0.44, 0.48], a range so tight the finding barely moves. The 40-row
estimate is bigger, near 0.61, and that is the trap: its interval stretches
from about 0.29 to 0.80, roughly fourteen times wider, pinning down almost
nothing about the true size. The flashier number is the weaker finding: it was
never the declared primary, it comes from an undeclared slice, and its interval
pins down almost nothing. Rank by the declaration and the design, never by the
sparkle; read interval width as one input to that judgment, not as the ordering
rule; and put the uncertainty in the same eye-span as the number every time it
appears.

*In plain terms, the picture says: the top line is the declared primary and has
earned the front of the note; the bottom line could be almost anything and must
wait at the back, labeled exploratory.*

## 3. The Note Spine: the Argument a Stranger Can Follow

**Guiding question:** *what does a research note have to argue in writing that
your notebooks and stand-ups were allowed to leave unsaid?*

Until this week your results lived in notebooks you could explain in person,
and in a dossier written for yourself. A **research note** is the first
artifact that must stand alone: a short, self-contained account of the study,
written for a stranger who has only the page. Nobody is beside it to fill the
silences, so every leap your voice used to carry, the prose now has to earn, in
order. And the order matters beyond this week: next week's poster is an
adaptation of this note, never the other way around. A poster is a compression,
and compressing a claim that never existed in bounded prose lets the poster's
space constraints author your claim for you.

The note's backbone is the **note spine**: problem, gap, question, design,
findings in their declared order, and the bounded headline claim, each linked to the
next by a sentence a reader can refuse to believe until you earn it. You have
built every piece already. Your M2 evidence map established the gap; your
Contract declared the question and design; your M8 audit ranked what survived.
The note argues the arrows between them.

> **A question that often comes up here:** *"If my dossier already holds all of
> this, why write the note at all?"* Because the dossier is written for you,
> and you are its most forgiving reader. The dossier records decisions in the
> order you made them; the note argues them in the order a stranger needs. The
> gap your dossier states in six words must show what prior work covered and
> what it left out. Writing is where research leaves your head and can first be
> misread, and the note is where you control the misreading in advance.

```python
# Draw the note spine: the argument a research note makes, in order.
# Expect six linked boxes; each arrow is a sentence the note must earn.
G = nx.DiGraph()
nodes = ["Problem", "Gap", "Question", "Design", "Findings", "Bounded\nclaim"]
pos = {n: (i * 1.6, 0.0) for i, n in enumerate(nodes)}
G.add_nodes_from(nodes)
edge_labels = {
    ("Problem", "Gap"): "motivates",
    ("Gap", "Question"): "sharpens into",
    ("Question", "Design"): "answered by",
    ("Design", "Findings"): "produces",
    ("Findings", "Bounded\nclaim"): "sized into",
}
G.add_edges_from(edge_labels.keys())

fig, ax = plt.subplots(figsize=(12, 3.2))
nx.draw_networkx_nodes(G, pos, node_color="#DCE6F1", edgecolors="#4C72B0",
                       linewidths=1.5, node_size=4200, ax=ax)
nx.draw_networkx_labels(G, pos, font_size=9, ax=ax)
nx.draw_networkx_edges(G, pos, arrowstyle="-|>", arrowsize=18, edge_color="#555555",
                       width=1.6, node_size=4200, ax=ax)
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=8,
                             label_pos=0.5, ax=ax)
ax.set_title("The note spine: findings ride in declared order, and the claim ships bounded")
ax.axis("off")
plt.tight_layout()
plt.show()

print("✓ Spine drawn: the problem motivates the gap, the gap sharpens into the")
print("  question, the design answers it, the findings arrive in declared order,")
print("  and the headline claim ships with its boundary inside the sentence.")
```

**Reading the diagram.** Read it left to right and you are reading the order
the note argues in. On the far right sits everything this lecture built: the
findings enter in their declared order: the predeclared primary first, the rest
by design validity and audit survival, the exploratory slice last. The spine
ends at a **bounded claim**, the headline sentence whose verb, reach, and
uncertainty all ride inside it.

*In plain terms, the picture says a research note is not six boxes. It is five
connections, and the connections are what your prose has to argue.*

**Optional depth: the evidence-density floor.** The spine's first two boxes,
problem and gap, rest on prior work, and prior work means sources a reader can
retrieve. **Evidence density** is how much of a paragraph rests on retrievable
sources versus bare assertion. Example: "as everyone knows" carries zero
density; a sentence resting on a source a reader can open carries some. The
next cell scores two four-sentence paragraphs on the same topic, one written as
confident assertion and one as sourced synthesis, by counting the sentences
that carry a retrievable marker. Expect a 0.00 and a 0.75.

```python
# OPTIONAL DEPTH — a tiny evidence-density scorer. A sentence "rests on a source"
# when it carries a retrievable marker [verified: ...]; density = sourced / all.
def evidence_density(paragraph):
    parts = paragraph.replace("!", ".").replace("?", ".").split(".")
    sentences = [s.strip() for s in parts if s.strip()]
    sourced = [s for s in sentences if "[verified" in s.lower()]
    n = len(sentences)
    return n, len(sourced), n - len(sourced), (len(sourced) / n if n else 0.0)

assertion = (
    "As everyone knows, people no longer trust their public institutions. "
    "Trust has obviously collapsed everywhere you look. "
    "It is common sense that nobody believes official numbers anymore. "
    "Clearly the situation is worse now than it has ever been."
)
synthesis = (
    "Reported trust in national institutions has fallen across several "
    "long-running surveys [verified: retrieved]. "
    "One widely cited 2000 study of civic life documented a decades-long decline "
    "in community-group membership [verified: retrieved]. "
    "A 2015 reproducibility project found that many published effects did not "
    "hold up under independent re-analysis [verified: retrieved]. "
    "Together these strands do not settle a single cause, but each rests on a "
    "source a reader can open."
)

for name, para in [("Assertion", assertion), ("Synthesis", synthesis)]:
    n, sourced, bare, dens = evidence_density(para)
    print(f"{name:10s}: {n} sentences, {sourced} sourced, {bare} bare "
          f"-> evidence density = {dens:.2f}")

n_a, s_a, b_a, d_a = evidence_density(assertion)
n_s, s_s, b_s, d_s = evidence_density(synthesis)
assert d_a == 0.0, "the assertion paragraph should score 0.0 density"
assert d_s >= 0.7, "the synthesis paragraph should score at least 0.70 density"
print(f"\n✓ Self-check passed: assertion density {d_a:.2f} ({b_a} bare claims), "
      f"synthesis density {d_s:.2f} ({s_s} sourced of {n_s}).")
```

**Reading the output.** The assertion paragraph scores 0.0: every sentence
sounds confident and not one rests on anything a reader could open. The
synthesis scores 0.75, and its fourth sentence is an honest summary that claims
no more than the three sources support. The scorer is crude on purpose. It
cannot judge whether a source says what you claim; it only asks the first
question a reviewer asks: is there anything here to retrieve at all? Density is
a floor a note's opening paragraphs must clear, never a proof they are good. A
paragraph that scores zero is an opinion in a lab coat.

## 4. The Claim-Evidence Table: No Row, No Claim

**Guiding question:** *can every sentence in your note name the evidence that
supports it, the check that verified it, and the boundary that limits it?*

> *"I point at one sentence and ask, 'where did this come from?' If you can
> walk me to the output that produced it and the check that verified it, I
> believe the rest of the page. If you cannot, I stop believing all of it."*
> — a **research-integrity reviewer**, reading a first-draft note

The **claim-evidence table** is the instrument that makes a note checkable.
Every claim in the note gets a row with four columns: the **claim** itself, the
**evidence** (the output that supports it, with its trace to the exact cell or
file that produced it), the **verification** (the real check that vouches for
it, such as your clean-restart rerun or the M8 pre-listed check it survived),
and the **boundary** (how far the claim reaches and no further). The table is
your claim ledger: it records what the research asserts, and it is a separate
instrument from the AI Research Ledger, which records what AI did.

One rule gives the table its teeth: **a claim with no row does not appear in
the note.** A claim you cannot fit in the table is a claim to cut or a check to
run, and both send you back before the note ships. Two failures have names
worth memorizing, because the M9 brief caps grades on both. An **untraceable
number** is a number with no path back to the output that produced it, typed
from memory or carried from an old draft. An **unretrievable source** is a
citation you cannot locate and open; a source a reader cannot retrieve supports
nothing, whoever suggested it.

> **A question that often comes up here:** *"My AI assistant told me the
> background fact. Can the evidence column just say that?"* No. The tool is not
> a source; it is a suggestion machine. A suggestion becomes evidence only when
> you retrieve the underlying work and read that it says what your sentence
> claims. Until then the honest entry is "unresolved," and an unresolved entry
> means the sentence it supports does not ship. This is the course's oldest
> rule wearing its writing clothes: every empirical claim traces to a real,
> retrievable source, and you are the one who retrieved it.

### 🛠️ Run the Study: audit the colleague's table

Run the cell. It rebuilds the colleague's claim-evidence table (a simulated
instructional artifact, reloading the teaching survey so the cell works on a
fresh kernel), then runs `audit_rows`, a transparent checker that asks of every
row only what a reviewer would ask: is there evidence, does it trace to an
output, is a real check named, is a boundary stated, and does the evidence rest
on something retrieved rather than remembered or merely suggested. Expect two
rows to pass and two to be flagged.

```python
# The colleague's claim-evidence table — a simulated instructional artifact.
lapop = load_course_data("lapop_brazil.csv")   # reload: works on a fresh kernel
assert lapop.shape == (10000, 10), "unexpected shape — flag this!"
print("✓ Loaded lapop_brazil.csv —", lapop.shape[0], "rows,", lapop.shape[1], "columns\n")

colleague_rows = [
    {"claim": "Police trust and military trust move together in this teaching sample",
     "evidence": "printed correlation r 0.46 with 95% interval [0.44, 0.48]",
     "trace": "the two-tier results cell (Lecture 1, section 2)",
     "verification": "clean rerun reproduced r and the interval",
     "boundary": "descriptive; teaching resample only, no population claim"},
    {"claim": "High system pride (6-7) goes with higher supreme-court trust than low pride (1-2)",
     "evidence": "two group means printed from the file: 5.34 versus 3.35",
     "trace": "the subgroup-means cell of the analysis notebook",
     "verification": "both group means recomputed by hand from the file",
     "boundary": "descriptive; these respondents; pride is not claimed as a cause"},
    {"claim": "Institutional trust has been declining across the region for a decade",
     "evidence": "AI-suggested source, never retrieved",
     "trace": "",
     "verification": "",
     "boundary": ""},
    {"claim": "About 70 percent report substantial military trust (5 or higher on 1-7)",
     "evidence": "remembered from an earlier draft",
     "trace": "",
     "verification": "",
     "boundary": "teaching resample"},
]

def audit_rows(rows):
    """Flag every claim-evidence row a reviewer could not check."""
    flags = {}
    for i, row in enumerate(rows, 1):
        ev, problems = row["evidence"].lower(), []
        if "never retrieved" in ev or ("suggest" in ev and "retriev" in ev):
            problems.append("rests on an unretrieved suggestion (not evidence yet)")
        if "remember" in ev or "earlier draft" in ev or "memory" in ev:
            problems.append("untraceable number (no path back to an output)")
        if not row["trace"]:
            problems.append("no trace to the output that produced it")
        if not row["verification"]:
            problems.append("no verification named")
        if not row["boundary"]:
            problems.append("no boundary stated")
        if problems:
            flags[i] = problems
    return flags

table = pd.DataFrame(colleague_rows)
table.index = range(1, len(table) + 1)
print(table[["claim", "evidence"]].to_string())
print()
flags = audit_rows(colleague_rows)
for i, problems in flags.items():
    print(f"Row {i} FLAGGED:")
    for p in problems:
        print(f"   - {p}")
print(f"\nRows passing the audit: {[i for i in table.index if i not in flags]}")
```

**Reading the output.** Rows 1 and 2 pass: each names its evidence, traces it
to a cell, records a real check, and states a boundary. Row 3 fails on every
column at once, because its support is a suggestion nobody retrieved: whatever
the sentence's truth, the note cannot carry it yet. Row 4 fails differently:
its number may even be right, but "remembered from an earlier draft" gives a
reviewer no path to check it, which is exactly what **untraceable** means. Note
what the auditor did not do: it never judged whether any claim is true. It
asked only whether a stranger could check each one, which is the floor a note
must clear before truth can even be discussed.

## 5. The Ledger-Derived Disclosure and the Note-v0 Lock

**Guiding question:** *what did AI receive and return while this note was
built, and does the written disclosure say so?*

An **AI disclosure** is the short statement in your note that records what you
delegated to AI tools and how you checked what came back. The rule that makes
it honest is mechanical: **the disclosure is derived from your ledger, never
written from memory.** Memory keeps the flattering rows. It remembers "AI
polished my wording" and forgets three things a reader is entitled to know:
what you **sent** to the tool (your draft and your results went out, not just
prose), what you **rejected** (a suggestion you refused is still AI
involvement), and what stays **unresolved** (a suggested source you could not
retrieve). Your ledger kept all three, which is why it, and not your memory,
writes the disclosure.

The **unresolved-source treatment**, this course's standing rule: a source you
cannot retrieve stays out of the evidence table and out of the note's support,
however plausible it sounds. You record it as unresolved in your disclosure,
and that record is a finding, not an embarrassment. The next cell shows the
whole discipline in ten lines: it derives the colleague's disclosure from the
ledger, then compares it with the one written from memory. Expect one ledger
row weakly reflected and two omitted entirely.

> **A question that often comes up here:** *"A full disclosure makes it sound
> like AI did a lot. Will honesty cost me?"* The opposite. Every researcher
> around you is using these tools; what separates work a reader can trust is
> whether the use was controlled and checked. A disclosure that names what was
> sent, what was rejected, and what stayed unresolved reads as competence,
> because it is the record of you directing the tool. The sentence that ships
> is yours, whatever drafted it, and the disclosure is where you prove you knew
> the difference.

```python
# Derive the disclosure FROM the ledger, then diff it against memory.
colleague_ledger = [
    {"task": "tighten the results paragraph",
     "sent": "the full draft note, results table included",
     "came_back": "cleaner prose, plus 'is related to' upgraded to 'drives'",
     "decision": "kept the prose; REJECTED the verb upgrade"},
    {"task": "suggest background sources on regional trust decline",
     "sent": "the note's background paragraph",
     "came_back": "candidate sources; the closest match could not be retrieved",
     "decision": "UNRESOLVED — the background sentence still leans on it"},
    {"task": "scan the claim-evidence table for empty cells",
     "sent": "the four-row table",
     "came_back": "two incomplete rows flagged",
     "decision": "accepted; repairs pending"},
]

memory_disclosure = "AI was used to polish the wording of this note."

print("The disclosure the ledger DERIVES (one sentence per row):")
for i, row in enumerate(colleague_ledger, 1):
    print(f"  {i}. AI was asked to {row['task']}; it received {row['sent']}; "
          f"it returned {row['came_back']}; decision: {row['decision']}.")

print(f"\nThe disclosure written from MEMORY:\n  \"{memory_disclosure}\"")

mem = memory_disclosure.lower()
mentioned = {i: any(k in mem for k in keys)
             for i, keys in [(1, ("wording", "polish", "tighten")),
                             (2, ("source",)),
                             (3, ("table",))]}
omitted = [i for i, ok in mentioned.items() if not ok]
print(f"\nLedger rows the memory version reflects : {[i for i, ok in mentioned.items() if ok]}")
print(f"Ledger rows it omits entirely           : {omitted}")
print("And even the reflected row omits what was SENT and what was REJECTED.")
```
