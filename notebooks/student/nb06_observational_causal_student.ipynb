## 1. Data Do Not Just Arrive: Settle the Acquisition Route

**Guiding question:** *before any file exists, what decides how data may reach
you at all?*

> *"You wrote that turnout in this county was 72 percent. Where did that come
> from? Not which website you copied it off. Who counted the ballots, and 72
> percent of what? Bring me the office that certified the count, or take the
> number out."*
> — a **thesis advisor**, reading a first draft's background section

That advisor is asking about arrival: how a number came to be in your hands.
Your project faces the same question one level up, and it is settled in
writing, not by luck.

- **Acquisition route**: the declared path by which your data come to exist and
  reach you. Example: "I collect my own survey inside the frame I declared," or
  "I use a city's openly licensed dataset," or "I simulate data of the shape I
  expect." The route is a design decision, so it gets written down like one.
- **Research Contract**: the dated, versioned declaration of your question,
  design, and analysis plan that your M3 milestone opened. Your Contract v0
  names your permission status, and it carries a measurement plan marked
  *provisional* on purpose. Lecture 2 settles it.

Your permission status decides the route, and there are exactly three.

**Cleared, or formal determination granted.** Collect as declared: name your
sampling frame, how you will reach those units, and what you will do about the
ones you cannot reach. Non-response is a design fact, not an accident.

**Pending.** Do everything that does not touch people: build and pilot the
instrument on yourself, write the analysis code against simulated data of the
shape you expect, and wait.

**No authority to ask, or not authorized.** Take the no-permission route, which
reaches every method in this course: published aggregate statistics, an openly
licensed dataset with no individual records, or data you simulate yourself.
Write down which you chose and what it costs you, because a simulated dataset
can demonstrate a method but cannot support a claim about the world.

> **A question that often comes up here:** *"My dataset is already public on
> the internet. Doesn't that settle the route?"* No. Public visibility is not
> permission. A dataset is governed by its license or terms of use, and those
> terms are a fact you record like any other. Openly licensed data are a
> perfectly good route, and the route still gets written down, because "I found
> it online" names no producer, no terms, and no method.

## 2. Provenance: The Chain of Hands Behind Every File

**Guiding question:** *what can a file testify about itself, and what can only
its history establish?*

A value with no traceable origin is a rumor with a decimal point. The
vocabulary that fixes this is short.

- **Data provenance**: the documented origin of a value or dataset and every
  hand it passed through before reaching you. Example: a turnout figure traced
  from a news story back through a state summary to the county office that
  certified the count.
- **Primary source**: the original record where a value was first produced.
  Example: the county's certified canvass. A **secondary source** re-reports a
  value it did not produce, like a newspaper table or a data aggregator. A
  value is only as trustworthy as the primary source at the end of its chain.
- **Data quality**: whether a value is fit for *your* specific use: real,
  produced by a method you can name, and defined the way your question needs.
  Example: a turnout rate over registered voters is high quality for a question
  about registered people, and the wrong number for a question about all adults.
- **Mischaracterized value**: a real number attached to the wrong definition or
  method. Example: a survey's self-reported turnout presented as an official
  count. The number is real; the label is false.

Whole datasets carry the same questions, column by column: who produced this
variable, who cleaned it, what did they drop, and does the definition behind
the column match your question? A file with no answers is not yet data. It is a
rumor in a spreadsheet. Time to interrogate the one that just arrived.

## 3. Verify by Opening: the Primary-Source Drill

**Guiding question:** *what has to happen before you may write the word
"verified" next to a number you did not produce?*

The record you just built ends at a source you have not opened. Closing that
gap has a name and four steps.

- **Retrieval-verification loop**: **Ask** a tool to surface a value and its
  source, **Retrieve** the primary source yourself, **Verify** that it exists
  and reports that value under that definition, and **Document** where you
  found it and when. Example: an AI answer names a state elections page, you
  open the page, you read the column header, and you write down the retrieval
  date.
- **Verified**: a value you have opened at its primary source and read in two
  lines, the line that *reports* it and the line that *defines* it. Example:
  the table cell showing 72 percent, plus the header saying "percent of
  registered voters." A citation you have not opened is a lead, not a source,
  however confident the tool that handed it to you.

Two lines, not a summary of two lines. The definition is where borrowed
numbers usually break: the same real count divides by three defensible
denominators and produces three different published figures, and the label
decides which question the number answers.

> **A question that often comes up here:** *"What if I do not have a borrowed
> number yet, or my source sits behind a paywall?"* Then you run the drill on
> the thing you do have: open the terms-of-use or documentation page of the
> dataset or aggregate table you intend to use, and read the terms. If it will
> not open, say so out loud and write "did not open" into the record. A blocked
> retrieval is a finding that directs your next hour, and it is worth far more
> than a citation you never tested.

## 4. The Permission Recheck, Minimization, and the Data-Management Record

**Guiding question:** *your permission covered the plan; does it cover what
actually landed in your folder?*

Your M3 permission status was granted to a *plan*. Files do not always honor
plans, as you just watched: two columns arrived that the plan never named.
Four habits close the gap.

- **Permission recheck**: holding your permission determination against the
  data as they actually arrived, not as you planned them. Example: a survey
  approved as anonymous arrives with signed names in a free-text box; the
  determination covered the plan, and the arrival broke it.
- **Data minimization**: keeping only the columns your declared analysis needs.
  Example: dropping the place label because no declared analysis uses it. A
  column with no analysis behind it is risk carried for nothing.
- **Re-identification check**: asking whether the columns you *kept*, after
  minimization, could plausibly intersect to pick out one person. Example: an
  identifier plus a 107-level place label is exactly the pair this check exists
  for.
- **Data-management record**: one compact line saying where your files live,
  who can open them, how they are backed up, how they are named, and how long
  you keep them. Example: "raw and cleaned copies in my course folder, shared
  with nobody, backed up nightly to my university drive, named
  `project_raw_YYYYMMDD.csv`, deleted at the end of the term."

Run the recheck as one written line with one of two outcomes: *still cleared*,
or *what changed and what I did about it*. In the file above, the honest line
reads something like: "an identifier and a 107-level place label arrived beyond
the plan; both are already anonymized labels, my declared analysis needs
neither, so I drop them and record the drop."

> **A question that often comes up here:** *"The file has no names in it. Isn't
> it anonymous?"* Not automatically. A few ordinary columns can intersect to
> single out one person, which is why the re-identification check runs on the
> columns you *keep*, after minimization, and why its result is recorded rather
> than assumed. "No names" is a fact about one column. "Cannot be
> re-identified" is a conclusion about all of them at once.

## 5. The Ladder: Concept, Construct, Indicator

**Guiding question:** *which specific number will you let stand in for the big
idea your question is about, and what does that swap cost?*

> *"Before you tell me the pilot made people more engaged, tell me exactly what
> you put a number on. 'Engaged' is a word. I need to know which question you
> asked, on what scale, and what that number quietly ignores."*
> — a **people-analytics lead**, reading the first draft of a workplace study

You never measure the idea. You measure a number that stands in for it, and
your finding is only as trustworthy as that swap. The swap has a name and a
ladder.

- **Operationalization**: turning an abstract idea into a specific, repeatable
  measurement procedure. Example: turning "trust in institutions" into "the
  answer to one 1-to-7 survey item about the national police."
- **Concept**: the abstract idea your question is about. Example: trust in
  institutions. Rich, fuzzy, and unmeasurable directly.
- **Construct**: the one specific, named facet of the concept you decide to
  measure. Example: trust in the national police, one institution among many.
- **Indicator**: the exact number a procedure produces. Example: the response
  to "how much do you trust the police?", recorded from 1 (none) to 7 (a lot).

Each rung visibly narrows the one above it. If you cannot see the narrowing,
you have skipped a rung rather than climbed it.

One more habit makes the ladder honest. For every indicator, write its
**meaning sentence**: one sentence saying what a single recorded value actually
asserts. Example: "a 4 on this item asserts that this respondent, on that day,
placed their trust in the police at 4 of 7." If you cannot write the sentence,
you do not yet know what your number means.

> **A question that often comes up here:** *"If this is real survey data, why
> can't I just say what Brazilians think?"* Because the file you governed on
> Monday is a **10,000-row resample with replacement** of the original
> interviews: rows drawn over and over, like names pulled from a hat and
> dropped back after every draw. The patterns are realistic and perfect for
> practice, and the precision is manufactured, so the caveat travels with every
> number you say out loud today: planning and practice, never population claims
> about Brazil.

The ladder climbs down to a single column. The first thing to do with an
indicator is look at all of it, not its average.

## 6. Reliability on Items, Never People

**Guiding question:** *what has to repeat before you may call a measure
consistent, and what does consistency still not buy you?*

- **Measurement reliability**: whether your instrument reads consistently when
  the measurement repeats, and it means nothing until you say *what* repeats.
  Repeat the **occasion**: ask the same people again later, when the construct
  should have held still. Repeat the **rater**: two trained coders score the
  same material and you compare. Repeat the **items**: for a scale with many
  questions about one construct, check that the questions agree with each other.
- **Split-half reliability**: the items version made concrete. Split the
  *items* into two comparable halves, score each half for every respondent, and
  check that the two halves agree. Example: a twelve-item scale gives each
  person an odd-item score and an even-item score, and high on one should mean
  high on the other. Two cautions ride along: the halves have to be comparable
  rather than odd-and-even by accident, and each half is only half as long as
  the real scale, so the raw agreement understates the full scale and a
  standard length correction closes the gap.

One warning saves a wasted afternoon: **you split items, never people.**
Comparing the average of one half of your *respondents* against the other half
tells you how much a group average wobbles between samples of people, which is
a real thing to know and a different question entirely. It says nothing about
whether your instrument reads consistently.

**Commit first:** the cell below runs both checks on the three-item trust
scale. Before you run it, predict which will look more reassuring, the items
check or the respondent split, and write one sentence on which one actually
speaks about the instrument.

```python
# Reliability done on ITEMS, and the respondent-splitting mistake, side by side.
trust_items = ["trust_police", "trust_military", "trust_supreme_court"]

# (a) The items check: do the scale's items agree with each other across people?
r_pm = lapop["trust_police"].corr(lapop["trust_military"])
r_pc = lapop["trust_police"].corr(lapop["trust_supreme_court"])
r_mc = lapop["trust_military"].corr(lapop["trust_supreme_court"])
mean_r = float(np.mean([r_pm, r_pc, r_mc]))
print("Agreement among the instrument's ITEMS (pairwise correlations):")
print(f"  police & military        : {r_pm:.2f}")
print(f"  police & supreme court   : {r_pc:.2f}")
print(f"  military & supreme court : {r_mc:.2f}")
print(f"  average item agreement   : {mean_r:.2f}")

# (b) The mistake: split RESPONDENTS in half and compare group means.
split_rng = np.random.default_rng(SEED)   # fresh generator: this cell stands alone
scale_score = lapop[trust_items].mean(axis=1)
order = split_rng.permutation(len(scale_score))
half = len(scale_score) // 2
gap = abs(scale_score.iloc[order[:half]].mean() - scale_score.iloc[order[half:]].mean())
print(f"\nSplit 10,000 RESPONDENTS in half instead: half-vs-half gap = {gap:.3f} points")

assert 0.35 < mean_r < 0.50, "item agreement drifted — investigate"
assert gap < 0.05, "the respondent-split gap should be tiny"
print("\n✓ Self-check passed. The items agree moderately (about 0.42): three questions")
print("  tracking one construct, noisily. The tiny respondent-split gap is NOT")
print("  reliability: it says a 5,000-person average is stable, which is a fact about")
print("  sampling. Items, never people.")
```

**Reading the output.** The items check says something about the instrument:
the three trust questions agree with each other at about 0.42 on average, which
reads as one construct measured noisily, and that is a verdict you can plan
around. The respondent split says nothing about the instrument at all, yet it
produces the far more flattering number, a gap of about 0.002. That is the trap
in full view: the wrong check looks more reassuring than the right one. If your
indicator is a single item measured once, say so honestly; "no defensible
reliability check is available for this measure and use" is a finding that
narrows your claim, not a blank to fill with the respondent split.

> **A question that often comes up here:** *"If my instrument is reliable,
> isn't it valid?"* No, and the bathroom scale is the standing counterexample:
> it reads your weight the same five times in a row, and weight is still a poor
> stand-in for fitness. Reliability says the number holds still. **Validity**
> asks whether the interpretation you put on that number, for the use you have
> in mind, is supported by evidence. Validity is never a stamp the instrument
> owns; it belongs to a claim, and changing the reading or the use means you
> need evidence again.

## 8. Version the Contract

**Guiding question:** *your measurement just got settled or changed; what does
your written record owe the reader?*

Your Contract v0 carried a measurement plan marked *provisional* on purpose.
This week settles it, and settling is a recorded event, because your
**estimand**, the exact quantity your analysis is trying to learn, is defined
through your indicator. Change the measure and the estimand quietly refers to
something else. That is why the rule is absolute: **any measurement change is a
Contract version**, a new dated, numbered entry with its reason, never a silent
edit of the last one.

A workable **Contract change log** line has four parts: the new version number,
the date you issue it, what changed, and why, written so a reader could
reconstruct your thinking. One worked example, from the workplace study above:

> **Contract v1** (dated the day you issue it). **Changed:** the engagement
> indicator settled from "provisional: an engagement survey, instrument to be
> chosen" to "the mean of one 1-to-7 recommend-this-workplace item." **Why:**
> the item was available, piloted, and repeatable, and its meaning sentence and
> items-based reliability plan are attached. **The estimand now refers to**
> willingness to recommend, one facet of engagement, not the whole concept.

One more line belongs in the version: **measurement error is uncertainty**.
State it alongside your sampling uncertainty, not instead of it. A perfectly
computed interval around a poorly measured quantity is still a poorly measured
quantity.

### A measurement your own instrument manufactured

Some numbers are mostly **artifact**, a value produced by the setup rather than
by the construct. Two artifacts do most of the damage in human measurement: a
**demand effect**, where the people being measured guess what result you want
and shift toward it, and an **instrument effect**, where the device or the
question itself moves the reading. Neither belongs to this week's ladder, and
both are worth seeing once, because they are how a perfectly consistent
instrument reads perfectly wrong.

**🏠 Optional depth.** Run this on your own if you want to go deeper.
**Before you ask:** the construct's true value below is 5.0 on a 0-to-10 scale,
and the naive design carries both artifacts. Write one number for where the
naive measurement lands before you run anything.

> 💡 **AI Prompt:** "Here is a simulation that measures a construct three ways:
> the true value, a naive design with a demand effect and an instrument effect,
> and a redesign that removes both: [paste the next cell]. Explain, step by
> step, which line introduces the demand effect and which introduces the
> instrument effect, and confirm that the redesign removes exactly those two
> terms. Then propose one additional redesign move I did not include."
>
> **After running, verify (counters *plausible-but-wrong-method*: a redesign
> that sounds principled can leave the artifact untouched):**
> - [ ] Confirm the naive and redesigned numbers it discusses match your
>   printout, and that the bias it names decomposes into demand plus
>   instrument, not an invented third term.
> - [ ] Check its extra redesign move: does it actually cut an artifact, or
>   just add noise? A move that touches neither artifact protects nothing.
> - [ ] Log this use in your AI Research Ledger: task, tool, decision, verification.

```python
# OPTIONAL: declare -> diagnose -> redesign a measurement (seed 464).
rng_meas = np.random.default_rng(SEED)
N = 1200
TRUE_MU     = 5.0     # the TRUE value of the construct on a 0-to-10 scale
DEMAND      = 1.2     # respondents shift UP toward the guessed hypothesis
INSTRUMENT  = 0.8     # a leading scale reads UP regardless of the construct
NOISE       = 2.0

# DECLARE: we want TRUE_MU. The naive reading = construct + demand + instrument + noise.
naive      = TRUE_MU + DEMAND + INSTRUMENT + rng_meas.normal(0, NOISE, N)
# REDESIGN: blind the respondents (cuts demand) + neutral scale (cuts instrument).
redesigned = TRUE_MU + rng_meas.normal(0, NOISE, N)

naive_est, redes_est = naive.mean(), redesigned.mean()
bias = naive_est - TRUE_MU

print("DECLARE : measure a construct whose true value is", TRUE_MU, "on a 0-to-10 scale.")
print("DIAGNOSE: the naive design (demand + instrument both active):")
print(f"          naive measurement      = {naive_est:.2f}")
print(f"          bias above the truth   = {bias:+.2f}   "
      f"(demand {DEMAND:+.1f} + instrument {INSTRUMENT:+.1f} = {DEMAND+INSTRUMENT:+.1f})")
print("REDESIGN: blind respondents (cut demand) + neutral scale (cut instrument):")
print(f"          redesigned measurement = {redes_est:.2f}   "
      f"(off the truth by {redes_est - TRUE_MU:+.2f})")

assert naive_est > TRUE_MU + 1.5, "naive design should read well above the truth"
assert abs(redes_est - TRUE_MU) < 0.4, "redesigned measurement should land near the truth"
assert naive_est - redes_est > 1.5, "the redesign should remove most of the artifact"
print(f"✓ Self-check passed: naive reads {naive_est:.2f} (bias {bias:+.2f});")
print(f"  redesigned reads {redes_est:.2f}, within {abs(redes_est - TRUE_MU):.2f} of 5.0.")
print("  Run the naive design again and it reads high again, consistently.")
print("  Consistent and wrong is a real combination.")
```

**Reading the output.** The naive design reported about 7.0 for a construct
whose true value is 5.0, and the entire gap is artifact: a demand effect near
1.2 plus an instrument effect near 0.8. The redesign, blinding plus a neutral
scale, landed near 5.0. Say it in validity language: the interpretation "this
population sits near 7 on the construct" was never supported, because the
number was construct plus demand plus instrument. Notice what the naive
instrument had going for it. Run it again and it reads high again, reliably.
That is the bathroom scale in simulation form.
