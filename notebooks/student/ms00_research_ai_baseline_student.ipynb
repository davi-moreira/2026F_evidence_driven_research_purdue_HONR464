you could actually obtain. "Price records showing menus tracked the overall
index" is a real answer. "Better data" is not.

## 2. The stretch: your candidates, with everything NEW marked

Widen the curiosity before you narrow it, and keep the trail.

Write **two candidate directions of your own first**. Then run the brainstorm
below and mark every element the tool introduced as **NEW**: any angle,
population, comparison, or variable that was not already in your own two. Then
list what you **kept** and what you **rejected**, each with its reason in one
sentence. "Rejected the second-town comparison: no access to its price records"
is a reason. "Not interesting" is not.

Your own two come first for a reason. Open the tool before you have written
anything down and its framing quietly becomes your framing, which makes every NEW
mark meaningless.

> 💡 **AI Prompt:** "I am scoping a one-semester research project. My
> curiosity is [paste your curiosity line], and here are the two directions I
> wrote myself, before asking you: (1) [first], (2) [second]. Widen this: propose
> six more candidate directions inside the same curiosity, and for each one name
> the angle, population, comparison, or variable that my own two did not already
> contain. Do not rank them and do not choose one for me."
>
> **After running, verify (counters *sycophantic agreement*: a partner that calls
> your own two excellent and adds nothing has flattered you, not stretched you):**
> - [ ] Mark as **NEW** every element that was not in your own two, in your own
>   words. If you cannot say what is new about a candidate, it is not new.
> - [ ] Keep or reject each candidate with a one-sentence reason you would defend
>   aloud. Access, time, and skills are reasons; taste is not.
> - [ ] Run the ownership test on anything you keep: would you still chase this if
>   every AI tool disappeared tomorrow? If not, go back one step.
> - [ ] Log this exchange in your AI Research Ledger below: task, tool, prompt,
>   output, decision, verification method, remaining concern.

```python
# The stretch — your own two FIRST, then what the brainstorm added.
# This list is the trail your grade reads, so keep the rejections in it.

OWN = [   # write these BEFORE you open any AI tool
    "EDIT ME: your own candidate direction 1",
    "EDIT ME: your own candidate direction 2",
]

# One dict per candidate that came back from the brainstorm.
#   new    True if it introduced an angle, population, comparison, or variable
#          that was not already in your own two
#   keep   True if you carry it forward, False if you reject it
#   reason one sentence, in your words, for that verdict
BRAINSTORMED = [
    {"candidate": "EDIT ME: a candidate the brainstorm proposed",
     "new": True, "keep": False, "reason": "EDIT ME: why you rejected it"},
    {"candidate": "EDIT ME: another candidate the brainstorm proposed",
     "new": True, "keep": True, "reason": "EDIT ME: why you kept it"},
]

print(f"Your own directions, written first ({len(OWN)}):")
for i, line in enumerate(OWN, 1):
    print(f"   {i}. {line}")
print()
print("The stretch, with every NEW element marked:")
for c in BRAINSTORMED:
    mark = "NEW" if c["new"] else "already mine"
    verdict = "KEPT" if c["keep"] else "REJECTED"
    print(f"   [{mark:>12}] {verdict:<8} {c['candidate']}")
    print(f"                  reason: {c['reason']}")
print()

kept = sum(1 for c in BRAINSTORMED if c["keep"])
thin = [c for c in BRAINSTORMED if "edit me" in c["reason"].lower()]
print(f"Kept: {kept}   Rejected: {len(BRAINSTORMED) - kept}")
if thin:
    print(f"⚠️ {len(thin)} verdict(s) still carry no reason. One sentence each, in your words.")
else:
    print("✓ Every candidate carries a verdict and a reason. That trail is the graded part.")
```
