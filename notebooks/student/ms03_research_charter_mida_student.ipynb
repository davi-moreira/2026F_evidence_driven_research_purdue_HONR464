## 2. The model behind the fields, and its DAG

State, in a short paragraph, **what the world would have to be like for your
objective to have an answer.** Name the **units**, the thing that can **vary**
across them, the **outcome**, and the other forces that could plausibly move that
outcome. This is a commitment to how the world *could* be, not yet a claim about
which picture is true.

Then draw that world as a **DAG**: a directed acyclic graph, a diagram where a
node is a variable and an arrow claims that one variable could directly influence
another. Yours needs **three nodes**: your treatment (or the thing that varies),
your outcome, and at least one **confounder**, a common cause with an arrow into
*both*. For each arrow, be ready to name the real-world mechanism it asserts.

> **A question that often comes up here:** *"My question is descriptive, and there
> is nothing to intervene on. Do I still draw a DAG?"* Yes, and the third node
> still earns its place. For a descriptive question the diagram shows what drives a
> unit into your data and what moves the thing you measure, and the confounder
> becomes the trait that bends both. A picture with only two nodes silently assumes
> nothing else moves the outcome, which is the assumption most descriptions get
> wrong.

> 💡 **AI Prompt:** "I am building a 3-node DAG for my project. My
> treatment, or the thing that varies, is [describe it], my outcome is [describe
> it], and my units are [describe them]. First, propose one **confounder**: a
> third variable that could be a common cause of *both*, and name the concrete
> mechanism by which it influences each. Second, phrase my inquiry as a single
> **estimand**: one sentence naming the exact quantity and the units it averages
> over. Do not decide my design for me."
>
> **After running, verify (counters the *plausible-but-wrong confounder*: a
> variable merely correlated with your outcome is not a confounder, and treating it
> as one aims your fix at the wrong arrow):**

>   arrow into BOTH the treatment and the outcome? If it only points at the
>   outcome, reject it or relabel it.
> - [ ] Read the estimand next to your plain-words inquiry. Does it name the same
>   quantity and the same units you meant, or did it quietly swap in an easier one?
> - [ ] Confirm the label matches the recipe yourself: does your number subtract

```python
# DAG scaffold — edit the three labels to match YOUR project, then run.
# These three names are the single source of truth for the diagram: rename them
# and every arrow, label, and position rebuilds itself. Export the figure and
# drop it into your Contract.

treatment  = "Treatment\n(EDIT ME)"
outcome    = "Outcome\n(EDIT ME)"
confounder = "Confounder\n(EDIT ME)"

G = nx.DiGraph()
G.add_edge(treatment, outcome)      # the relationship you WANT to estimate
G.add_edge(confounder, treatment)   # the confounder steers who gets treated ...
G.add_edge(confounder, outcome)     # ... and moves the outcome on its own

pos = {treatment: (0, 0), outcome: (2, 0), confounder: (1, 1.15)}
effect_edges   = [(treatment, outcome)]
backdoor_edges = [(confounder, treatment), (confounder, outcome)]

fig, ax = plt.subplots(figsize=(9, 4.5))
nx.draw_networkx_nodes(G, pos, node_size=5200, node_color="#e8eef6",
                       edgecolors="#33475b", linewidths=1.5, ax=ax)
nx.draw_networkx_labels(G, pos, font_size=9, ax=ax)
nx.draw_networkx_edges(G, pos, edgelist=effect_edges, arrowstyle="-|>",
                       arrowsize=20, width=2.0, edge_color="#1a7f37",
                       node_size=5200, ax=ax)
nx.draw_networkx_edges(G, pos, edgelist=backdoor_edges, arrowstyle="-|>",
                       arrowsize=20, width=1.6, edge_color="#b3261e",
                       style="dashed", node_size=5200,
                       connectionstyle="arc3,rad=0.08", ax=ax)
ax.set_title("Your model: green is the relationship you want, "
             "red dashed is the confounder's two arrows", fontsize=11)
ax.axis("off")
plt.show()

print("Now name the mechanism each arrow asserts, in one clause each.")
```

## 4 and 5. The diagnosis, and the redesign record

A design earns a claim only when it is **diagnosed**, not merely argued for. Build
a world where you set the true answer yourself, run the design many times on it,
and read three numbers:

- **Bias**: the average of your estimate minus the truth across runs, **counting
  direction**. A tilt that more data cannot fix. Report it signed, because
  "biased" alone does not say whether you are over- or under-stating.
- **Wobble**: how wide the spread of estimates is from run to run. More data
  shrinks it, and so do sharper measurement and better-balanced groups.
- **Detection**: how often the design would clear the decision rule you declared,
  when the effect you assumed is really there. Power is a property of a whole
  triple: the **statistical test**, its threshold, and the assumed effect. Write
  the test and the threshold down or the number means nothing.

Then **change exactly one thing**, the thing your diagnosis points at, and
diagnose again. More units for a wobble or detection problem; a changed
comparison, assignment, or sampling frame for a bias problem. "Collect more data"
for a confounded design is the classic wrong fix, and it buys a very precise
estimate of the wrong quantity. Record both diagnoses side by side, then write
**the honest call**: run it as redesigned, redesign again, or narrow the claim.

> **A question that often comes up here:** *"My diagnosis says the design will
> usually miss. Do I have to change the question?"* No. Weak data never change what
> a question asks. What you write is that the question is causal and **not yet
> identified by this design**, which is an honest state and a real answer.
> Shrinking the question to fit the data you happen to have is the move to avoid.

```python
# Diagnosis and redesign — declare a design shaped like YOURS, then read its
# signed bias, its wobble, and how often it clears your declared decision rule.
# Seeded, so every number here reruns identically.

DESIGN = {
    "assumed_effect": 2.0,      # the effect you are assuming is really there
    "n_per_group": 15,          # units in each condition
    "outcome_noise": 8.0,       # unit-to-unit wobble in the outcome
    "confounder_strength": 6.0,  # how hard the confounder pushes on the outcome
    "assignment": "randomized",  # "randomized" (a coin flip) or "self-selected"
}

# EDIT ME — change exactly ONE thing, the one your diagnosis points at.
REDESIGN = dict(DESIGN, n_per_group=60)

# EDIT ME — your declared decision rule. Power means nothing without it.
DECLARED_TEST = "two-sided difference in means, |t| > 1.96 (about the 0.05 level)"
THRESHOLD = 1.96

def diagnose(design, reps=1500, seed=SEED, threshold=THRESHOLD):
    d = np.random.default_rng(seed)          # own reproducible stream
    n = design["n_per_group"] * 2
    ests, cleared = np.empty(reps), np.zeros(reps, dtype=bool)
    for r in range(reps):
        prior = d.normal(0, 1, n)                                  # the confounder
        Y0 = 50 + design["confounder_strength"] * prior + d.normal(0, design["outcome_noise"], n)
        Y1 = Y0 + design["assumed_effect"]
        if design["assignment"] == "randomized":
            Z = np.zeros(n, dtype=int)
            Z[d.permutation(n)[:n // 2]] = 1                        # coin flip: half treated
        else:
            Z = (d.random(n) < 1 / (1 + np.exp(-1.5 * prior))).astype(int)
        Y = np.where(Z == 1, Y1, Y0)                                # observe one world only
        est = Y[Z == 1].mean() - Y[Z == 0].mean()                   # answer strategy
        se = np.sqrt(Y[Z == 1].var(ddof=1) / max((Z == 1).sum(), 1)
                     + Y[Z == 0].var(ddof=1) / max((Z == 0).sum(), 1))
        ests[r], cleared[r] = est, se > 0 and abs(est) / se > threshold
    return {"bias": ests.mean() - design["assumed_effect"],
            "wobble": ests.std(ddof=1), "detection": cleared.mean()}

declared, redesigned = diagnose(DESIGN), diagnose(REDESIGN)

print(f"Declared decision rule: {DECLARED_TEST}")
print(f"{'':<12}{'DECLARED':>12}{'REDESIGNED':>14}")
FORMATS = {"bias": lambda v: f"{v:+.3f}",      # signed: direction is the point
           "wobble": lambda v: f"{v:.3f}",     # a spread, so never signed
           "detection": lambda v: f"{v:.1%}"}
for k, label in [("bias", "bias (signed)"), ("wobble", "wobble"),
                 ("detection", "detection")]:
    fmt = FORMATS[k]
    print(f"{label:<12}{fmt(declared[k]):>12}{fmt(redesigned[k]):>14}")
print()
print("Bias is the average of estimate minus truth, counting direction: a lean no")
print("extra data can fix. Wobble shrinks with more units. Detection is how often")
print("this design clears the rule above when the assumed effect is really there.")
print("\nName the worst of the three for YOUR design, say what it does to your")
print("answer, and write the honest call: run it, redesign again, or narrow the claim.")
```
