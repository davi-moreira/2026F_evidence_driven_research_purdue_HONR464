## 🎯 Definition of Done

You are done with this sprint when your submission carries all nine parts,
applied to **your own project**. Verification here is five checks, not one:
the restart must match, every reported number must trace to its cell, the
uncertainty itself must reproduce, two key numbers must survive re-derivation,
and the required auditor must find no leak and no unlicensed generalization.

1. **The pipeline.** Seeded (`SEED = 464`), running top to bottom from data to
   result. Three rules govern it: it implements the analysis you **declared**
   and nothing else, it loads data from the source your M5 governance record
   documents, and every number it reports traces to a data cell and a line of
   code. Delegate the writing of code as freely as you like, then run the two
   checks that cover it: the **line review** (read every returned line and say
   in your own words what it does; any line you cannot explain gets rewritten
   or removed) and the **known-answer test** (run a piece of code on a tiny
   input whose right answer you already know). Both live next to the code they
   verify, and each earns a ledger row.
2. **The result, with its uncertainty.** The one result your declared analysis
   produces, in the form your route and Contract specified, with the interval,
   spread, standard error, or held-out range your route provides. If your
   route predicts, report held-out performance beside the declared baseline it
   has to beat. Then read the number in one sentence: what it says, for which
   units, and what it does not cover.
3. **The clean restart and the environment record.** Restart your runtime so
   memory is empty, then Restart & Run All. Record the environment (one cell
   printing your Python version, your library versions, and your data files
   with the date each was retrieved), the headline numbers before and after,
   and the verdict. "Close enough" is not a category: a moved digit has a
   cause, and finding it is a real result about your own evidence machine.
4. **The claim-to-output trace.** A table pointing every number you plan to
   report to the specific cell that produces it: the sentence, the number, the
   cell. A figure with no cell behind it is an **untraceable number**, and it
   caps your Verification score.
5. **Two independent re-derivations, against a tolerance you declare first.**
   Two key numbers, each reached twice by unrelated paths: by hand from a
   small table, with a different library, or from the raw definition written
   fresh in a new cell. Write your **tolerance** as a number before you run
   the second route ("agreement means matching to three decimal places"),
   because a tolerance chosen after you see the gap always turns out wide
   enough.
6. **The leakage audit.** The **Prediction & Leakage Auditor** run (the
   required GenAI Studio reviewer for M6). On a prediction route it hunts
   **data leakage**, a feature whose value is only settled at or after the
   outcome it is supposed to predict. On every other route it hunts
   out-of-sample or generalization claims in your language. Every flag fixed
   or refuted by a check in your own pipeline.
7. **The URC abstract, gated.** Roughly 150 to 250 words describing your
   project as it now stands: the question, what you measure, the analysis your
   pipeline executes, the verified result with its uncertainty, and the
   boundary around the claim. The gate has one rule: the abstract may use only
   the claims your current verified result licenses.
8. **AI Research Ledger rows.** One row for every AI-assisted step, each with
   a named verification method.
9. **The version line and the dossier update line.** The artifact opens with
   **Book Milestone 7, version 1 (first reproducible analysis)**, dated, with
   its reason; it ends with the dossier line naming where the verified
   notebook, its trace, and the gated abstract now live.

> **A question that often comes up here:** *"My numbers matched on the first
> restart. Am I done?"* A match is the entry ticket, not the milestone. A
> pipeline can reproduce a wrong number perfectly, run after run. The
> claim-to-output trace catches the number you misquoted in your prose, the
> re-derivation catches the bug both runs shared, and the leakage audit
> catches the sentence that quietly claims more than the pipeline computed.
> All five checks together are what verified means here.

```python
# Pipeline skeleton + known-answer harness — make this YOUR pipeline.
# The stand-in loader below simulates data of a declared shape so the cell
# runs; your job this week is to replace it with a loader that reads the
# source your M5 governance record documents, and to make declared_analysis()
# implement EXACTLY the analysis your Contract declares. Nothing else.

def load_data(rng):
    # EDIT ME — replace with your documented M5 source. The stand-in draws a
    # two-group dataset so the skeleton runs end to end.
    group = np.repeat([0, 1], 60)
    outcome = np.where(group == 1, 3.4, 3.0) + rng.normal(0, 0.9, size=120)
    return group, outcome

def declared_analysis(group, outcome, rng, n_boot=2000):
    # EDIT ME — your declared analysis, and nothing else. The stand-in:
    # difference in group means with a seeded bootstrap interval.
    diff = outcome[group == 1].mean() - outcome[group == 0].mean()
    boots = []
    for _ in range(n_boot):
        idx = rng.integers(0, len(outcome), size=len(outcome))
        g, y = group[idx], outcome[idx]
        if g.min() == g.max():
            continue  # a resample with one group cannot compute the contrast
        boots.append(y[g == 1].mean() - y[g == 0].mean())
    lo, hi = np.percentile(boots, [2.5, 97.5])
    return diff, (lo, hi)

group, outcome = load_data(rng)
diff, (lo, hi) = declared_analysis(group, outcome, rng)
print(f"Result: {diff:.3f}  (95% interval {lo:.3f} to {hi:.3f})")
print("Read it in one sentence: what it says, for which units, what it does not cover.")

# --- Known-answer test: a tiny input you can work by hand -------------------
tiny_group   = np.array([0, 0, 0, 1, 1, 1])
tiny_outcome = np.array([1.0, 2.0, 3.0, 2.0, 4.0, 6.0])   # means 2 and 4
tiny_diff, _ = declared_analysis(tiny_group, tiny_outcome,
                                 np.random.default_rng(SEED))
assert abs(tiny_diff - 2.0) < 1e-12, "known-answer test FAILED"
print("✓ Known-answer test passed: hand-computed 4 − 2 = 2, function returns",
      tiny_diff)
print("  Now replace both EDIT ME bodies with your own project's source and")
print("  declared analysis, and rerun top to bottom.")
```

```python
# Verification toolkit — the five checks, in the order you run them.
# Edit the EDIT ME blocks against YOUR pipeline: (1) the environment record,
# (2) the before/after restart comparison, (3) the tolerance you DECLARE
# before the second route runs, (4) two independent re-derivations.
import sys
import matplotlib

print("— 1. Environment record —")
print(f"  Python {sys.version.split()[0]} · numpy {np.__version__} · "
      f"matplotlib {matplotlib.__version__}")
# EDIT ME — one line per data file, with the date you retrieved it.
data_files = {"EDIT ME (data file)": "EDIT ME (retrieval date)"}
for f, when in data_files.items():
    print(f"  data: {f} · retrieved {when}")

# EDIT ME — your headline numbers as reported BEFORE the restart, then run
# Restart & Run All and fill in what printed AFTER. Use the digits you report.
before = {"headline": 0.380, "interval_lo": 0.050, "interval_hi": 0.710}
after  = {"headline": 0.380, "interval_lo": 0.050, "interval_hi": 0.710}

print("— 2. Before / after the clean restart —")
stable = True
for k in before:
    match = bool(np.isclose(before[k], after[k], atol=0.0005))
    stable = stable and match
    print(f"  {k}: {before[k]:.3f} → {after[k]:.3f}  "
          + ("✓ match" if match else "⚠️ MOVED — find the cause before any claim goes forward"))
print("  Verdict:", "stable to the reported digit" if stable
      else "NOT stable: hidden state, an out-of-order cell, or an unseeded draw")
print("  The interval reproduces too, not just the point estimate.")

# EDIT ME — DECLARE the tolerance now, before you look at any gap.
tolerance = 0.001   # "agreement means the two routes match to three decimals"
print(f"— 3. Tolerance declared BEFORE the runs: {tolerance} —")

print("— 4. Two independent re-derivations —")
# Number 1: the pipeline's route (numpy group means) against plain arithmetic.
g = np.array([0, 0, 0, 1, 1, 1]); y = np.array([1.0, 2.0, 3.0, 2.0, 4.0, 6.0])
pipeline_1 = y[g == 1].mean() - y[g == 0].mean()
independent_1 = (2.0 + 4.0 + 6.0) / 3 - (1.0 + 2.0 + 3.0) / 3
# Number 2: a share, computed by the pipeline's route and from the raw definition.
flags = np.array([1, 0, 1, 1, 0, 0, 1, 0])
pipeline_2 = flags.mean()
independent_2 = sum(1 for v in flags if v == 1) / len(flags)

for label, a, b in [("number 1 (group difference)", pipeline_1, independent_1),
                    ("number 2 (share flagged)", pipeline_2, independent_2)]:
    gap = abs(a - b)
    verdict = "✓ agree inside the declared tolerance" if gap <= tolerance \
        else "⚠️ DISAGREE — you just found the bug this milestone exists to catch"
    print(f"  {label}: {a:.4f} vs {b:.4f} · gap {gap:.4f} · {verdict}")
print("  Imitate this for TWO of your own key numbers, on routes that share")
print("  none of the original code, and keep the tolerance you wrote first.")
```

**Round 2: the Auditor attacks your claims.**
Submit your verified draft to the **Prediction & Leakage Auditor**, the
required GenAI Studio reviewer for M6 (full briefing in
`genai_studio/roles/prediction_leakage_auditor.md`). On a prediction route it

each flag with the **timing check** (is the value known before the prediction

check is dropped or re-timed, however much it helps your score. On every other
route the Auditor hunts sentences that quietly extend your result to units,
times, or populations your design never reached; every flagged sentence gets
bounded or cut, and a confirmed "I make no unlicensed prediction claim" is
itself a ledgered result. Write your fix or your refutation for every flag,
each settled by a check in your own pipeline.

**Round 3: attack the abstract at the gate.**
Read your abstract aloud against the gate's one rule: it may use only the
claims your current verified result licenses. Hunt the three failure modes by
name: a causal promise your route does not license, a reach to units or
populations your data never touched, and a number stated as settled that no
stress test has touched yet.
