```python
# Pipeline skeleton + known-answer harness — make this YOUR pipeline.
# The stand-in loader below simulates data of a declared shape so the cell
# runs; your job this week is to replace it with a loader that reads the
# source your M5 governance record documents, and to make declared_analysis()
# implement EXACTLY the analysis your Contract declares. Nothing else.

def load_data(rng):
    # EDIT ME — replace with your documented M5 source. The stand-in draws a
    # two-group dataset so the skeleton runs end to end.
    group = np.repeat([0, 1], 60)
    outcome = np.where(group == 1, 3.4, 3.0) + rng.normal(0, 0.9, size=120)
    return group, outcome

def declared_analysis(group, outcome, rng, n_boot=2000):
    # EDIT ME — your declared analysis, and nothing else. The stand-in:
    # difference in group means with a seeded bootstrap interval.
    diff = outcome[group == 1].mean() - outcome[group == 0].mean()
    boots = []
    for _ in range(n_boot):
        idx = rng.integers(0, len(outcome), size=len(outcome))
        g, y = group[idx], outcome[idx]
        if g.min() == g.max():
            continue  # a resample with one group cannot compute the contrast
        boots.append(y[g == 1].mean() - y[g == 0].mean())
    lo, hi = np.percentile(boots, [2.5, 97.5])
    return diff, (lo, hi)

group, outcome = load_data(rng)
diff, (lo, hi) = declared_analysis(group, outcome, rng)
print(f"First result: {diff:.3f}  (95% interval {lo:.3f} to {hi:.3f})")
print("Label, in writing: PROVISIONAL pending the clean-restart verification (M7).")

# --- Known-answer test: a tiny input you can work by hand -------------------
tiny_group   = np.array([0, 0, 0, 1, 1, 1])
tiny_outcome = np.array([1.0, 2.0, 3.0, 2.0, 4.0, 6.0])   # means 2 and 4
tiny_diff, _ = declared_analysis(tiny_group, tiny_outcome,
                                 np.random.default_rng(SEED))
assert abs(tiny_diff - 2.0) < 1e-12, "known-answer test FAILED"
print("✓ Known-answer test passed: hand-computed 4 − 2 = 2, function returns",
      tiny_diff)
print("  Now replace both EDIT ME bodies with your own project's source and")
print("  declared analysis, and rerun top to bottom.")
```
