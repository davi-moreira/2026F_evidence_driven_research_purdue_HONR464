## 1. Two Streams of Feedback, Three Verdicts

**Guiding question:** *your claim was tested in public and rerun by a stranger;
what do you owe each piece of feedback that came back?*

> *"Do not show me a note that changed everywhere the criticism was loud. Show
> me a note that changed exactly where the evidence demanded it, and a record
> that tells me why the rest stayed."*
> — a **thesis advisor**, opening the revision meeting

Since your note was locked, two streams of evidence arrived that no self-check
can produce. The **public test** is what strangers asked and objected to at the
poster session: real readers, no coaching, reacting to the claim itself. The
**cold run** is what a peer found rerunning your package alone: every place your
materials made them stop, and every number that did not come back as written.
Both streams are evidence about your work. Neither comes with its verdict
attached.

**Feedback adjudication** is the discipline of giving every criticism an
explicit verdict backed by a check, instead of an instinct backed by a mood.
Three verdicts, and only three:

- **Accept**: the criticism is right, so the note changes, and the record says
  what moved. Example: the write-up says 4 points, the code prints 3.41, so the
  write-up changes to the computed number.
- **Rebut**: the criticism does not survive a check, so the note stands, and the
  record carries a reason you own. Example: "this is not new" contests novelty,
  not evidence.
- **Bound**: the claim survives but shrinks, so the boundary moves and the
  caveat goes in the same sentence. Example: an honest interval that reaches
  zero retires the word "clearly."

> **A question that often comes up here:** *"Isn't accepting every criticism the
> humble thing to do?"* It is the lazy thing to do, and it is not humble. A note
> that softens everything the moment anyone objects is not sized to its
> evidence; it is sized to its critics. Humility here means running the check.
> If the check says the criticism is right, you change. If it says the criticism
> is wrong, you keep the sentence and write down why. Both verdicts take the
> criticism more seriously than reflexive agreement ever could.

### 🛠️ Run the Study: adjudicate against printed numbers

Run the two cells below. The first prints the **evidence record**: the numbers
this note's earlier runs already produced, each carried forward with the studio
that produced it. Example: the headline gap of 3.41 points travels with the
first-analysis studio attached to it, so anyone can go back and read the run.
Adjudication reuses that record; it does not compute the note again. **Expect**
five lines, a headline of 3.41 points against a poster that printed 4, a
weighted alternative sitting lower, and two intervals that disagree about
whether zero is in range. The second cell prints the adjudication table and
self-checks the four numbers the verdicts lean on.

```python
# The evidence record for the sample note: every number the five criticisms
# touch, carried forward from the run that produced it. Nothing is recomputed
# here. Adjudication reuses printed evidence; re-deriving it is a different job,
# and it was already done and checked at the studios named in the last column.
diff_pts = 3.41       # headline gap, unweighted, in percentage points
naive_lo, naive_hi = 1.00, 5.81      # 95% interval, people treated as independent
diff_w_pts = 2.83     # gap recomputed with the shipped weights
cluster_lo, cluster_hi = -0.07, 6.89  # 95% interval, whole streets resampled
poster_pts = 4.00     # the number the poster printed
n_streets = 613       # streets the package assigned canvassing by

record = pd.DataFrame([
    ["headline gap, unweighted", f"{diff_pts:.2f} points",
     "first-analysis studio: the estimate, with a standard error of 1.23"],
    ["95% interval, people independent", f"[{naive_lo:.2f}, {naive_hi:.2f}]",
     "first-analysis studio: the uncertainty statement shipped with the note"],
    ["gap with the shipped weights", f"{diff_w_pts:.2f} points",
     "reproduce-and-package studio: the weighting the write-up never disclosed"],
    ["95% interval, whole streets", f"[{cluster_lo:.2f}, {cluster_hi:.2f}]",
     f"reproduce-and-package studio: the cold run, resampling all {n_streets} streets"],
    ["the number on the poster", f"{poster_pts:.2f} points",
     "the poster itself: no run has ever printed it"],
], columns=["quantity", "as printed then", "where it was established"])
print(record.to_string(index=False))
```

**Reading the output.** Five lines printed, and each one belongs to a criticism.
The analysis produced 3.41 points, so the poster's 4 was a rounding the write-up
never earned. The weighted version is 2.83, lower, which means a choice nobody
disclosed was moving the number. The interval that treats people as independent
sits entirely above zero, while the one that resamples whole streets reaches down
to about zero. Notice the third column. None of this is a discovery today: every
line was printed and checked at the studio that produced it, and that is what
makes it usable as evidence now. Your job for the rest of this block is not to
find these numbers again. It is to decide what each criticism is owed now that
they sit on the table together.

```python
# The adjudication table: every verdict next to the check that decided it.
adjudication = pd.DataFrame([
    ["1. poster says 4, code prints 3.41", "public",
     f"the record holds {diff_pts:.2f}", "ACCEPT: report the computed number"],
    ["2. 'old news, nothing new'", "public",
     "no number in the note is contested", "REBUT: novelty is not the claim"],
    ["3. shipped weights never applied", "cold run",
     f"weighted gap {diff_w_pts:.2f} vs {diff_pts:.2f}", "ACCEPT: disclose the choice, report both"],
    ["4. street clustering ignored", "cold run",
     f"interval [{cluster_lo:.2f}, {cluster_hi:.2f}] reaches about zero",
     "BOUND: keep the estimate, gain the honest interval"],
    ["5. 'should have studied mail'", "public",
     "asks for a different study", "REBUT: a future direction, not a flaw"],
], columns=["criticism", "stream", "the check (printed evidence)", "verdict"])
print(adjudication.to_string(index=False))

# Self-check: the carried record actually supports each verdict. These test the
# relationships the verdicts lean on, which is what a reused record has to prove.
assert abs(poster_pts - diff_pts) > 0.05, "no gap between poster and record, so no accept"
assert diff_w_pts < diff_pts, "the weighting has to move the estimate for this to matter"
assert naive_lo > 0, "the people-independent interval should sit above zero"
assert cluster_lo < 0 < cluster_hi, "the street interval should reach across zero"
print(f"\n✓ Self-check passed: verdicts rest on printed numbers "
      f"({diff_pts:.2f}, {diff_w_pts:.2f}, and a lower bound of {cluster_lo:.2f}).")
```

**Reading the evidence.** Read the verdict column against the check column, and
notice what never appears: how loud the criticism was. The devastating-sounding
"old news" objection is a rebut, because no check contradicts any number in the
note, and the reason gets written down and owned. The quiet clustering line is
the week's biggest verdict, a bound: the estimate survives, but the claim must
now carry an interval that reaches about zero, so "clearly positive" is no
longer a sentence the evidence licenses. Criticism 1 is the cheapest accept in
research: the write-up moves to the number the code prints. Now compare all five
against the verdicts you committed before anything ran, and mark every mismatch.

## 2. The Adjudication Record and Claim–Evidence Table v2

**Guiding question:** *how does a reader tell a disciplined revision from a
quiet rewrite?*

The verdicts now become two artifacts, and both open your M14 submission.

The **adjudication record** is a dated account of the revision that opens the
note. For each substantive change it says what changed, which public question or
cold-run finding drove it, and what it cost or bought the claim. It also lists
every criticism you declined, each with the reason you own. Example: one line
records that the headline gained its street-respecting interval because of the
clustering finding; another records that the novelty objection was rebutted, and
why. Revision without a record is just editing. Revision with one is evidence
about your own work.

The **claim–evidence table v2** is the table your note already shipped with, one
row per claim (claim, evidence, trace, boundary), now updated so every revised
sentence re-traces to its row. When a claim moved, its row moved with it, and no
caveat falls off between versions. Here is the headline row before and after
this adjudication:

| version | claim | evidence | boundary |
|---|---|---|---|
| v0 row | "Canvassing raised turnout by 4 points." | the poster figure | none stated |
| v2 row | "Canvassing raised turnout by about 3.4 points in these wards (95% street-respecting interval, about 0 to 6.9)." | the rerun output above | interval reaches zero; weighting moves the estimate to 2.83 |

Now the naming rule, and it is worth saying slowly, because two ladders meet in
one submission. The course calls the note you are building **research note v1**,
and the book's ladder calls that same artifact **Book Milestone 9, version 2**.
One object, two labels, both on your submission. The package you repair
tomorrow is **Book Milestone 11, version 2**, because version 1 already exists
and the peer cold run is the external test that told you what to fix.

> **A question that often comes up here:** *"Why version numbers at all? It is
> just my note."* Because versions are how a claim's history stays inspectable.
> A reader who can compare the old headline with the new one, through the
> adjudication record, can see the system working: a public test and a cold run
> reached the work, and the claim moved exactly where the evidence pushed it. A
> quiet rewrite hides that history, and a claim whose history is hidden asks for
> trust it has not earned.

## 3. Cold-Run Repairs and the Five Package Parts

**Guiding question:** *what does a stranger need to rebuild your numbers, and
what did your cold runner's questions tell you is still missing?*

> *"Your peer asked you three questions before your package would run. That is
> the review. Each question is a line your documentation does not have, located
> for free. Write the lines, not the excuses."*
> — a **journal reviewer**, on what a cold run is for

A cold runner's question is answered **in the package**, never only in
conversation. The **repair log** is where that happens: every question your peer
had to ask becomes a line in the package, with the file and the sentence that
now answers it. Repairs land in one of the **five package parts**, the anatomy
your M14 package ships:

- The **README**: the stranger's entry point. What this is, how to run it, and
  what result to expect, including the headline number and its uncertainty.
- The **data or its access recipe**: the files themselves where terms allow;
  otherwise the exact route a reader follows to obtain them, with version and
  provenance. An **access recipe** is the step-by-step path to data you may not
  ship, and your permissions decide which of the two your package carries.
- The **seeded code**: the notebook that passes restart-and-run-all with
  `SEED = 464`, no manual fixes.
- The **environment record**: the library versions a rerun needs, written down,
  not remembered.
- The **licence and permissions note**: what may be shared and reused, stated
  explicitly, with your decision log of by-hand choices and their reasons.

Your reading names the same bundle by its contents rather than by what a
stranger opens first: a runnable notebook, a data-provenance note, a fixed seed,
a decision log, and an AI-use ledger. Nothing is lost in the translation.
Provenance rides in the access recipe, the seed rides in the code, and the
decision log rides in the licence and permissions note. The five parts above are
the shipping names your milestone grades.

Packages break in boringly consistent ways, and the auditor below hunts the
**five package sins**: a **hard-coded path** that exists only on your laptop, a
**missing seed** that moves every run, a **by-hand edit** no clean run
reproduces, an **undocumented exclusion** with no logged reason, and **stale or
unversioned data** a reader cannot reobtain.

> **A question that often comes up here:** *"My peer already ran my package once.
> Why does this week need a package step at all?"* Because the cold run was the
> diagnosis, not the cure. Your peer's report is a list of exactly where your
> materials made a stranger stop. This week is where each stop becomes a repair,
> the five parts are reassembled around the fixed code, and the whole thing is
> proven again. A diagnosed package that ships unrepaired wasted the one
> stranger it will ever get for free.

### 🛠️ Run the Study: audit a package for the five sins

Run the three cells below. The first defines `audit_capsule`, a transparent
checker that scans source lines for the five sins. The second runs it on a
sinful capsule and a clean one. The third self-checks both results. **Expect**
the sinful capsule to trip all five checks, the clean one to trip none, and the
self-check to print the line the whole lecture turns on. Keep the auditor: your
own package's key lines go through it this week.

```python
# A tiny capsule auditor. Each check is one thing a stranger's eyes would catch.
# It scans a list of source LINES (strings) — paste your own package's key lines in.
def audit_capsule(lines):
    text = "\n".join(lines).lower()
    sins = []
    # Sin 1 — a hard-coded absolute path: dies on every machine but its author's.
    if any(("/users/" in ln.lower()) or ("/home/" in ln.lower()) for ln in lines):
        sins.append("hard-coded path")
    # Sin 2 — no fixed seed anywhere: any sample/split/shuffle moves every run.
    if not any(k in text for k in ("seed", "random_state", "default_rng")):
        sins.append("missing seed")
    # Sin 3 — a value patched by hand: no clean run will reproduce it.
    if any((".loc[" in ln and "=" in ln) or ("by hand" in ln.lower())
           or ("manually" in ln.lower()) for ln in lines):
        sins.append("by-hand edit")
    # Sin 4 — an exclusion/filter with no logged reason nearby.
    filtered = [ln for ln in lines
                if (".drop(" in ln) or (".dropna(" in ln)
                or ("df[df[" in ln.replace(" ", ""))]
    if filtered and not any("reason" in ln.lower() for ln in lines):
        sins.append("undocumented exclusion")
    # Sin 5 — data with no shareable provenance: a reader can't reobtain the inputs.
    loads_csv = any("read_csv" in ln for ln in lines)
    shareable = any(("http" in ln.lower()) or ("load_course_data" in ln) for ln in lines)
    versioned = any(("version" in ln.lower()) or ("license" in ln.lower())
                    or ("licence" in ln.lower()) for ln in lines)
    if loads_csv and not shareable and not versioned:
        sins.append("stale/unversioned data")
    return sins

print("✓ audit_capsule defined — five checks, one list of flags.")
```

```python
# A deliberately sinful capsule — every line breaks reproducibility on purpose.
sinful = [
    'df = pd.read_csv("/Users/rivera/Desktop/thesis/turnout_FINAL_v3.csv")',
    'draw = df.sample(200)                    # nothing fixes the shuffle',
    'df = df[df["ward"] == "F"]               # kept one ward only; no note on why',
    'df.loc[41, "age"] = 33                   # patched this row by hand post-survey',
    '# data pulled sometime last spring; nobody logged the download date',
]

# A clean capsule — same study, nothing a stranger would have to guess.
clean = [
    "SEED = 464",
    "rng = np.random.default_rng(SEED)",
    'df = load_course_data("foos_etal.csv")   # access recipe + licence in the README',
    'df = df[df["marked_register_2014"].notna()]   # reason: complete cases; in the decision log',
    "draw = df.sample(200, random_state=SEED)",
]

print("Sinful capsule flags:", audit_capsule(sinful))
print("Clean  capsule flags:", audit_capsule(clean))
```

```python
# Self-check: the sinful capsule trips all five checks, the clean one none.
assert audit_capsule(sinful) == ["hard-coded path", "missing seed", "by-hand edit",
                                 "undocumented exclusion", "stale/unversioned data"], \
    "auditor drifted — the sinful capsule should trip all five checks"
assert audit_capsule(clean) == [], "clean capsule should trip no checks"
print("✓ Self-check passed: 5 sins in the sinful capsule, 0 in the clean one.")
print("  Reminder: a clean pass proves the capsule is RUNNABLE, never that it is CORRECT.")
```

**Reading the evidence.** The sinful capsule trips all five checks; the clean
one trips none. But read the last printed line again, because it carries the
lesson of the whole package week. Zero flags means **runnable**, not
**correct**, and not even complete: the auditor reads lines, so it cannot see
that your README's step 2 names no data source, that a logged reason is a bad
reason, or that your seed is fixed on the wrong sample. Those are exactly the
failures a cold human run finds, which is why the puzzle's README sailed past
every automated check and still died at the first cell. The auditor gets your
package to the starting line. The stranger's run is the race.

## 4. The Second Clean Run

**Guiding question:** *after the repairs, what single piece of evidence lets the
package claim it is reusable?*

Every repair you just made has a side effect: it makes your previous cold-run
record obsolete. A **stale run** is a run record that predates any change to a
number, a claim, the data, the code, the environment record, or the instructions
a stranger would follow. The record vouches for a package that no longer exists.
So the week closes on one non-negotiable rule: **after your last change, run the
package clean once more.** A fresh environment, your written instructions only,
restart-and-run-all, and a dated record of the result. A changed package with a
stale run is not done, however good the changes were.

The **second clean run** must bring back two things, not one: the headline
number, and its **uncertainty statement**. An interval that does not reproduce
is a discrepancy to explain and fix, exactly like a moved point estimate. This
week that bites for real, because your adjudication just changed the headline's
interval, so a rerun that only checks the estimate would miss the very sentence
your revision most needs to vouch for.

Two runs are in play this week, and they are not the same event. Your own clean
rerun happens after tonight's last change to your package. Your partner's run at
Friday's studio is external, performed by someone who is not you. Whichever one
is newer than your last change is the record that ships.

> **A question that often comes up here:** *"The only thing I changed after my
> cold run was one README sentence. Surely that does not count?"* It counts,
> because the README is part of what a stranger executes: instructions are code
> that runs on people. If the sentence you changed altered no step, the rerun
> will be cheap and will say so on the record; if it did alter a step, the rerun
> is the only thing that would have caught it. Either way the rule holds,
> because the alternative is a judgment call made by the one person who cannot
> test it: the author, who already knows what every sentence means.
