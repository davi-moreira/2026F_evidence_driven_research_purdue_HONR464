## 1. Two Streams of Feedback, Three Verdicts

**Guiding question:** *your claim was tested in public and rerun by a stranger;
what do you owe each piece of feedback that came back?*

> *"Do not show me a note that changed everywhere the criticism was loud. Show
> me a note that changed exactly where the evidence demanded it, and a record
> that tells me why the rest stayed."*
> — a **thesis advisor**, opening the revision meeting

Since your note v0 was locked, two streams of evidence arrived that no
self-check can produce. The **public test** is what strangers asked and objected
to at the conference: real readers, no coaching, reacting to the claim itself.
The **cold run** is what a peer found rerunning your package alone: every place
your materials made them stop, and every number that did not come back as
written. Both streams are evidence about your work. Neither comes with its
verdict attached.

**Feedback adjudication** is the discipline of giving every criticism an
explicit verdict backed by a check, instead of an instinct backed by a mood.
Three verdicts, and only three:

- **Accept**: the criticism is right, so the note changes, and the change log
  records what moved and why. Example: the write-up says 4 points, the code
  prints 3.41, so the write-up changes to the computed number.
- **Rebut**: the criticism does not survive a check, so the note stands, and the
  change log records the reason you own. Example: "this is not new" is a
  complaint about novelty, not about whether the evidence supports the claim.
- **Bound**: the criticism does not kill the claim, but it shrinks what the
  claim may say, so the boundary moves and the caveat goes in the same
  sentence. Example: an honest interval that reaches zero turns "clearly
  positive" into a bounded sentence that names the interval.

> **A question that often comes up here:** *"Isn't accepting every criticism the
> humble thing to do?"* It is the lazy thing to do, and it is not humble. A note
> that softens everything the moment anyone objects is not sized to its
> evidence; it is sized to its critics. Humility here means running the check.
> If the check says the criticism is right, you change. If it says the criticism
> is wrong, you keep the sentence and write down why. Both verdicts take the
> criticism more seriously than reflexive agreement ever could.

```python
# Rebuild every number the five criticisms touch, from the package data.
pkg = load_course_data("foos_etal.csv")
assert pkg.shape == (8375, 5), "unexpected shape — flag this as a lineage break!"
print("✓ Loaded the package data —", pkg.shape[0], "rows,", pkg.shape[1], "columns")

y = "marked_register_2014"
canv, notc = pkg[pkg.treat == 1][y], pkg[pkg.treat == 0][y]
diff = canv.mean() - notc.mean()
diff_pts = diff * 100                     # criticism 1: the computed headline

w_canv = np.average(pkg.loc[pkg.treat == 1, y], weights=pkg.loc[pkg.treat == 1, "weights"])
w_notc = np.average(pkg.loc[pkg.treat == 0, y], weights=pkg.loc[pkg.treat == 0, "weights"])
diff_w_pts = (w_canv - w_notc) * 100      # criticism 3: the weighted alternative

# Criticism 4: the interval, naive (people independent) vs street-respecting.
n1, n0 = len(canv), len(notc)
naive_se = np.sqrt(canv.var(ddof=1) / n1 + notc.var(ddof=1) / n0)
naive_lo, naive_hi = (diff - 1.96 * naive_se) * 100, (diff + 1.96 * naive_se) * 100

tmp = pd.DataFrame({
    "street": pkg["street"],
    "yt": pkg[y] * (pkg.treat == 1), "nt": (pkg.treat == 1).astype(int),
    "yc": pkg[y] * (pkg.treat == 0), "nc": (pkg.treat == 0).astype(int),
})
per = tmp.groupby("street")[["yt", "nt", "yc", "nc"]].sum()
yt, nt = per["yt"].to_numpy(float), per["nt"].to_numpy(float)
yc, nc = per["yc"].to_numpy(float), per["nc"].to_numpy(float)
K = len(per)

rng_boot = np.random.default_rng(SEED)    # fresh seed: reproducible on its own
boot = np.empty(400)
for b in range(400):
    idx = rng_boot.integers(0, K, size=K)          # resample whole streets
    boot[b] = yt[idx].sum() / nt[idx].sum() - yc[idx].sum() / nc[idx].sum()
cluster_se = boot.std()
cluster_lo, cluster_hi = (diff - 1.96 * cluster_se) * 100, (diff + 1.96 * cluster_se) * 100

print(f"\n  headline gap (computed)   : {diff_pts:.2f} points   (the poster wrote 4)")
print(f"  weighted gap (undisclosed): {diff_w_pts:.2f} points")
print(f"  NAIVE   95% interval      : [{naive_lo:5.2f}, {naive_hi:5.2f}] points")
print(f"  STREET  95% interval      : [{cluster_lo:5.2f}, {cluster_hi:5.2f}] points  ({K} streets)")

# Figure: the two intervals on one axis, each in its own hue and labeled directly.
fig, ax = plt.subplots(figsize=(9, 3.2))
ax.errorbar(diff_pts, 1.0, xerr=1.96 * naive_se * 100, fmt="o", color="#DD8452",
            capsize=6, elinewidth=2.5, label="naive (people independent)")
ax.errorbar(diff_pts, 0.0, xerr=1.96 * cluster_se * 100, fmt="s", color="#4C72B0",
            capsize=6, elinewidth=2.5, label="street-respecting (whole streets)")
ax.axvline(0, color="gray", linestyle=":", linewidth=1.5, label="no effect (zero)")
ax.set_yticks([0.0, 1.0])
ax.set_yticklabels(["street-\nrespecting", "naive"])
ax.set_ylim(-0.6, 1.6)
ax.set_xlabel("Turnout gap (percentage points), 95% interval")
ax.set_title("The cold-run finding: the honest interval reaches down to about zero")
ax.legend(loc="lower right", fontsize=8)
plt.tight_layout()
plt.show()
```

*In plain terms, the figure shows one estimate with two uncertainty statements:
the naive interval floats safely above zero, while the street-respecting one
reaches down to about zero, and only the second matches how the design actually
assigned canvassing.*

```python
# The adjudication table: every verdict next to the check that decided it.
adjudication = pd.DataFrame([
    ["1. poster says 4, code prints 3.41", "public",
     f"rerun prints {diff_pts:.2f}", "ACCEPT: report the computed number"],
    ["2. 'old news, nothing new'", "public",
     "no number in the note is contested", "REBUT: novelty is not the claim"],
    ["3. shipped weights never applied", "cold run",
     f"weighted gap {diff_w_pts:.2f} vs {diff_pts:.2f}", "ACCEPT: disclose the choice, report both"],
    ["4. street clustering ignored", "cold run",
     f"interval [{cluster_lo:.2f}, {cluster_hi:.2f}] reaches ~zero",
     "BOUND: keep the estimate, gain the honest interval"],
    ["5. 'should have studied mail'", "public",
     "asks for a different study", "REBUT: a future direction, not a flaw"],
], columns=["criticism", "stream", "the check (printed evidence)", "verdict"])
print(adjudication.to_string(index=False))

# Self-check: the four numbers that decide the verdicts are what we said they are.
assert abs(diff_pts - 3.41) < 0.05, "headline drifted — investigate the lineage"
assert abs(diff_w_pts - 2.83) < 0.05, "weighted gap drifted — investigate"
assert naive_lo > 0, "the naive interval should sit above zero"
assert cluster_lo < 0.1, "the street-respecting interval should reach about zero"
print(f"\n✓ Self-check passed: verdicts rest on printed numbers "
      f"({diff_pts:.2f}, {diff_w_pts:.2f}, and a lower bound of {cluster_lo:.2f}).")
```

## 2. The Note Change Log and the Claim–Evidence Table v2

**Guiding question:** *how does a reader tell a disciplined revision from a
quiet rewrite?*

The verdicts now become two artifacts, and both open your M14 submission.

The **note change log**, which the milestone brief calls the revision record, is
a dated account of the revision: for each substantive change, what changed,
which public question or cold-run finding drove it, and what it cost or bought
the claim. It also lists every criticism you declined, each with the reason you
own. Example: one line records that the headline gained its street-respecting
interval because of the clustering finding; another records that the novelty
objection was rebutted, and why. Revision without a record is just editing.
Revision with one is evidence about your own work.

The **claim–evidence table v2** is the same table your note v0 shipped with,
one row per claim (claim, evidence, verification, boundary), now updated so
every revised sentence re-traces to its row. When a claim moved, its row moved
with it, and no caveat falls off between versions. Here is the headline row
before and after this adjudication:

| version | claim | evidence | boundary |
|---|---|---|---|
| v0 row | "Canvassing raised turnout by 4 points." | the poster figure | none stated |
| v2 row | "Canvassing raised turnout by about 3.4 points in these wards (95% street-respecting interval, about 0 to 6.9)." | the rerun output above | interval reaches zero; weighting moves the estimate to 2.83 |

One naming rule, stated plainly because two ladders meet here: the course calls
the artifact you are building **research note v1**, and the book's milestone
ladder calls the same artifact **Book Milestone 9, version 2**. One object, two
labels, both on your submission.

> **A question that often comes up here:** *"Why version numbers at all? It is
> just my note."* Because versions are how a claim's history stays inspectable.
> A reader who can compare v0's headline with v1's, through the change log, can
> see the system working: a public test and a cold run reached the work and the
> claim moved exactly where the evidence pushed it. A quiet rewrite hides that
> history, and a claim whose history is hidden asks for trust it has not
> earned.

## 3. Cold-Run Repairs and the Five Package Parts

**Guiding question:** *what does a stranger need to rebuild your numbers, and
what did your cold runner's questions tell you is still missing?*

> *"Your peer asked you three questions before your package would run. That is
> the review. Each question is a line your documentation does not have, located
> for free. Write the lines, not the excuses."*
> — a **journal reviewer**, on what a cold run is for

The puzzle's README fails at step 2 for the oldest reason in the genre: it never
says where the data comes from. Nothing in the package names the file, its
source, or how to obtain it, so "run all cells" dies at the first `read_csv`.
The author can answer the question out loud, and the package is still broken,
because the next stranger has to ask it again. That is the rule this section
runs on: a cold runner's question is answered **in the package**, never only in
conversation. The **repair log** records each question from your M13 cold run
and the line or file that now answers it.

Repairs land in one of the **five package parts**, the anatomy your M14 package
assembles:

- The **README**: the stranger's entry point. What this is, how to run it, and
  what result to expect, including the headline number and its uncertainty.
- The **data or its access recipe**: the files themselves where terms allow;
  otherwise the exact route a reader follows to obtain them, with version and
  provenance. An **access recipe** is the step-by-step path to data you may not
  ship, and your permissions decide which of the two the package carries.
- The **seeded code**: the notebook that passes restart-and-run-all with
  `SEED = 464`, no manual fixes.
- The **environment record**: the library versions a rerun needs, written down,
  not remembered.
- The **licence and permissions note**: what may be shared and reused, stated
  explicitly, with your decision log of by-hand choices and their reasons.

Packages break in boringly consistent ways, and the auditor below hunts the
**five package sins**: a **hard-coded path** that exists only on your laptop, a
**missing seed** that moves every run, a **by-hand edit** no clean run
reproduces, an **undocumented exclusion** with no logged reason, and **stale or
unversioned data** a reader cannot reobtain.

> **A question that often comes up here:** *"My peer already ran my package at
> M13. Why does M14 need a package step at all?"* Because the cold run was the
> diagnosis, not the cure. Your peer's report is a list of exactly where your
> materials made a stranger stop; M14 is where each stop becomes a repair, the
> five parts are assembled around the fixed code, and the whole thing is proven
> again. A diagnosed package that ships unrepaired wasted the one stranger it
> will ever get for free.

Run the three cells below. The first defines `audit_capsule`, a transparent

```python
# A tiny capsule auditor. Each check is one thing a stranger's eyes would catch.
# It scans a list of source LINES (strings) — paste your own package's key lines in.
def audit_capsule(lines):
    text = "\n".join(lines).lower()
    sins = []
    # Sin 1 — a hard-coded absolute path: dies on every machine but its author's.
    if any(("/users/" in ln.lower()) or ("/home/" in ln.lower()) for ln in lines):
        sins.append("hard-coded path")
    # Sin 2 — no fixed seed anywhere: any sample/split/shuffle moves every run.
    if not any(k in text for k in ("seed", "random_state", "default_rng")):
        sins.append("missing seed")
    # Sin 3 — a value patched by hand: no clean run will reproduce it.
    if any((".loc[" in ln and "=" in ln) or ("by hand" in ln.lower())
           or ("manually" in ln.lower()) for ln in lines):
        sins.append("by-hand edit")
    # Sin 4 — an exclusion/filter with no logged reason nearby.
    filtered = [ln for ln in lines
                if (".drop(" in ln) or (".dropna(" in ln)
                or ("df[df[" in ln.replace(" ", ""))]
    if filtered and not any("reason" in ln.lower() for ln in lines):
        sins.append("undocumented exclusion")
    # Sin 5 — data with no shareable provenance: a reader can't reobtain the inputs.
    loads_csv = any("read_csv" in ln for ln in lines)
    shareable = any(("http" in ln.lower()) or ("load_course_data" in ln) for ln in lines)
    versioned = any(("version" in ln.lower()) or ("license" in ln.lower())
                    or ("licence" in ln.lower()) for ln in lines)
    if loads_csv and not shareable and not versioned:
        sins.append("stale/unversioned data")
    return sins

print("✓ audit_capsule defined — five checks, one list of flags.")
```

```python
# A deliberately sinful capsule — every line breaks reproducibility on purpose.
sinful = [
    'df = pd.read_csv("/Users/rivera/Desktop/thesis/turnout_FINAL_v3.csv")',
    'draw = df.sample(200)                    # nothing fixes the shuffle',
    'df = df[df["ward"] == "F"]               # kept one ward only; no note on why',
    'df.loc[41, "age"] = 33                   # patched this row by hand post-survey',
    '# data pulled sometime last spring; nobody logged the download date',
]

# A clean capsule — same study, nothing a stranger would have to guess.
clean = [
    "SEED = 464",
    "rng = np.random.default_rng(SEED)",
    'df = load_course_data("foos_etal.csv")   # access recipe + licence in the README',
    'df = df[df["marked_register_2014"].notna()]   # reason: complete cases; in the decision log',
    "draw = df.sample(200, random_state=SEED)",
]

print("Sinful capsule flags:", audit_capsule(sinful))
print("Clean  capsule flags:", audit_capsule(clean))
```

```python
# Self-check: the sinful capsule trips all five checks, the clean one none.
assert audit_capsule(sinful) == ["hard-coded path", "missing seed", "by-hand edit",
                                 "undocumented exclusion", "stale/unversioned data"], \
    "auditor drifted — the sinful capsule should trip all five checks"
assert audit_capsule(clean) == [], "clean capsule should trip no checks"
print("✓ Self-check passed: 5 sins in the sinful capsule, 0 in the clean one.")
print("  Reminder: a clean pass proves the capsule is RUNNABLE, never that it is CORRECT.")
```

**Reading the evidence.** The sinful capsule trips all five checks; the clean
one trips none. But read the last line of the self-check again, because it
carries the lesson of the whole package week. Zero flags means **runnable**, not
**correct**, and not even complete: the auditor reads lines, so it cannot see
that your README's step 2 names no data source, that a logged reason is a bad
reason, or that your seed is fixed on the wrong sample. Those are exactly the
failures a cold human run finds, which is why the puzzle's README sailed past

## 4. The Second Clean Run

**Guiding question:** *after the repairs, what single piece of evidence lets the
package claim it is reusable?*

Every repair you just made has a side effect: it makes your previous cold-run
record obsolete. A **stale run** is a cold-run record that predates any change
to the package: a number, a claim, the data, the code, the environment record,
or an instruction a stranger would follow. The record vouches for a package
that no longer exists. So the week closes on one non-negotiable rule: **after
your last change, run the package cold once more.** A clean environment, your
written instructions only, restart-and-run-all, and a dated record of the
result. A changed package with a stale run is not done, however good the
changes were.

The **second clean run** must bring back two things, not one: the headline
number, and its **uncertainty statement**. An interval that does not reproduce
is a discrepancy to explain and fix, exactly like a moved point estimate. This
week that bites for real: your adjudication just changed the headline's
interval, so a rerun that only checks the estimate would miss the very sentence
your revision most needs to vouch for.

> **A question that often comes up here:** *"The only thing I changed after my
> cold run was one README sentence. Surely that does not count?"* It counts,
> because the README is part of what a stranger executes: instructions are code
> that runs on people. If the sentence you changed altered no step, the rerun
> will be cheap and will say so on the record; if it did alter a step, the rerun
> is the only thing that would have caught it. Either way the rule holds,
> because the alternative is a judgment call made by the one person who cannot
> test it: the author, who already knows what every sentence means.
