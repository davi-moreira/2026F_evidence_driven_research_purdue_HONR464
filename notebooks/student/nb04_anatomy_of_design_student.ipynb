## 1. The Four Parts of Every Research Design

**Guiding question:** *what are the four parts every research design must have,
and how do you check that they all point at the same quantity?*

> *"Before you collect a single data point, walk me through four things: the
> world you are assuming, the exact number you want out of it, how the data will
> come to you, and how you will turn that data into your number. Leave one out
> and you do not have a design yet. You have a hope."*
> — a **thesis advisor**, on the first meeting about your project

Every research design, from a two-person survey to a national experiment, can be
described in four parts. Blair, Cooper, Coppock, and Humphreys named that
framework **MIDA**, one letter each, in "Declaring and Diagnosing Research
Designs" (*American Political Science Review*, 2019); Blair, Coppock, and
Humphreys developed it at book length in *Research Design in the Social
Sciences* (2023, free at
[book.declaredesign.org](https://book.declaredesign.org/)). You are using their
four labels here, in plain language, on the survey team's own declaration.

- **Model (M)**: your written-down picture of how the world could work, meaning
  which things exist and what could affect what. Example: "trust in the police
  runs from 1 to 7, and it varies from place to place."
- **Inquiry (I)**: the exact quantity you want from that world, named before any
  data arrive. Example: "the average `trust_police` score over the 10,000 rows in
  this file."
- **Data strategy (D)**: every procedure that makes your data exist, meaning who
  gets sampled, who gets which condition when you assign one, and how each
  outcome is measured. Example: "draw 60 rows at random, without replacement."
- **Answer strategy (A)**: the whole procedure that turns those data into an
  answer, meaning how you prepare them, the estimate you compute, how you state
  the uncertainty around it, and how you read what comes back. Example: "take
  their mean and report an interval around it."

The four are not a checklist you fill in any order. They have to **align**, which
means your data and answer strategies actually reach the inquiry your model makes
askable. Here is the sentence form you will reuse all week, and again on Friday:
**"Because my data strategy samples X, my answer strategy's number speaks to Y."**
If X and Y do not connect when you read them out loud, the design is not aligned
yet.

One warning before the picture, because it is the place this framework gets
misused. Alignment does not let your data rewrite your question. If your inquiry
is causal and your data strategy cannot reach it, the honest label is **causal,
currently unidentified**: the question stays causal, and you say plainly that
this design does not answer it. You never quietly relabel it descriptive because
the data came out weak.

**Before you run the next cell:** expect four labeled boxes in a row, joined by
arrows, with one caption about alignment. Nothing is computed here.

```python
# A plain schematic of the four parts of a research design (MIDA).
fig, ax = plt.subplots(figsize=(11, 3.2))
parts = [
    ("M", "Model", "the world your\nquestion assumes"),
    ("I", "Inquiry", "the exact number\nyou want"),
    ("D", "Data strategy", "how the data\ncome to exist"),
    ("A", "Answer strategy", "data turned into\nan estimate"),
]
x_centers = [1.5, 4.0, 6.5, 9.0]
for (letter, name, blurb), xc in zip(parts, x_centers):
    ax.add_patch(plt.Rectangle((xc - 1.0, 0.4), 2.0, 1.7, facecolor="#DCE6F1",
                               edgecolor="#4C72B0", linewidth=1.6))
    ax.text(xc, 1.72, letter, ha="center", va="center", fontsize=20, fontweight="bold")
    ax.text(xc, 1.28, name, ha="center", va="center", fontsize=11, fontweight="bold")
    ax.text(xc, 0.78, blurb, ha="center", va="center", fontsize=8.5)

for xc in x_centers[:-1]:
    ax.annotate("", xy=(xc + 1.5, 1.25), xytext=(xc + 1.0, 1.25),
                arrowprops=dict(arrowstyle="-|>", color="#555555", lw=1.6))

ax.text(5.25, 2.55, "the four parts must ALIGN — all point at the same quantity",
        ha="center", va="center", fontsize=10, style="italic", color="#333333")
ax.set_xlim(0, 10.5)
ax.set_ylim(0, 3.0)
ax.axis("off")
plt.tight_layout()
plt.show()

print("✓ MIDA drawn: Model -> Inquiry -> Data strategy -> Answer strategy.")
print("  Today declares all four and asks whether the answer strategy can reach")
print("  the inquiry. Wednesday diagnoses how a declared design actually behaves.")
```

a row, and each one has to serve the same quantity. The arrows run left to right
because that is the order you write them in, not the order you think of them in.

> **A question that often comes up here:** *"Isn't 'the data strategy' just how I

## 2. Estimand, Estimator, Estimate

**Guiding question:** *when a study is run twice and returns two different
numbers, which of the three things in play actually moved?*

Three words get confused constantly, and separating them clears up most of the
confusion in this whole area. Read them against the survey team's own puzzle.

- An **estimand** is the quantity you want, defined in the world before you touch
  any outcome data. Example: the average `trust_police` score over the 10,000
  rows in this file. It would have a value even if nobody ever ran a survey.
- An **estimator** is the recipe you apply to data to guess the estimand.
  Example: draw 60 rows at random and take their mean. A recipe is a procedure,
  so it is the thing that can be repeated.
- An **estimate** is what one run of the recipe returns. Example: 4.58. One
  number, from one run, on one day.

**The estimand does not move. The estimate does. The distance between those two
facts is where uncertainty lives.**

One bridge back to MIDA. The **inquiry** is the question your design asks; the
**estimand** is the value that question has in the world. Same idea, two
vocabularies, and both appear on your Contract v0.

**A word about this file, because it is exactly this week's lesson.**

10,000 rows in this file**. It is not "trust in the police in Brazil." An
estimand is only as good as the population, setting, and time you attach to it,
and here you can say all three precisely.

### 🛠️ Run the Study: put 2,000 runs of the same recipe on screen

Run the cell below. It loads the file, computes the estimand, runs the estimator
2,000 times, and then, separately, checks how often a 95% interval built from one
single run actually catches the estimand.

**Before you run, expect:** an estimand near 4.4, a first run and a second run
that disagree, a pile centred close to the estimand, and a catching rate close
to but not exactly 95%.

```python
# The declared population for this lecture: the 10,000 rows in this file.
lapop = load_course_data("lapop_brazil.csv")
trust = lapop["trust_police"].to_numpy(dtype=float)
estimand = trust.mean()
print(f"✓ Loaded lapop_brazil.csv: {lapop.shape[0]} rows x {lapop.shape[1]} columns")

# The estimator: draw 60 rows at random, without replacement, and take the mean.
rng_pile = np.random.default_rng(SEED)
pile = np.array([rng_pile.choice(trust, size=60, replace=False).mean()
                 for _ in range(2000)])

# A second, separate question: does a 95% interval built from ONE run catch the
# estimand as often as it promises? Width is estimated from that one run only.
rng_cov = np.random.default_rng(SEED)
caught = 0
for _ in range(2000):
    s = rng_cov.choice(trust, size=60, replace=False)
    se_one = s.std(ddof=1) / np.sqrt(60)
    lo, hi = s.mean() - 1.96 * se_one, s.mean() + 1.96 * se_one
    caught += lo <= estimand <= hi

print(f"estimand (mean trust_police over these 10,000 rows): {estimand:.3f}")
print(f"run 1 said {pile[0]:.3f}   run 2 said {pile[1]:.3f}")
print(f"centre of the pile        : {pile.mean():.3f}")
print(f"spread of the pile        : {pile.std(ddof=1):.3f}")
print(f"middle 95% of the pile    : {np.percentile(pile, 2.5):.3f}"
      f" to {np.percentile(pile, 97.5):.3f}")
print(f"share of 95% intervals that caught the estimand: {caught / 2000:.1%}")

fig, ax = plt.subplots()
ax.hist(pile, bins=35, color="#4C72B0", edgecolor="white")
ax.axvline(estimand, color="black", linestyle="--", linewidth=2,
           label=f"estimand = {estimand:.3f}")
ax.set_xlabel("Estimated mean trust_police, one value per 60-person run")
ax.set_ylabel("Number of runs")
ax.set_title("2,000 runs of the same recipe on the same declared population")
ax.legend()
plt.tight_layout()
plt.show()
```

**Reading the output.** Three facts, and they are the three things uncertainty
language exists to say.

**Where the pile sits.** The centre is 4.421, almost exactly the estimand of
4.427. This recipe aims true.

**How wide it is.** The spread is 0.263, which is the typical distance between
one run and the centre. That is why run 1 came back 4.583 and run 2 came back
4.233. Neither is a mistake. Both are what this recipe does.

**What range covers nearly all of it.** The middle 95% of the pile runs from
3.900 to 4.933. And the last line is the one that matters most: intervals built
from a single run caught the estimand 93.6% of the time, close to the 95% the
recipe promises, and worth saying honestly rather than rounding up.

*Figure note: a histogram of 2,000 estimates, piled around a black dashed line
marking the estimand.*

## 3. What One Run Cannot Tell You

**Guiding question:** *what exactly does the 95% in a confidence interval
describe, and what does it definitely not describe?*

Two names for two of the numbers you just printed.

- The **standard error** is the typical distance between one run's estimate and
  the centre of the pile. Example: 0.263 in the printout above. It says how wide
  the pile is. It does not say how wrong today's estimate is, and a tightly
  packed pile can still sit centred on the wrong quantity.
- A **confidence interval** is a range built by a recipe that catches the
  estimand a stated share of the time, when the same recipe is repeated. Example:
  the recipe above caught the estimand in 93.6% of 2,000 runs.

Now the sentence to refuse, written next to the true one, because the obvious
paraphrase of that definition is false.

| | |
|---|---|
| **Warranted** | "Under its assumptions, this procedure catches the estimand in about 95 of every 100 repeated surveys." |
| **Refused** | "There is a 95% chance the true average is between 3.90 and 4.93." |

The second sentence fails because the computed range is fixed and the estimand is
fixed, so there is no chance left to assign between them. What has the 95%
property is the **recipe**, and only while its assumptions hold. Anyone who
believes the flattering version will also start reading a single interval as a
verdict, and that habit is hard to unlearn later.

## 4. Sixty Interviews, Six Municipalities

**Guiding question:** *if the estimand, the arithmetic, and the number of
interviews all stay the same, how can the uncertainty statement change?*

Back to part (b) of the puzzle. The team keeps the same estimand, the same
recipe, and the same 60 interviews. They change only **how** those 60 are
collected: 10 people in each of 6 municipalities instead of 60 people scattered
across the file. That is a change to the **data strategy**, and to nothing else.

One term you need first. **Dependence** means that learning one observation
changes what you should expect about another. Example: people interviewed in the
same municipality share a police force, a neighbourhood, and a local reputation,
so the second person you interview there tells you less that is new than a fresh
person from somewhere else would.

The arithmetic does not notice. The procedure's honesty does.

**Before you run, expect:** the naive interval, which treats the 60 as 60
independent respondents, to catch the estimand well under the 95% it promises.
Write down your guess for how far under.

```python
# The same estimand, the same arithmetic, a different data strategy:
# 6 municipalities drawn at random, then 10 respondents inside each one.
muni = {name: g["trust_police"].to_numpy(dtype=float)
        for name, g in lapop.groupby("municipality")}
muni_names = np.array(sorted(muni))
print(f"municipalities available: {len(muni_names)}")

rng_c = np.random.default_rng(SEED)
naive_caught = cluster_caught = 0
cluster_est = np.empty(2000)
for r in range(2000):
    picked = rng_c.choice(muni_names, size=6, replace=False)
    block = np.array([rng_c.choice(muni[p], size=10, replace=False) for p in picked])

    flat = block.ravel()                       # pretend: 60 independent respondents
    se_flat = flat.std(ddof=1) / np.sqrt(60)
    naive_caught += abs(flat.mean() - estimand) <= 1.96 * se_flat

    town = block.mean(axis=1)                  # honest: 6 independent municipalities
    se_town = town.std(ddof=1) / np.sqrt(6)
    cluster_caught += abs(town.mean() - estimand) <= 1.96 * se_town
    cluster_est[r] = town.mean()

n_eff = (trust.std(ddof=1) / cluster_est.std(ddof=1)) ** 2
print(f"treating 60 respondents as independent : {naive_caught / 2000:.1%} caught the estimand")
print(f"treating 6 municipalities as the unit  : {cluster_caught / 2000:.1%} caught the estimand")
print(f"these 60 clustered respondents carry about the information of "
      f"{n_eff:.0f} independent ones")
```

**Reading the output.** The naive interval promised 95% and delivered 81.0%. It
is not a little optimistic. It is wrong about one case in five, and **nothing in
that printout would have told the team**. Treating the 6 municipalities as the
independent unit brings it to 90.1%, better by a lot and still shy of 95, which
is worth saying rather than rounding away. With only 6 groups, the recipe's usual
1.96 multiplier is too narrow.

Two honest caveats. First, the effective information here, about 29 independent
respondents out of 60, comes from these data. It is not "divide by the group
size," and a different dependence structure gives a different answer. Second, the
direction of the correction is not automatic: here people in one municipality
resemble each other, which widens the pile, but a design that deliberately
balances within groups can move the pile the other way.

> **A question that often comes up here:** *"How would I know my own data have
> this problem?"* Ask how the data were collected, not what they look like. If
> your units were selected in groups, measured repeatedly, or connected to each

## 6. Station A: Rerun the Diagnosis

**Guiding question:** *how do you find out whether a declared design works,
before you spend a semester running it?*

> *"Do not bring me a design you have only argued for. Bring me one you have
> already run a thousand times in simulation, so you can tell me how often it
> lands on the truth and how often it embarrasses you. I trust a diagnosed
> design. I do not trust a confident one."*
> — a **skeptical methodologist** on your committee, before you collect any data

To **declare** a design is to write all four MIDA parts as something a computer
can run: a model that generates fake data, an inquiry it can compute on that fake
data, a data strategy that samples and assigns, and an answer strategy that
produces an estimate. Once declared, you can **diagnose** it: run the whole
design many times on simulated worlds and watch how its estimate behaves. Three
numbers start that story, and each answers a different worry.

- **Bias** is the average of your errors across the runs, counting direction:
  each estimate minus the truth, averaged, so misses that cancel count as zero.
  Example: the truth is 2 and five runs return 0, 1, 2, 3, 4; the errors average
  to zero, so bias is zero even though only one run landed on 2. A tilt built
  into the design does not shrink when you collect more of the same data.
- **Wobble** is how much the estimate moves from one run to the next, measured
  below as the standard deviation of the run estimates. Example: estimates
  scattering from -6 to +10 wobble a lot; ones hugging 1.8 to 2.2 barely wobble.
  More of the same kind of data usually tightens it.
- **Power** is how often the design would actually detect the effect, and it
  means nothing until you say how you would decide. The rule below is printed in
  the same cell as the number, because a power claim without its rule is exactly
  the flaw this audit scores.

Bias asks *are we aimed at the truth?* Wobble asks *how steady is our aim?*
Power asks *will we actually hit the target often enough to bother?*

### 🛠️ Run the Study: diagnose the colleague's declared design

Run the cell below. It declares the mentoring world, runs the design 2,000 times,
and prints the decision rule and its threshold beside the power number.

**Before you run, expect:** bias close to zero, because a coin flip is fair, and
a power number nowhere near 80%.

```python
# Declare the mentoring design, then diagnose it by simulation.
EFFECT = 2.0   # the true mentor effect planted in this world, in belonging points

def diagnose(n_per_arm, assignment="randomized", effect=EFFECT, reps=2000, seed=SEED):
    # Run a declared design `reps` times; return every run's estimate plus its
    # bias, wobble (SD of the run estimates), and power (share of runs whose
    # difference-in-means test clears the 1.96 rule, i.e. the 5% threshold).
    rng_d = np.random.default_rng(seed)          # own stream, reproducible on its own
    n = n_per_arm * 2
    ests = np.empty(reps)
    detected = np.zeros(reps, dtype=bool)
    for r in range(reps):
        prior = rng_d.normal(0, 1, n)                   # prior campus involvement
        Y0 = 50 + 6 * prior + rng_d.normal(0, 8, n)     # belonging without a mentor
        Y1 = Y0 + effect                                # with a mentor (constant nudge)
        if assignment == "randomized":
            Z = np.zeros(n, dtype=int)
            Z[rng_d.permutation(n)[:n // 2]] = 1        # a coin flip: exactly half mentored
        else:                                           # confounded: involved -> mentor
            Z = (rng_d.random(n) < 1 / (1 + np.exp(-1.5 * prior))).astype(int)
        Y = np.where(Z == 1, Y1, Y0)                    # observe only the assigned world
        est = Y[Z == 1].mean() - Y[Z == 0].mean()       # answer strategy: difference in means
        se = np.sqrt(Y[Z == 1].var(ddof=1) / (Z == 1).sum()
                     + Y[Z == 0].var(ddof=1) / (Z == 0).sum())
        ests[r] = est
        detected[r] = abs(est) / se > 1.96              # would this run 'find' an effect?
    return ests, ests.mean() - effect, ests.std(ddof=1), detected.mean()

TRUE_ATE = EFFECT
ests_small, bias_s, var_s, power_s = diagnose(n_per_arm=10)

print("THE TEST AND THE THRESHOLD, stated before the number:")
print("  test      : difference in means, called detected when the estimate is more")
print("              than 1.96 times its own estimated wobble away from zero")
print("  threshold : the usual 5% setting, read across repetitions, which means a")
print("              design whose TRUE effect is zero still cries effect in about")
print("              5 runs in 100")
print()
print("Declared design diagnosed over 2000 runs (10 first-years per arm):")
print(f"  true effect (the answer key)     : {TRUE_ATE:.2f} belonging points")
print(f"  BIAS  (avg estimate minus truth) : {bias_s:+.3f}    near 0 means on target")
print(f"  WOBBLE (spread of estimates, SD) : {var_s:.3f}    how much the answer moves")
print(f"  POWER (share of runs that detect): {power_s:.1%}    under the rule above")

fig, ax = plt.subplots()
ax.hist(ests_small, bins=30, color="#4C72B0", edgecolor="white")
ax.axvline(TRUE_ATE, color="black", linestyle="--", linewidth=2, label=f"true effect = {TRUE_ATE:.0f}")
ax.axvline(0, color="gray", linestyle=":", linewidth=1.5, label="no effect (0)")
ax.set_xlabel("Estimated mentor effect, one value per run (belonging points)")
ax.set_ylabel("Number of runs")
ax.set_title("The declared design: centred on the truth, and far too wide to trust one run")
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# Self-check: the declared design is unbiased yet badly underpowered.
assert abs(bias_s) < 0.5, "a randomized design should be ~unbiased"
assert power_s < 0.15, "the small design should be badly underpowered"
assert 4.0 < var_s < 5.0, "estimate spread drifted — is the seed 464?"
print(f"✓ Self-check passed: bias is {bias_s:+.2f} (on target), yet the design catches the "
      f"real effect only {power_s:.0%} of the time.\n  Aimed at the truth, far too wide to "
      f"trust any single run.")
```

**Reading the diagnosis.** The estimates centre almost exactly on 2, so the
design is unbiased: its aim is true. Yet the wobble is 4.503, so a single run
could easily land at -5 or +8, and power comes out at 9.3% under the rule printed
above. The design is honest and useless at the same time. It points at the truth
on average and almost never gets close enough on any one try to prove it.

That is the audit's first verified finding. The contract's "80% power" is not a
rounding error: **the declared design detects the effect about 9 times in 100,
under a rule the contract never states.** A power number without its test and its
threshold is not a diagnosis at all.

*Figure note: 2,000 estimates piled between roughly -12 and +15, straddling both
the black dashed true effect and the grey dotted zero line, which is what low
power looks like.*

### 🛠️ Run the Study: the redesign, and the warning that rides with it

Now the second half of the station. The next cell keeps everything fixed and
raises the sample to 600 per arm. It also runs a **different** design as a
warning: the coin flip is gone and mentors are taken up by the already-involved
first-years, which is exactly the sign-up rule the colleague's data strategy
describes. That design also has 600 per arm, so its estimates are just as steady.
Watch where they steady around.

**Before you run, expect:** power well above 90% for the bigger randomized
design, and a confounded pile that is tight but sitting in the wrong place.

> **A question that often comes up here:** *"If more data fixes the small design,
> why not just collect more data for everything?"* Because more data only cures
> wobble, never bias. A confounded design with a huge sample gives you a very
> precise estimate of the wrong number. Power climbs, the spread shrinks, and
> every tight run agrees on an answer that is off. Size buys steadiness, not
> truth.

```python
# Redesign 1: more units (600 per arm), everything else unchanged.
ests_big, bias_b, var_b, power_b = diagnose(n_per_arm=600)
# A different design: same big sample, but CONFOUNDED (the already-involved sign up).
ests_conf, bias_c, var_c, power_c = diagnose(n_per_arm=600, assignment="confounded")

print("Redesign 1 -- more units (600 per arm), still randomized:")
print(f"  bias {bias_b:+.3f}    wobble(SD) {var_b:.3f}    power {power_b:.1%}")
print("\nA confounded design -- big sample, mentors taken up by the already-involved:")
print(f"  bias {bias_c:+.3f}    wobble(SD) {var_c:.3f}    power {power_c:.1%}")
print(f"  its estimates centre on {ests_conf.mean():.2f} when the truth is {TRUE_ATE:.0f}")

fig, ax = plt.subplots(figsize=(10, 5))
ax.hist(ests_small, bins=30, color="#4C72B0", alpha=0.55, edgecolor="white",
        label="small randomized (10 per arm)")
ax.hist(ests_big, bins=30, color="#55A868", alpha=0.85, edgecolor="white",
        label="big randomized (600 per arm)")
ax.hist(ests_conf, bins=30, color="#DD8452", alpha=0.85, edgecolor="white",
        label="big but confounded (600 per arm)")
ax.axvline(TRUE_ATE, color="black", linestyle="--", linewidth=2,
           label=f"true effect = {TRUE_ATE:.0f}")
ax.axvline(0, color="gray", linestyle=":", linewidth=1.3)
ax.set_xlabel("Estimated mentor effect (belonging points)")
ax.set_ylabel("Number of runs")
ax.set_title("More units narrows the spread; confounding shifts the centre")
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# Self-check: more units rescues power without adding bias; confounding does not.
assert abs(bias_b) < 0.3 and power_b > 0.85, "more units should rescue power with no new bias"
assert bias_c > 4.0 and power_c > 0.95, "the confounded design should be badly biased yet 'sure'"
print(f"✓ Self-check passed: more units lifted power from {power_s:.0%} to {power_b:.0%} and shrank "
      f"the wobble from {var_s:.2f} to {var_b:.2f}, bias still near 0.\n  The confounded design is "
      f"off by {bias_c:+.1f} points and 'detects' it {power_c:.0%} of the time: confident and wrong.")
```

**Reading the redesign.** More units did exactly what the diagnosis predicted.
Power jumped from 9.3% to 93.5% and the wobble collapsed from 4.503 to 0.574, all
while bias stayed near zero. That is wobble cured by size. The confounded design
tells the cautionary half. Its estimates are just as tight, but they cluster
around 8.37 instead of the true 2, off by 6.366 belonging points, and it
"detects" that wrong answer on essentially every run.

That is the audit's second verified finding. The colleague's diagnosis calls the
sign-up group's extra involvement "noise" and proposes to fix it by collecting
more data. It is not noise. It is **signed bias**: a tilt with a direction, which
more data makes more precise rather than less wrong. Power also yields to sharper
measurement, better-balanced groups, and a stronger treatment contrast, not to
sample size alone. Bias yields to none of them.

## 7. Station B: Rerun the Permission Line

**Guiding question:** *what does "the data are public and we removed the names"

Your project has a **permission status**: a statement of whether you may collect,
who decided, and what you are waiting on. There are four, and your job is to land
on exactly one and write it down.

- **Cleared** means no formal determination is needed and you can say why in one
  sentence. Example: you are analyzing a published national statistics table with

- **Formal determination required** means a competent authority has to rule

- **Pending** means you asked and are waiting. Work that does not touch people or
  their data may continue.
- **Not authorized, stop** means you may not proceed as planned. Stopping is a
  design decision, not a failure.

The **competent authority** is whoever, at your institution, is empowered to make
that ruling: an ethics committee, a review board, a data protection officer, a
supervisor holding delegated authority. Find out which one covers course-based
projects at your institution before you need it.

**De-identification** means removing enough information that a person cannot
reasonably be picked out of the data. Deleting names is not de-identification,
because the remaining columns can still single someone out. The next cell shows
what that means on the exact file the colleague's contract points at.

**Before you run, expect:** party alone to single out almost nobody, and party
plus age plus census tract to single out almost everybody.

```python
# The colleague's second data source: a public voter-file extract, names already
# absent. One call prints the whole re-identification picture.
voters = load_course_data("la_voter_file.csv")
print(f"✓ Loaded la_voter_file.csv: {voters.shape[0]} rows x {voters.shape[1]} columns")
print(f"  distinct census tracts among {len(voters)} rows: {voters['census_tract'].nunique()}")

def share_unique(frame, cols):
    """Share of rows whose combination of `cols` appears exactly once."""
    return float((~frame.duplicated(subset=list(cols), keep=False)).mean())

print("\nshare of rows that are ALONE in their combination:")
for cols in [["party"], ["party", "age"], ["census_tract"],
             ["party", "age", "census_tract"]]:
    print(f"  {' + '.join(cols):32s} {share_unique(voters, cols):6.1%}")

# Who carries the risk? Compare the least common parties against the largest one.
top3 = voters["party"].value_counts().index[:3].tolist()
rare = voters[~voters["party"].isin(top3)]
dem = voters[voters["party"] == "DEM"]
print(f"\nrows outside the three largest parties ({', '.join(top3)}): {len(rare)}")
print(f"  unique on party + age + tract : {share_unique(rare, ['party', 'age', 'census_tract']):.1%}")
print(f"rows in the largest party (DEM): {len(dem)}")
print(f"  unique on party + age + tract : {share_unique(dem, ['party', 'age', 'census_tract']):.1%}")

# Minimisation: the fix is not more deleting. Coarsen what you keep.
coarse = pd.DataFrame({
    "party": voters["party"],
    "age_band": pd.cut(voters["age"], bins=[17, 29, 44, 64, 120],
                       labels=["18-29", "30-44", "45-64", "65+"]).astype(str),
    "tract_prefix": voters["census_tract"] // 10_000,
})
print("\nafter coarsening (age in four bands, only the tract's leading digits):")
print(f"  party + age band + tract prefix : "
      f"{share_unique(coarse, ['party', 'age_band', 'tract_prefix']):.1%}")
print(f"  age band + tract prefix         : "
      f"{share_unique(coarse, ['age_band', 'tract_prefix']):.1%}")
```

**Reading the output.** Party alone singles out 0.4% of rows. Party and age
together, 9.7%. Census tract alone, 59.4%, because 776 distinct tracts sit inside
1,000 rows. Put all three together and **99.6% of rows are alone in their
combination**, in a file with no names in it.

Then look at who carries that risk. Among the 159 rows outside the three largest
parties, all 159 are unique. Among the 509 rows in the largest party, 99.6% are.
The protection failed for nearly everyone, and it failed completely for the
smallest groups.

The fix is not more deleting. **Data minimisation** means collecting only what
your declared analysis requires, so the risky combination never exists. Coarsen
what you keep and the picture changes: age in four bands plus the tract's leading
digits, with party, drops to 19.9%; without party, to 1.2%.

## 9. What Your Diagnosed Design Can Claim: Identify or Associate

**Guiding question:** *when is it honest to say your treatment causes the
outcome, and when must you stop at "is associated with"?*

The diagnosis just drew a bright line between the randomized designs and the
confounded one, and Contract v0 is graded on drawing it.

- **Identify a cause**: your design pins the outcome gap on the treatment and
  rules out the other explanations, so you can defend the word *causes*. Example:
  the coin flip cut the link from prior involvement to who gets a mentor.
- **Association**: two things move together, but your design cannot rule out a
  common cause, so you can only say they are related. Example: in the confounded
  design, prior involvement lifted both who signed up and who felt they belonged.
- **Silent scope change**: quietly upgrading your wording from *is associated
  with* to *causes* without the design that earns it.

Here is the rule for your Contract's claim boundary, and it has one edge people
get wrong. A design that only shows an association does not license causal words,
no matter how large the sample or how tight the wobble. **But a causal question
your data cannot reach stays causal.** Its honest label is *causal, currently
unidentified*. It is never relabeled descriptive because the data came out weak.
Shrinking the question to fit the evidence is a silent rewrite of your research.
