## 🎯 Definition of Done

You are done with this sprint when your submission carries all six parts below,
built from **your own** finished project. These are floors, not ceilings: clearing
them is necessary, never sufficient, and the grade lives in how well each is argued.

1. **The final research chapter.** Your M14 note grown into the whole arc, with
   every verdict from your adjudication record already in the prose: the question
   and its **inquiry declaration** (compass **kind** and **reach**, why that
   position fits the question's words, what crossing your design licenses, and what
   it therefore cannot establish); the **headline claim at its exact boundary** with
   its uncertainty in the same sentence, any exploratory finding labeled and placed
   after it; **uncertainty and limitations throughout**, next to each claim they
   qualify; and **every claim traced and verified**, pointing to a real retrievable
   source or a reproducible computation, with how it was checked. The notebook
   behind the chapter passes **restart-and-run-all**, so every headline number
   reruns from it.
2. **The final release audit and the release decision.** You ran this audit at M11
   on a project that was still moving; you run it now against the finished artifact,
   marked with its date and the reason this version exists. Four parts: the
   **stale-run check** (anything changed since the run that vouches for the package
   means you rerun and record the new date), the **claim sweep** (every quantitative
   claim resolved to an output your *latest* rerun produced, or cut, or rewritten to
   what the output supports), the **audit rows** marked **clear**, **pending**, or
   **blocking**, and then the **release decision** itself: release, or withhold
   pending the repair you name. Plus your **stopping rule** (one sentence on why you
   stop here, phrased around a check you actually ran) and your **next-study
   agenda**, written as *directions*, never as findings.
3. **The dossier and its manifest.** Close your dossier with an inventory marking
   every component **present**, **absent with a reason**, or **not applicable**:
   contract versions, evidence registry, permission determinations, measurement
   documentation, analysis and claim ledgers, AI Research Ledger, artifacts, package,
   and revision history. Every present row sends a reader to the file or section
   where the thing actually lives.
4. **The AI-management portfolio.** Drawn from your full-semester ledger and written
   so a skeptic who has never seen your project can follow how you directed your AI
   team. Four parts: the **decomposition** (each scoped role, whether it ran in
   parallel or sequentially, the one node you kept human, and who chose each next
   step), the **conflict-and-override log** (at least one real conflict, diagnosed as
   real disagreement, correlated error, or false consensus, with the human override
   you made and its non-AI verification), the **never-automate list**, and the
   **independence check** showing that an agreement that mattered was worth more than
   a head-count.
5. **The Evidence Defense, live.** Ten minutes in three movements plus five minutes of
   cross-examination: **claims** (about three minutes), **choices** (about four), and
   **verification** (about three, and it is the spine, not an afterthought). One
   graceful concession identified in advance from your ledger, and the three forbidden
   moves rehearsed out: inventing a claim under pressure, un-conceding by insistence,
   and blaming the AI. Full format in
   `project/final_dossier/evidence_defense_protocol.md`.
6. **The finalized full-semester AI Research Ledger,** complete and reconciled inside
   the portfolio, with at least one **escalation row** where you overrode an AI output.

Plus the **closing dossier line** naming where each finalized piece lives.

> **A question that often comes up here:** *"My research note already argued the
> finding. Isn't the chapter just the note with a portfolio stapled on?"* No. The note
> argued one finding to a reader who had the page. The chapter argues the whole arc,
> and the portfolio proves you stayed in charge of the tools that helped you write it.
> A reader cannot tell a verified claim from a lucky guess by reading the claim. The
> portfolio and the ledger are where that check lives.

```python
# Setup + release-and-portfolio self-check — run this cell, then edit the inputs.
import numpy as np

SEED = 464  # course number — keeps this notebook's Monte Carlo reproducible
rng = np.random.default_rng(SEED)

# --- EDIT ME (1) — your release audit, one disposition per row. Use only
# "clear", "pending", or "blocking". A blocking row is not an opinion about
# polish: it is a reason the work may not leave your hands yet.
audit_rows = {
    "permissions":              "clear",
    "promises to participants": "clear",
    "AI disclosure":            "clear",
    "package matches its run":  "blocking",   # the run that vouches for it is stale
    "reproducibility":          "pending",
    "claim boundaries":         "clear",
    "uncertainty stated":       "clear",
}
named_repair = "rerun the package clean after the last edit, then re-check the sweep"

# --- EDIT ME (2) — the claim sweep: every quantitative claim in your chapter, and
# whether you resolved it to an output your LATEST rerun produced.
claims_in_chapter = 9
claims_resolved_to_latest_rerun = 8

# --- EDIT ME (3) — your AI review team, as you ACTUALLY ran it this semester.
n_roles = 4               # how many AI roles read your chapter or a key claim
share_one_framing = True  # True if they share one base model AND one prompt framing
p_catch = 0.5             # honest guess: chance ONE role catches a real flaw on its own
rho_blind = 0.40          # chance the team shares a blind spot (0 only if truly independent)

# --- EDIT ME (4) — your full-semester AI Research Ledger, in two counts.
ledger_rows_total = 12   # total AI-assisted rows across the whole semester
escalation_rows = 1      # rows where YOU overrode an AI output (a human override)

# ---- everything below is DERIVED from the four edits above ----
# One source of truth per number: change an input once, the whole readout follows.

# CHECK 1 — the release decision, read off your own dispositions.
blocking = [k for k, v in audit_rows.items() if v == "blocking"]
pending  = [k for k, v in audit_rows.items() if v == "pending"]
unswept  = claims_in_chapter - claims_resolved_to_latest_rerun
print("CHECK 1  release audit")
print(f"         rows: {len(audit_rows)} | blocking: {len(blocking)} | pending: {len(pending)}")
print(f"         claim sweep: {claims_resolved_to_latest_rerun} of {claims_in_chapter} "
      f"claims resolved to your latest rerun")
if blocking or unswept > 0:
    print("         -> DECISION: WITHHOLD pending a named repair.")
    for k in blocking:
        print(f"            blocking row: {k}")
    if unswept > 0:
        print(f"            {unswept} claim(s) unresolved: cut each one or rewrite it")
        print("            to what your latest output actually supports.")
    print(f"            your named repair: {named_repair}")
    print("            Writing this up as 'essentially ready' is the euphemistic")
    print("            release the rubric caps at Beginning.")
else:
    print("         -> DECISION: RELEASE. Say in one sentence why, and keep the")
    print("            pending rows visible as limitations rather than silent.")

# CHECK 2 — the independence check. Simulate 5000 chapters, each with one hidden flaw.
# A "false consensus" = all your roles say "fine" while a real flaw is present.
n_ch = 5000
caught = (rng.random((n_ch, n_roles)) < p_catch)
if share_one_framing:
    blind = rng.random(n_ch) < rho_blind        # shared blind spot: all roles miss together
    caught = caught & (~blind[:, None])
false_consensus_rate = (~caught.any(axis=1)).mean()

print("\nCHECK 2  independence of your AI team")
print(f"         roles = {n_roles}, share one framing = {share_one_framing}")
print(f"         chance a unanimous 'looks fine' is actually a MISSED flaw: "
      f"{false_consensus_rate:.1%}")
if share_one_framing and false_consensus_rate > 0.15:
    print("         -> high false-consensus risk. Agreement here is close to ONE voice,")
    print("            not several. Your portfolio needs an independent check you ran,")
    print("            or two roles you made genuinely independent, not the head-count.")
else:
    print("         -> lower risk, but 'lower' is not 'zero'. Name the independent")
    print("            check behind any agreement you lean on in the chapter.")

# CHECK 3 — the escalation floor. A semester of AI use with zero overrides is a flag.
print("\nCHECK 3  escalation floor")
print(f"         ledger rows = {ledger_rows_total}, overrides logged = {escalation_rows}")
if escalation_rows == 0:
    print("         -> ZERO overrides. Per the brief this is a warning sign, not a badge:")
    print("            either AI touched nothing that mattered, or a boundary slipped by.")
    print("            Do not manufacture one. Take a load-bearing output you kept and")
    print("            show the independent check that earned its place.")
else:
    print(f"         -> {escalation_rows} override(s) logged. Make sure at least one carries")
    print("            a NON-AI verification in its row, and a diagnosis: real")
    print("            disagreement, correlated error, or false consensus.")

print("\nAll three are FLOORS, never proof: a clean readout means your audit and")
print("portfolio are well-formed, not that your claim is right. The cross-examination")
print("and the cold rerun are the proof.")
```

### 🤝 AI Research Partner

AI is your arm and your research assistant in this final sprint, not your brain. This
week the skill is different from every week before it: you are not tasking one
assistant, you are **managing a small team** and deciding for yourself when its roles
disagree. More roles do not mean more truth. Three roles that quietly share one wrong
assumption will agree with great confidence and walk you off a cliff.

**Good to delegate here:**
- Proposing a **decomposition** of your chapter into scoped roles, so you can decide
  which are worth the hand-off and which you supervise.
- Rehearsing the hardest cross-examination questions, or red-teaming a paragraph
  **you** already wrote, so you can prepare an answer from a real ledger row.
- Ranking your limitations or listing the inputs a cold replicator would need, so you
  can check the ranking against your own printout.

**Never delegate (these are yours to decide, no matter how good the roles get):**
- The **release decision**, and which claims your chapter can defend and where each
  claim must stop.
- Which sources are real and the words you put your name on.
- Every **human override**: when an AI output would *settle* rather than *inform* a
  decision that carries your name, you take the pen back. The public defense is yours.

**Commit first, then ask.** Write your own answer before you open any role, and save a
copy of your chapter before a long multi-role session, so you can compare what you
wrote against what the roles changed. Verify what comes back, and log it in the **AI
Research Ledger** below. The full never-delegate list is in
[`ai_resources/human_responsibility_checklist.md`](https://github.com/davi-moreira/2026F_evidence_driven_research_purdue_HONR464/blob/main/ai_resources/human_responsibility_checklist.md),
and the rule for taking the pen back is in
[`ai_resources/escalation_protocol.md`](https://github.com/davi-moreira/2026F_evidence_driven_research_purdue_HONR464/blob/main/ai_resources/escalation_protocol.md).

The AI prompt below is your delegable sprint assist: it turns your finished
chapter into a hostile cross-examiner so you can rehearse the live defense. It speeds up
the tedious part, generating hard questions, so you can spend your time on the part that
is yours, answering each from a verified ledger row. Everything it hands back is a
proposal you test against reality, never a verdict.

> 💡 **AI Prompt:** "Act as a hostile but fair cross-examiner at a research defense.
> Here is my headline claim, its stated boundary, my compass position, and my three
> weakest ledger rows: [paste them]. Ask me the five hardest questions you can, each
> grounded in a specific claim or verification of mine, ranked from most to least
> dangerous. Do not soften them, and do not answer them for me. Then name the single
> question you think I am least prepared for."
>
> **After running, verify (a delegable assist is only as good as your check against
> reality; counters *sycophantic agreement*: a partner that reassures you has rehearsed
> your ego, not your defense):**
> - [ ] If every question is gentle or it calls your claim "well-supported," push back:
>   "assume I am overclaiming, find the weakest link." Weak questions are a red flag,
>   not a passing grade.
> - [ ] For each question, rehearse your answer from a **real ledger row**, not from
>   memory. If no row backs your answer, that is the row to build before you defend, not
>   a sentence to improvise in the room.
> - [ ] Pick your **graceful concession** from this list: the one question where the
>   honest answer names a smaller true claim. Conceding the right point on time is
>   scored as strength.
> - [ ] Log this use in your AI Research Ledger: task, tool, decision, verification.

### 🗡️ Red-Team Exchange

The final studio's red-team block. Pair with one partner and run two rounds, with an AI
reviewer as a third voice. The role for this milestone is the **AI Research Team
Orchestrator**, which helps you plan how you split your chapter across roles and how you
logged their conflicts (quick card in
[`ai_resources/agent_role_cards.md`](https://github.com/davi-moreira/2026F_evidence_driven_research_purdue_HONR464/blob/main/ai_resources/agent_role_cards.md);
full briefing in
[`genai_studio/roles/ai_research_team_orchestrator.md`](https://github.com/davi-moreira/2026F_evidence_driven_research_purdue_HONR464/blob/main/genai_studio/roles/ai_research_team_orchestrator.md)).
It is a planning aid: it proposes a decomposition and a way to log conflicts, and it
hands every consequential decision back to you. It cannot resolve a disagreement, and it
never runs your notebook.

**First, the Orchestrator pass (before the rounds).** Paste your portfolio's
decomposition and one conflict-and-override entry. It returns whether your roles were
wired so an early error could flow downstream unchecked, and whether your logged
agreement could have been correlated by construction. Every flag is a **suspect you test
yourself** against your ledger, not a verdict.

**Round 1: cross-examine the defense (about 15 minutes).** Your partner plays a hostile
examiner. They take your headline claim and press its boundary: "show me the printout
that says this, not a bigger version of it." You run your three movements, claims then
choices then verification, and they force one **graceful concession**. You concede
precisely and name the smaller true claim, or you defend from a ledger row. Watch for
the three forbidden moves in your own answers: inventing a new claim, un-conceding by
insistence, and blaming the AI.

**Round 2: rerun the notebook cold, then read the audit (about 15 minutes).** Your
partner opens your chapter's notebook on a fresh kernel and runs **restart-and-run-all**
with no help from you, logging whether your headline number comes back and where, if
anywhere, it broke. Then they read your release audit and check one thing: does every
quantitative claim in the chapter resolve to an output *that* rerun produced, and does
any blocking row hide behind soft language? A number that reruns for you and not for
them is a finding you fix tonight, because M15 is terminal and there is no Monday.

**Mind the correlated-error trap.** If your AI rehearsed your questions and the
Orchestrator calls your wiring sound, that agreement is not proof your claim holds.
Neither tool ran your notebook. Only your partner's cold rerun and your own verified
ledger settle whether the number reproduces and the claim stands.

**What to log.** Write down the concession the cross-examination forced, the strongest
question you could not answer from a ledger row (and the row you will build), and
whether the cold rerun reproduced your headline number. A challenge that changed nothing
still earns a line saying why it did not land. That record feeds your portfolio.

> **A question that often comes up here:** *"What if my partner and the Orchestrator
> disagree about whether my claim is bounded?"* Your evidence settles it, not either of
> them. The Orchestrator read only your plan and can miss a naked claim; your partner
> read your page but not your full ledger. Say in one sentence which read your own
> printout and claim ledger support, and why. That adjudication is stronger proof you
> understand your chapter than agreeing with either would be.
