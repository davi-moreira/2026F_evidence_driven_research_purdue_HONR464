### 🧩 Research Puzzle

A visitor gives you thirty seconds. A judge gives you two minutes. Both hear a
version of the same research.

Here is the trap. Qualifiers are long and verbs are short. When you cut for time,
the qualifier goes first because it is where the words are, and what survives is
a stronger sentence than the one you started with. Nobody decides to overstate.
The clock does it.

So the question for this lecture is: **if your short version says more than your
long version licenses, which one did you lie in?**

## 1. Three Architectures, Not One Speech Cut Three Ways

Each length is a different job. A thirty-second pitch is not a two-minute pitch
with the middle removed.

| Length | Ceiling | The job |
|---|---|---|
| **30 seconds** | ≤ 75 words | Earn the next ninety seconds |
| **90 seconds** | ≤ 200 words | The finding, with its bound |
| **2 minutes** | ≤ 300 words | The finding, the bound, and the choice you would defend |

**Word ceiling.** A stand-in for a clock you do not control. Example: at a normal
speaking pace, 75 words runs about thirty seconds. If your "thirty-second"
version is 110 words, it runs closer to fifty, and by then your visitor has read
the board themselves and stopped listening to you.

**Boundary sentence.** The one line that says where your evidence stops. It
appears at **all three lengths**, in the same words. It is the first thing people
cut and the last thing they should.

**Spoken upgrade.** A verb, quantifier, or scope word that gets stronger when the
version gets shorter. Example: "is higher among" becomes "raises"; "in this
sample" quietly disappears. The upgrade is almost never deliberate, which is
exactly why you check for it mechanically rather than trusting your ear.

> **A question that often comes up here:** *"Isn't a thirty-second version
> supposed to be simpler?"* Simpler, yes. Broader, no. Simplifying means fewer
> ideas said plainly. Broadening means the same idea said about more of the
> world than you looked at. The two feel identical while you are writing and are
> completely different when someone repeats you.

### 🛠️ Run It Live: counting words and catching the upgrade

The next cell does two things a human ear does badly. It counts, and it compares
two versions of the same claim word by word against a list of strength markers.

Two worked versions sit in the cell so the scan has something to work on. Swap
in your own drafts as soon as they exist, then run it again.

```python
# Two versions of the same finding, as they are usually first drafted.
long_version = (
    "We looked at whether neighbourhoods with regular patrols report higher "
    "trust in police. Using survey data from ten thousand respondents, trust "
    "is about one point higher on a seven-point scale among people who also "
    "rate their government as responsive. This is an observational comparison, "
    "so it does not establish that patrols caused the difference; responsive "
    "government may drive both. In this sample, the gap is well located, with "
    "an interval of roughly plus or minus six hundredths of a point."
)
short_version = (
    "Patrols raise trust in police by about one point on a seven-point scale. "
    "It is a big effect and it shows up clearly in a large national survey."
)

# Strength markers: words that assert more than an association.
STRONG = {"raise", "raises", "raised", "cause", "causes", "caused", "effect",
          "impact", "drives", "boosts", "proves", "always", "every", "clearly",
          "significantly", "big", "substantial"}
# Hedges and scope markers: the words compression eats first.
HEDGE = {"associated", "association", "observational", "does", "not", "may",
         "about", "roughly", "interval", "sample", "suggests", "appears",
         "correlation", "correlated"}

def scan(text, label):
    words = re.findall(r"[a-z']+", text.lower())
    strong = sorted({w for w in words if w in STRONG})
    hedge = sorted({w for w in words if w in HEDGE})
    return {"version": label, "words": len(words),
            "strength markers": ", ".join(strong) or "—",
            "hedges/scope": ", ".join(hedge) or "—"}

report = pd.DataFrame([scan(long_version, "2-minute (≤300)"),
                       scan(short_version, "30-second (≤75)")])
print(report.to_string(index=False))
print()

lw = {w for w in re.findall(r"[a-z']+", long_version.lower()) if w in STRONG}
sw = {w for w in re.findall(r"[a-z']+", short_version.lower()) if w in STRONG}
gained = sorted(sw - lw)
lost_hedges = sorted({w for w in re.findall(r"[a-z']+", long_version.lower()) if w in HEDGE}
                     - {w for w in re.findall(r"[a-z']+", short_version.lower()) if w in HEDGE})
print("Strength words the SHORT version gained:", ", ".join(gained) or "none")
print("Hedge/scope words the SHORT version dropped:", ", ".join(lost_hedges) or "none")
print()
print("Compression may drop blocks. It may not drop bounds." if gained or lost_hedges
      else "No upgrade detected on this word list.")
```

## 3. The Questions You Can Predict

Most of the questions you will get on Tuesday are derivable today, from your
design, without anyone asking them.

**The predictable question set.** The three questions a given board always
attracts, derived from what the design cannot do. Example: an observational
comparison always attracts "but could something else explain that?"; a
convenience sample always attracts "would this hold anywhere else?"; a
self-reported outcome always attracts "do people actually do what they say?"

**Bound request against attack.** A **bound request** wants to know where your
evidence stops. An **attack** claims your evidence is wrong. They sound similar
and need opposite answers: a bound request is answered by stating the boundary
you already wrote; an attack is answered by naming the check you already ran.
Reading the type correctly is most of the skill.

**The smaller true claim.** What you offer when the honest answer is that your
design cannot settle the question. Example: "My design cannot tell you whether
the programme caused this. What it does establish is that among the people I
observed, the gap is this size and it is not explained by the two measures I was
able to adjust for."

**Audience-directed questions.** The questions *you* ask the visitor. They turn a
monologue into a conversation and, more usefully, they tell you what the person
actually heard. Example: "Before I go on, what did you take the main finding to
be?"

> **A question that often comes up here:** *"Doesn't saying 'my design cannot
> settle that' make me look like I did not do enough?"* It reads as the opposite
> to anyone who knows the field, because the sentence is only available to
> someone who knows exactly what their design does. What reads as weak is a
> confident answer that a follow-up question dismantles.

### 🛠️ Run It Live: typing the questions a real board attracted

Below is a set of questions collected at a poster session, together with the
board's design. The next cell types each one and reports what the mix says about
which part of the board is doing the least work.

Replace the list with the questions your panel actually asks you today, and
rerun. The rerun is the raw material for your question strategy.

```python
# Questions asked at a board reporting an observational trust gap.
questions = [
    "Could responsive government be causing both of these?",
    "How far does this reach beyond the people you surveyed?",
    "What is that interval on the headline, exactly?",
    "Did you check whether the result holds if you drop the youngest respondents?",
    "Isn't self-reported trust a pretty weak measure?",
    "Would this replicate in another country?",
    "Which comparison produced the one-point number?",
    "Are you claiming patrols work?",
]

# A question is a BOUND REQUEST when it asks how far the claim reaches;
# an ATTACK when it asserts something is wrong or missing.
BOUND_CUES  = ("how far", "beyond", "reach", "replicate", "another",
               "exactly", "which comparison", "what is that")
ATTACK_CUES = ("isn't", "weak", "could ", "causing", "did you check",
               "claiming", "wrong", "but ")

typed = []
for q in questions:
    low = q.lower()
    bound = any(c in low for c in BOUND_CUES)
    attack = any(c in low for c in ATTACK_CUES)
    kind = ("bound request" if bound and not attack else
            "attack" if attack and not bound else "ambiguous")
    typed.append({"question": q[:52] + ("…" if len(q) > 52 else ""), "type": kind})

qdf = pd.DataFrame(typed)
print(qdf.to_string(index=False))
print()
print(qdf["type"].value_counts().to_string())
print()
print("Ambiguous rows are the ones you must read live, from the visitor's face,")
print("and they are the ones worth rehearsing.")
```
