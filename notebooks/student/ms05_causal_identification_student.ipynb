## 🎯 Definition of Done

You are done with this sprint when your submission carries all six parts of
the governance record, applied to **your own project**:

1. **The acquisition and provenance record.** Your **acquisition route**
   declared in writing from your M3 permission status (cleared: name the
   frame and the plan for unreached units; pending: build and pilot without
   touching people; no permission: aggregates, open data, or simulation, with
   the cost stated), one **provenance** entry per dataset and borrowed number
   (who produced it, how, when, under what terms), and the **permission
   recheck**: the M3 determination held against the data as they actually
   arrived, with one declared outcome.
2. **The measurement plan.** The concept → construct → indicator ladder for
   every measured concept, a **meaning sentence** per indicator saying what a
   single recorded value asserts, one **reliability** check planned on the
   instrument's items (never on split respondents), and **validity** argued
   as interpretation and use, with the construct-indicator gap named.
3. **Column minimization and the re-identification check.** Every kept column
   tied to the analysis your Contract declares, what you dropped or coarsened
   recorded, and the re-identification check run: could someone plausibly
   combine your kept columns to pick out an individual?
4. **The Contract revision.** The settled operationalization issued as a new,
   dated, numbered Research Contract version with a usable reason, and
   measurement error stated alongside sampling uncertainty, not instead of
   it.
5. **AI Research Ledger rows.** One row for every AI-assisted step, each with
   a named verification method against the actual file or page.
6. **The dossier update line.** Where your data and measurement documentation
   component now lives, named by file or section.

> **A question that often comes up here:** *"My data have not arrived yet, or
> I have no permission to collect any. How do I govern data I do not have?"*
> You govern the route your permission status assigns. Pending means you
> pilot the instrument on yourself and write your analysis code against
> simulated data of the shape you expect, while you wait. No authority to ask
> means the no-permission route: published aggregates, an openly licensed
> dataset, or data you simulate yourself, with the cost written down. A
> simulated dataset can demonstrate a method, but it cannot support a claim
> about the world.

> 💡 **AI Prompt:** "Act as a privacy attacker. Here are the columns
> I plan to keep in my dataset: [list them, e.g., residence hall, major,
> hometown, response date, a 1-to-5 scale score]. Propose the column
> combinations most likely to pick out a single individual when intersected,
> ranked by risk, and say what outside information an attacker would combine
> them with. Do not reassure me."
>
> **After running, verify (counters *sycophantic agreement*: a tool that calls
> your table safe has reviewed your hopes, not your columns):**
> - [ ] Run the re-identification probe below on your own kept columns and
>   compare its result against the tool's ranking.
> - [ ] For each risky combination, decide the fix yourself: drop the column,
>   coarsen it, or defend keeping it against your declared analysis.
> - [ ] Record what you dropped or coarsened in the governance record.
> - [ ] Log this use in your AI Research Ledger: task, tool, decision,
>   verification.

```python
# Governance self-check — edit the EDIT ME blocks to match YOUR project.
# Three checks in one cell: (1) provenance entries with missing fields flagged,
# (2) column minimization against your declared analysis, (3) a seeded
# re-identification probe on a simulated table shaped like yours.

# EDIT ME — one provenance entry per dataset and per borrowed number.
provenance = [
    {"source": "EDIT ME (dataset or number)", "producer": "", "method": "",
     "date": "", "terms": ""},
]

# EDIT ME — every column you plan to KEEP, mapped to the declared analysis
# that needs it. An empty string means no analysis behind it: a flag.
kept_columns = {
    "hall":     "EDIT ME (which declared analysis needs this?)",
    "major":    "",
    "hometown": "",
    "score":    "the primary outcome of my declared analysis",
}

# EDIT ME — for the probe: plausible category counts for your quasi-identifier
# columns (how many distinct values each could take in your data).
quasi_levels = {"hall": 3, "major": 12, "hometown": 40}
n_rows = 200   # about the size of your expected dataset

print("— Provenance check —")
for e in provenance:
    missing = [k for k, v in e.items() if k != "source" and not str(v).strip()]
    mark = "⚠️ incomplete: missing " + ", ".join(missing) if missing else "✓ complete"
    print(f"  {e['source']}: {mark}")

print("— Minimization check —")
for col, analysis in kept_columns.items():
    ok = analysis.strip() and not analysis.startswith("EDIT ME")
    print(f"  {col}: " + (f"✓ tied to: {analysis}" if ok
                          else "⚠️ no declared analysis behind it — drop, coarsen, or justify"))

print("— Re-identification probe (seeded simulation) —")
qcols = [c for c in quasi_levels if c in kept_columns]
sim = {c: rng.integers(0, quasi_levels[c], size=n_rows) for c in qcols}
combos = list(zip(*(sim[c] for c in qcols)))
unique_share = sum(1 for c in combos if combos.count(c) == 1) / n_rows
print(f"  kept quasi-identifiers: {qcols}")
print(f"  share of simulated rows UNIQUE on their intersection: {unique_share:.0%}")
print("  A high share means a few columns can pick out one person even with")
print("  no names. Coarsen or drop until the share is defensible, then record")
print("  the change and rerun.")
```

### 🗡️ Red-Team Exchange

The attack runs on your own governance record, at the studio with your AI
assistant. Two rounds.

**Round 1: attack the provenance.**
Ask your AI what your dataset contains, how a variable is coded, or what its
licence allows, and then catch it: hold every answer against the file itself.
Count the rows, read the codebook, read the licence page. Any claim the file
does not confirm is the week's central trap made visible: a fluent
description of data the tool has never seen. Record one place the
against-the-file check corrected or confirmed the tool.

**Round 2: attack the measurement plan.**
Press on your own ladder the way a reviewer will. Does each rung narrow the
one above it? Can you say the meaning sentence for every indicator without
notes? Is the reliability check on items, or did a respondent-split sneak in?
Where does the construct-indicator gap threaten the interpretation you put on
the scores? And if an instrument artifact worries you (a wording nudge, a
this-day mood effect), keep the verbs straight: only a redesign of the
instrument can prevent or remove the artifact. A sensitive diagnostic detects
it. A post-hoc check merely updates your concern. Rewrite whatever you could
not defend aloud.

**What to log.** The strongest hit from each round and what you changed in
response. An attack that changed nothing still gets a line saying why it did
not land.

> **A question that often comes up here:** *"The licence hunt and the privacy
> attack both came back clean. Am I done?"* Clean is a result only if the
> checks were real: the licence page opened and read, the probe run on your
> actual kept columns. A clean bill from an unrun check is the exact failure
> this milestone exists to catch.
