## 🎯 Definition of Done

You are done with this sprint when your submission carries all seven parts of
the governed-data record, assembled as one document, applied to **your own
project**:

1. **The acquisition route and the provenance record.** Your **acquisition
   route** is the declared path by which your data come to exist, and your M4
   permission status decides it. *Cleared*: name your sampling frame, how you
   reach those units, and what you do about the ones you cannot reach.
   *Pending*: do everything that does not touch people, piloting the
   instrument on yourself and writing analysis code against simulated data of
   the expected shape. *No authority to ask*: take the no-permission route
   (published aggregates, openly licensed data with no individual records, or
   simulation) and state its cost. Then one **provenance** entry per dataset
   and per borrowed number: who produced it, how, when, under what terms, with
   a retrieval date on every entry, and any missing link written as "unknown"
   rather than filled with a plausible guess.
2. **The data-management record.** Where the raw file sits, where the working
   copy sits, who besides you can open either, and how the two are kept
   distinct. Every column you kept tied to the analysis your Contract
   declares, with what you dropped or coarsened since M4 recorded. And the
   **re-identification check** rerun on the columns that actually arrived,
   with what it found and what you changed in response.
3. **The route-specific permission recheck.** Your M4 determination held
   against the data as they actually arrived, in the form your route requires.
   *Cleared*: did the units you reached stay inside what you were cleared to
   collect, and did a new field appear that the determination never covered?
   *Pending*: did anything you did this week touch a person? *No permission*:
   do the aggregates, open data, or simulated data you used carry terms that
   permit your use, and did you state the cost to your claim? Close with one
   declared outcome. A status that moved to **not authorized** is a blocking
   gate: stop and bring it to the studio before anything else.
4. **The measurement specification.** The concept → construct → indicator
   ladder for every measured concept, each rung visibly narrowing the one
   above it and the construct-indicator **gap** named. A **meaning sentence**
   per indicator, saying what a single recorded value asserts. One
   **reliability** check run on items, occasions, or raters, never on split
   respondents, with its result reported; where no such check exists for your
   measure and use, write that down as the honest finding it is and say what
   you narrowed. And **validity** argued as interpretation and use: the
   interpretation, the use, the strongest rival reading that would produce the
   same number, and the boundary sentence naming what your indicator misses
   and who it never reaches.
5. **The Contract version.** Your operationalization was provisional at M4 by
   design, and it is provisional no longer. Issue the settled specification as
   the next dated, numbered version of your **Research Contract** (v2, if M5
   issued v1) with a reason a reader could reconstruct. Three lines are
   mandatory: the measurement fields settled into the Contract, the
   provisional M4 field named and its change explained, and measurement error
   stated alongside your sampling uncertainty rather than instead of it.
6. **AI Research Ledger rows.** One row for every AI-assisted step, each with
   a verification method run against the actual file or page.
7. **The dossier update line.** Where your data and measurement documentation
   component now lives, named by file or section.

> **A question that often comes up here:** *"My data have not arrived yet, or
> I have no permission to collect any. How do I govern data I do not have?"*
> You govern the route your permission status assigns. Pending means you
> pilot the instrument on yourself and write your analysis code against
> simulated data of the shape you expect, while you wait. No authority to ask
> means the no-permission route: published aggregates, an openly licensed
> dataset, or data you simulate yourself, with the cost written down. A
> simulated dataset can demonstrate a method, but it cannot support a claim
> about the world.

### 🤝 AI Research Partner

AI is your arm and your research assistant in this sprint, not your brain.
Task it with the mechanical work, keep every judgment for yourself.

**Good to delegate here:**
- Hunting the licence or terms-of-use page for a dataset you name, for you to
  open and read yourself.
- Drafting the provenance table's format, or proposing indicator wordings for
  you to judge.
- Listing what an indicator might fail to capture, and the rival readings that
  could produce the same number.
- Red-teaming your re-identification recheck: proposing column combinations
  that could pick out an individual.

**Never delegate (these are yours to decide):**
- Whether a measure measures your concept. Choosing measures and judging data
  quality is a never-delegate decision.
- Which columns your declared analysis actually needs, and which get dropped.
- The one trap that governs the whole week: an assistant can describe a
  dataset it has never seen, fluently and wrongly. Every AI claim about your
  data gets verified against the file itself.

Whatever you delegate, verify what comes back and log it in your **AI

>

> generate rivals that sound serious and do not apply to your setting):**
> - [ ] Keep only the rivals that are live in your own setting, and say in one

> - [ ] Write the strongest survivor into your validity argument, with the

> - [ ] Write the boundary sentence yourself: what your indicator misses, and

```python
# Governance self-check — edit the EDIT ME blocks to match YOUR project.
# Four checks in one cell: (1) provenance entries with missing fields flagged,
# (2) the data-management record, (3) column minimization against your declared
# analysis, (4) a seeded re-identification probe on a table shaped like yours.

# EDIT ME — one provenance entry per dataset and per borrowed number. Writing
# "unknown" in a field counts as complete: an unknown producer is a finding
# about your evidence, and it belongs on the page.
provenance = [
    {"source": "EDIT ME (dataset or borrowed number)", "producer": "",
     "method": "", "produced": "", "terms": "", "retrieved": ""},
]

# EDIT ME — where your data live now and who can reach them.
data_management = {
    "raw file location":       "EDIT ME (where the untouched file sits)",
    "working copy location":   "",
    "who else can open them":  "",
    "raw and working kept distinct by": "",
}

# EDIT ME — every column that ACTUALLY ARRIVED and you plan to keep, mapped to
# the declared analysis that needs it. An empty string means no analysis behind
# it: a flag.
kept_columns = {
    "hall":     "EDIT ME (which declared analysis needs this?)",
    "major":    "",
    "hometown": "",
    "score":    "the primary outcome of my declared analysis",
}

# EDIT ME — for the probe: plausible category counts for your quasi-identifier
# columns (how many distinct values each could take in your data).
quasi_levels = {"hall": 3, "major": 12, "hometown": 40}
n_rows = 200   # about the size of your expected dataset

print("— Provenance check —")
for e in provenance:
    missing = [k for k, v in e.items() if k != "source" and not str(v).strip()]
    mark = "⚠️ incomplete: missing " + ", ".join(missing) if missing else "✓ complete"
    print(f"  {e['source']}: {mark}")

print("— Data-management record —")
for field, value in data_management.items():
    ok = value.strip() and not value.startswith("EDIT ME")
    print(f"  {field}: " + (f"✓ {value}" if ok
                            else "⚠️ not filled in — a stranger could not act on this record"))

print("— Minimization check —")
for col, analysis in kept_columns.items():
    ok = analysis.strip() and not analysis.startswith("EDIT ME")
    print(f"  {col}: " + (f"✓ tied to: {analysis}" if ok
                          else "⚠️ no declared analysis behind it — drop, coarsen, or justify"))

print("— Re-identification probe (seeded simulation) —")
qcols = [c for c in quasi_levels if c in kept_columns]
sim = {c: rng.integers(0, quasi_levels[c], size=n_rows) for c in qcols}
combos = list(zip(*(sim[c] for c in qcols)))
unique_share = sum(1 for c in combos if combos.count(c) == 1) / n_rows
print(f"  kept quasi-identifiers: {qcols}")
print(f"  share of simulated rows UNIQUE on their intersection: {unique_share:.0%}")
print("  A high share means a few columns can pick out one person even with")
print("  no names. Coarsen or drop until the share is defensible, then record")
print("  the change in the data-management record and rerun.")
```

```python
# Reliability harness — reliability lives in your INSTRUMENT, not in your
# respondents. The seeded simulation below builds a three-item scale, then runs
# the two checks side by side so you can see which one reads the instrument.
# Left: agreement AMONG THE ITEMS, a defensible reliability check. Right: a
# split of respondents into halves, which is reassuring and tells you nothing
# about whether your instrument reads consistently.

n = 160
true_belonging = rng.normal(0, 1, size=n)          # what the scale is aiming at
noise = 0.8                                        # per-item reading noise
items = np.column_stack([true_belonging + rng.normal(0, noise, size=n)
                         for _ in range(3)])

# Check 1 — agreement among the items (on items, as required).
pairs = [(0, 1), (0, 2), (1, 2)]
inter_item = [float(np.corrcoef(items[:, a], items[:, b])[0, 1]) for a, b in pairs]
mean_r = float(np.mean(inter_item))

# Check 2 — the respondent split that measures sampling, not the instrument.
half = rng.permutation(n)
scale_score = items.mean(axis=1)
mean_a = float(scale_score[half[: n // 2]].mean())
mean_b = float(scale_score[half[n // 2:]].mean())

fig, axes = plt.subplots(1, 2, figsize=(11, 4.2))
axes[0].scatter(items[:, 0], items[:, 1], s=14, color="#1a7f37", alpha=0.7)
axes[0].set_title(f"On the ITEMS: item 1 vs item 2 (r = {inter_item[0]:.2f})",
                  fontsize=10)
axes[0].set_xlabel("item 1 reading"); axes[0].set_ylabel("item 2 reading")
axes[1].bar(["respondent half A", "respondent half B"], [mean_a, mean_b],
            color="#b3261e", alpha=0.75)
axes[1].set_title("On the RESPONDENTS: two halves' mean scores", fontsize=10)
axes[1].set_ylabel("mean scale score")
plt.tight_layout()
plt.show()

print("Reading the output.")
print(f"  Average agreement among the three items: r = {mean_r:.2f}. That is a")
print("  reliability reading: the items are reading the same thing.")
print(f"  Respondent halves: {mean_a:.2f} vs {mean_b:.2f}. Whatever that gap is,")
print("  it is beside the point: splitting respondents reads sampling wobble,")
print("  and it would read the same way with items that agreed on nothing.")
print("  Report the LEFT check. If your indicator is a single item measured")
print("  once, no defensible reliability check is available for that measure")
print("  and use: write that down, then narrow the claim or change the")
print("  instrument.")
```

### 🗡️ Red-Team Exchange

The attack runs on your own governed-data record, at the studio with your AI
assistant. Two rounds.

**Round 1: attack the provenance.**
Ask your AI what your dataset contains, how a variable is coded, or what its
licence allows, and then catch it: hold every answer against the file itself.
Count the rows, read the codebook, read the licence page. Any claim the file
does not confirm is the week's central trap made visible, a fluent description
of data the tool has never seen. Record one place the against-the-file check
corrected or confirmed the tool, and one place you wrote "unknown" instead of
a guess.

**Round 2: attack the measurement specification.**
Press on your own ladder the way a reviewer will. Does each rung narrow the
one above it? Can you say the meaning sentence for every indicator without
notes? Is the reliability check on items, occasions, or raters, or did a
respondent-split sneak in? What is the strongest rival reading of your
indicator, and does your validity argument answer it? And if an instrument
artifact worries you (a wording nudge, a this-day mood effect), keep the verbs
straight: only a redesign of the instrument can prevent or remove the
artifact. A sensitive diagnostic detects it. A post-hoc check merely updates
your concern. Rewrite whatever you could not defend aloud.

**What to log.** The strongest hit from each round and what you changed in
response. An attack that changed nothing still gets a line saying why it did
not land.
