### 🧩 Research Puzzle

A visitor stops at your board. They give you ninety seconds of attention that
they did not plan to give you, and most of that time their eyes are moving, not
your mouth.

Everything your project knows has to survive that compression. The dangerous part
is that compression is not neutral. When you cut words, the words that survive
tend to be the strong ones, because qualifiers are long and verbs are short. A
finding that was carefully bounded in your research note arrives on the board as
a claim that reaches further than your design licenses, and nobody decided to do
that. The layout did it.

So the question for this lecture is not "how do I make a nice poster". It is:
**what does my board make a stranger conclude, and is that what my evidence
says?**

## 1. The Reading Compass

A poster is read in an order, and the order is not the order you wrote it in.

**Reading compass.** The positions a visitor's eye actually travels on a board,
in sequence, and what belongs in each. Example: on a standard three-column URC
board, the eye lands first on the upper-left headline zone, sweeps right across
the top, and only then drops into the columns. That first landing is where your
finding belongs, not where your literature review belongs.

**The ninety-second read.** Everything a visitor gets before they decide to stay
or walk. It is not a summary of your poster. It is whatever their eye collected
on its own, in order, without you talking.

Two consequences follow, and both are checkable:

**The eye-span rule.** A claim and the uncertainty that qualifies it are read
together, or they are not read together at all. **Eye-span** means the region a
reader takes in without moving their head. Example: a headline that says
"turnout was 6 points higher" with "±2.4 points" on the next line is one claim;
the same headline with the interval in a bottom-right footnote is two separate
objects, and most visitors will only ever read one of them.

**The traceability rule.** Every number printed on the board reaches a named cell
in a named notebook. A **claim-to-output trace** is that mapping, written down.
A number with no run behind it is a fabricated number, whatever route it took
onto the poster.

> **A question that often comes up here:** *"Isn't 'fabricated' a bit strong for
> a number I copied out of my own write-up?"* It is the right word, and not as an
> accusation. The point is about what you can defend, not about what you
> intended. If a reader asks where a number came from and the honest answer is
> "from a sentence I wrote earlier", you cannot show them the run, and neither
> can you.

### 🛠️ Run It Live: the traceability sweep on a worked board

Here is a worked poster. Someone built it from `lapop_brazil` during a busy week,
and its four printed values are recorded in an export manifest. The next cell
recomputes each value from the data as it stands now, and compares.

This is the sweep you will run on your own board tonight. The only difference is
that yours has your numbers in it.

```python
# The worked board's export manifest: what the poster PRINTS.
board = pd.DataFrame([
    {"block": "Headline",  "label": "Trust gap (responsive vs not)", "printed": 0.99},
    {"block": "Headline",  "label": "Interval half-width",           "printed": 0.06},
    {"block": "Data",      "label": "Respondents rating responsive high", "printed": 4519},
    {"block": "Result",    "label": "Mean trust, low-responsive group",   "printed": 3.90},
])
board
```

```python
# Recompute each printed value from the data the board claims to rest on.
lapop = load_course_data("lapop_brazil.csv")

high = lapop.loc[lapop["govt_responsive"] >= 4, "trust_police"]
low  = lapop.loc[lapop["govt_responsive"] <= 2, "trust_police"]

gap = high.mean() - low.mean()
# Standard error of a difference in means, then a 95% half-width.
se = np.sqrt(high.var(ddof=1) / len(high) + low.var(ddof=1) / len(low))
half = 1.96 * se

recomputed = {
    "Trust gap (responsive vs not)":        round(gap, 2),
    "Interval half-width":                  round(half, 2),
    "Respondents rating responsive high":   int(len(high)),
    "Mean trust, low-responsive group":     round(low.mean(), 2),
}

board["recomputed"] = board["label"].map(recomputed)
board["matches"] = np.isclose(board["printed"], board["recomputed"], atol=0.005)
board["disposition"] = np.where(board["matches"], "clear", "BLOCKING")
print(board[["block", "label", "printed", "recomputed", "disposition"]].to_string(index=False))
print()
n_bad = int((~board["matches"]).sum())
print(f"Blocking items: {n_bad} of {len(board)}")
print("A board with any BLOCKING row does not lock." if n_bad else "Board clear to lock.")
```

## 2. Where the Headline Goes

A number that traces is not yet a poster. The second half of this lecture is
placement.

Draw your board as **six blocks on paper before you open the template**. A poster
laid out inside the software takes the shape the software offers. A poster laid
out on paper takes the shape the argument needs, and then you make the software
serve it.

The six blocks, in the order a visitor meets them:

1. **The question**, in the words a non-specialist would use.
2. **The headline finding**, with its uncertainty inside the same eye-span.
3. **The boundary sentence**: where your evidence stops.
4. **The design and the data**: what you did, and to what.
5. **The result panel**: one figure that carries the finding, not four that
   surround it.
6. **The disclosure and the citations**: what AI did, and what you read.

Notice what is not on the list. There is no literature review block, no
methods appendix, and no acknowledgement banner big enough to be read from two
metres. Space on a board is the scarcest thing you have.

> **A question that often comes up here:** *"What if my design genuinely needs
> explaining before the finding makes sense?"* Then the design block moves up and
> the finding follows it, and you have made a real choice you can defend. What
> does not work is keeping both at full length. A board that explains everything
> in order is a paper printed large, and nobody reads a paper standing up.

## 3. Findings, Preferences, and the Wrong-Inference Test

Criticism is only useful if it can be acted on, argued with, and checked later.
That gives us a hard line.

**A finding** names three things: the block it lives in, the defect, and the
change that would fix it. Example: "The three-colour result panel makes the
non-significant bar read as a second finding; use one colour and label the bar."

**A preference** names none of them. Example: "The colours are a bit much."
A preference cannot be repaired, cannot be rejected on the merits, and cannot be
audited by anyone later. It is not criticism. It is a mood.

**The wrong-inference test.** What did you, the reader, conclude from this board
alone, and is that what the author claimed? This is the single most valuable
thing a reviewer produces, and it is the one that requires no expertise in the
author's topic at all. You do not need to know their field to report honestly
what their board made you believe.

**Calibration.** Naming what is genuinely sound, specifically. A sheet that is
all defects tells the author nothing about what to protect while revising, and a
reviewer who never calibrates is a reviewer whose criticism cannot be trusted.
This is required on every sheet, not encouraged.

> **A question that often comes up here:** *"What if I genuinely cannot find
> anything sound on a board?"* Look again at a smaller scale. A clearly worded
> question, one honest interval, a figure whose axis is labelled: these are
> ordinary and they are also real. A reviewer who cannot name one is usually
> reading for defects rather than reading the board.

### 🛠️ Run It Live: sorting a returned sheet

Here is a returned review sheet from a previous board. Some of these lines are
findings and some are preferences. The next cell scores each one mechanically
against the three parts a finding must have, so you can see how thin most first
drafts of criticism are.

The classifier is deliberately crude. It looks for a named block, a stated
defect, and a proposed repair. It cannot judge whether the criticism is *right*,
which is your job, and it will not pretend otherwise.

```python
# A returned review sheet: eight lines, as written.
sheet = [
    "Result panel: the non-significant bar is the same colour as the significant one, so it reads as a second finding; use one colour and label it.",
    "The colours are a bit much.",
    "Headline: the verb 'raises' claims causation the observational design cannot support; change to 'is higher among'.",
    "Nice figure!",
    "Data block: sample size is given but the exclusion rule is not, so a reader cannot tell who is missing; add the one-line exclusion rule.",
    "I would have used a different font.",
    "Boundary sentence is present but sits below the fold in the third column; move it under the headline.",
    "This is really interesting work.",
]

BLOCK_WORDS = ("headline", "result", "data", "boundary", "panel", "block",
               "figure", "column", "title", "disclosure")
REPAIR_WORDS = ("use ", "add ", "change ", "move ", "cut ", "label ",
                "replace ", "state ")
DEFECT_WORDS = ("so ", "because ", "cannot ", "reads as", "does not",
                "claims ", "not given", "missing")

rows = []
for line in sheet:
    low = line.lower()
    has_block  = any(w in low for w in BLOCK_WORDS)
    has_defect = any(w in low for w in DEFECT_WORDS)
    has_repair = any(w in low for w in REPAIR_WORDS)
    parts = sum([has_block, has_defect, has_repair])
    rows.append({"line": line[:58] + ("…" if len(line) > 58 else ""),
                 "block": has_block, "defect": has_defect, "repair": has_repair,
                 "verdict": "finding" if parts == 3 else "preference"})

sheet_df = pd.DataFrame(rows)
print(sheet_df.to_string(index=False))
print()
n_find = int((sheet_df["verdict"] == "finding").sum())
print(f"Findings: {n_find} of {len(sheet_df)}  ·  Preferences: {len(sheet_df) - n_find}")
print("A sheet is worth the reviewer's hour only in proportion to its findings.")
```
