## 1. Three Buckets Before You Open the Tool

**Guiding question:** *which pieces of your project may be handed to a tool,
which must be handed out and then checked, and which never leave your hands?*

> *"A research assistant here can run the assay, plot the readings, and format
> the table, honestly faster than I would. What they do not do is choose the
> question, decide what the readings mean, or sign the claim. Learn where that
> line sits, because it does not move just because the assistant got faster."*
> — a **principal investigator**, briefing a new lab member

Your puzzle already found the buckets. Look again at the three pieces of that
creek sentence, because each one sits in a different place.

- **Safe to delegate** means low stakes and easy to check, so a wrong answer
  costs almost nothing. Example: reformatting three plates of raw numbers into
  one tidy table, where a value in the wrong column is caught at a glance.
- **Delegate then verify** means fine to hand out, but confirmed by a real check
  before you trust it. Example: asking a tool to locate a published threshold,
  then opening the agency's actual document and reading the figure yourself.
- **Never delegate** means a judgment that stays yours no matter how capable the
  tool becomes. Example: deciding whether ten samples on one day earn the word
  "unsafe" for a whole creek all winter.

Sorting your whole task list into those three buckets, before you start, is a
**delegation triage**. Example: you write out the twelve things your project
needs, and beside each one you write which bucket it belongs in and why.

The third bucket has a name worth memorizing. A **never-delegate decision** is
any call that defines the science itself: which question you pursue, what your
evidence licenses, the ethics of the work, the boundary of your claim, and the
uncertainty you report. Example: a tool may list five ways to measure water
quality, but choosing the one you will defend is yours.

Now sort the creek sentence aloud. The mean of ten samples is a recomputation,
so it goes in delegate then verify. The published criterion is a document you
open, so it also goes in delegate then verify, with a different check. The word
"unsafe" is a claim about a whole creek from one day of sampling, so it never
leaves your hands.

> **A question that often comes up here:** *"Is 'delegate then verify' just a
> polite way of saying do the work twice?"* No. Verifying is almost always
> cheaper than producing. Recomputing one mean takes a minute; collecting the
> samples took a morning. What you are buying with that minute is the right to
> put the number in a sentence with your name on it.

## 2. The Route a Delegated Task Travels

**Guiding question:** *once you decide to hand something out, what fixed route
does it travel before you are willing to believe it?*

The buckets say what may leave your hands. They do not say what happens to it
while it is gone. That is what this section fixes.

First, three names for the way a careful reader gets fooled. You cannot guard
against a danger you cannot name.

- **Automation bias** is over-trusting an automated system and stopping the
  check precisely because it is automated. Example: following a navigation app
  into a closed road because the machine said turn.
- **False confidence** is when fluent, detailed output makes you feel sure of
  something you never actually verified. Example: a polished paragraph makes a
  wrong statistic feel solid, so you quote it.
- **Illusion of understanding** is mistaking a smooth explanation for real
  comprehension. Example: you nod along to a line-by-line walkthrough of a cell,
  then find you could not defend it to a skeptic.

All three are cheap to fall into and cheap to beat, but only if the check is a
fixed habit rather than a decision you make when something happens to feel off.
That is what the rest of this section installs.

**SDIIVDD** is a **checklist**, which is a fixed sequence of steps you run every
time so you never skip the one that bites. Its name is the first letter of each
step: **Specify, Delegate, Interrogate, Inspect, Verify, Document, Defend**.
Example: a pilot runs a preflight checklist not because they forgot how to fly,
but because the skipped item is the one that costs.

Here is each step, once, with what it actually asks of you.

1. **Specify.** Write the exact task and your own expected answer before you
   ask. Example: "I expect the new page to win by a few percent, not by half."
2. **Delegate.** Hand out the well-specified piece, not your judgment.
3. **Interrogate.** Ask the draft what it assumes. Example: "when you report
   turnout, is the denominator registered voters or every adult?"
4. **Inspect.** Read what the tool actually produced, not its summary of itself.
5. **Verify.** Confirm the result with a real, independent check.
6. **Document.** Log the tool, the task, the prompt, and the check.
7. **Defend.** State the result in your own words, with the tool out of the room.

Two of the seven stay human no matter how good the tool gets: **Specify** and
**Defend**. They are the human bookends, and everything you hand off sits safely
between them. That is exactly why the room can hand off the middle at all.

**What the next cell does:** draws the seven steps as one chain, left to right.
Expect seven boxes, with the two human bookends drawn in a darker outline, and a
✓ line naming the everyday shorthand.

```python
# The route a delegated task travels: Specify -> Delegate -> Interrogate ->
# Inspect -> Verify -> Document -> Defend. Specify and Defend are the human bookends.
steps = ["Specify", "Delegate", "Interrogate", "Inspect", "Verify", "Document", "Defend"]
G = nx.DiGraph()
G.add_edges_from([(steps[i], steps[i + 1]) for i in range(len(steps) - 1)])
pos = {s: (i, 0) for i, s in enumerate(steps)}

# The human-owned bookends get a darker outline AND are named in the title, so the
# meaning never rests on color alone. Verify, where trust is earned, is filled apart.
fill = ["#B7C9E2", "#DCE6F1", "#DCE6F1", "#DCE6F1", "#55A868", "#DCE6F1", "#B7C9E2"]
edgecol = ["#1F3B6E", "#4C72B0", "#4C72B0", "#4C72B0", "#2F6B45", "#4C72B0", "#1F3B6E"]

fig, ax = plt.subplots(figsize=(12, 3))
nx.draw_networkx_nodes(G, pos, node_color=fill, edgecolors=edgecol,
                       linewidths=2.0, node_size=3400, ax=ax)
nx.draw_networkx_labels(G, pos, font_size=8.5, ax=ax)
nx.draw_networkx_edges(G, pos, arrowstyle="-|>", arrowsize=16, edge_color="#555555",
                       width=1.5, node_size=3400, ax=ax)
ax.set_title("SDIIVDD: the route a delegated task travels, human at both ends (Specify, Defend)")
ax.set_xlim(-0.6, 6.6)
ax.set_ylim(-0.6, 0.6)
ax.axis("off")
plt.tight_layout()
plt.show()

print("✓ Chain drawn. Shorthand: Ask (Specify + Delegate), then Verify (Interrogate")
print("  + Inspect + Verify), then Document. Defend closes it, and it is yours.")
```

**Reading the output.** *In plain terms, the chain says: you set the task, hand
out the work, argue with the draft, look at what it really did, confirm it with a
real check, log it, and defend it yourself. The two darker circles at the ends
are the steps you never hand to a tool.*

## 3. The Record and the Name on the Work

**Guiding question:** *what has to be written down, and who answers for the
sentence when someone asks how you know?*

> *"When something in this paper turns out to be wrong, I do not ask which
> software wrote it. I ask whose name is on it. That person answers for every
> line."*
> — a **journal editor**, to a first-time author

The route you just ran produces two things you keep: a corrected reading, and a
record of how you got it. The record has a name. Your **AI Research Ledger** is
a running table with eight fields, one row per exchange: task delegated · tool
used · prompt · output summary · decision · verification method · remaining
concern · responsible researcher. Example: one row saying a tool listed the
claims in a drafted paragraph, that you rejected both, and that you recomputed
turnout from the file yourself.

One rule about the ledger surprises people, so it is worth stating plainly. A
delegation that happened before the ledger existed still gets a row. The date a
ledger opens does not erase an earlier delegation. So if you brainstormed with a
tool in Week 1, that exchange becomes row one, marked as written after the fact.

Two more terms finish the picture, and they are about you rather than the tool.
**Intellectual ownership** means being the author of, and answerable for, every
claim under your name, whatever tool helped produce it. Example: if a drafted
paragraph says a creek is unsafe and it is wrong, the mistake is yours.
**Accountability** is the part that shows up in public: being the person who
answers for the work when it is questioned. Example: at your defense, "the tool
computed it" is not an answer, and "I recomputed it from my own measurements, and
here is the check" is.

The habit that makes ownership visible is an **AI-use disclosure**: a short,
honest note saying which tool did which task and how you checked it. Here is how
it goes wrong. Asked to describe its own role, a tool wrote: *"AI was used only
for minor language editing."* It had drafted the entire results paragraph,
including the number and the word "unsafe". The disclosure sounds professional
and it is false. You catch it with a record the tool cannot see, which is your
ledger. Lay the one-line disclosure beside your rows and the mismatch is plain.
Where ledger and disclosure disagree, the ledger wins, and you rewrite the
disclosure. Never let a tool describe its own role.

> **A question that often comes up here:** *"Do I really have to log a use where
> nothing went wrong?"* Yes, and that row is one of the most valuable ones you
> will write. A verification whose outcome is "nothing changed, I reran the mean
> myself and got the same number" is a real result on record. Without it, nobody
> can tell the difference between a check you ran and a check you skipped.

## 4. The Inquiry Compass: Kind and Reach

**Guiding question:** *what KIND of answer does your question ask for, and how
far is it trying to reach?*

> *"Before I approve a single method, I make you answer two questions about your
> question. What kind of answer does it want, and for whom must that answer
> hold?"*
> — a **thesis advisor**, at your first committee meeting

The **inquiry compass** is the map that sorts any research question by exactly
two things. Together they place the question at one of four named positions, and
that position fixes what its answer may ever claim.

**Axis 1, KIND.**

- A **descriptive question** asks what the world is or was, with no intervention
  needed. Example: "what share of first-years skip breakfast?" You look and
  count.
- A **causal question** asks what would change if someone intervened, which
  forces a comparison to a world that did not happen. Example: "does a free
  breakfast raise attendance?" You have to imagine the same people without it.

**Axis 2, REACH.**

- **Data at hand**: only the units you actually observed.
- **A population**: a larger group beyond your data that your sampling reaches.
- **Unseen cases**: units you have not observed yet, usually future ones.

Cross the two and you get the four positions. **Description** is descriptive
about the data at hand. **Generalization** is descriptive about a population, and
it must pay for that reach with a sampling design. **Prediction** is descriptive
about unseen cases, forecasting a new case from a pattern. **Causal reasoning**
is the causal kind, and it is the only position that earns the word "because".

> **A question that often comes up here:** *"Prediction feels causal to me. If a
> model predicts dropout from attendance, doesn't attendance cause dropout?"* No,
> and this is the trap the compass exists to catch. A predictor only needs a
> pattern that travels to new cases. It can predict beautifully using a feature
> that causes nothing, as long as that feature rides along with the outcome.
> Performance is not understanding. Prediction sits in the descriptive half for
> exactly this reason.

**What the next cell does:** draws the compass as one grid, kind down the side
and reach across the top. Expect four labeled cells across two rows, with one
grey cell marked "(rare)".

```python
# The inquiry compass on one grid: KIND (rows) x REACH (columns).
fig, ax = plt.subplots(figsize=(9.5, 4.7))
reaches = ["Reach:\ndata at hand", "Reach:\na population", "Reach:\nunseen cases"]
desc_cells = {0: "Description", 1: "Generalization", 2: "Prediction"}

for cx in range(3):
    ax.add_patch(mpatches.Rectangle((cx, 1), 1, 1, facecolor="#EAF1F8",
                                    edgecolor="#33506E", linewidth=1.3))
    ax.text(cx + 0.5, 1.5, desc_cells[cx], ha="center", va="center",
            fontsize=11, fontweight="bold")

# Causal kind: one position, drawn across the reaches it can hold.
ax.add_patch(mpatches.Rectangle((0, 0), 2, 1, facecolor="#F2E4D3",
                                edgecolor="#33506E", linewidth=1.3))
ax.text(1.0, 0.5, "Causal reasoning", ha="center", va="center",
        fontsize=11, fontweight="bold")
ax.add_patch(mpatches.Rectangle((2, 0), 1, 1, facecolor="#E8E8E8",
                                edgecolor="#BBBBBB", linewidth=1.0))
ax.text(2.5, 0.5, "(rare)", ha="center", va="center", fontsize=9, color="#777777")

ax.text(-0.12, 1.5, "Descriptive", ha="right", va="center", fontsize=11, rotation=90)
ax.text(-0.12, 0.5, "Causal", ha="right", va="center", fontsize=11, rotation=90)
for cx in range(3):
    ax.text(cx + 0.5, 2.14, reaches[cx], ha="center", va="center", fontsize=9)

ax.set_xlim(-0.65, 3.05)
ax.set_ylim(-0.15, 2.4)
ax.axis("off")
ax.set_title("The inquiry compass: KIND (rows) x REACH (columns)")
plt.tight_layout()
plt.show()

print("✓ Compass drawn. Read any question by asking two things:")
print("  KIND  -> descriptive or causal?   REACH -> data at hand, a population, or unseen?")
```

**Reading the output.** *In plain terms, the grid says every question lands in
exactly one box. The top row is the descriptive half, split by how far it
reaches. The bottom row is the causal half, which is defined by the kind of
answer it wants rather than by reach.*

Now the rule this week is graded on, and it is the one most often broken. **The
kind is fixed by the question's own words, never by your data.** A causal
question your design cannot isolate stays causal, and its honest status is
"causal, currently unidentified". The descriptive result you can report sits
beside it, never in place of it.

> **A question that often comes up here:** *"If my data cannot answer a causal
> question, should I just rewrite it as a descriptive one?"* Only if a
> descriptive question is genuinely what you now want to ask. Your current result
> may well be descriptive while your causal question sits unanswered beside it.
> Write "currently unidentified" next to the causal one, then decide whether to
> redesign or to change questions. Both are respectable. Quietly swapping the
> question while keeping the causal wording is not.

One more hazard, which your puzzle already found. A **double-barreled question**
is one sentence that secretly asks two questions of different kinds. Example:
"how common is burnout, and does night work cause it?" It cannot be answered as
one, and the fix is always to split it into two clean questions, each with its
own position.

## 8. Optional Depth: Sorting by Words Versus by the Answer

**Guiding question:** *why does classifying a question by its giveaway words
fail, and what does it fail on?*

Two tools finish the compass skill.

- **Giveaway words** are the words that hint at a position: "cause", "effect",
  and "raise" lean causal; "predict", "will", and "next" lean prediction; "all"
  and "population" lean generalization. They are hints you check, never the
  verdict, because a question can wear a word it does not mean.
- A **naive keyword sorter** is a classifier that only scans for those words. It
  stands in for a hurried tool that classifies by surface pattern.

The cells below deal a deck of eight questions in a reproducible shuffled order
and put the keyword sorter's guess next to the position each question actually
holds.

**🏠 Optional depth.** Run this on your own if you want to go deeper.
**Before you ask:** commit your own position for question 5 and question 7 on
paper first, so you can tell whether your AI assistant reasons or just
pattern-matches.

> 💡 **AI Prompt:** "Here are eight research questions [paste the eight question
> texts from the next cell]. For each, tell me its KIND (descriptive or causal)
> and its REACH (data at hand, a population, or unseen cases), and name the
> single word that most influenced you. Put it in a table. Then, for any question
> whose answer would only need a pattern and not an intervention, say so
> explicitly."
>
> **After running, verify (counters *silent scope change*: a partner that quietly
> upgrades a descriptive question to a causal one has answered a different
> question than you asked):**
> - [ ] Compare the table to the "true position" column the cell printed. Every
>   disagreement is a question to reread against the two definitions, because one
>   of you is wrong.
> - [ ] Flag any question it calls causal that never mentions an intervention. If
>   the question really does ask what an intervention would change, it stays
>   causal even when your data cannot answer it: that status is "causal,
>   currently unidentified".
> - [ ] Log this use in your AI Research Ledger: task, tool, decision,
>   verification method, remaining concern.

```python
# A deck of eight sample questions with their reasoned-out true positions.
deck = pd.DataFrame([
    (1, "In our file of 10,000 interviews, what share of respondents rate trust in the police above 4?", "Description"),
    (2, "What percentage of all registered voters in the state would support ranked-choice voting?",      "Generalization"),
    (3, "Which of this year's applicants will still be enrolled three years from now?",                   "Prediction"),
    (4, "Does giving first-year majors a free breakfast raise their exam scores?",                        "Causal reasoning"),
    (5, "What share of campus buildings have effective ventilation during class hours?",                  "Description"),
    (6, "How many first-years skipped breakfast, and does skipping it lower their exam scores?",          "Double-barreled"),
    (7, "Do most adults across the country trust scientists more than elected officials?",                "Generalization"),
    (8, "Which of next year's incoming first-years will leave before their second year?",                 "Prediction"),
], columns=["id", "question", "true_position"])

# A naive keyword sorter: it only scans for giveaway words, in a fixed priority.
CAUSAL_WORDS  = ["cause", "effect", "affect", "impact", "raise", "lower", "reduce", "increase", "improve", "because"]
PREDICT_WORDS = ["will ", "predict", "forecast", " next ", "from now", "going to"]
POP_WORDS     = ["all registered", "percentage of all", "every citizen"]

def keyword_sort(text):
    t = text.lower()
    if any(w in t for w in CAUSAL_WORDS):
        return "Causal reasoning"
    if any(w in t for w in PREDICT_WORDS):
        return "Prediction"
    if any(w in t for w in POP_WORDS):
        return "Generalization"
    return "Description"

# Deal the deck in a reproducible shuffled order (seed 464), then classify.
deal = rng.permutation(deck.index.to_numpy())
deck = deck.loc[deal].reset_index(drop=True)
deck["keyword_guess"] = deck["question"].apply(keyword_sort)
deck["match"] = np.where(deck["keyword_guess"] == deck["true_position"], "ok", "MISS")

print("Dealt order (seed 464). Keyword sorter vs the true position:")
print(deck[["id", "true_position", "keyword_guess", "match"]].to_string(index=False))

n_ok = (deck["match"] == "ok").sum()
missed = sorted(deck.loc[deck["match"] == "MISS", "id"].tolist())
print(f"\nKeyword sorter got {n_ok} of {len(deck)} right ({n_ok / len(deck):.1%}).")
print(f"It missed question ids: {missed}")
```

```python
# Self-check: the keyword sorter's failures are the lesson, so pin them down.
assert n_ok == 5, "keyword sorter should get exactly 5 of 8"
assert missed == [5, 6, 7], "the misses should be questions 5, 6, and 7"
print("✓ Self-check passed: 5 of 8 correct; the three misses are 5, 6, and 7.")
print("  Q5 carries 'effective' (the sorter sees 'effect' and cries causal).")
print("  Q6 is double-barreled (the sorter grabs one half and drops the other).")
print("  Q7 is a population question with no keyword the brittle list knows.")
```

```python
# Plot each single-position question on the compass, marking hits and misses.
xy = {"Description": (0.5, 1.5), "Generalization": (1.5, 1.5),
      "Prediction": (2.5, 1.5), "Causal reasoning": (1.0, 0.5)}

fig, ax = plt.subplots(figsize=(9.5, 4.7))
for cx in range(3):
    ax.add_patch(mpatches.Rectangle((cx, 1), 1, 1, facecolor="#F4F8FC",
                                    edgecolor="#9BB6D2", linewidth=1.0))
ax.add_patch(mpatches.Rectangle((0, 0), 2, 1, facecolor="#FBF3E9",
                                edgecolor="#9BB6D2", linewidth=1.0))
for name, (x, y) in xy.items():
    ax.text(x, y + 0.34, name, ha="center", va="center", fontsize=9, color="#33506E")

single = deck[deck["true_position"] != "Double-barreled"].copy()
for pos_name, grp in single.groupby("true_position"):
    x0, y0 = xy[pos_name]
    xs = np.linspace(-0.22, 0.22, len(grp)) if len(grp) > 1 else [0.0]
    for dx, (_, r) in zip(xs, grp.iterrows()):
        hit = r["match"] == "ok"
        ax.scatter(x0 + dx, y0 - 0.05, s=230,
                   marker="o" if hit else "X",
                   color="#4C72B0" if hit else "#DD8452",
                   edgecolors="black", linewidths=0.8, zorder=3)
        ax.text(x0 + dx, y0 - 0.05, str(r["id"]), ha="center", va="center",
                fontsize=8, color="white", zorder=4, fontweight="bold")

ax.scatter([], [], marker="o", s=120, color="#4C72B0", edgecolors="black", label="sorter got it right")
ax.scatter([], [], marker="X", s=120, color="#DD8452", edgecolors="black", label="sorter missed")
ax.text(2.5, 0.5, "Q6 fuses two boxes\n(needs a split)", ha="center", va="center",
        fontsize=8.5, color="#8a5a2b", style="italic")
ax.set_xlim(-0.15, 3.05)
ax.set_ylim(-0.1, 2.15)
ax.axis("off")
ax.set_title("Where the keyword sorter lands each question (circle = right, X = wrong)")
ax.legend(loc="lower left", fontsize=8)
plt.tight_layout()
plt.show()

print("✓ Plotted. Shape carries the result: a circle is a hit, an X is a miss.")
```

**Reading the output.** *The picture pins each question to its true box on the
compass. A circle means the word-scanner landed it correctly; an X means it
missed. The two X marks (questions 5 and 7) plus the off-grid note for question 6
are the whole lesson: surface words are not the position.*

> - [ ] Log this use in your AI Research Ledger: task, tool, decision,

```python
# A double-barreled question comes apart into two clean, single-kind questions.
G2 = nx.DiGraph()
pos2 = {"fused": (0.0, 1.0), "descr": (2.0, 1.75), "causal": (2.0, 0.25)}
G2.add_edges_from([("fused", "descr"), ("fused", "causal")])
labels = {
    "fused":  "How common is burnout,\nand does night work cause it?\n(two questions in one)",
    "descr":  "How common is burnout?\n(Description)",
    "causal": "Does night work\ncause burnout?\n(Causal reasoning)",
}
colors = {"fused": "#E8E8E8", "descr": "#EAF1F8", "causal": "#FBF3E9"}

fig, ax = plt.subplots(figsize=(9.5, 4.3))
nx.draw_networkx_edges(G2, pos2, arrowstyle="-|>", arrowsize=20, edge_color="#555555",
                       width=1.6, node_size=15000, ax=ax)
nx.draw_networkx_nodes(G2, pos2, nodelist=list(labels),
                       node_color=[colors[n] for n in labels],
                       edgecolors="#33506E", linewidths=1.4, node_size=15000,
                       node_shape="s", ax=ax)
nx.draw_networkx_labels(G2, pos2, labels=labels, font_size=8.5, ax=ax)
ax.set_title("Split a double-barreled question into one descriptive and one causal question")
ax.axis("off")
plt.tight_layout()
plt.show()

print("✓ Split drawn. One fused sentence becomes two clean questions,")
print("  each with its own single compass position. Now each half can be answered.")
```

**Reading the output.** *The diagram shows one fused sentence branching into two
arrows, one to a descriptive question and one to a causal question. Neither box
can hold both, which is why the fused sentence had to be split before it could be
answered.*
