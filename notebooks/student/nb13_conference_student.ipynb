## 1. The Lock Froze the Board, Not the Claim

*(minutes 9–14)*

**Guiding question:** *what did the lock actually freeze, and what is still
moving underneath your poster?*

> *"The board is the smallest part of what you are bringing tomorrow. I can read
> it in forty seconds. What I cannot read is whether the numbers on it still come
> out of your code, and whether you can say where the claim stops without me
> asking twice."*
> — your **thesis advisor**, the day before the conference

A **poster lock** is the deadline after which the artifact cannot change. Example:
once the file goes to print, a typo in your headline stays a typo through the
whole session. The lock is a real constraint and it is also a narrow one. It
froze the layout, the wording, the figures, and the printed numbers. It froze
nothing else.

Everything on the other side of that line is still live: your notebook, your data
file, the code that produced each value, the claim you are willing to defend, and
every answer that will come out of your mouth tomorrow. Those are the pieces
today is about.

**Testing an artifact in its actual mode** means checking it the way its audience
will meet it, not the way you built it. Example: a poster gets scanned from a
few feet away by someone walking past, so you test it by standing it up and
walking past it, never by zooming in on your laptop. Today you run that test on
the two things the lock could not certify: your spoken spine, and whether your
printed numbers still reproduce.

### The dress rotation: both spines, one boundary sentence

Your pitch has two lengths and they share one skeleton. A **spine** is the fixed
order of beats you speak, at whatever length you have. The **30-second spine** is
hook, claim, boundary, invitation, with no evidence beats. The **3-minute spine**
is the same skeleton with two or three evidence beats added.

The beat that gets dropped under pressure is always the same one. Your
**boundary sentence** is the single line that says where your claim stops.
Example: "this compares people who chose the compact toolbar with people who did
not, so I cannot say the toolbar caused the difference." Say it at BOTH lengths,
word for word. A 30-second version that drops the boundary is not a shorter
claim, it is a different one.

**How the rotation runs.** Each person stands their board for 90 seconds: the
30-second spine, then the 3-minute spine compressed into what is left. Before
you stand, write down the one question type you most fear. Before you visit
someone, write down the type you will ask, drawn from the card below, and ask
that type. Afterward, each presenter writes one line comparing the type they
feared with the type they actually received.

A visitor is not there to be nice. The **visitor role card** spreads five kinds
of question across the boards you visit, so you feel each kind before a stranger
asks it for real:

- **method**: "how exactly did you estimate that?"
- **alternative-explanation**: "could something you did not measure explain the gap?"
- **so-what**: "what changes if you are right?"
- **generalization**: "would this hold beyond the sample you studied?"
- **AI-use**: "did you use AI in the analysis, and how did you check it?"

Keep that card. It is the vocabulary you will tag questions with tomorrow, and
the column you will count them in on Friday.

> **A question that often comes up here:** *"Isn't rehearsing this late too
> late to change anything?"* It is far too late to change your design, and that
> is not what this is for. It is for finding the piece that is not ready while
> there is still a tonight: a spine that runs long, a boundary you keep dropping,
> a question that makes you stall, a number you cannot trace on the spot. Small,
> fixable, and invisible until you run it cold.

### 🛠️ Run It Live: the package-current check on a locked board

**🔴 Live in class: we run this one together.**

**Before you run it:** the worked board reports four values from a survey of
10,000 respondents in Brazil. It compares support for the political system
between people high and low on an institutional-trust index. You just predicted
whether anything moved on your own board. Predict here too, in one word: how
many of these four do you expect to still reproduce?

The first cell loads the data and states the manifest exactly as it was printed.
The second recomputes each value from the code the notebook has now.

```python
# The locked board's export manifest: the four values printed on the poster,
# written down exactly as they appear on it. Nothing here can change.
printed = {
    "N (respondents)":     "10,000",
    "Trust-index mean":    "4.16",
    "High-trust group n":  "5,210",
    "Gap in mean support": "1.98  95% CI [1.90, 2.05]",
}

lapop = load_course_data("lapop_brazil.csv")
print(f"✓ Loaded lapop_brazil.csv — {lapop.shape[0]:,} rows x {lapop.shape[1]} columns")
print()
print("PRINTED ON THE LOCKED BOARD")
for label, value in printed.items():
    print(f"  {label:<21} {value}")
```

**Reading the output.** Four values, one board, no way to change any of them.
The index runs from 1 to 7 and averages a respondent's trust in institutions;
the gap compares mean support for the political system between the high-trust
group and everyone else. That is the artifact. Now the check.

```python
# Rebuild each printed value from the code this notebook has NOW.
# One thing changed after the export: govt_pride left the trust index during the
# measurement review, because pride in the government is an attitude toward the
# people currently in office, not trust in an institution. That is a validity
# decision, not a mistake. It is also the only change.
INDEX_NOW       = ["trust_police", "trust_military", "trust_supreme_court"]
INDEX_AT_EXPORT = INDEX_NOW + ["govt_pride"]

def board_numbers(items):
    """Recompute the four printed values from a given set of index items."""
    idx  = lapop[items].mean(axis=1)
    high = idx > idx.mean()                       # the high-trust group
    a = lapop.loc[high,  "support_political_system"]
    b = lapop.loc[~high, "support_political_system"]
    d  = a.mean() - b.mean()
    se = np.sqrt(a.var(ddof=1) / len(a) + b.var(ddof=1) / len(b))
    return {
        "N (respondents)":     f"{len(lapop):,}",
        "Trust-index mean":    f"{idx.mean():.2f}",
        "High-trust group n":  f"{int(high.sum()):,}",
        "Gap in mean support": f"{d:.2f}  95% CI [{d - 1.96 * se:.2f}, {d + 1.96 * se:.2f}]",
    }

check = pd.DataFrame({"PRINTED (locked board)": printed,
                      "RECOMPUTED (code now)":  board_numbers(INDEX_NOW)})
check["verdict"] = np.where(check.iloc[:, 0] == check.iloc[:, 1], "MATCH", "STALE")
print(check.to_string())

n_stale = int((check["verdict"] == "STALE").sum())
print()
print(f"{n_stale} of {len(check)} printed values no longer reproduce from the current code.")

# Where does the discrepancy live? Rebuild with the items the index had at export.
print()
print("Rebuilt with the four items the index carried at export time:")
for label, value in board_numbers(INDEX_AT_EXPORT).items():
    verdict = "matches the board" if value == printed[label] else "still differs"
    print(f"  {label:<21} {value:<26} ({verdict})")
```

**Reading the output.** Three of the four printed values no longer reproduce.
The sample size is unchanged, but the index mean, the size of the high-trust
group, and the headline gap all moved, and the gap moved by a third of a point
with an interval that no longer overlaps the printed one.

The second block is the part that matters more. Rebuild the index with the four
items it carried at export time and all four printed values come back exactly.
So the three discrepancies are not three problems. They are one choice, the
membership of the index, showing up in three places. That is what a good check
buys you: a location, not just an alarm.

*In plain terms: the board did not change, the code did, and one edit explains
everything that moved.*

## 3. Feedback Is Data. Code It Like Data.

*(minutes 31–50)*

**Guiding question:** *what do you owe a reader when the artifact cannot be
reprinted, and how does every question you get tomorrow become something you can
count?*

Two things get formalized here, and then the room stops.

**The package-current gate.** In its release-audit wording: any change to a
number, a claim, the data, the code, the environment record, or the instructions
a stranger would follow, made after the run that produced the artifact, makes
that run stale. Each row of a release audit carries one of three
**dispositions**:

- **clear**: the check ran and nothing has moved since.
- **pending**: something moved, the repair is named, and the repair has not
  happened yet.
- **blocking**: something moved that changes what the artifact asserts, and it
  must be repaired or disclosed before the work goes out.

**The locked-artifact corollary.** When the artifact cannot be reprinted, the
repair is disclosure at the point of presentation, plus a dated version reason
written into your record. Never a quiet reprint. Never silence. Your colleague's
sticker was not the sin. The missing record was.

The second formalization is where tomorrow's questions go. **Feedback as data**
means treating audience questions as observations about your project, to be
recorded and counted rather than merely felt. Example: "would this hold on
another campus?" is neither an insult nor a compliment. It is a data point about
which claim a stranger will not grant you for free.

The tool is **coding**, in the qualitative-research sense: sorting free-form
questions into a small set of labeled categories so you can count them. Every
question on your sheet gets three codes:

- **Type**: the five kinds from your visitor role card, plus *other*.
- **Target**: did it aim at your **claims**, meaning whether your result is true
  or properly bounded, or your **communication**, meaning whether a listener
  could follow you?
- **Recurrence**: how many times you got a version of it. Asked once, it is an
  anecdote. Asked three times by three different people, it is a finding about
  your project.

The target code does the real work, because the two signals demand different
repairs. A **communication signal** says your telling failed: "I couldn't tell
which line was the comparison group." Your figure failed, not your evidence. A
**claims signal** says your evidence was doubted: "would this hold outside your
sample?" The fix may change what you are allowed to claim. A third bucket,
**noise**, is a reaction with no question and nothing to fix, such as "nice
colors." 

> **A question that often comes up here:** *"A comment feels like it hits both my
> claims and my communication at once. Which column does it go in?"* Tag the target
> where the **fix** lives, not where the sting lives. "Your comparison group might
> differ from your treatment group" stings like a communication problem, because it
> sounds like you explained badly. But it doubts your evidence, so the fix is in
> your claim, and it codes to **claims**. When in doubt, ask: would a clearer poster
> make this question go away? If yes, communication. If a clearer poster leaves the
> doubt standing, claims.

```python
# A routing diagram: every audience question flows to one of three places.
# (networkx is imported once in the setup cell.)
G = nx.DiGraph()
pos = {
    "An audience\nquestion":              (0.0, 1.0),
    "CLAIMS signal\n(probe the evidence)":     (3.2, 1.9),
    "COMMUNICATION signal\n(fix the telling)": (3.2, 1.0),
    "NOISE\n(log it, no fix)":                 (3.2, 0.1),
}
edges = [
    ("An audience\nquestion", "CLAIMS signal\n(probe the evidence)"),
    ("An audience\nquestion", "COMMUNICATION signal\n(fix the telling)"),
    ("An audience\nquestion", "NOISE\n(log it, no fix)"),
]
G.add_edges_from(edges)

fig, ax = plt.subplots(figsize=(10, 4.8))
node_colors = ["#DCE6F1", "#4C72B0", "#55A868", "#B0B0B0"]
nx.draw_networkx_nodes(G, pos, node_color=node_colors, edgecolors="#333333",
                       linewidths=1.5, node_size=6600, ax=ax)
nx.draw_networkx_labels(G, pos, font_size=8.5, ax=ax)
nx.draw_networkx_edges(G, pos, arrowstyle="-|>", arrowsize=18, edge_color="#555555",
                       width=1.6, node_size=6600, ax=ax)
edge_labels = {
    ("An audience\nquestion", "CLAIMS signal\n(probe the evidence)"): "doubts your evidence",
    ("An audience\nquestion", "COMMUNICATION signal\n(fix the telling)"): "could not follow you",
    ("An audience\nquestion", "NOISE\n(log it, no fix)"): "no question, nothing to fix",
}
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=8,
                             label_pos=0.55, ax=ax)
ax.set_title("Every audience question routes to one of three targets")
ax.axis("off")
plt.tight_layout()
plt.show()

print("✓ Routing diagram drawn.")
print("  Claims and communication demand DIFFERENT fixes, so they never share a column.")
print("  Noise (warmth, an aside, no question) is logged but changes nothing.")
```

**Reading the diagram.** Every question you are asked flows to exactly one of
three targets. A **claims** question doubts your evidence and may change your
claim. A **communication** question could not follow you and is fixed in the
telling. **Noise** is a reaction with no question and nothing to fix. The danger
is piling all three into one warm blur, because then the recurring claims doubt
hides inside the applause.

*In plain terms, the picture says sort every question by what it would take to
answer it: better evidence, a clearer poster, or nothing at all.*

**Before the next cell:** it is a read-only illustration, not your data. It holds
a worked tally standing in for one afternoon at a board, seven distinct questions
with recurrence counts. Watch two numbers as it prints: which target carries more
weight once every row is summed, and which single question was asked most often.
Expect them to disagree.

```python
# A worked coded tally — seven distinct questions from one afternoon at a board.
# Read it; do not edit it. You code your OWN questions in the edit-safe scaffold
# in the M15 studio workbook, whose summaries recompute from your rows.
tally = pd.DataFrame([
    {"question": "Would this hold on a different campus or sample?",      "type": "generalization",          "target": "claims",        "count": 3},
    {"question": "Could something you didn't measure explain the gap?",   "type": "alternative-explanation", "target": "claims",        "count": 3},
    {"question": "So what changes if you're right?",                      "type": "so-what",                 "target": "claims",        "count": 2},
    {"question": "I couldn't tell which line was your comparison group.", "type": "other",                   "target": "communication", "count": 4},
    {"question": "How exactly did you estimate that number?",             "type": "method",                  "target": "claims",        "count": 2},
    {"question": "Did you use AI in the analysis, and did you check it?", "type": "AI-use",                  "target": "claims",        "count": 1},
    {"question": "Your title font was hard to read from a few feet away.","type": "other",                   "target": "communication", "count": 2},
])

total     = int(tally["count"].sum())
by_target = tally.groupby("target")["count"].sum().sort_values(ascending=False)
loudest   = tally.loc[tally["count"].idxmax()]

print(f"✓ {len(tally)} distinct questions, {total} asked in total")
print()
print("By TARGET (weighted by recurrence):")
print(by_target.to_string())
print()
print(f"Most-repeated single question ({int(loudest['count'])}x, target = {loudest['target']}):")
print(f'  "{loudest["question"]}"')
print()
print("Recurring CLAIMS doubts (asked 3 or more times):")
for _, row in tally[(tally["target"] == "claims") & (tally["count"] >= 3)].iterrows():
    print(f"  {row['count']}x  [{row['type']}]  {row['question']}")
```

**Reading the output.** Claims questions outnumber communication questions, 11 to
6, once every row is summed. Yet the single loudest item, asked four times, is a
communication fix: nobody could tell which line was the comparison group. That is
real and worth fixing, and it repairs a figure, not a claim.

The findings are the two claims doubts tied at three each, generalization and
alternative-explanation. Neither is soothed by the fact that the afternoon felt
warm. Say the tension out loud: the loudest ask was cosmetic while the recurring
doubts aimed at the evidence.

**4. At three other boards: the gallery-walk audit.** Auditing at least THREE
peer posters is required and graded. A **gallery-walk audit** is a structured
read of someone else's board from the viewer's position, not the author's.
At each board write three things:

- the peer's central claim, as printed;
- the evidence actually shown for it;
- the one place the two do not line up.

Then rank the hit by **how far the error travels**, meaning how much of the rest
of the board a reader stops trusting once they catch it. A claim-boundary
overreach travels farthest, because a reader who catches one causal verb the
design never bought starts doubting everything else. An unlabeled axis travels a
short distance: it is real, and it is local.

Those three ranked notes are your Friday portfolio entries. A gap you cannot
point at on the board in front of you never enters the portfolio.

> **A question that often comes up here:** *"Isn't sharp criticism of a classmate
> unkind?"* Vague praise leaves your peer with nothing to act on. Naming exactly
> where their evidence stops short, on a stated axis, is the most useful thing a
> reader can hand them, and it is exactly what you want in return. Kind but
> critical means you attack the claim, never the person.
