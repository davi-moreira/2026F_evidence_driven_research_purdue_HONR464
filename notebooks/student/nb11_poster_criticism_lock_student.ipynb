## 1. From Note to Board: What the Poster May Lose

**Guiding question:** *your note is locked; what may its poster adaptation drop,
and what must survive every cut?*

> *"I read your note last night and your poster this morning, and they made two
> different claims. The board dropped the limitation to save space, and what was
> left said more than your evidence does. A shorter version of a claim is fine.
> A bigger one wearing the same numbers is not."*
> — a **thesis advisor** comparing the two artifacts

This week is dossier-first: **the note exists, and the poster adapts it.** A
**poster adaptation** is a re-staging of an existing bounded claim for a new
audience and a 90-second read, not a new piece of research writing. Example:
your note's careful paragraph becomes one headline, one figure, and one visible
bound. The rule that governs every cut:

- **The poster may lose:** secondary results, methods prose, most of the
  numbers, robustness detail, and the literature. A stranger at a poster hall
  does not need them to receive the claim.
- **The poster may never lose:** the headline's boundary verb, the uncertainty
  in the same eye-span as the claim, the trace from every shown number to the
  cell that produced it, and the permissions your data promises rest on.

The instrument that makes the cut honest is the **adaptation-loss statement**:
a short written paragraph naming what your poster leaves out of the note and
why the claim survives the loss. Example: "the poster drops the three secondary
outcomes and the robustness table; the headline association, its interval, and
its descriptive boundary are unchanged from note sentence 1." If you cannot
write that paragraph, the board is not an adaptation. It is a new claim.

One more line belongs on the board itself. The **preliminary-edition label** is
a visible line naming the poster as a preliminary presentation of work in
progress. Example: "Preliminary edition: public test of a project that
completes at the end of the semester." Your project's full package closes after
the conference, so the label is simply true, and it is a strength, not an
apology: it tells a stranger exactly what stage of evidence they are reading.

> **A question that often comes up here:** *"If the poster may only repeat the
> note, why make one at all?"* Because the two artifacts do different jobs. The
> note argues to a reader who sits with it; the poster earns and survives a
> stranger's 90 seconds. Adaptation is real work: choosing the one figure, the
> one headline, and the one visible bound that carry the note's claim to
> someone who will never read the note. What adaptation never does is add
> material the note cannot back.

## 2. The Visual-Honesty Audit

**Guiding question:** *does your board's figure say exactly what the note's
number says, and can every reader receive it?*

A **visual-honesty audit** is a structured pass over a poster where you run a
fixed checklist instead of glancing and admiring. It turns vague reactions into
located, fixable findings. The core is **the four audits**, four ways a board
fails, each a question you ask of one specific poster:

- **Claim-boundary audit.** Does the headline stay inside the poster's compass
  position, and does every claim have a note sentence behind it? A **claim
  boundary** is the line between what the evidence licenses and what it does
  not. Example: an observational project may say "is associated with," never
  "causes." The boundary lives in the headline verb.
- **Figure-honesty audit.** Does the figure's visual impression match its number?
  **Figure honesty** means the picture cannot say more than the data. Example: an
  axis that starts just below the data draws a tiny gap as a canyon.
- **Read-path audit.** Can a stranger find the headline in 90 seconds? The **read
  path** is the order a reader's eye actually travels. Example: the eye should
  reach the claim before the methods box, not after.
- **Uncertainty audit.** Is the bound visible, or buried? **Uncertainty on the
  page** means the interval or caveat sits in the same eye-span as the claim.
  Example: a "±0.7" that lives only in a footnote is doing hidden work.

A **hit** is a specific, located finding with a fix: "the y-axis starts just below
the shorter bar, so a gap of about half a point looks like a canyon; run it 1 to
5." A review with no hits did not spare the poster. It failed it.

Two more checks ride alongside, and they decide whether your claim reaches
*every* reader, not just a fast sighted one.

- **Visual design.** How the layout guides the eye: size, position, and contrast.
  The one number you want remembered gets the biggest type and the top-center
  slot. Example: the headline claim, not the university logo, owns the center.
- **Accessibility.** Whether the poster's message reaches readers with different
  perception. **Accessibility** means the claim is legible to a color-blind reader
  and to a screen-reader user, not only to someone who sees every color and leans
  in close. Example: two series told apart by red versus green alone vanish for a
  red-green color-blind reader, who is roughly 1 in 12 men.

The fix for the color trap has a name worth knowing. **Redundant encoding** means
carrying the same distinction in more than one channel, so no single channel is
load-bearing. Example: label each bar with its value and give each a distinct
pattern, so color is a bonus, not the only signal.

> **A question that often comes up here:** *"Accessibility sounds like a
> compliance checkbox. Is it really an evidence issue?"* It is the same issue as
> every other audit. Your claim is only as strong as the readers it reaches. A
> figure that encodes its result in color alone makes a claim that a color-blind
> reviewer literally cannot receive, which is an overclaim by exclusion. Reaching
> every reader is not politeness. It is the claim doing its job.

```python
# One note sentence, adapted into a board figure two ways from the SAME pilot.
# LEFT = the dramatic adaptation (the flaws the audit must catch).
# RIGHT = "as the note reads" (honest axis, labels, uncertainty, reachable colors).

# A small satisfaction pilot on a 1-to-5 scale, two program cohorts.
scale = np.array([1, 2, 3, 4, 5])
p_a = np.array([0.16, 0.22, 0.30, 0.20, 0.12])   # cohort A: lower satisfaction
p_b = np.array([0.08, 0.14, 0.26, 0.26, 0.26])   # cohort B: a little higher
a = rng.choice(scale, size=200, p=p_a)
b = rng.choice(scale, size=200, p=p_b)

mean_a, mean_b = a.mean(), b.mean()
ci_a = 1.96 * a.std(ddof=1) / np.sqrt(len(a))     # 95% interval half-widths:
ci_b = 1.96 * b.std(ddof=1) / np.sqrt(len(b))     # the uncertainty the honest panel shows
gap = mean_b - mean_a

# One source of truth for the truncation: a floor just below the shorter bar.
floor = min(mean_a, mean_b) - 0.05
exaggeration = (mean_b - floor) / (mean_a - floor)

print(f"Note sentence: cohort B about half a point higher, descriptive, intervals shown.")
print(f"Cohort A mean = {mean_a:.2f}   Cohort B mean = {mean_b:.2f}   gap = {gap:.2f} on a 1-to-5 scale")
print(f"Truncated at {floor:.2f}, the taller bar is drawn {exaggeration:.1f}x the shorter one.")

fig, axes = plt.subplots(1, 2, figsize=(12, 4.8))

# LEFT: the dramatic adaptation. Bring the audits to it YOURSELF in the exercise
# below; these comments only say what the code does, never which audit each
# choice trips.
axL = axes[0]
axL.bar(["Program A", "Program B"], [mean_a, mean_b],
        color=["#D62728", "#2CA02C"])              # two bars told apart by red vs green only
axL.set_ylim(floor, mean_b + 0.05)                 # y-axis floored just below the shorter bar
axL.set_title("Program B BOOSTS satisfaction")     # title worded the way a program might pitch it
axL.set_ylabel("Satisfaction")
# no interval drawn, no value labels, no full scale shown.

# RIGHT: as the note reads — honest, and reachable by every reader.
axR = axes[1]
bars = axR.bar(["Program A", "Program B"], [mean_a, mean_b],
               yerr=[ci_a, ci_b], capsize=6,
               color=["#4C72B0", "#DD8452"],        # colorblind-distinguishable pair
               hatch=["", "//"], edgecolor="white")  # redundant encoding: pattern backs up color
axR.set_ylim(1, 5)                                   # honest scale: the real 1-to-5 range
for bar, m in zip(bars, [mean_a, mean_b]):
    axR.annotate(f"{m:.2f}", (bar.get_x() + bar.get_width() / 2, m + 0.10),
                 ha="center", fontsize=11)           # direct labels: a screen reader gets the number
axR.set_ylabel("Mean satisfaction (1 = low, 5 = high)")
axR.set_title("Program A vs B: means with 95% intervals")

plt.tight_layout()
plt.show()

# Self-check: the flaws and the honest reading are both really present.
assert mean_b > mean_a, "cohort B should read a little higher in this pilot"
assert exaggeration > 3, "the truncated axis should visibly inflate the gap"
assert axes[0].get_ylim()[0] > 1, "the left axis is truncated by design"
assert axes[1].get_ylim() == (1.0, 5.0), "the right axis shows the honest full scale"
print("✓ Same numbers, adapted two ways: the left board changes the apparent finding; the right one repeats the note.")
```

**Reading the output.** The printout is the note's evidence base: two means about
half a point apart, and the factor by which the truncated axis inflates that gap.
*The figure shows one honest set of numbers drawn two ways: on the left a red
and green pair of bars on a chopped axis under a causal title, and on the right
the same means on the full 1-to-5 scale, labeled, patterned, and carrying their
intervals. The left picture lies with true numbers; the right one repeats the

## 3. Poster Criticism and the Fix Log

**Guiding question:** *when the room criticizes your board, what turns the hits
into a poster that is actually better?*

> *"At the Expo I point at one number and ask, 'where did this come from?' If you
> can walk me to the line of code that made it, I believe the whole board. If you
> cannot, I stop believing all of it."*
> — a **research-integrity judge** working the poster hall

Criticism only improves a poster if it lands somewhere. **Poster criticism** is
the audit of Lecture 1 run by other eyes: classmates and reviewers walk your
board with the four audits plus accessibility and hand you located hits. What
you keep is the **fix log**: a written record where every hit gets three
entries: the hit, located; the fix, applied once; and the check that confirms
the fix worked. Example: "hit: y-axis starts at 78; fix: full score range;
check: re-rendered, gap now reads at true scale." A hit with no fix-log row is
a flaw you have agreed to ship.

The fix log follows one discipline: **fixed once, then re-audited.** You do not
negotiate with a hit, and you do not patch it three different ways. You apply
the smallest defensible fix, re-run the audit that caught it, and close the row.

> **A question that often comes up here:** *"My criticism round found nothing.
> Am I done?"* Almost never. A round with no hits usually means the reviewers
> glanced and admired, so run the checklist again yourself, audit by audit. The
> rule of the room is **hard on the poster, warm on the person**: every hit a
> friendly room finds today is a hit a hostile stranger never gets to find at
> the Expo. You are giving each other a free stranger's eyes.

## 4. The Two Gates and the Lock

**Guiding question:** *before your poster becomes permanent, what must you
prove, and what does your own proof still not establish?*

The lock is terminal: once the file goes for printing, the Expo audience reads
exactly that version, with no chance to fix a number afterward. So two blocking
gates stand in front of it. A **blocking gate** is a check that stops the lock
until it passes: not advice, not a suggestion, a stop. Example: a poster whose
numbers did not reproduce this week does not lock, however beautiful the board.

**Gate 1: author self-reproduction.** You open a fresh runtime, press Restart
and Run All on the notebook behind your poster, and confirm that **every number
on the board reappears**. The record of that run is your **self-reproduction
log**: the date, the environment, and each board number matched to its freshly
printed value. Example: "fresh runtime today; headline gap 0.52 reproduced;
both means and both intervals reproduced." The log must be *fresh*: a change to
the code, the data, or the board after the run makes it stale, and a stale log
does not pass the gate.

Be precise about what this gate is not. **Independent reproduction** means
someone who is not you obtains your result from your materials alone, with you
silent. Your own rerun is not that: your presence, your environment, and your
habits ride along. Author self-reproduction catches stale numbers and broken
traces, and it is exactly the right gate before a lock. The independent test,
a cold run by other hands, comes later in the course, and nothing this week
licenses you to claim it.

**Gate 2: the release preflight.** Four checks, each answered in writing with
evidence, before the file leaves your hands:

- **Permissions.** May every element on the board be shown publicly: the data,
  the quotes, the images, under the promises your data rests on?
- **Disclosure.** Is the AI-use disclosure block on or with the poster, and does
  it match your AI Research Ledger?
- **Reproducibility.** Is the self-reproduction log fresh, with nothing changed
  since the run?
- **Claim boundary.** Does the headline, and every claim on the board, sit
  inside the note's license, with the preliminary-edition label visible?

Two habits from earlier weeks carry the preflight's evidence. **Claim-evidence
traceability** means every number on the board points to the exact notebook cell
that produced it; a number you typed from memory or carried from an old draft is
an **untraceable number**, and it cannot go on the board. **Citation
verification** means every source the poster names resolves to a real,
retrievable work you opened yourself.

> **A question that often comes up here:** *"Why gate the lock twice? If the
> numbers reproduce, isn't the poster honest?"* No, and the colleague case shows
> half of the problem: numbers can be stale. The preflight shows the other half:
> a board can reproduce perfectly and still break a permission, hide its AI use,
> or carry a headline the note never licensed. Arithmetic honesty and release
> honesty are different checks, and the lock waits for both.

### 🛠️ Run the Self-Reproduction Check: fresh numbers versus the board

The cell rebuilds the pilot from the seed, recomputes the three numbers the
draft board displays, and prints a match table. It is the gate in miniature:
the same comparison you will run on your own poster, number by number.

**🔴 Live in class: we run this one together.**
**Before you ask:** write your own one-line verdict on the colleague's lock
claim from the puzzle, so the tool cannot write it for you.

> 💡 **AI Prompt:** "Here are my four release-preflight entries with their
> evidence lines: [paste your permissions, disclosure, reproducibility, and
> claim-boundary entries]. Act as a strict release auditor. For each check, say
> whether the evidence PROVES the check or merely asserts it, name the single
> weakest evidence line, and tell me what a stranger would demand to see before

>
> **After running, verify (counters *sycophantic agreement*: an auditor that

> - [ ] Map each finding to exactly one of the four checks. A finding that fits
>   none, or that demands a brand-new analysis, is outside the preflight's job;
>   reject it.
> - [ ] Check its reproducibility verdict against your own log's date: if
>   anything on the board changed after the run, the log is stale, whatever the
>   auditor said.
> - [ ] If it passes all four with no weakest line, push back: "assume one entry
>   is only an assertion; name it." An audit with zero findings is a red flag.
> - [ ] Log this use in your AI Research Ledger: task, tool, decision, verification.

```python
# The self-reproduction gate in miniature: recompute every board number from a
# clean start and compare it to what the draft board carries.

# Clean start: rebuild the pilot from the seed, exactly as the notebook defines it.
rng_fresh = np.random.default_rng(SEED)
scale2 = np.array([1, 2, 3, 4, 5])
pa2 = np.array([0.16, 0.22, 0.30, 0.20, 0.12])
pb2 = np.array([0.08, 0.14, 0.26, 0.26, 0.26])
a2 = rng_fresh.choice(scale2, size=200, p=pa2)
b2 = rng_fresh.choice(scale2, size=200, p=pb2)
fresh = {
    "Mean, Program A": round(a2.mean(), 2),
    "Mean, Program B": round(b2.mean(), 2),
    "Headline gap":    round(b2.mean() - a2.mean(), 2),
}

# What the draft board displays. The means were updated during drafting; the
# headline gap was copied from an early draft and never refreshed.
board = {
    "Mean, Program A": fresh["Mean, Program A"],
    "Mean, Program B": fresh["Mean, Program B"],
    "Headline gap":    0.61,   # carried from an old draft of the analysis
}

print(f"{'Board number':<18}{'on the board':>14}{'fresh rerun':>13}{'match?':>9}")
mismatches = []
for name in board:
    ok = board[name] == fresh[name]
    if not ok:
        mismatches.append(name)
    print(f"{name:<18}{board[name]:>14}{fresh[name]:>13}{('✓' if ok else '✗ FAIL'):>9}")

print()
print(f"Gate verdict: {len(mismatches)} board number(s) failed to reproduce -> the lock is BLOCKED.")

# Self-check: the gate catches exactly the stale headline number.
assert mismatches == ["Headline gap"], "the stale gap should be the one failure"
assert abs(board["Headline gap"] - fresh["Headline gap"]) > 0.03, "the drift is real, not rounding"
print("✓ The fresh rerun caught a stale headline number before a stranger did.")
```

**Reading the output.** Both means match their fresh values, so most of the board
is honest, and the notebook itself runs green. The headline gap fails: the board
says 0.61 and the clean start prints 0.52, because the number was copied from an
early draft and the analysis moved on without it. *In plain terms, the table
says a poster can be built from a working notebook and still display a number
that notebook no longer produces. Only a fresh rerun, compared line by line,
catches it, and that is exactly what the gate demands.*
