## 1. The Cold Run: Does Your Number Come Back Without You?

**Guiding question:** *handed nothing but the files a study shipped, can you get
the reported number back out?*

> *"I do not care that your figure is pretty or that the code ran on your laptop.
> Hand me your files and let me get the same number back out. If I cannot, you do
> not have a result yet. You have a story."*
> — a **principal investigator** reading a first-year researcher's lab report

Two words sound alike and mean different things. **Reproduction** is getting the
same number from the *same* data and code a study shipped. Example: you open
someone's notebook, run it top to bottom, and the same turnout gap comes back.
**Replication** is getting a similar result from a *new* study or new data.
Example: a second team runs its own canvassing experiment and sees a similar
effect. This week is about reproduction. You test the package, not the world.

What you test is a **reproducibility package**: the full bundle a study ships so
someone else can rebuild its numbers, meaning the data, the code, the run order,
any random seed, and the write-up. The one value the main claim rests on is the
**headline number**. Example: "canvassing raised turnout by 4 points."

You rebuild that first with a **cold run**, also called
**restart-and-run-all**: you clear the notebook's memory and run every cell from
the top, with no manual fixes and no cells run out of order. Example: Runtime,
Restart and run all, and the headline value reappears. If a number only shows up
when you run cells in a special order you remember, it is not reproducible.

The package you are about to reproduce is a real one: a UK get-out-the-vote field
experiment in which whole streets were canvassed door to door.

> **A question that often comes up here:** *"If the package reproduces, why keep
> going?"* Because a clean run proves the code executes, and nothing more. It
> does not prove the write-up matches the computation, that the choices behind
> the number were disclosed, or that the reported certainty was earned. A
> successful reproduction is the moment you earn the right to audit. It is the
> start of the review, not the verdict.

### 🛠️ Run the Study: cold-reproduce the package's headline

Run the cell below. It is the package's analysis from a clean start: load the
data, split on the canvassing flag, and difference the two turnout rates. Read
the printed values before you read anything else.

**Expect** two turnout rates a few points apart, and a headline gap that lands
near 3.4 rather than on 4.

```python
# THE SHIPPED PACKAGE'S ANALYSIS, reproduced cold from a clean start.
# `treat` = the person's street was canvassed door to door.
# `marked_register_2014` = 1 if the person voted.
pkg = load_course_data("foos_etal.csv")
assert pkg.shape == (8375, 5), "unexpected shape — stop and check the file"
print("✓ Loaded the shipped package —", pkg.shape[0], "rows,", pkg.shape[1], "columns")
print("  Columns:", list(pkg.columns))

y = "marked_register_2014"
canvassed = pkg[pkg["treat"] == 1][y]      # streets that were canvassed
not_canv = pkg[pkg["treat"] == 0][y]       # streets that were not

print(f"\n  canvassed    : {len(canvassed):>5} people, turnout {canvassed.mean() * 100:.2f}%")
print(f"  not canvassed: {len(not_canv):>5} people, turnout {not_canv.mean() * 100:.2f}%")

diff = canvassed.mean() - not_canv.mean()  # THE HEADLINE NUMBER
diff_pts = diff * 100
print(f"\n  HEADLINE GAP = {diff_pts:.2f} percentage points")
print(f"  The write-up says 4 points. The package says {diff_pts:.2f}. Hold that gap.")

# Self-check: the package really does reproduce, and really does not print 4.
assert abs(diff_pts - 3.41) < 0.05, "the headline drifted — investigate before continuing"
assert canvassed.mean() > not_canv.mean(), "canvassed streets should show higher turnout"
print("\n✓ The package reproduces. Finding #1 is already on the table.")
```

**Reading the output.** The package reproduces cleanly. Canvassed streets voted
at 50.66 percent, not-canvassed streets at 47.25 percent, and the headline gap
comes back at **3.41 percentage points**. That is a genuine success: the number
is not trapped in the author's head, it regenerates from the files. Now read the
write-up again. It says **4 points**, and the code says 3.41.

That mismatch has a name. **Claims-versus-computation agreement** is the check
that every sentence a write-up asserts is backed by a value the code actually
prints. Example: a write-up saying "4 points" beside a cell printing 3.41 fails
the check. This is finding #1.

> **A question that often comes up here:** *"The gap between 3.41 and 4 is tiny.
> Is flagging it just pedantic?"* On its own a rounding stretch is small, and
> notice it is not even rounding: 3.41 rounds to 3.4. What the mismatch tells you
> is that the write-up quotes a value the package does not produce. If that got
> rounded up quietly, what else got smoothed? The small mismatch earns you the
> right to ask the two bigger questions coming next: which undisclosed choice
> moved this number, and whose independence is the interval assuming.

### 🛠️ Run It Live: the alternative specification nobody declared

The package ships a `weights` column and never applies it. A **weight** is a
per-person number that corrects for who was easier or harder to reach, so the
sample better matches the population it stands for. Example: a person from a
hard-to-reach street counts for more than one easy-to-reach neighbor.

An **alternative specification** is a different but equally defensible way to
compute the same headline, run to see whether the answer depends on a choice
nobody disclosed. Example: the same gap, weighted instead of unweighted.

Before it runs, check the thing that licenses reporting a spread at all: both
versions must still answer the same question about the same people. They do here.

**Expect** the gap to move, and watch how far.

```python
# ALTERNATIVE SPECIFICATION: the same gap, using the weights the package shipped.
def weighted_mean(values, weights):
    """A weighted average: each person counts in proportion to their weight."""
    return np.average(values, weights=weights)


w_canv = weighted_mean(pkg.loc[pkg.treat == 1, y], pkg.loc[pkg.treat == 1, "weights"])
w_notc = weighted_mean(pkg.loc[pkg.treat == 0, y], pkg.loc[pkg.treat == 0, "weights"])
diff_w_pts = (w_canv - w_notc) * 100

print(f"  unweighted gap (what the write-up reported): {diff_pts:.2f} points")
print(f"  weighted gap   (weights the write-up skipped): {diff_w_pts:.2f} points")
print(f"\n  The headline MOVES by {diff_pts - diff_w_pts:.2f} points on a choice")
print("  the write-up never named.")

# Self-check: the undisclosed choice moves the headline by a real amount.
assert abs(diff_w_pts - 2.83) < 0.05, "the weighted gap drifted — investigate"
assert diff_pts - diff_w_pts > 0.3, "the two specifications should differ noticeably"
print("\n✓ Finding #2: the headline depends on an undeclared decision.")
```

**Reading the output.** Same data, same question, one defensible change: the gap
slides from **3.41** points unweighted to **2.83** points weighted, a swing of
**0.57 points**. Neither number is a lie, and neither is automatically the true
one. The finding is that the headline depends on a choice the write-up never
disclosed, so the honest report is the spread across both, never the friendlier
end of it. You have now moved the number by changing a *choice*. Next you push on
an *assumption*, and that one does more damage.

### 🛠️ Run It Live: reproduce the uncertainty, not only the point estimate

The package reports its gap as if all 8,375 people were separate facts. They are
not. Canvassing happened **by street**, so neighbors share a canvasser, a
neighborhood, and a turnout habit.

Two terms carry this. A **cluster** is a group of observations more alike inside
the group than across groups, so it carries less independent information than its
row count suggests. Example: 20 neighbors on one canvassed street are not 20
independent facts about canvassing. The **unit of randomization** is the thing
that actually got assigned to treatment. Example: here the street was assigned,
not the person.

The next cell sizes the uncertainty twice: the naive way, treating people as
independent, and the cluster-aware way, resampling whole streets. **Predict
before you run it:** will the cluster-aware interval be narrower, about the same,
or wider, and will it still sit above zero? Say it out loud, then run.

```python
# NAIVE uncertainty: treat all 8,375 people as independent observations.
n1, n0 = len(canvassed), len(not_canv)
naive_se = np.sqrt(canvassed.var(ddof=1) / n1 + not_canv.var(ddof=1) / n0)
naive_lo, naive_hi = (diff - 1.96 * naive_se) * 100, (diff + 1.96 * naive_se) * 100

# CLUSTER-AWARE uncertainty: resample whole STREETS, not individuals (seed 464).
tmp = pd.DataFrame({
    "street": pkg["street"],
    "yt": pkg[y] * (pkg.treat == 1), "nt": (pkg.treat == 1).astype(int),
    "yc": pkg[y] * (pkg.treat == 0), "nc": (pkg.treat == 0).astype(int),
})
per = tmp.groupby("street")[["yt", "nt", "yc", "nc"]].sum()
yt, nt = per["yt"].to_numpy(float), per["nt"].to_numpy(float)
yc, nc = per["yc"].to_numpy(float), per["nc"].to_numpy(float)
K = len(per)

boot = np.empty(400)
for b in range(400):
    idx = rng.integers(0, K, size=K)                 # resample whole streets
    boot[b] = yt[idx].sum() / nt[idx].sum() - yc[idx].sum() / nc[idx].sum()
cluster_se = boot.std()
cluster_lo, cluster_hi = (diff - 1.96 * cluster_se) * 100, (diff + 1.96 * cluster_se) * 100

print(f"  streets (clusters): {K}  "
      f"({int((per['nt'] > 0).sum())} canvassed, {int((per['nc'] > 0).sum())} control)")
print(f"  NAIVE   95% interval: [{naive_lo:5.2f}, {naive_hi:5.2f}] points")
print(f"  CLUSTER 95% interval: [{cluster_lo:5.2f}, {cluster_hi:5.2f}] points")
print(f"\n  Respecting the streets makes the interval about "
      f"{cluster_se / naive_se:.2f} times wider.")
print(f"  The point estimate did not move: it is still {diff_pts:.2f} points.")

# Plot both intervals against zero. Each is drawn with its own marker AND
# labeled directly on the axis, so nothing is encoded by color alone.
fig, ax = plt.subplots(figsize=(9, 3.2))
ax.errorbar(diff_pts, 1.0, xerr=1.96 * naive_se * 100, fmt="o", color="#DD8452",
            capsize=6, elinewidth=2.5, label="naive (people independent)")
ax.errorbar(diff_pts, 0.0, xerr=1.96 * cluster_se * 100, fmt="s", color="#4C72B0",
            capsize=6, elinewidth=2.5, label="cluster-aware (whole streets)")
ax.axvline(0, color="gray", linestyle=":", linewidth=1.5, label="no effect (zero)")
ax.set_yticks([0.0, 1.0])
ax.set_yticklabels(["cluster-aware", "naive"])
ax.set_ylim(-0.6, 1.6)
ax.set_xlabel("Turnout gap (percentage points), 95% interval")
ax.set_title("Same estimate, honest uncertainty: the cluster-aware interval reaches zero")
ax.legend(loc="lower right", fontsize=8)
plt.tight_layout()
plt.show()

# Self-check: honoring the clustering widens the interval down to zero.
assert cluster_se > 1.3 * naive_se, "the cluster SE should be clearly larger"
assert naive_lo > 0, "the naive interval should sit above zero"
assert cluster_lo < naive_lo, "the honest interval must reach lower than the naive one"
print("✓ Finding #3: the point estimate held; the honest uncertainty did not.")
```

**Reading the output.** This is the finding with teeth. The naive interval, which
treats every person as a separate fact, runs from **1.00 to 5.81 points** and sits
safely above zero, so it reads as a clear effect. Resample whole **streets**,
which is what the design actually did, and the interval runs from **-0.07 to
6.89** across **613 streets** (447 canvassed, 166 control). That is about
**1.45 times** wider, with the lower edge at zero. The point estimate barely
moved. The honest uncertainty around it exploded.

*In plain terms, the picture shows one estimate with two different honesty
levels: the top interval borrows certainty from rows that are not independent,
and the bottom one earns its certainty from the 613 streets that actually got
assigned.*

## 2. What the Green Run Did Not Prove

**Guiding question:** *what has to be true before you may say a result was
verified?*

Three findings are now on the table: a write-up that quotes a value the code does
not print, a headline that moves 0.57 points on an undisclosed weighting choice,
and an interval that reaches zero once the streets are counted. None of them
showed up as an error message.

Two more ideas finish the toolkit, and both decide what you are allowed to say.

**Classify missing values before you fill any of them.** A blank cell is two
different things. It can be *never recorded*, meaning the value exists in the
world and nobody wrote it down. Or it can be *never existed*, meaning an event
made the measurement impossible. That second case has a name: a **post-treatment
event** is something that happens after treatment, that treatment itself may have
caused. Example: a canvassed household moves away before the election, so there
is no turnout value to record at all. If treatment could have caused the event,
report the event by group first, and label whatever comparison remains as a
**complete-case contrast**: a comparison among the units that are still there.
Never call it "the effect among the people who stayed." And never carry the last
recorded value forward to patch a gap, because that answers a different question
under an assumption you never stated.

**Your own rerun is a proxy, not a verdict.** **Independent reproduction** means
someone who is not you obtains your result from your materials alone, with you
silent. Your own rerun is not that. The honest label for a rerun you did yourself
is **solo proxy**: a clean-environment run you performed, recorded as
"solo proxy; external cold run pending." Example: you reran your notebook in a
fresh session from your written instructions only, and you wrote that exact
label on the record.

Two standing rules close this lecture, and they govern everything that follows.

- **Author self-reproduction is never independent verification.** It is the right
  gate before a lock, and it is not a claim about anyone else's ability to run
  your work.
- **You reproduce the uncertainty statement, not only the point estimate.** A
  point estimate that comes back while its interval doubles has not reproduced in
  any sense a reader cares about.

## 3. Package It for a Stranger: The Capsule and Its Five Sins

**Guiding question:** *what has to be in the folder before someone who is not you
can rebuild your numbers?*

> *"I don't want to hear that it runs on your laptop. Hand me the folder, let me
> clear everything, and let me press run. If your headline number does not come
> back on my machine, you do not have a result yet. You have a memory of one."*
> — a **replication editor**, opening the folder attached to your submission

A **reproducibility capsule** is everything a stranger needs to rebuild your
numbers and nothing they would have to guess. Example: a folder holding your
notebook, your data or a route to it, and a short record of every by-hand choice.
It has five parts.

- A **runnable notebook** that survives restart-and-run-all with no manual fixes.
- A **data-provenance note**: where each dataset came from, its version, and how
  it may be used. Example: the archive URL, the retrieval date, and the licence.
- A **fixed seed**, a starting number that makes every random step return the
  same values on every run. Example: `SEED = 464` feeding each resample.
- A **decision log**: the by-hand choices that shaped the result, each with its
  reason. Example: every row you dropped, and why.
- An **AI-use ledger**: every tool, its task, and how you verified its output.

Capsules break in boringly predictable ways, which is good news, because a
predictable failure is a catchable one. Named, they are **the five package sins**.

| Sin | What it is | The repair |
|---|---|---|
| **Hard-coded path** | a file location that exists on one machine only | load through a shared route the reader also has |
| **Missing seed** | no starting number, so every run gives different values | pin `SEED = 464` before anything random |
| **By-hand edit** | a fix you made once, by clicking, that no clean run repeats | move the repair into code the clean run runs too |
| **Undocumented exclusion** | rows dropped with no logged rule and no reason | log the rule and its reason, or declare it in advance |
| **Stale data** | a file a reader cannot reobtain | record the source, the version, and the retrieval date |

Hold onto the one line that carries this whole lecture: a capsule with zero flags
is **runnable**, never proven **correct**. The scan gets you to the starting line.
A person who reruns you cold is the race.

> **A question that often comes up here:** *"My capsule scan came back with zero
> flags. Am I done?"* No, and the distinction is the point. A clean scan says
> every repair happens in code a stranger repeats, which is exactly what makes
> your work checkable. It says nothing about whether the choices inside are good
> ones. Your seed can be pinned on the wrong subset. A bad exclusion, cleanly
> logged, is still a bad exclusion. Runnable is the condition that lets someone
> find your mistake, not proof that you did not make one.

### 🛠️ Run It Live: the five-sin scan, part one

One analysis, shipped two ways. The sinful version loads a path only its author
has. The clean version loads through the shared route this course ships. The
headline is the same either way: the turnout gap between voters under 30 and
voters 30 and over in a 1,000-person voter-file extract.

The cell then does the thing the colleague's record claims to have done. It
builds a **match table**: every number the capsule's README reports, set beside
the value the fresh run just printed, with a mark for each.

**Expect** the sinful load to stop with a `FileNotFoundError` before any analysis
happens, and expect the match table to come back clean.

```python
# SIN 1, the hard-coded path: a location that exists on one machine only.
SINFUL_PATH = "/content/drive/MyDrive/turnout_clean.csv"
try:
    cap = pd.read_csv(SINFUL_PATH)
    print("loaded from the author's own drive")
except (FileNotFoundError, OSError) as err:
    print("SIN 1 (hard-coded path) ->", type(err).__name__)
    print("  ", SINFUL_PATH)
    print("   No stranger has that path, so cell 1 stops the run before any")
    print("   analysis question is even reachable.")

# THE REPAIR: load through the shared route every reader also has.
cap = load_course_data("la_voter_file.csv")
print("\n✓ Loaded the capsule data —", cap.shape[0], "rows,", cap.shape[1], "columns")

under30 = cap[cap.age < 30]
over30 = cap[cap.age >= 30]
age_gap_pts = (over30.voted_2012.mean() - under30.voted_2012.mean()) * 100

print(f"\n  overall 2012 turnout      : {cap.voted_2012.mean() * 100:.2f}%")
print(f"  under 30                  : {under30.voted_2012.mean() * 100:.2f}%  (n = {len(under30)})")
print(f"  30 and over               : {over30.voted_2012.mean() * 100:.2f}%  (n = {len(over30)})")
print(f"  HEADLINE AGE GAP          : {age_gap_pts:.2f} percentage points")

# THE MATCH TABLE: what the capsule's README reports vs what this run prints.
readme = {"Turnout, under 30": 47.87, "Turnout, 30 and over": 63.18,
          "Headline age gap": 15.30}
fresh = {"Turnout, under 30": round(under30.voted_2012.mean() * 100, 2),
         "Turnout, 30 and over": round(over30.voted_2012.mean() * 100, 2),
         "Headline age gap": round(age_gap_pts, 2)}

print(f"\n{'Reported number':<24}{'in the README':>16}{'fresh run':>12}{'match?':>9}")
mismatches = []
for name in readme:
    ok = readme[name] == fresh[name]
    if not ok:
        mismatches.append(name)
    print(f"{name:<24}{readme[name]:>16.2f}{fresh[name]:>12.2f}"
          f"{('✓' if ok else '✗ FAIL'):>9}")

print(f"\n  Match table verdict: {len(mismatches)} number(s) failed to reproduce.")

assert mismatches == [], "every README number should reproduce here"
print("  Every number matched. The release claim is still not settled.")
```

**Reading the output.** Two separate things happened, and keeping them apart is
the whole lecture. The sinful load raised a `FileNotFoundError` on the first cell,
which is what a stranger meets before they can even reach a question about the
analysis. The clean load worked, and the headline age gap came back at **15.30
percentage points**: 47.87 percent turnout among the 188 voters under 30 against
63.18 percent among the 812 voters aged 30 and over.

Then the match table returned **zero mismatches**. Every number the README reports
is a number this run prints. Read that carefully. A clean match table is exactly
what a **self-reproduction** check produces when it passes, and it still says
nothing about who ran it. The colleague's record has a clean match table too, and
it uses that table to claim independent reproduction. Those are different

### The scan continues: the sins you cannot see in a stack trace

Sin 1 announced itself with an error. The next two never do.

The sinful version bootstraps its interval with **no seed**, so its numbers move
between runs. A **bootstrap** here means resampling the 1,000 voters with
replacement many times and reading the spread of the resulting gaps. Example: 2,000
resamples, then the middle 95 percent of the answers.

The sinful version also keeps only voters registered with the two largest parties
and never logs the rule. That is an **undocumented exclusion**.

**Expect** the seeded runs to match each other exactly, the unseeded runs to
differ, and the unlogged filter to move the headline by more than a point.

```python
# SIN 2, the missing seed, and SIN 4, the undocumented exclusion.
AGE = cap.age.to_numpy()
VOTE = cap.voted_2012.to_numpy()


def boot_interval(generator, draws=2000):
    """95% interval for the age gap, resampling voters with replacement."""
    out = np.empty(draws)
    n = len(AGE)
    for b in range(draws):
        idx = generator.integers(0, n, size=n)
        a, v = AGE[idx], VOTE[idx]
        out[b] = (v[a >= 30].mean() - v[a < 30].mean()) * 100
    return np.percentile(out, [2.5, 97.5])


seeded_1 = boot_interval(np.random.default_rng(SEED))
seeded_2 = boot_interval(np.random.default_rng(SEED))
loose_1 = boot_interval(np.random.default_rng())      # no seed pinned
loose_2 = boot_interval(np.random.default_rng())      # no seed pinned

print("  SEEDED (SEED = 464), run 1 : "
      f"[{seeded_1[0]:.2f}, {seeded_1[1]:.2f}] points")
print("  SEEDED (SEED = 464), run 2 : "
      f"[{seeded_2[0]:.2f}, {seeded_2[1]:.2f}] points")
print("  UNSEEDED, run 1            : "
      f"[{loose_1[0]:.2f}, {loose_1[1]:.2f}] points")
print("  UNSEEDED, run 2            : "
      f"[{loose_2[0]:.2f}, {loose_2[1]:.2f}] points")
print(f"  the unseeded lower edge moved {abs(loose_1[0] - loose_2[0]):.2f} points "
      "between two runs of the same cell")

# SIN 4: keep only the two largest parties, with no logged rule and no reason.
kept = cap[cap.party.isin(["DEM", "REP"])]
gap_kept = (kept.loc[kept.age >= 30, "voted_2012"].mean()
            - kept.loc[kept.age < 30, "voted_2012"].mean()) * 100

print(f"\n  headline with every voter          : {age_gap_pts:.2f} points "
      f"(n = {len(cap)})")
print(f"  headline after the unlogged filter : {gap_kept:.2f} points "
      f"(n = {len(kept)})")
print(f"  an undeclared exclusion moved the headline by "
      f"{age_gap_pts - gap_kept:.2f} points")

assert np.allclose(seeded_1, seeded_2), "a pinned seed must give the same interval"
assert abs(gap_kept - 11.64) < 0.05, "the filtered headline drifted"
print("\n✓ Two sins, neither of which ever raises an error.")
```

**Reading the output.** The two seeded runs return the identical interval, because
a pinned seed makes every random step repeat. The two unseeded runs do not, and
they will not match the values printed here when you run this yourself. That is
the sin: a reader who sees one unseeded interval has no way to know how much of
it was the data and how much was the draw.

The undocumented exclusion is worse, because it moves the answer. Keeping only
the two largest parties drops the sample from 1,000 voters to 693 and slides the
headline age gap from **15.30** points to **11.64**, a swing of **3.66 points**
that nobody declared. The filter may even be defensible. The sin is that it was
invisible, so a reader cannot tell whether the headline is a finding or the
residue of an unstated choice.

The clean capsule handles all of this in code a stranger repeats: it loads through
the shared route, pins `SEED = 464`, records the relabel in the decision log,
applies any party filter only as a declared and logged rule, and ships the source,
the retrieval date, and the terms of use.

## 4. The Two Blocking Gates and the Lock

**Guiding question:** *before your poster becomes permanent, what must you prove,
and what does your own proof still not establish?*

The lock is terminal. Once the file goes for printing, the conference audience
reads exactly that version, with no chance to fix a number afterward. So two
gates stand in front of it. A **blocking gate** is a check that stops the lock
until it passes. Not advice, not a suggestion, a stop. Example: a board whose
numbers did not come back this week does not lock, however good it looks.

**Gate 1: author self-reproduction.** You open a fresh runtime, press Restart and
Run All on the notebook behind your poster, follow your written instructions
only, and confirm that every number on the board reappears. The record is your
**self-reproduction log**: the date, the environment, and each board number
matched to its freshly printed value, exactly like the match table you ran
earlier. Two rules govern it. The **freshness rule** says a change to the code,
the data, the environment record, the instructions, or the board after the run
makes the log stale, and a stale log does not pass. And the label rule you already
know: the log says "solo proxy; external cold run pending," because your own
rerun is not independent reproduction.

Reproduce the uncertainty statement too. A point estimate that comes back while
its interval was never rechecked has half reproduced.

**Gate 2: the release preflight.** Four checks, each answered in writing with an
evidence line rather than an assertion, before the file leaves your hands.

- **Permissions.** May every element on the board be shown publicly, under the
  promises your data rests on?
- **Disclosure.** Is the AI-use disclosure block on or with the poster, and does
  it match your AI Research Ledger?
- **Reproducibility.** Is the self-reproduction log fresh, with nothing changed
  since the run?
- **Claim boundary.** Does every claim on the board sit inside what your evidence
  licenses, with the preliminary-edition label visible? The
  **preliminary-edition label** is a visible line naming the board as a
  preliminary presentation of work still in progress. Example: "Preliminary
  edition: public test of a project that completes at the end of the semester."
  It is a strength, not an apology, because it tells a stranger which stage of
  evidence they are reading. It never excuses a number that failed a gate.

Two habits from earlier weeks supply that evidence. **Claim-evidence
traceability** means every number on the board points to the exact cell that
produced it, and a number you typed from memory or carried from an old draft is an
**untraceable number** that cannot go on the board. **Citation verification**
means every source the poster names resolves to a real, retrievable work you
opened yourself.

> **A question that often comes up here:** *"Why gate the lock twice? If the
> numbers reproduce, isn't the board honest?"* No, and your own board shows why.
> Gate 1 catches the number you copied from an early draft, the one your notebook
> no longer produces. Gate 2 catches everything arithmetic cannot see: a
> participant quote you never got permission to show, a missing disclosure block,
> a headline verb your evidence never licensed. Arithmetic honesty and release
> honesty are separate checks, and the lock waits for both.
