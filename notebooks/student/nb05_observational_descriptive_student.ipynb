## 1. The Route Is Not the Topic: Three Signals Pick a Pathway

**Guiding question:** *two projects share a topic; what decides that they need
two different research designs?*

> *"Do not tell me your topic. Tell me your route: what your question asks, what
> your data let you do, and what claim you will be licensed to defend when the
> work is done. Topics are cheap. Routes are commitments."*
> — a **thesis advisor**, opening the semester's first design meeting

Your **Research Contract** is the written declaration you built at M4: your
question, your model-inquiry-data-answer design, your permission status, and
your claim boundary, versioned like software so every later change is visible.
Example: Contract v0 says WHAT you will study; this week's v1 adds the pathway,
HOW your evidence will earn its claim. Three signals, and only three, pick that
pathway:

- **Kind**: whether the question's words ask what *is* (descriptive) or what
  *would change under an intervention* (causal). Example: "what share of
  buildings exceed the threshold" is descriptive; "does the program reduce
  skipped meals" is causal.
- **Reach**: which cases the answer is about: the data at hand, a wider
  population, or cases nobody has seen yet. Example: "in the dorms I surveyed"
  reaches the data at hand; "among all first-years" reaches a population.
- **Data strategy**: whether you **observe** the world as you find it or
  **assign** conditions yourself. Example: comparing dorms that adopted a
  program is observing; flipping a coin over which dorms get it is assigning.

**Prediction** stands apart. When the question's object is unseen cases and the
answer is graded only by how well a forecast scores on cases it never studied,
prediction is its own answer objective, not a variant of the other four.

**The kind rule, the one guard this week enforces.** A causal question whose
data cannot yet identify it keeps its kind. Its honest status is **"causal,
currently unidentified"**: the question stays causal, the design does not yet
reach it, and the gap is stated in the open. Relabeling it descriptive because
the data are weak rewrites your research without admitting it. Changing the
question is a human decision, recorded openly as a Contract revision, never a
silent reclassification. Hold the rule for now and keep your verdict on Project B
where you wrote it; section 3 settles that case against the matrix.

> **A question that often comes up here:** *"Isn't the route just another name
> for the method?"* No, and the difference is the whole week. A method is
> machinery: a regression, a classifier, a survey instrument. A route is a
> license structure: what the question asks, what the data strategy permits, and
> what claim the evidence can then carry. Two projects can run the same
> regression on different routes and be licensed to say entirely different
> things. Pick the route first and the method follows the question; pick the
> method first and the claims tend to follow the toolbox.

### 🛠️ Run the Study: deal the five cards

Run the five card cells in order. After each one, answer aloud before moving on:

**Card 1 · Observational descriptive.** The question asks *how much, how many,
or how distributed*. You observe the world as you find it. Data:
`lapop_brazil`, a real survey of trust in institutions in Brazil. One caveat
travels with every number: this file is a 10,000-row resample with replacement
of the original interviews, realistic in its patterns but manufactured in its
precision, so it serves planning and practice, never substantive claims about
Brazil. Expect a table of counts across a 1-to-7 trust scale and three summary
numbers that disagree with each other interestingly.

**Four boxes keep a description honest**, and this route's whole reach argument
lives in them.

- **Target population**: everyone your question is about. Example: every adult
  in Brazil, if that is what your question says.
- **Accessible population**: the part of the target you could reach in
  principle, given time, money, and access. Example: adults reachable by an
  in-person interview in the sampled municipalities.
- **Sampling frame**: the concrete list, database, or feed you actually draw
  from. Example: the survey's household listing for those municipalities. This
  is the box that misbehaves, because people can be missing from it, listed
  twice, or listed but outside your target altogether.
- **Sample**: the people you actually measured, drawn from the frame and no
  wider. Example: the rows in the file below.

Read those four in order and your claim stops where the boxes stop.

```python
# ROUTE CARD 1 — observational descriptive: describe what you actually reached.
lapop = load_course_data("lapop_brazil.csv")
assert lapop.shape == (10000, 10), "unexpected shape — flag this!"
print("✓ Loaded lapop_brazil.csv —", lapop.shape[0], "rows,", lapop.shape[1], "columns")
print("  Route: OBSERVE the world as found; ask how much / how many / how distributed.")
print()

tp = lapop["trust_police"]
counts = tp.value_counts().sort_index()
print("Trust in the police (1 = none, 7 = a lot) — counts per scale point:")
print(counts.to_string())
print(f"\nSignature output: mean = {tp.mean():.2f}, median = {tp.median():.0f}, "
      f"most common answer = {tp.mode().iloc[0]:.0f}")
print("Card boundary: describes these respondents; a population claim needs a")
print("sampling-and-response warrant, and this teaching resample licenses none.")

assert abs(tp.mean() - 4.43) < 0.02, "mean drifted — investigate"
assert tp.mode().iloc[0] == 7, "mode should be 7"
print("\n✓ Card 1 self-check passed: mean 4.43, median 5, mode 7 — three honest summaries, three stories.")
```

**Reading the output.** The mean says 4.43, a lukewarm middle; the counts say
the room is split, with 7 the single most common answer and a real lump of 1s.
An honest description reports the shape, not just the average, and it stops at
the respondents actually reached until a sampling and response warrant pays for
more. In box language: this output describes the sample, and every step out
toward the target population has to be paid for at the frame. The claim verb
this card licenses is **describe**.

**Card 2 · Observational causal.** The question asks *whether one thing causes
another*. You still only observe, so the route lives or dies on an
**identification argument**: a defensible reason the comparison you observe
stands in for the intervention you never ran. Without one, a **back door** can
be at work: a third trait feeding both groups being compared, manufacturing a
difference no cause produced. Data: `cliningsmith_etal` records applicants to a
real visa lottery for the Hajj pilgrimage. The lottery is as-if random among
applicants, which closes the back doors for that pool. One honesty note before
you run it: the file records who *won* the lottery, not who actually traveled,
so the comparison below estimates the **intention-to-treat effect**: the causal
effect of winning the visa lottery on these applicants, whatever each winner did
next. A winner who stayed home, for example, still counts as a winner. Expect
two group means and a difference with a 95% interval.

```python
# ROUTE CARD 2 — observational causal: observe, but with an identification argument.
hajj = load_course_data("cliningsmith_etal.csv")
assert hajj.shape == (958, 8), "unexpected shape — flag this!"
print("✓ Loaded cliningsmith_etal.csv —", hajj.shape[0], "rows,", hajj.shape[1], "columns")
print("  Route: OBSERVE, plus an identification argument (a real visa lottery).")
print()

won = hajj.loc[hajj.success == 1, "views"]
lost = hajj.loc[hajj.success == 0, "views"]
diff = won.mean() - lost.mean()
se = np.sqrt(won.var(ddof=1) / len(won) + lost.var(ddof=1) / len(lost))
ci_lo, ci_hi = diff - 1.96 * se, diff + 1.96 * se
print(f"Views-toward-other-groups index: won lottery {won.mean():.2f} (n={len(won)}), "
      f"lost lottery {lost.mean():.2f} (n={len(lost)})")
print(f"Signature output: difference {diff:+.2f}, 95% interval [{ci_lo:+.2f}, {ci_hi:+.2f}]")
print("Card boundary: the causal effect of WINNING THE LOTTERY (the intention-to-treat")
print("effect) for these applicants; the effect of actually performing Hajj needs the")
print("instrumental-variables machinery the route chapter describes. Unadjusted")
print("classroom version, and silent on mechanism.")

assert len(won) == 510 and len(lost) == 448, "arm sizes changed — investigate"
assert diff > 0 and ci_lo > 0, "expected a positive difference whose interval clears zero"
print("\n✓ Card 2 self-check passed: the lottery bought a causal reading (of winning) no software could.")
```

**Reading the output.** Winners score +0.47 higher, interval [+0.16, +0.79].
Because the lottery is as-if random among applicants, that difference carries a
causal reading, for applicants, about winning: it is the intention-to-treat
effect of lottery success, not the effect of performing Hajj itself. The file
never records who traveled, and the original study reaches the pilgrimage's own
effect only through the instrumental-variables machinery the route chapter
describes. Same observing posture as Card 1; a different license, bought by the
lottery, not by the code. The claim verb is **cause**, bounded to the applicant
pool and to winning. Without leverage like this, the route's honest status is
exactly the kind rule's: causal, currently unidentified.

**Card 3 · Experimental descriptive.** The question asks *how much of a property
exists, measured cleanly*. You assign conditions, a genuine intervention, and
you use the assignment as a measuring instrument. Data: `bonilla_tillery`
randomly assigned each respondent one of four identity frames for a social
movement, then recorded support on a 0-to-1 scale; the random frames probe how
support responds when the words change. Expect four frame means with standard
errors and a gap smaller than its own sampling uncertainty.

**A measurement can be confident and still wrong**, because the setup is one of
the ingredients. Two artifacts do most of the damage, and they are the two your
milestone asks you to name.

- A **demand effect** is when the people being measured guess what you are
  looking for and drift toward it. Example: a respondent who senses the survey
  favors one frame answers warmer to that frame than they would at home.
- An **instrument effect** is when the wording or the device moves the reading
  on its own. Example: one frame is phrased in longer, warmer words, so the
  scale reads higher for reasons that have nothing to do with the identity.

Both threaten **construct validity**, which is how well your number measures the
concept you named rather than something adjacent. Example: a "support" score
that mostly tracks reading ease is a valid number about something, but not about
support. Randomizing who sees what does nothing to a bias that hits every
respondent alike, so you design these out separately.

```python
# ROUTE CARD 3 — experimental descriptive: assign, in order to MEASURE.
bt = load_course_data("bonilla_tillery.csv")
assert bt.shape == (849, 10), "unexpected shape — flag this!"
print("✓ Loaded bonilla_tillery.csv —", bt.shape[0], "rows,", bt.shape[1], "columns")
print("  Route: ASSIGN conditions, an intervention used here as a measuring instrument.")
print()

by_frame = bt.groupby("Z")["blm_support"].agg(["mean", "std", "count"])
by_frame["se"] = by_frame["std"] / np.sqrt(by_frame["count"])
print("Mean support (0-1) under each randomly assigned identity frame:")
print(by_frame[["mean", "se", "count"]].round(3).to_string())

gap = by_frame["mean"].max() - by_frame["mean"].min()
se_gap = float(np.sqrt(by_frame["se"].max() ** 2 + by_frame["se"].min() ** 2))
print(f"\nSignature output: high-minus-low frame gap = {gap:.3f} "
      f"(about {gap / se_gap:.1f} standard errors of that gap: smaller than its sampling uncertainty)")
print("Card boundary: HIGH support under every randomly assigned frame; this experiment")
print("did not detect a frame effect of this size, which is not the same as showing")
print("the frames are equivalent.")

assert by_frame["mean"].min() > 0.79, "every frame should read as high support"
assert gap < 0.05 and gap < 2 * se_gap, "the frame gap should be smaller than its sampling uncertainty"
print("\n✓ Card 3 self-check passed: assignment everywhere, and the verb is still 'measure'.")
```

**Reading the output.** The four frame means sit between 0.805 and 0.842, and the
largest arm gap, 0.037, is smaller than its own sampling uncertainty. The frames
were randomly assigned, so they are a real intervention, used here to measure
support under each wording. What that licenses is precise and modest: this
experiment did not detect a frame effect of that size. It does not license "the
frames are equivalent." An **equivalence claim** is a positive claim that any
effect is smaller than a margin you declared in advance, for example "any frame
effect larger than 0.05 points matters, and our interval rules that out," and
this card declared no margin. The claim verb is **measure**, and the standing
course line holds: experimental assignment does not automatically imply a causal
inquiry. Randomization also bought nothing against the two artifacts above,
because a demand effect or an instrument effect would sit in all four arms at
once. That is why construct validity is this route's warrant.

**Card 4 · Prediction.** The question asks *what will happen to cases nobody has
seen*. You fit a forecast and grade it only on cases it never studied. Data:
`la_voter_file`, a real voter-file extract whose 2012 turnout column makes an
honest practice target. Today the card only signs the contract: it names the
target, computes the **baseline** (the dumbest honest rule any model must beat),
and declares the split.

**The split has three roles, not two.** The **training set** is the data each
candidate model learns from; example: three feature lists all fit here. The
**selection set** is separate data used to choose among those candidates;
example: you compare the three and keep one. The **final holdout** is locked
away and opened exactly once, after every choice is settled; example: only the
winner is ever scored on it. The middle role exists because the best score among
many tries is flattered by luck, so the score that crowned a model overstates
what it will do on data it has never met. No model is fit this week. Expect a
base rate and the three role sizes.

```python
# ROUTE CARD 4 — prediction: name the target, the baseline, and the THREE-role split.
voters = load_course_data("la_voter_file.csv")
assert voters.shape == (1000, 4), "unexpected shape — flag this!"
print("✓ Loaded la_voter_file.csv —", voters.shape[0], "rows,", voters.shape[1], "columns")
print("  Route: FIT and honestly SCORE a forecast; graded only on unseen cases.")
print()

base_rate = voters["voted_2012"].mean()
print(f"TARGET = voted_2012.  Class balance: {voters['voted_2012'].value_counts().to_dict()}")
print(f"Signature output: baseline to beat = {base_rate:.1%} "
      f"(always guessing the majority answer, 'voted')")

# Declare all THREE roles NOW, before any model exists (seed 464).
final_holdout = rng.choice(len(voters), size=250, replace=False)
rest = np.setdiff1d(np.arange(len(voters)), final_holdout)
selection = rng.choice(rest, size=250, replace=False)
training = np.setdiff1d(rest, selection)
print(f"\nThe split, declared before any model exists (seed {SEED}):")
print(f"  training  : {len(training):4d} rows — every candidate learns here")
print(f"  selection : {len(selection):4d} rows — the candidates are compared here")
print(f"  final hold: {len(final_holdout):4d} rows — locked, opened exactly once")
print("Card boundary: a forecast says WHO, never WHY, and never what an intervention")
print("would change; its score means nothing until it beats the baseline on the final")
print("holdout, and a score read off the selection set is flattered by the comparison.")

assert abs(base_rate - 0.603) < 0.001, "base rate drifted — investigate"
assert len(final_holdout) == 250 and len(set(final_holdout)) == 250, \
    "the final holdout must hold 250 distinct rows"
assert len(training) == 500 and len(selection) == 250, "the three roles must sum to 1000"
assert not (set(training) & set(selection)) and not (set(training) & set(final_holdout)) \
    and not (set(selection) & set(final_holdout)), "the three roles must not overlap"
print("\n✓ Card 4 self-check passed: contract signed, three roles disjoint, no model touched.")
```

**Reading the output.** The baseline is 60.3%: always guessing "voted" is right
six times in ten for free, so any future model must beat that, on the 250
final-holdout rows it never saw. The 250 selection rows are where candidates
compete, and their winning score is not the number you report. Prediction's
honesty is structural, contract first and scoring later, and its claim verb is
**forecast**, earned only if the final-holdout score beats the baseline.

**Card 5 · Experimental causal.** The question asks *whether an intervention
works*. You assign conditions to evaluate them: chance decides who is treated,
the arms become comparable, and a plain difference carries a causal reading for
these units. Data: `foos_etal`, a real UK get-out-the-vote field experiment in
which chance assigned voters to a contact arm or a control arm. Expect turnout
by arm and a difference with a 95% interval.

**One honesty line rides with this route.** **Attrition** is a unit's outcome
going missing after assignment. Example: a voter moves away, so no 2014 turnout
is ever recorded for them. When outcomes go missing in a way tied to what the
outcome would have been, the difference you can still compute is a
**complete-case contrast**, a comparison of whoever happened to be measured in
each arm. Example: if contact kept the least-likely voters engaged enough to be
matched in the register, the two measured groups are no longer the same kind of
people. A complete-case contrast is not the effect for everyone enrolled, and it
does not belong to any named group of people who stayed either, because that
reading needs assumptions you would have to state and defend. So report
**retention by arm**, the share of
each arm whose outcome you actually observed, before any effect. Uneven
retention is a warning light that tells you where to dig, never a verdict on its
own.

```python
# ROUTE CARD 5 — experimental causal: assign, in order to EVALUATE an intervention.
foos = load_course_data("foos_etal.csv")
assert foos.shape == (8375, 5), "unexpected shape — flag this!"
print("✓ Loaded foos_etal.csv —", foos.shape[0], "rows,", foos.shape[1], "columns")
print("  Route: ASSIGN conditions by chance to test whether an intervention works.")
print()

t1 = foos.loc[foos.treat == 1, "marked_register_2014"]
t0 = foos.loc[foos.treat == 0, "marked_register_2014"]
diff = t1.mean() - t0.mean()
se = np.sqrt(t1.mean() * (1 - t1.mean()) / len(t1)
             + t0.mean() * (1 - t0.mean()) / len(t0))
ci_lo, ci_hi = diff - 1.96 * se, diff + 1.96 * se
print(f"2014 turnout: contacted {t1.mean():.1%} (n={len(t1)}), control {t0.mean():.1%} (n={len(t0)})")
print(f"Signature output: difference {diff * 100:+.1f} points, "
      f"95% interval [{ci_lo * 100:+.1f}, {ci_hi * 100:+.1f}] points")

# Retention by arm goes FIRST in a real write-up; it is printed second here only
# because the difference is what the card is about.
ret1 = foos.loc[foos.treat == 1, "marked_register_2014"].notna().mean()
ret0 = foos.loc[foos.treat == 0, "marked_register_2014"].notna().mean()
print(f"Retention by arm: contacted {ret1:.1%} measured, control {ret0:.1%} measured")
print("Card boundary: causal for THESE units; the unweighted classroom version,")
print("not the study's published headline estimate. Retention is complete in both")
print("arms here, so this difference speaks for everyone assigned; with outcomes")
print("missing it would be a complete-case contrast instead.")

assert len(t1) == 6104 and len(t0) == 2271, "arm sizes changed — investigate"
assert diff > 0 and ci_lo > 0, "expected a positive effect whose interval clears zero"
assert ret1 == 1.0 and ret0 == 1.0, "outcomes went missing — read this as a complete-case contrast"
print("\n✓ Card 5 self-check passed: chance built the comparison, so the difference reads causally.")
```
