## 1 and 2. The candidate list and the search log

Seed the candidate list three ways and say which route produced each row: **your
own search** in a database or repository you queried, **one AI sweep** (the prompt
below), and **hand snowballing**, following the reference list of a source you
already trust and the newer work that cites it. Mark one row as your **lead
source**, the one whose reference list you actually walked. A dropped row stays
visible, marked *could not retrieve*, because the attempt is part of your record.

A **search log** is the record of how you looked, not of what you found. Log the
empty searches. A query that returns nothing usable is a result, and it is the
only evidence that separates a real gap from a search you gave up on. A log with
only successes cannot support any claim about what is missing.

> 💡 **AI Prompt:** "I am building the evidence base for this question:
> [paste your declared question]. Name up to eight published works most likely to
> speak to it, each with authors, year, exact title, and venue, so that I can
> retrieve and verify each one myself. For each, say in one line what it claims,
> and mark how confident you are that the work exists. Do not summarize a
> literature for me."
>
> **After running, verify (counters *confident fabrication*: a real-sounding
> author, year, and venue for a paper that was never written):**
> - [ ] Search each exact title in quotes in your library portal or a scholarly
>   index. Only the ones you open go anywhere near your registry.
> - [ ] Every candidate that will not resolve stays on the candidate list marked
>   *could not retrieve*, with the search you ran beside it. That row is evidence
>   about the tool, and it is worth keeping.
> - [ ] Read the claim in the source itself before you write it down. A tidy
>   one-line claim is where a paraphrase quietly becomes a distortion.
> - [ ] Log this sweep in your AI Research Ledger below: task, tool, prompt,
>   output summary, decision, verification method, remaining concern.

```python
# Candidate list and search log — edit both tables, then run.
# Rows never disappear from CANDIDATES: a source you could not retrieve stays,
# marked, because the attempt is part of your record.

STATUSES = {"verified", "could not retrieve", "pending"}
ROUTES = {"own search", "AI sweep", "snowballing"}

CANDIDATES = [
    {"cite": "EDIT ME: author, year, title, venue", "route": "own search",
     "status": "verified", "lead": True},
    {"cite": "EDIT ME: a name the AI sweep handed you", "route": "AI sweep",
     "status": "could not retrieve", "lead": False},
    {"cite": "EDIT ME: a source from a reference list you walked", "route": "snowballing",
     "status": "pending", "lead": False},
]

# One line per search you ran, the empty ones included.
SEARCH_LOG = [
    {"where": "EDIT ME: the database or repository",
     "terms": "EDIT ME: the exact terms you typed",
     "result": "EDIT ME: what came back and what did not",
     "date": "EDIT ME: the date you ran it"},
    {"where": "EDIT ME: another place you looked",
     "terms": "EDIT ME: the exact terms",
     "result": "nothing usable retrieved",
     "date": "EDIT ME: the date you ran it"},
]

cand = pd.DataFrame(CANDIDATES)
print("Candidate list")
print(cand.to_string(index=False))
print()
print("Status tally:", dict(cand["status"].value_counts()))

problems = []
bad_status = sorted(set(cand["status"]) - STATUSES)
bad_route = sorted(set(cand["route"]) - ROUTES)
if bad_status:
    problems.append(f"status values outside the three allowed: {bad_status}")
if bad_route:
    problems.append(f"route values outside the three seeding routes: {bad_route}")
if cand["lead"].sum() != 1:
    problems.append("mark exactly one row as your lead source")
if not set(cand["route"]) >= ROUTES:
    problems.append(f"seed all three routes; missing {sorted(ROUTES - set(cand['route']))}")

empties = [s for s in SEARCH_LOG if "nothing" in s["result"].lower()
           or "no hits" in s["result"].lower()]
print(f"\nSearch log: {len(SEARCH_LOG)} searches, {len(empties)} of them empty.")
if not empties:
    problems.append("no empty search logged; a log of successes cannot support a gap")

if problems:
    print("\n⚠️ Not finished yet:")
    for p in problems:
        print(f"   - {p}")
else:
    print("\n✓ Three routes seeded, one lead source marked, and the empty searches kept.")
```

## 3. The evidence registry, and the retrievability audit

The registry holds only the rows that survived verification, and it needs **at
least six**. Verification has exactly one meaning here: you retrieved the source
and read it. An AI summary of a paper is not the paper, and a row built from one
does not belong in the registry.

Each row carries five fields: the **full citation**; **where you retrieved it**,
meaning the concrete database, journal, or repository page where you opened it;
its **central claim in your own words**, which is its claim and not its topic;
its **evidence type**, meaning the kind of claim its evidence actually supports
(**descriptive**, a **generalization**, **predictive**, or **causal**); and its
**licence and terms** for any dataset, recorded before you plan any analysis on it.

> **A worked registry row:** *Blair, Coppock & Humphreys (2023),* Research Design
> in the Social Sciences, *ch. 4, retrieved from book.declaredesign.org and read
> in full.* **Central claim, my words:** a research design is situated in a
> literature by naming the inquiry it answers, not by listing what came before.
> **Evidence type:** descriptive, a framework claim rather than a population
> generalization. That row proves retrieval, states the claim plainly, and types
> the evidence honestly.

Then run the **retrievability audit**: take each verified row and try to open its
source again from the citation alone, the way a stranger with only your
bibliography would. A row you cannot reopen goes back to *pending*, whatever you
remember about reading it.

```python
# Evidence registry — one dict per source YOU retrieved and read.
# reopened: the retrievability audit. Set it True only after you have reopened
# the source from the citation alone. A False sends the row back to pending.

TARGET = 6
EVIDENCE_TYPES = {"descriptive", "generalization", "predictive", "causal"}

REGISTRY = [
    {"cite": "EDIT ME: author, year, title, venue",
     "where_retrieved": "EDIT ME: the database, journal, or repository page you opened",
     "claim": "EDIT ME: its central claim, in your own words",
     "evidence_type": "descriptive",
     "licence": "n/a (not a dataset)",
     "reopened": False},
    {"cite": "EDIT ME: author, year, title, venue",
     "where_retrieved": "EDIT ME: where you opened it",
     "claim": "EDIT ME: its central claim, in your own words",
     "evidence_type": "generalization",
     "licence": "EDIT ME: who may use these data, and under what terms",
     "reopened": False},
    # keep adding rows until at least six survive the audit
]

reg = pd.DataFrame(REGISTRY)

def unverified(row):
    blob = f"{row['cite']} {row['where_retrieved']} {row['claim']}".lower()
    return ("edit me" in blob) or ("ai told me" in blob) or ("ai summary" in blob)

reg["placeholder"] = reg.apply(unverified, axis=1)
reg["bad_type"] = ~reg["evidence_type"].isin(EVIDENCE_TYPES)
reg["counts_as_verified"] = (~reg["placeholder"]) & (~reg["bad_type"]) & reg["reopened"]

show = reg[["cite", "evidence_type", "reopened", "counts_as_verified"]].copy()
show["cite"] = show["cite"].str.slice(0, 44)
print("Registry")
print(show.to_string(index=False))
print()

verified = int(reg["counts_as_verified"].sum())
back_to_pending = reg.loc[~reg["reopened"], "cite"].tolist()
print(f"Rows that survive the retrievability audit: {verified} of a required {TARGET}.")
if back_to_pending:
    print(f"→ {len(back_to_pending)} row(s) not reopened from the citation alone, "
          "so they return to pending until you can open them again.")
if verified >= TARGET:
    print("✓ Six or more rows a stranger could reopen from your bibliography alone.")
else:
    print(f"⚠️ {TARGET - verified} more retrieved and reopened source(s) needed. That is a")
    print("   finding, not a dead end: run the next lead and log the search either way.")
```

## 4. The evidence map, and your bounded gap sentence

The map sorts your verified claims into three zones, and every claim on it answers
to a registry row.

- **Settled**: claims your verified sources agree on. Say how many sources back
  each one, because two sources agreeing is not the same as ten.
- **Contested**: claims your verified sources genuinely disagree about, with one
  sentence per disagreement naming what could make both sides right, such as a
  different population, measure, time window, or design.
- **Missing**: the quiet spot, where your verified sources say nothing at all.

Contested territory is uncertainty you **inherit**. Where the field disagrees,
your project starts out unsure too, and that has to show up in what you later
claim. Then write your **gap** as one bounded sentence, in this shape: *"Across
the sources I retrieved and verified, no source establishes ___ for ___."* Not
"nothing is known about ___", which is a claim about the whole world you cannot
make from your desk.

A **genuine gap** is a question your verified sources actually leave unanswered,
confirmed by looking and logged. An **unfinished search** is a question you simply
stopped looking for. They feel identical from your desk, and only the search log
tells them apart.

```python
# Evidence map — fill the two tables, then run. The figure draws itself from
# them, so an edit to a claim or a relation redraws the map with no other change.

# Nodes: one row per claim, each traceable to a registry row.
#   zone: "settled", "contested", or "missing"
CLAIMS = [
    {"id": "C1", "claim": "EDIT ME: a claim your sources agree on",
     "zone": "settled", "n_sources": 3},
    {"id": "C2", "claim": "EDIT ME: a claim your sources disagree about",
     "zone": "contested", "n_sources": 2},
    {"id": "C3", "claim": "EDIT ME: the spot where your sources say nothing",
     "zone": "missing", "n_sources": 0},
]

# Edges: how two claims relate. "agrees", "contradicts", or "qualifies".
RELATIONS = [
    {"src": "C1", "dst": "C2", "relation": "contradicts"},
    {"src": "C2", "dst": "C3", "relation": "qualifies"},
]

GAP_SENTENCE = ("Across the sources I retrieved and verified, no source "
                "establishes EDIT ME for EDIT ME.")

ZONE_X = {"settled": 0.0, "contested": 1.0, "missing": 2.0}
COLOR = {"agrees": "#1a7f37", "contradicts": "#b3261e", "qualifies": "#33475b"}

G = nx.DiGraph()
pos, seen = {}, {z: 0 for z in ZONE_X}
for c in CLAIMS:
    G.add_node(c["id"])
    pos[c["id"]] = (ZONE_X.get(c["zone"], 1.0), -seen.get(c["zone"], 0))
    seen[c["zone"]] = seen.get(c["zone"], 0) + 1
for r in RELATIONS:
    if r["src"] in pos and r["dst"] in pos:
        G.add_edge(r["src"], r["dst"], relation=r["relation"])

fig, ax = plt.subplots(figsize=(9, 4.5))
nx.draw_networkx_nodes(G, pos, node_size=2200, node_color="#e8eef6",
                       edgecolors="#33475b", linewidths=1.5, ax=ax)
nx.draw_networkx_labels(G, pos, font_size=10, ax=ax)
for rel, color in COLOR.items():
    edges = [(u, v) for u, v, d in G.edges(data=True) if d["relation"] == rel]
    if edges:
        nx.draw_networkx_edges(G, pos, edgelist=edges, edge_color=color, width=1.8,
                               arrowstyle="-|>", arrowsize=18, node_size=2200,
                               connectionstyle="arc3,rad=0.12", ax=ax)
for zone, x in ZONE_X.items():
    ax.text(x, 0.8, zone.upper(), ha="center", fontsize=11, color="#33475b")
ax.set_title("Your evidence map: green agrees, red contradicts, dark qualifies",
             fontsize=11)
ax.set_xlim(-0.7, 2.7)
ax.set_ylim(-max(seen.values(), default=1) - 0.2, 1.3)
ax.axis("off")
plt.show()

for c in CLAIMS:
    backing = f"{c['n_sources']} source(s)" if c["n_sources"] else "no source"
    print(f"{c['id']} [{c['zone']:<9}] {backing}: {c['claim']}")
print(f"\nGap sentence: {GAP_SENTENCE}")
if "contradicts" not in {r["relation"] for r in RELATIONS}:
    print("⚠️ No contradiction drawn. Look again before you smooth the field into "
          "a consensus it does not have.")
if "nothing is known" in GAP_SENTENCE.lower():
    print("⚠️ That is an absolute-absence claim. Bound it to what you retrieved.")
```

### 🗡️ Red-Team Exchange

This studio is a **gallery walk**: your registry and your map go up, the room
circulates and leaves challenges, and you carry one into the artifact before you
submit. Run two rounds with a partner, and bring your AI assistant in as a third
voice, briefed as a hostile reviewer rather than a helper.

**Round 1: attack the registry.**
Your partner picks two rows and asks the question that decides this milestone:
show me where you retrieved this, and state its claim in your own words. Any row
you cannot defend by naming its retrieval comes off the registry until you open it
again. A tool can flag which citations look unverifiable, and that flag is a
prompt to check, never a verdict: a clean-looking citation nothing flagged is not
confirmed either.

**Round 2: attack the gap.**
Your partner tries to close your gap by naming a source, yours or theirs, that
might already cover it. If one of your own verified sources fills it, that is the
most important find of the block, and the gap moves to where the map is genuinely
quiet. You defend the gap against your search log, or you concede and re-place it.

**What to log.** Pick **one** challenge from the walk, work it into the registry,
the map, or the gap sentence *before* you submit, and write one or two sentences
naming the challenge and what you changed. A challenge that changed nothing still
gets a line saying why it did not land.

> **A question that often comes up here:** *"Two of my verified sources flatly
> contradict each other. Is one of them wrong?"* Usually neither. Ask what would
> make them *both* right: a different population, measure, time window, or method.
> That question is where new gaps are born, so put the contradiction in the
> contested zone instead of smoothing it away.
