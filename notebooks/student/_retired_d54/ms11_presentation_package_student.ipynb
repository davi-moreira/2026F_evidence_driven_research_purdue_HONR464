## 🎯 Definition of Done

**A release audit runs before the release, never after it.** Once strangers
read your board and hear your claim, the work is public, and no audit run
afterward calls any of it back. You are done when your submission carries all
six parts, applied to **your own project**.

1. **The review-management record.** The AI team you directed for the closing
   review, in five parts: the roles and their one-sentence scopes (at minimum
   a **worker**, a **critic**, and an **auditor**, who checks the record
   rather than the claim), the wiring and why each role's input existed before
   it ran, the **escalation rules written first**, the one **human node** you
   never delegated, and an honest **capability declaration** saying who chose
   each next step. Record one **load-bearing pair**: two roles on the same
   question, where something on your poster depended on the answer.
2. **The independence check.** At least one named place where **agreement was
   not verification**: name the correlation (same model, same prompt, same
   source, or the same summary of your work standing in for the work), say
   what the agreement was actually worth (usually one opinion, repeated), and
   name the human check you ran instead.
3. **The release audit.** Seven rows, each marked **clear**, **pending**, or
   **blocking**, each supported by a sentence saying what you checked:
   permissions, participant promises, disclosure, package matches the cold
   run, reproducibility, claim boundaries, uncertainty. **If anything moved
   since your last clean rerun, that rerun is stale**, and the repair is a
   rerun rather than an argument.
4. **The release decision.** **Release**, or **withhold pending a named
   repair**, written as a dated, numbered version with a reason a reader could
   use to reconstruct your thinking. Never a soft release: "mostly fine" is
   not a disposition, and a blocking row described gently is still blocking.
5. **Dossier v1 and its manifest.** Every component marked **present**,
   **absent with reason**, or **not applicable**, with a location for each
   present one: contract versions, evidence registry, permissions,
   measurement documentation, analysis and claim ledgers, the AI Research
   Ledger, artifacts, the reproducibility package with its cold-run record,
   and the revision history.
6. **The stopping rule, the next-cycle agenda, the ledger rows, and the
   dossier line.** One paragraph saying why you stop here rather than running
   one more check. Then two or three agenda items, each written as a
   **direction**: the question the next study should ask, the unresolved
   limitation it comes from, and the evidence that would settle it.

> **A question that often comes up here:** *"My poster is locked and printed.
> What is actually left to decide?"* The paper is decided; the claims are not.
> Release is a decision about what you assert in public, under whose
> permission, from which version of the evidence. You can still add a caveat
> to what you say, withhold a figure whose permission is unclear, or correct a
> number that moved after your last clean run.

### 🤝 AI Research Partner

This is the sprint where you stop being a user of one tool and start
**directing a team**. A **directed AI team** is two or more scoped roles you
assign, sequence, and adjudicate yourself, rather than one open-ended
conversation.

**Good to delegate here:**
- The worker's job: drafting or checking something concrete, such as a
  manifest draft from your file list.
- The critic's job: attacking the worker's output on grounds you state.
- The auditor's job: reading your own record and asking whether what you say
  happened is what the logs show.

**Never delegate (these are yours to decide):**
- The **release decision**. That is the natural human node, and naming it
  explicitly is part of the record.
- Every audit **disposition**. "Clear" with no sentence behind it is an
  assertion, and this course grades checks rather than assertions.
- The **next-cycle agenda**. The claims a next study would test are exactly
  the ones your current evidence has not earned.

One trap this milestone exists to catch: **agreement among correlated
reviewers is not independent verification.** Two reviewers are correlated when
they share a model, a prompt, or a source, so their mistakes repeat instead of
cancelling. Verify by a route that does not pass through the reviewer. Log
every step in your **AI Research Ledger** below; the full never-delegate list
lives in `ai_resources/human_responsibility_checklist.md`.

> 💡 **AI Prompt:** "You are the **critic** in a review bench I am
> directing. Here is the worker output I am checking, and here is the claim on
> my poster that depends on it: '[paste both].' Attack the worker's output on
> three stated grounds: is any number in it unsupported by what I pasted, does
> any sentence reach past my stated boundary, and is any part of it simply
> repeating my own summary back to me? Quote the exact words for each. Do not
> propose a decision, and do not tell me the work looks fine."
>
> **After running, verify (this is the independence check, and it is required
> content rather than an optional confession):**
> - [ ] Name the correlation before you weigh the answer: same model, same
>   prompt, same source document, or your summary standing in for your work.
> - [ ] Write what the agreement or the disagreement was worth. Two roles that
>   nod at the same summary are one opinion, delivered twice.
> - [ ] Run the human check instead: recompute the number, open the cited
>   source, or rerun the cell. A check whose path avoids the thing being
>   checked is the only kind that adds evidence.
> - [ ] Log this use in your AI Research Ledger: task, tool, decision,
>   verification.

> 💡 **AI Prompt:** "Here are the unresolved issues documented in my
> project: the limitations my audit could not clear, the warrant gaps in my
> design, and the data I could not obtain: '[paste them].' For each, write the
> next study's question, the documented issue it comes from, and the evidence
> that would settle it. Write **directions only**: no sentence may assert what
> the next study would find, and no item may cite a source I have not listed."
>
> **After running, verify (this counters the *agenda that is really a
> finding*, a claim you did not test dressed as a plan):**
> - [ ] Cut any item that asserts an outcome. "The effect is probably stronger
>   in larger firms" is a finding; "test whether the pattern holds where
>   hiring is centralized, using records I could not obtain" is a direction.
> - [ ] Confirm each item ties back to something documented in this cycle, so

> - [ ] Check that no source appeared that you did not supply, and retrieve
>   any that did before it enters the artifact.

### 🛠️ Run It Live: the audit, the decision, and the agenda

The scaffold below runs your audit as a set of dispositions rather than a
feeling. Cell one takes the seven rows, applies the stale-run rule, and prints
the verdict. Cell two reads your dossier manifest and scans your agenda for
items that are really findings. Run them on the defaults first, then edit the
values at the top with **your own** audit and rerun.

```python
# The seven-row release audit — edit the EDIT ME blocks with YOUR checks.
# A disposition with no supporting sentence is worth nothing, so each row
# carries both. The stale-run rule is applied for you: if anything moved since
# your last clean rerun, the package row is blocking until you repeat the run.

# EDIT ME — what has changed since your last clean rerun of the package.
moved_since_cold_run = ["figure 2 filter"]   # empty list if nothing moved

# EDIT ME — one row per check: (disposition, the sentence saying what you did).
audit = {
    "permissions":          ("clear", "reread my determination; nothing identifiable is on the board"),
    "participant promises": ("clear", "reread the recruitment text; the use I am presenting is the use I promised"),
    "disclosure":           ("clear", "the AI-use block on the board names tool, task, and verify step"),
    "package matches cold run": ("clear", "EDIT ME — say what you compared, by content"),
    "reproducibility":      ("pending", "EDIT ME — the run that produces the numbers I will present"),
    "claim boundaries":     ("clear", "every public claim reread against my compass position"),
    "uncertainty":          ("clear", "the bound travels with the estimate on the board and in what I say"),
}

if moved_since_cold_run:
    audit["package matches cold run"] = (
        "blocking",
        f"these moved since the last clean rerun: {moved_since_cold_run}; "
        "the repair is a rerun, not an argument")

MARKS = {"clear": "✓", "pending": "○", "blocking": "⚠️"}
print("— The release audit —")
unsupported = []
for row, (state, why) in audit.items():
    if "EDIT ME" in why or not why.strip():
        unsupported.append(row)
    print(f"  {MARKS[state]} {row:26s} {state:9s} · {why}")

print("\n— The dispositions —")
blocking = [r for r, (s, _) in audit.items() if s == "blocking"]
pending  = [r for r, (s, _) in audit.items() if s == "pending"]
if unsupported:
    print(f"  ⚠️ unsupported dispositions: {unsupported}")
    print("     'Clear' with no sentence behind it is an assertion, not a check.")
print(f"  blocking: {blocking or 'none'}   ·   pending: {pending or 'none'}")

print("\n— The release decision —")
if blocking:
    print("  WITHHOLD pending a named repair. Name the exact repair, the row")
    print("  that blocked, and the one-line plan that turns this into a")
    print("  release. Withholding is a fully creditable outcome.")
elif pending or unsupported:
    print("  NOT YET. Finish each named repair and support every disposition,")
    print("  then rerun this cell.")
else:
    print("  RELEASE. Write it as a dated, numbered version with its reason,")
    print("  naming which package version and which claim set go public.")
print("  A soft release is not an option: 'mostly fine' is not a disposition.")
```

```python
# The dossier v1 manifest and the next-cycle agenda.
# Part 1: a manifest entry marked present must say where it lives. Part 2: an
# agenda item is a DIRECTION, never a finding your evidence has not earned.

# EDIT ME — mark each component present / absent with reason / not applicable,
# and give a location for every present one.
manifest = {
    "contract versions":        ("present", "dossier §1, v0 through v3"),
    "evidence registry":        ("present", "dossier §2"),
    "permissions":              ("present", "dossier §3"),
    "measurement documentation": ("present", "dossier §4"),
    "analysis and claim ledgers": ("present", "dossier §5"),
    "AI Research Ledger":       ("present", "dossier §6"),
    "artifacts (poster, note)": ("present", "dossier §7"),
    "package and cold-run record": ("present", ""),          # EDIT ME
    "revision history":         ("absent with reason", "rebuilt from version lines this week"),
}

# EDIT ME — two or three items, each: question · documented issue · evidence.
agenda = [
    "Test whether the pattern holds where the schedule is centrally set, "
    "using the staffing records my permission did not cover this term.",
    "The effect is probably stronger in the larger halls.",
]

print("— Dossier v1 manifest —")
for item, (state, where) in manifest.items():
    if state == "present" and not where.strip():
        print(f"  ⚠️ {item:30s} present, but a reader cannot find it. Name "
              "the file or section.")
    else:
        print(f"  {'✓' if state == 'present' else '·'} {item:30s} {state}"
              + (f" ({where})" if where.strip() else ""))
print("  An honest 'absent with reason' beats a checkbox that is not true.")

print("\n— Next-cycle agenda: direction or finding? —")
FINDING_CUES = ["probably", "likely", "is stronger", "is weaker", "we know",
                "the effect is", "tends to", "must be", "clearly"]
DIRECTION_CUES = ["test whether", "ask whether", "would settle", "measure",
                  "collect", "compare", "replicate"]
for i, item in enumerate(agenda, 1):
    low = item.lower()
    finding = [c for c in FINDING_CUES if c in low]
    direction = [c for c in DIRECTION_CUES if c in low]
    if finding:
        print(f"  {i}. ⚠️ reads as a FINDING ({finding}). Rewrite it as the "
              "question, the issue it")
        print("     comes from, and the evidence that would settle it.")
    elif direction:
        print(f"  {i}. ✓ reads as a direction. Confirm it names the documented "
              "issue it comes from.")
    else:
        print(f"  {i}. ○ neither cue found. Read it aloud: is it a plan, or a "
              "claim in disguise?")
```

### 🗡️ Red-Team Exchange

The Friday studio is where the bench gets wired, the audit gets run, and the
decision gets made before you leave. Two rounds.

**Round 1: the bench, and the independence check.**
Run your worker and your critic on one **load-bearing pair**, a question where
something on your poster depends on the answer, then diagnose the result
rather than accepting it. If the roles agreed, name what they shared before
you count the agreement: a model, a prompt, a source, or your own summary
standing in for your work. Then run the check whose path avoids the reviewer
entirely: recompute the number, open the source, rerun the cell. Record what
it found, and record any override with its reason.

**Round 2: a peer audits your audit.**
Your partner takes the **auditor** role on your record: not "is your claim
right" but "is what you say happened what your logs show". They press on three
places. The stale-run row, answered by date and by content. Any disposition
whose supporting sentence is missing or vague. And any softening language in
the decision. Then they read your agenda and challenge every item that asserts
what a next study would find.

**What to log.** The load-bearing pair with your diagnosis, the independence
check with what it found, every override, and the strongest hit your auditor
landed. A challenge that changed nothing still gets a line saying why it did
not land.

> **A question that often comes up here:** *"If three roles all pass my claim,
> why is that not stronger evidence than one?"* Because they may all be
> repeating the same input. Three readers who share a model and a prompt fail
> in the same places at the same time, so their agreement measures their
> similarity rather than your claim's strength. Two genuinely different checks
> that disagree are worth more than three identical ones that nod.
