## 🎯 Definition of Done

You are done with this sprint when your submission carries two things applied to
**your own project**: a **poster draft 1** and a **research audit** whose every
number reruns from a seeded notebook. The audit is the graded core, and it has a
four-part spine:

1. **Pre-listed checks.** The **three** robustness or placebo checks you commit to
   *before* you look at any result, each tied to the reviewer attack it answers.
   Name each by what it varies (the sample, the measurement, the specification, or
   the metric), or as a placebo test or a leave-one-out influence check.
2. **What survived.** Your headline claim after the checks ran, worded to carry its
   **range** and its **compass boundary** (descriptive or causal), with a note on
   any check that moved the estimate and by how much.
3. **The verified AI-review trail.** The real flaws your human and multi-model AI
   reviewers found, each marked **confirmed or refuted by a data check**, plus the
   single most confident wrong flag you caught.
4. **The remaining limitation.** The one weakness no check could fix, stated as
   expertise rather than hidden.

Your **poster draft 1** shapes the same evidence into question, evidence, and one
bounded headline claim, with every number traceable to a cell and the uncertainty
visible. Plus **AI Research Ledger rows** and the **dossier update line**.

> **A question that often comes up here:** *"My estimate is still rough. Is it
> ready to audit?"* Audit it anyway. A rough number you have honestly tried to
> break beats a polished one you have not. The audit grades how hard you pushed and
> what you learned, not whether the finding turned out big. A check that made your
> estimate shrink is evidence you did the work.

```python
# Pre-listed robustness grid — edit the EDIT ME block to point at YOUR M8 estimate.
# It computes ONE headline gap under a SMALL grid of specifications you commit to
# BEFORE looking, then prints the direction and a span WITHIN each panel. Every
# label flows from a single source of truth, so renaming a column in ONE place
# updates the whole grid; there is no second hardcoded name to fall out of sync.

# --- Single source of truth: name your columns and labels ONCE. -----------
GROUP_COL   = "trust_group"      # EDIT ME — column that splits high vs low on your key variable
OUTCOME_COL = "support"          # EDIT ME — the number your headline is about
NC_COL      = "civic_knowledge"  # EDIT ME — a NEGATIVE-CONTROL outcome your cause could not touch
HIGH        = "high"             # EDIT ME — label for the high group
LOW         = "low"              # EDIT ME — label for the low group
PLACEHOLDER = True               # EDIT ME — set False once YOUR M8 data are wired in

# --- PLACEHOLDER data: replace this whole block with YOUR M8 data. ---------
# A seeded stand-in so the grid runs before your real data are wired in. Swap it
# for load_course_data(...) or your own CSV, keeping the names above. This
# placeholder is NOT evidence about anything; it only proves the pipeline runs.
# It builds in one honest complication: the high-trust group skews OLDER, and
# age itself pushes both outcomes up, so part of any raw gap here is age. The
# built-in trust effect on the headline outcome is +0.50 at any fixed age, and
# exactly 0 on the negative-control outcome.
n_per = 400                                   # EDIT ME — your per-group size
age = np.clip(np.concatenate([rng.normal(52, 12, n_per),
                              rng.normal(38, 12, n_per)]), 18, 90)
df = pd.DataFrame({
    GROUP_COL:   [HIGH] * n_per + [LOW] * n_per,
    "age":       age,
    OUTCOME_COL: 0.02 * (age - 45)
                 + np.concatenate([rng.normal( 0.25, 1.0, n_per),
                                   rng.normal(-0.25, 1.0, n_per)]),
    "ideology_extreme": rng.random(2 * n_per) < 0.15,   # a subgroup you may drop
    "speeder":          rng.random(2 * n_per) < 0.05,   # finished implausibly fast
    NC_COL:      0.02 * (age - 45) + rng.normal(0, 1.0, 2 * n_per),  # NO trust term
})
# --------------------------------------------------------------------------

# --- Two estimators, because "adjust" is not a row label. It is real code. --
def headline_gap(data, col=OUTCOME_COL):
    hi = data.loc[data[GROUP_COL] == HIGH, col]
    lo = data.loc[data[GROUP_COL] == LOW,  col]
    return hi.mean() - lo.mean()

def age_band_standardized_gap(data):
    # Compare the groups INSIDE each age band, then average the within-band
    # gaps weighted by pooled band size. This is AGE-BAND STANDARDIZATION: it
    # compares the groups at a common age-band distribution. It is NOT an
    # exact same-age comparison — inside a coarse band the two groups can
    # still differ in age, so some age difference survives. Finer bands (or a
    # declared model in age) tighten it, at the price of sparser cells.
    bands = pd.cut(data["age"], [18, 35, 50, 65, 90], include_lowest=True)
    gaps_, w_ = [], []
    print("  band support (hi/lo counts — a sparse cell weakens the comparison):")
    for band, chunk in data.groupby(bands, observed=True):
        hi = chunk.loc[chunk[GROUP_COL] == HIGH, OUTCOME_COL]
        lo = chunk.loc[chunk[GROUP_COL] == LOW,  OUTCOME_COL]
        flag = "  ⚠ SPARSE — treat this band's contribution with suspicion"             if min(len(hi), len(lo)) < 5 else ""
        print(f"    {str(band):16s}: {len(hi):3d} / {len(lo):3d}{flag}")
        if len(hi) and len(lo):
            gaps_.append(hi.mean() - lo.mean())
            w_.append(len(chunk))
    return float(np.average(gaps_, weights=w_))

# --- Your PRE-LISTED grid: commit this list BEFORE reading any estimate. ---
# Reporting only the flattering corner is a specification search. Reporting the
# whole grid is robustness. Same code, opposite integrity.
#
# Each row declares its PANEL and its ESTIMATOR. A span may only be computed
# WITHIN a panel: rows that change the population, the outcome definition, or
# the question itself (raw vs age-standardized) answer different questions, so
# mixing them into one min-max range compares unlike numbers. A panel holding
# one row is an estimate standing on its own, never a span.
#
# PANEL DECLARATION for the primary panel: "all rows" and "drop speeders" both
# target the same population — genuine responses — under two pre-listed rules
# for what counts as genuine (trust every row vs drop implausibly fast ones).
# Same target, same outcome, same estimator: that is what makes them share a
# span. An exclusion that changed WHO you are asking about (like the ideology
# row) gets its own panel instead.
q_lo, q_hi = df[OUTCOME_COL].quantile(0.02), df[OUTCOME_COL].quantile(0.98)
specs = {
    # name                     (panel, data, estimator)
    "all rows":               ("genuine responses / raw gap", df, headline_gap),
    "drop speeders":          ("genuine responses / raw gap",
                               df.loc[~df["speeder"]], headline_gap),
    "drop ideology extremes": ("ideology-moderate subset / raw gap",
                               df.loc[~df["ideology_extreme"]], headline_gap),
    "trim outer 2%":          ("genuine responses / trimmed gap",
                               df.loc[df[OUTCOME_COL].between(q_lo, q_hi)],
                               headline_gap),
    "age-band-standardized":  ("genuine responses / age-band-standardized gap",
                               df, age_band_standardized_gap),
}
gaps = {name: (panel, est(data)) for name, (panel, data, est) in specs.items()}

# --- Null check. These groups were NOT randomly assigned, and the high-trust
# group is OLDER by construction, so the groups are not exchangeable: a plain
# label shuffle would build a null world you never had, and comparing your gap
# with it would answer nothing. Use a NEGATIVE CONTROL instead: an outcome
# your cause could not touch that still runs through the same age and
# measurement pathways. It produces its OWN estimate, not a pile — read how
# far it lands from zero.
nc_gap = headline_gap(df, col=NC_COL)

for name, (panel, g) in gaps.items():
    print(f"  {name:24s}: gap = {g:+.3f}   [panel: {panel}]")

print("\nSpans, one per panel (never across panels):")
for panel in dict.fromkeys(p for p, _ in gaps.values()):
    vals = [g for pn, g in gaps.values() if pn == panel]
    if len(vals) == 1:
        print(f"  {panel:42s}: {vals[0]:+.3f}  (single specification -> its own"
              " estimate, not a span)")
    else:
        print(f"  {panel:42s}: {min(vals):+.3f} to {max(vals):+.3f}")
print("  -> a span measures how much your CHOICES moved it, not sampling uncertainty")

raw, adj = gaps["all rows"][1], gaps["age-band-standardized"][1]
print(f"\nRaw vs age-band-standardized: {raw:+.3f} vs {adj:+.3f} (moved {adj - raw:+.3f})")
print("  -> age is predictive AND imbalanced here, so standardizing moves the")
print("     estimate; the two answer different questions and BOTH get reported.")
print("     The placeholder BUILT IN a fixed-age trust gap of +0.50: the band")
print("     estimate lands near it but not on it, because coarse bands leave")
print("     residual age differences inside each band.")
if PLACEHOLDER:
    # placeholder-only sanity check — delete once your own data are wired in
    assert abs(adj - raw) > 0.10, "age adjustment should move the placeholder gap"

print(f"\nNegative control ({NC_COL}; true trust effect is 0 by construction):")
print(f"  contrast = {nc_gap:+.3f}")
print("  -> clearly nonzero: the raw machinery + the age composition invent a")
print("     'trust gap' on an outcome trust cannot touch. That concern attaches")
print("     to the raw headline too, and it is exactly the pathway the")
print("     age-band-standardized row addresses.")
print("\n✓ Ran on PLACEHOLDER data — swap in your M8 estimate before you report anything.")
```
