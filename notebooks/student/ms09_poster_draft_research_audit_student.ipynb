## 🎯 Definition of Done

You are done with this sprint when your submission carries all six parts,
applied to **your own project**. One rule governs every one of them: the
artifact adapts the note, and it never outruns it.

1. **The venue contract.** Seven lines, one per field, written before you
   build anything: audience, medium, limits, required elements,
   accessibility, submission rule, question channel. Close it with one
   sentence naming what your claim must lose to fit this format.
2. **The transformation memo and the content map.** The memo says what
   expands, what compresses, what moves, what becomes visual, and what becomes
   spoken. The map says *where* the claim, the evidence, the boundary, the
   uncertainty, and the disclosure will live, as locations rather than
   intentions.
3. **The artifact, in both branches.** The **poster draft** built to the
   contract and carrying a visible **preliminary** label, with values labeled
   directly on every figure and axes that encode magnitude honestly. And the
   **spoken spine**: a 30-second hook (hook, claim, boundary, invitation) and
   a 3-minute walk (the same skeleton plus two or three evidence beats and one
   honest caveat), both built boundary-first.
4. **The mode test and the gallery-walk audit.** Read the board at gallery
   distance and time the spine aloud on someone outside your field. Then audit
   your own draft across five lenses (claim boundary, figure honesty, read
   path, uncertainty, accessibility), rank the hits by how far the error
   travels, revise once in that order, and record what you fixed and what you
   deliberately left.
5. **The required AI review and the defense rehearsal.** Run the **Poster
   Critic**, adjudicate every untraceable-number flag and claim-boundary flag
   with a check you can show, and record at least one flag you refuted with
   the check that refuted it. Then write five hard, fair questions, answer
   each in two sentences, and mark each **defend**, **concede**, or
   **do-not-know**, with your uncertainty-and-limitations statement in **ULN
   form** (uncertainty, limitation, next step).
6. **The version line, the ledger rows, and the dossier line.** The artifact
   opens with **Book Milestone 10, version 1**, dated, with its reason. One
   ledger row per AI-assisted step. One closing line naming where the
   contract, the map, the artifact, and the rehearsal record now live.

> **A question that often comes up here:** *"My poster and my note say the
> same things. Isn't the poster just the note, shrunk?"* No, and the
> difference is the mode. A note is read at arm's length by someone who can
> reread a sentence. A poster is scanned from a few feet away by someone who
> will not, and a talk is heard once with no rewind. Deciding what expands,
> what compresses, what becomes visual, and what becomes spoken is the work.

### 🤝 AI Research Partner

AI is your arm and your research assistant in this sprint, not your brain.
Task it with the mechanical work, keep every judgment for yourself.

**Good to delegate here:**
- Compressing a paragraph you already wrote into a candidate headline block,
  checked afterward against your boundary column.
- Running the Poster Critic against your draft, then generating hostile
  questions a visitor could ask.
- Drafting the transformation memo's inventory, and scanning a figure for a
  distinction carried in color alone.

**Never delegate (these are yours to decide):**
- The headline's verb and reach under compression. The shorter sentence is
  where verbs get upgraded, and a shorter sentence is not a bigger finding.
- Whether an AI flag is real. A flag becomes real when a check in your own
  material confirms it, never because the reviewer sounded certain.
- The **defend**, **concede**, or **do-not-know** call on every question. A
  concede is a boundary you say out loud before a stranger finds it for you.

### 🛠️ Run It Live: check the contract, the adaptation, and the spine

Run the scaffold below on its defaults first to see what it does. Then edit
the values at the top with **your own** contract, your own inventories, and
your own spoken spine, and rerun. Everything downstream is derived from those
edits, so there is one source of truth per number: paste once at the top and
the whole readout stays consistent.

```python
# Contract, adaptation, and spine scaffold — edit ONLY the blocks below.
# Part 1: is your venue contract complete, with the sentence naming what the
# claim must lose? Part 2: does anything on the artifact fail to appear in the
# M9 note? Part 3: does each spoken layer fit its window at a calm pace?

# EDIT ME — one line per contract field. Several lines are fixed for you: a
# printed poster, presented in a hall with live questions, submitted for print
# by the terminal deadline on your course platform.
contract = {
    "audience":         "EDIT ME (who reads or listens, and what they know)",
    "medium":           "printed poster, presented in a hall",
    "limits":           "EDIT ME (the board size, and the minutes at the board)",
    "required elements": "EDIT ME (what the venue insists on)",
    "accessibility":    "EDIT ME (low vision, color-blindness, screen reader)",
    "submission rule":  "print file to the venue channel by the terminal deadline",
    "question channel": "live questions at the board",
}
must_lose = "EDIT ME (one sentence: what your claim must lose to fit this format)"

# EDIT ME — every load-bearing element you plan to put on the artifact, and
# every element your M9 note already contains.
artifact_elements = ["headline claim", "interval", "boundary sentence",
                     "method line", "figure 1", "limitations",
                     "new side comparison"]
note_elements     = ["headline claim", "interval", "boundary sentence",
                     "method line", "figure 1", "limitations",
                     "robustness detail", "references"]

# EDIT ME — your spoken spine, and the pace you actually speak at.
HOOK_30S = ("In the halls my survey reached, residents who joined a floor "
            "event reported belonging about a third of a point higher on a "
            "five-point scale, give or take a tenth. That is an association "
            "in my sample, not evidence that the events caused it. Come see "
            "where it holds and tell me what would break it.")
WALK_3MIN = HOOK_30S + " " + ("EDIT ME: the same skeleton with two or three "
                              "evidence beats and one honest caveat.")
WORDS_PER_MIN = 130   # a calm conference pace; raise it if you speak fast

# ---- derived below; do not retype any number, read it from the edits above --
print("— Part 1: the venue contract —")
missing = [k for k, v in contract.items() if "EDIT ME" in v]
print(f"  fields written: {len(contract) - len(missing)} of {len(contract)}"
      + (f"; still open: {', '.join(missing)}" if missing else ""))
print("  what the claim must lose: " + ("⚠️ not written yet"
      if "EDIT ME" in must_lose else "✓ written, and a reviewer will test it"))

print("\n— Part 2: the adaptation rule —")
orphans = [e for e in artifact_elements if e not in note_elements]
if orphans:
    print(f"  ⚠️ on the artifact but not in the note: {orphans}")
    print("     Build from the note and the claim-to-evidence table, and let")
    print("     empty space stay empty.")
else:
    print("  ✓ every artifact element appears in the note")
dropped = [e for e in note_elements if e not in artifact_elements]
print(f"  compressed out of the artifact: {dropped}")
BOUNDARY_WORDS = ("limit", "boundary", "uncertain", "interval", "caveat")
lost_bounds = [e for e in dropped if any(w in e.lower() for w in BOUNDARY_WORDS)]
if lost_bounds:
    print(f"  ⚠️ {lost_bounds} is the boundary, not detail. An adaptation that")
    print("     drops it is a different claim, not a shorter version.")
else:
    print("  ✓ what left is detail. Read the list yourself to be sure.")

print("\n— Part 3: the spoken spine —")
for label, text, window in (("30-second hook", HOOK_30S, 30),
                            ("3-minute walk", WALK_3MIN, 180)):
    words = len(text.split())
    secs = round(words / WORDS_PER_MIN * 60, 1)
    fit = "fits" if secs <= window else "OVER: cut words, never the boundary"
    print(f"  {label:15s}: {words:3d} words -> {secs:5.1f} s of {window} s  ({fit})")
print("  Now say both aloud on someone outside your field and time them. The")
print("  clock in the room settles it, not this estimate.")
```

### 🗡️ Red-Team Exchange

The Friday studio is an **artifact criticism gallery**. Your peers read your
draft the way your venue's audience will, from the distance that audience will
use. Two rounds, then one revision pass.

**Round 1: the gallery walk.**
Post the draft and step back to gallery distance. A peer reads it cold across
the five lenses: claim boundary, figure honesty, read path, uncertainty, and
accessibility. Does the headline land? Does the eye find the figure without
being told? Does the boundary reach the reader at the same moment as the
number? Does any distinction live in color alone? Write down every hit, then
rank the hits by how far the error would travel. A causal verb your design
never earned travels furthest, because it changes what the whole board
asserts. A crowded margin travels least.

**Round 2: the defense rehearsal, with the Poster Critic.**
Run the required **Poster Critic** (full briefing in
`genai_studio/roles/poster_critic.md`) on your text and figure descriptions,
and adjudicate every flag with a check you can show, including one you refute.
Then deliver the 30-second hook cold while a peer plays a hurried visitor and
throws the five hardest fair questions they can. Call **defend**, **concede**,
or **do-not-know** out loud on each, and close with your ULN statement.

**What to log.** The strongest hit from each round, your single revision pass
in ranked order, and what you deliberately left. Every question you marked
do-not-know becomes your agenda for M13, written as one list.

> **A question that often comes up here:** *"A peer and the Poster Critic
> disagree about what matters most. Whose flag do I fix first?"* You are the
> judge, not either voice. Fix the one your own material confirms is real: the
> number you cannot point to a cell for, or the line that outruns your
> boundary column. Adjudicating the disagreement against your own evidence is
> stronger proof you own the artifact than deferring to either voice.
