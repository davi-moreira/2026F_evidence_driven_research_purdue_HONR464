## 🎯 Definition of Done

You are done with this sprint when your submission carries all four artifacts
plus the ledger and dossier line, applied to **your own project**:

1. **The research note v0.** One to two pages for a stranger who has only the
   page, opening with a dated version line and its reason. Five moves in
   order: the question exactly as your Research Contract declares it (with
   its compass position), the method in one plain-language paragraph, the
   results with uncertainty **in the same eye-span** as every number, the
   **bounded headline claim** (one sentence carrying its compass kind and its
   range, every verb sized to what the design established), and the
   limitations, led by the one your M8 audit could not fix, then retrievable
   references.
2. **The claim-evidence table.** Every claim in the note gets a row with four
   columns: claim, evidence, verification, boundary. This is your **claim
   ledger**, a separate instrument from the AI Research Ledger. The rule it
   enforces: a claim with no row does not appear in the note.
3. **The AI disclosure statement.** Drafted **from your ledger, never from
   memory**: what you delegated, what came back, what you *sent* to tools,
   and what you did not verify.
4. **The poster content brief.** One page deciding what travels from the note
   to the poster: the headline claim word for word, the one figure that
   carries it, and the numbers that will appear, each with its cell trace.
   Nothing appears in the brief that does not appear in the note; the poster
   itself is a labeled-preliminary adaptation, and its craft waits for M10.
5. **AI Research Ledger rows.** One row for every AI-assisted step, each with
   a named verification method.
6. **The dossier update line.** Where note v0, the claim-evidence table, and
   the disclosure now live, named by file or section.

> **A question that often comes up here:** *"The Expo is the deadline that
> counts. Shouldn't I build the poster first and write the note after?"* No,
> and the reason is protective. A poster is a compression, and compressing a
> claim that does not yet exist in defensible written form lets the poster's
> space constraints author your claim for you. Write the note first, and the
> poster can only ever say less than the note, never more.

### 🤝 AI Research Partner

AI is your arm and your research assistant in this sprint, not your brain.
Task it with the mechanical work, keep every judgment for yourself.

**Good to delegate here:**
- Red-teaming your headline sentence and generating hostile questions about
  the note.
- Tightening prose you already wrote, checked against the claim-evidence
  table afterward.
- Assembling the disclosure draft from ledger rows you paste in.

**Never delegate (these are yours to decide):**
- The headline's verb and range. A verb one size too strong turns an honest
  finding into an **overclaim**, and the compression to a headline is
  exactly where verbs get upgraded.
- Which claims get a table row, and which claims get cut. A claim you cannot
  fit in the table is a claim to cut or a check to run.
- The standing rule holds whatever drafted a sentence: the sentence that
  ships is yours.

Whatever you delegate, verify what comes back and log it in your **AI
Research Ledger** below. See `ai_resources/human_responsibility_checklist.md`
for the full never-delegate list.

> 💡 **AI Prompt:** "Here is my bounded headline claim, which I wrote
> myself: '[paste it].' My design is [route], my data cover [units], and my
> M8 audit licensed the range [A to B]. Act as a hostile reader who has only
> this sentence. Attack it three ways: is any verb stronger than the design
> established, does the sentence quietly reach units or populations my data
> never touched, and does it carry its range or hide behind one flattering
> number? Do not rewrite it for me."
>
> **After running, verify (counters the *verb upgrade under compression*: 'is
> associated with' quietly becoming 'drives' as the sentence shrinks):**
> - [ ] Check each attack against your claim-evidence table's boundary
>   column, not against your hopes.
> - [ ] Reword the headline yourself; confirm the compass kind and the range
>   survived the rewrite.
> - [ ] Confirm the same sentence, word for word, is what enters the poster
>   content brief.
> - [ ] Log this use in your AI Research Ledger: task, tool, decision,
>   verification.

> 💡 **AI Prompt:** "Here are the rows of my AI Research Ledger for
> this project: [paste them]. Draft an AI disclosure statement from these
> rows only. It must cover what I delegated, what the tools produced, what I
> *sent* to the tools, and anything the rows show I did not verify. If a row
> is too vague to support a sentence, list it as a gap instead of writing
> around it. Do not add anything the rows do not contain."
>
> **After running, verify (a disclosure drafted from memory is the pitfall
> this component exists to prevent):**
> - [ ] Check every sentence of the draft against a ledger row; cut any
>   sentence with no row behind it.
> - [ ] Confirm the statement includes what you sent to tools, not only the
>   polished outputs.
> - [ ] Fill the gaps the draft exposed: a vague row gets fixed in the
>   ledger, not papered over in the disclosure.
> - [ ] Log this use in your AI Research Ledger: task, tool, decision,
>   verification.

```python
# Claim-and-verb checker — edit the EDIT ME blocks to match YOUR note.
# Part 1 holds your claim-evidence table: every claim in the note needs a row
# with a real verification and a boundary. Part 2 reads your headline against
# your declared compass kind and hunts the classic verb upgrades.

# EDIT ME — one row per claim that will appear in the note.
claim_table = [
    {"claim": "EDIT ME (a sentence from your note)",
     "evidence": "EDIT ME (the output that supports it)",
     "verification": "EDIT ME (the real check: a clean-restart trace, a pre-listed audit check)",
     "boundary": "EDIT ME (how far it reaches, and no further)"},
]

# EDIT ME — your headline sentence and your compass kind.
headline = "EDIT ME (one sentence carrying its compass kind and its range)"
compass_kind = "descriptive"   # or "causal", only if your route earned it

print("— Claim-evidence table check —")
for i, row in enumerate(claim_table, 1):
    problems = [k for k, v in row.items() if "EDIT ME" in v or not v.strip()]
    vague = row["verification"].strip().lower() in {"verified", "checked", "yes"}
    status = "⚠️ incomplete: " + ", ".join(problems) if problems else \
             ("⚠️ vague verification — name the real check" if vague else "✓ complete")
    print(f"  row {i}: {status}")
print("  The rule the table enforces: a claim with no row does not appear in")
print("  the note.")

print("— Headline verb check —")
causal_verbs = ["drives", "causes", "improves", "boosts", "increases",
                "reduces", "leads to", "results in"]
hits = [v for v in causal_verbs if v in headline.lower()]
if compass_kind == "descriptive" and hits:
    print(f"  ⚠️ causal verb(s) {hits} in a descriptive headline — downgrade to")
    print("     'is associated with' unless your route earned the causal verb")
elif "EDIT ME" in headline:
    print("  ⚠️ headline still a placeholder — paste your real sentence")
else:
    print("  ✓ no classic verb upgrade found; still read it against your route")
has_range = any(ch.isdigit() for ch in headline)
print("  Range present in the sentence:", "✓ looks like it" if has_range
      else "⚠️ no number found — a bounded headline carries its range")
```

### 🗡️ Red-Team Exchange

The attack runs at the Friday studio, where your note meets its first
strangers. Two rounds.

**Round 1: the room attacks the note.**
Peers read your note v0 with only the page, the way a stranger will. They
hunt the two classic failures: the **verb upgrade** (a sentence asserting
more than the design established) and the claim with no table row beneath
it. For every hit, the fix is the table's rule: cut the claim, run the
check, or downgrade the verb. Your AI assistant joins as a second reader
with the headline prompt above.

**Round 2: shape the poster content brief.**
Decide, at the studio, what travels to the poster: the headline word for
word, the one figure, and the numbers with their cell traces. Read the brief
against its two rules: nothing appears in it that does not appear in the
note, and every number keeps its trace; a figure with no cell behind it is
an **untraceable number**, and the Expo will ask. Mark the poster as a
labeled-preliminary adaptation; its craft questions wait for M10.

**What to log.** The strongest hit from each round and what you changed. An
attack that changed nothing still gets a line saying why it did not land.

> **A question that often comes up here:** *"My results section is honest;
> why does the eye-span rule matter?"* Because readers read sentences, not
> sections. A number in one line and its interval three lines later is, for
> most readers, a number with no interval. Same sentence or same line, every
> time.
