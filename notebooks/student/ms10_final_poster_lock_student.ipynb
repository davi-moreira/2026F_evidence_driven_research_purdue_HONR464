## 🎯 Definition of Done

Two things land together this week, and the link between them is the whole
lesson. **The numbers that lock onto the poster are the numbers that just
reproduced from the clean package.** The order below is the order of the work.

1. **The reusable research package.** Six parts: the seeded notebook
   (`SEED = 464`, top to bottom with no manual step), the data or its written
   **access recipe** (source, exact route, retrieval date, any registration),
   the environment record, the README whose first three lines say what the
   project asks, what it found, and how to run everything, the licence with
   its reuse boundary, and the dated frozen manifest.
2. **The cold-run record.** A run made in a clean environment from your
   written instructions only, in four parts: the run, the question log (every
   place the runner had to ask you something), the fixes you added in
   response, and the honest label. Working alone, that label is **"solo proxy;
   external cold run pending"**, and it never softens into an external run.
3. **Gate 1, the author self-reproduction log (blocking).** A dated, fresh
   runtime, Restart and Run All, with every poster number copied beside the
   value **its own cell** produced, each pair marked matched or not matched,
   and every mismatch repaired. Every estimate carries its bound into that
   comparison, because **you reproduce the uncertainty, not only the point
   estimate**. A number your design gives no bound to, such as how many cases
   the analysis covers, still needs its cell.
4. **Gate 2, the release preflight (blocking).** Permissions, disclosure,
   reproducibility, and claim boundary, each marked **clear**, **pending**, or
   **blocking**. A pending row is a repair you name and finish. A blocking row
   stops the lock. **The poster does not lock without both gate records.**
5. **The locked poster, built on the reproduced numbers.** Print-ready to the
   URC spec and legible at print scale, with one headline claim inside its
   compass position, every number traced to a cell in the package that just
   reran, the uncertainty visible in the same eye-span as the claim, and
   **redundant encoding** so no distinction is carried by color alone. It
   ships with a signed final-lock checklist and an AI-use disclosure block,
   and every citation resolves to a real work you opened yourself.
6. **AI Research Ledger rows and the dossier line.** One row per AI-assisted
   step, each naming the run or the check that tested the output against
   reality, plus one closing line locating the package, the cold-run record,
   the gate records, and the poster.

> **A question that often comes up here:** *"I ran everything last week and it
> worked. Why run it again on the day I lock the poster?"* Because a package
> is only current as of its last clean run. You fixed a figure, changed a
> filter, and the number under the headline quietly moved. A judge who points
> at a number is really asking when you last saw it come out of the machine,
> and this milestone makes the answer "today".

### 🛠️ Run It Live: reproduce, then decide whether the poster locks

The scaffold below is the lock discipline made runnable. Part 1 checks the
package and its manifest. Part 2 is the **number-by-number trace**: a real
board carries several different numbers, so each row names the cell that
produced it and is compared against **that** cell, not against one shared
value. A row also says whether your design licenses a **bound** for it, meaning
the interval or range you may attach to an estimate, such as a 95% interval. An
estimate is checked with its bound. A count, a year, or an axis endpoint has no
bound to check, and that is not a defect. Parts 3 and 4 are the two blocking
gates, and the final line is the verdict: lock, or do not lock.

**Expect this from the defaults.** The demonstration board carries four
numbers: two different estimates, one sample count with no bound, and one
figure carried over from an earlier draft. Three rows should trace back to
their own cells. The stale row should fail, the package should report its
missing licence, and the verdict should read DO NOT LOCK YET. Run it as it
stands, then edit the EDIT ME blocks with your own package and rerun.

```python
# Package, reproduction, and the two blocking gates — edit the EDIT ME blocks.
# One source of truth per number: ONE cell computes it, the board prints it, and
# the trace below compares that printed number against ITS OWN cell.

# EDIT ME — Part 1: your package contents, marked present or absent.
package = {
    "seeded notebook":      True,
    "data or access recipe": True,
    "environment record":   True,
    "README (asks / found / how to run)": True,
    "licence with reuse boundary": False,   # EDIT ME
    "dated frozen manifest": True,
}
manifest_date = "frozen at lock day"   # EDIT ME — the date you froze it
cold_run_label = "solo proxy; external cold run pending"   # EDIT ME if external

# --- Part 2, the CELL side. Swap these stand-ins for YOUR computation. A real
# board carries several DIFFERENT numbers, so this side is a registry: one entry
# per cell that produces a number, holding what that cell produced and the bound
# it produced when the design licenses one.
n_analysis = 800                                  # EDIT ME — your data
hi = rng.normal(3.4, 1.0, n_analysis // 2)
lo = rng.normal(2.7, 1.0, n_analysis // 2)
gap = hi.mean() - lo.mean()
gap_se = np.sqrt(hi.var(ddof=1) / (n_analysis // 2)
                 + lo.var(ddof=1) / (n_analysis // 2))
share = (rng.random(n_analysis) < 0.42).mean()    # a SECOND, different estimate
share_se = np.sqrt(share * (1 - share) / n_analysis)

def interval(point, se):
    # The 95% interval this design licenses around one estimate.
    return (point - 1.96 * se, point + 1.96 * se)

def rounded(bound):
    # The bound as the board would print it, or None when there is none.
    return None if bound is None else (round(bound[0], 2), round(bound[1], 2))

cell_outputs = {   # cell id -> what that ONE cell just produced
    "A1 · analysis sample": {"value": float(n_analysis), "bound": None},
    "A2 · trust gap":       {"value": gap,   "bound": interval(gap, gap_se)},
    "A3 · share agreeing":  {"value": share, "bound": interval(share, share_se)},
}

PRINT_TOL = 0.005   # the board prints two decimals, so match to two decimals

# EDIT ME — every number you plan to PRINT: one row per number, each naming the
# cell it came from and whether the design licenses a bound for it. The last row
# is left stale on purpose, to show the warning path.
poster_numbers = [
    {"label": "analysis sample", "printed": 800, "printed_bound": None,
     "source": "A1 · analysis sample", "requires_bound": False},
    {"label": "trust gap", "printed": round(gap, 2),
     "printed_bound": rounded(interval(gap, gap_se)),
     "source": "A2 · trust gap", "requires_bound": True},
    {"label": "share agreeing", "printed": round(share, 2),
     "printed_bound": rounded(interval(share, share_se)),
     "source": "A3 · share agreeing", "requires_bound": True},
    {"label": "old draft gap", "printed": round(gap + 0.40, 2),
     "printed_bound": rounded(interval(gap + 0.40, gap_se)),
     "source": "A2 · trust gap", "requires_bound": True},
]

def trace(row):
    # Compare ONE printed number against the cell that produced it.
    src = cell_outputs.get(row["source"])
    if src is None:
        return False, ("⚠️ names no cell in the package: a number with nothing "
                       "behind it cannot go on the board")
    if abs(row["printed"] - src["value"]) > PRINT_TOL:
        return False, (f"⚠️ does NOT reproduce: that cell now gives "
                       f"{src['value']:,.2f}. Repair it before the lock")
    if row["printed_bound"] is None:
        if row["requires_bound"]:
            return False, ("⚠️ bound missing: an estimate without its bound is "
                           "half the finding")
        return True, "✓ matches its own cell, and this number licenses no bound"
    if src["bound"] is None:
        return False, "⚠️ prints a bound its own cell never produced"
    moved = max(abs(row["printed_bound"][0] - src["bound"][0]),
                abs(row["printed_bound"][1] - src["bound"][1]))
    if moved > PRINT_TOL:
        return False, (f"⚠️ value matches but the bound moved: that cell now "
                       f"gives {src['bound'][0]:+.2f} to {src['bound'][1]:+.2f}")
    return True, "✓ value and bound both came back from its own cell"

print("— Part 1: the package —")
absent = [k for k, v in package.items() if not v]
print(f"  parts present: {len(package) - len(absent)} of {len(package)}"
      + (f"; missing: {', '.join(absent)}" if absent else ""))
print(f"  manifest: {manifest_date}   ·   cold-run label: {cold_run_label}")

print("\n— Part 2: the trace, number by number, each against its own cell —")
for cell_id, out in cell_outputs.items():
    shown = ("no bound" if out["bound"] is None else
             f"{out['bound'][0]:+.2f} to {out['bound'][1]:+.2f}")
    print(f"  {cell_id:22s} produced {out['value']:>8,.2f}   ({shown})")
reproduced = []
for row in poster_numbers:
    ok, note = trace(row)
    reproduced.append(ok)
    shown = ("no bound printed" if row["printed_bound"] is None else
             f"{row['printed_bound'][0]:+.2f} to {row['printed_bound'][1]:+.2f}")
    print(f"\n  {row['label']}: board prints {row['printed']:,.2f} ({shown}), "
          f"traced to cell {row['source']}")
    print(f"      {note}")

print("\n— Part 3: Gate 1, author self-reproduction log (blocking) —")
gate1 = all(reproduced) and not absent
blocked_by = [r["label"] for r, ok in zip(poster_numbers, reproduced) if not ok]
print(f"  rows that traced: {sum(reproduced)} of {len(reproduced)}"
      + (f"; still to repair: {', '.join(blocked_by)}" if blocked_by else ""))
print("  fresh runtime, Restart and Run All, dated, every number matched "
      "against its own cell: "
      + ("✓ clear" if gate1 else "⚠️ not yet"))
print("  This log says the numbers are current. It is not independent "
      "verification.")

# EDIT ME — Part 4: the release preflight, taken immediately before the lock.
preflight = {
    "permissions":     "clear",    # clear / pending / blocking
    "disclosure":      "clear",
    "reproducibility": "clear" if gate1 else "blocking",
    "claim boundary":  "pending",  # EDIT ME
}
print("\n— Part 4: Gate 2, release preflight (blocking) —")
for row, state in preflight.items():
    mark = {"clear": "✓", "pending": "○", "blocking": "⚠️"}[state]
    print(f"  {mark} {row:16s}: {state}")
gate2 = all(s == "clear" for s in preflight.values())

print("\n— The verdict —")
print("  LOCK THE POSTER" if gate1 and gate2 else
      "  DO NOT LOCK YET: name each repair, finish it, and rerun this cell.")
print("  A blocking gate is not scored, so no strength elsewhere averages it "
      "away.")
```

**Reading the output.** Three rows trace and one does not, which is the
whole point of checking row by row. The sample count clears without a bound,
because a count of cases is not an estimate and your design gives it no
interval. The two estimates clear with their bounds, and they clear against
*different* cells: the trust gap against the cell that computed the gap, the
share against the cell that computed the share. The old draft figure fails,
because the value moved when the analysis reran and the new one never reached
the board. That one failing row holds Gate 1 shut, and the printout names it so
you know exactly what to repair.

> **A question that often comes up here:** *"My board prints the year of the
> survey and the number of cases. Do those need intervals too?"* No. Set
> `requires_bound` to `False` for them and they pass on the value alone. The
> flag exists so the trace demands a bound exactly where your design licenses
> one, and never turns a legitimate count into a false failure.

### 🗡️ Red-Team Exchange

The Friday production block. The order matters: the package and its cold run
come first, because the poster's numbers are their output.

**Round 1: the cold run.**
Trade packages with a partner. They open a clean environment, follow your
README and nothing else, and run it top to bottom. You write down every
question they have to ask you, including the ones whose answers are obvious to
you, because obvious-to-you is exactly what documentation exists to replace.
Every question becomes a documentation line. Then record which numbers came
back identical and which did not. A cold run that changes nothing usually
means the log was not kept.

**Round 2: the gates and the board.**
Restart your own runtime, run everything, and fill Gate 1 by copying each
poster number beside the value **its own cell** produced, with the bound
alongside wherever your design licenses one. Then take
Gate 2 immediately before the lock, marking permissions, disclosure,
reproducibility, and claim boundary clear, pending, or blocking. While that
runs, your partner attacks the board: point at a number and ask for its cell,
click a citation and watch it load, read the headline against its compass
position, and check whether any distinction lives in color alone.

**What to log.** The question log, the fixes, both gate records, and the
single strongest hit on the board with what you changed. A pending row you
finished still gets a line saying what the repair was.

> **A question that often comes up here:** *"My partner ran my package and it
> worked. Can I call that independent reproduction?"* Yes, and name them and
> the date. That is exactly what an external cold run is, and it is stronger
> than your own rerun. Reserve the "solo proxy" label for the case where you
> ran it yourself after a break, and never soften that label into the other
> one.
