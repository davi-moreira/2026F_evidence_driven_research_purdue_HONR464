## 🎯 Definition of Done

You are done with this sprint when your **Replication and Red-Team Report** on the
package you were assigned carries all eight parts, each pointing at something the
package shipped or a number your own run produced.

1. **The intake and privacy check.** Three lines before you run anything: the
   package identifier, the date you received it, and your confirmation that it is
   identity-stripped and cleared. Two rules bind from there: do not try to identify
   or contact the author, and paste only cleared package contents into any AI tool.
2. **Environment recreation.** Rebuild the software conditions the code needs from
   the record inside the package (language version, libraries and their versions,
   the random seed), in a clean session. A fresh Colab runtime is the simplest clean
   room you have. Log what the package pinned, what you had to guess, and whether
   any guess changed a result. A missing environment record is a finding on its own.
3. **The cold run.** **Restart-and-run-all** means clearing the session's memory and
   running every cell from the top with no manual intervention, following only the
   written instructions inside the package. Log three things: whether it ran (with
   every break traced to the **exact cell**), what you had to do anyway (written as
   the instruction the package should have carried), and a signed one-line
   attestation that no author help reached you.
4. **The output comparison, including the uncertainty.** One table: the point
   estimate as **your** run produced it beside the reported value, the interval or
   standard error beside its reported version, and the **claim-to-output trace**
   putting each written sentence next to the output meant to back it. Reproducing a
   point estimate while ignoring its uncertainty is a half-reproduction. A package
   that reports no uncertainty at all has a reporting gap, and that is a finding.
5. **The undocumented-instruction log.** Every place you had to guess, infer, or
   would have had to ask: the question you would have asked, the guess you made
   instead, and whether the guess changed the result.
6. **The red-team of the claims.** **One alternative specification** (an equally
   defensible way to compute the same headline, reported both ways with the size of
   the swing) and **one hidden assumption** the analysis leans on quietly, each
   labeled as it **moves** the estimate, **widens** its uncertainty, or **breaks**
   the claim.
7. **The required Reproducibility Auditor pass, verified by your own run.** Which of
   its flags survived *your* run, which you dropped, which you added, and the
   correlated-error trap named.
8. **The repair handoff.** Findings ranked by **threat to the claim, not ease of
   fixing**, each with the one check that would confirm or dissolve it, with your
   undocumented-instruction log attached. Then the single most-threatening weakness,
   with its settling check, posted to the async board, plus one substantive reply to
   a classmate.

Plus your **AI Research Ledger** rows, the **version line** (replication and
red-team report, version 1, dated, with its reason, named as Book Milestone 11
practice on another researcher's package), and the **dossier update line**.

> **A question that often comes up here:** *"The package reproduced with a green
> check. Isn't the sprint basically over?"* No, that is where it starts. A clean run
> proves the code executes; it says nothing about whether the write-up's sentence
> matches the output. Reproduction earns you a trustworthy number; parts 4 through 8
> are the red-team, and that is where the grade lives.

### 🛠️ Run It Live: organize your reproduced numbers

Run the scaffold below on its default numbers first to see what it does. Then edit
the **one dict** and the **one list** at the top with the figures from *your* assigned
package, after your own cold run, and rerun. Everything downstream derives from those
two edits, so there is a single source of truth for every number: change a value once
at the top and the whole readout stays consistent. The scaffold covers three parts of
the report (the output comparison, the alternative specification, and the hidden
assumption) and leaves the ranking to you.

```python
# --- Reproduction scaffold — edit ONLY the dict and the list below.
# Everything after them is DERIVED, so there is one source of truth per number.
# Plug in the figures from YOUR assigned package AFTER your own restart-and-run-all.

MY_PACKAGE = {
    "claim_label":      "canvassing raised turnout",         # EDIT ME: the headline claim, in words
    "reported_value":   4.00,   # EDIT ME: the number the WRITE-UP states
    "reported_interval": None,  # EDIT ME: the interval the WRITE-UP reports, or None if it reports none
    "reproduced_value": 3.41,   # EDIT ME: the number YOUR run produced
    "alt_spec_label":   "weighted instead of unweighted",    # EDIT ME: your alternative specification
    "alt_spec_value":   2.83,   # EDIT ME: the headline under that alternative choice
    "units":            "percentage points",
}

# Per-cluster estimates from YOUR run: one number per independent group (a street,
# a site, a class — whatever your package's clustering unit is). EDIT ME with your
# package's per-cluster differences. The defaults are the sample package's fifteen
# street clusters, and they average to the 3.41 above, so the cell runs before you do.
cluster_estimates = [14.2, -8.4, 8.1, 0.9, -5.3, 16.4, 2.6, -2.4, 6.8, 3.2,
                     -9.3, 9.5, 1.1, 4.3, 9.4]

# ---- derived below; do not retype any number, read it from the two edits above ----
u = MY_PACKAGE["units"]

# 1) Claim-to-output: does the write-up's number match your run?
claim_gap = MY_PACKAGE["reported_value"] - MY_PACKAGE["reproduced_value"]
matched = abs(claim_gap) < 0.005
print(f"CLAIM-TO-OUTPUT TRACE for '{MY_PACKAGE['claim_label']}'")
print(f"  write-up states  : {MY_PACKAGE['reported_value']:.2f} {u}")
print(f"  your run produced: {MY_PACKAGE['reproduced_value']:.2f} {u}")
print(f"  gap = {claim_gap:+.2f} {u}  -> " +
      ("MATCH" if matched else
       "MISMATCH: the write-up reports a number your run does not produce"))

# 1b) The uncertainty half of the comparison. A point estimate alone is a
#     half-reproduction, and a reported number with NO interval is itself a finding.
if MY_PACKAGE["reported_interval"] is None:
    print("  uncertainty      : the write-up reports NONE -> log this as a reporting gap")
else:
    r_lo, r_hi = MY_PACKAGE["reported_interval"]
    print(f"  uncertainty      : write-up reports [{r_lo:+.2f}, {r_hi:+.2f}] {u} "
          f"-> compare it to the interval below, not just the point estimate")

# 2) Alternative specification: how far does one undisclosed choice move the number?
spec_swing = MY_PACKAGE["reproduced_value"] - MY_PACKAGE["alt_spec_value"]
print(f"\nALTERNATIVE SPECIFICATION ({MY_PACKAGE['alt_spec_label']})")
print(f"  as reported: {MY_PACKAGE['reproduced_value']:.2f} {u}  |  "
      f"alternative: {MY_PACKAGE['alt_spec_value']:.2f} {u}  |  "
      f"swing = {spec_swing:+.2f} {u}")

# 3) Hidden assumption: honor the clustering. Resample WHOLE clusters (seed 464),
#    not individual rows, and read whether the honest interval reaches zero.
arr = np.array(cluster_estimates, dtype=float)
K = len(arr)
boot = np.array([rng.choice(arr, size=K, replace=True).mean() for _ in range(2000)])
lo, hi = np.percentile(boot, [2.5, 97.5])
reaches_zero = lo <= 0 <= hi
print(f"\nHIDDEN ASSUMPTION (clustering): cluster-aware 95% interval over {K} clusters")
print(f"  [{lo:+.2f}, {hi:+.2f}] {u}  -> " +
      ("reaches zero: the clear-effect reading weakens once clusters count"
       if reaches_zero else "stays above zero"))

# 4) Threat-ranking template — YOU write the order; the numbers above are the evidence.
print("\nRANK BY THREAT (most dangerous first) — replace with YOUR ordering:")
print("  1. <weakness that could BREAK the claim>   (evidence: a number above)")
print("  2. <weakness that MOVES the estimate>       (evidence: a number above)")
print("  3. <weakness that only WIDENS uncertainty>  (evidence: a number above)")
print("\n✓ Scaffold ran on the default numbers. Now edit MY_PACKAGE and")
print("  cluster_estimates with YOUR package's figures and rerun.")
```

### 🗡️ Red-Team Exchange

M16 has no studio. The red-team block moves to the **async course discussion board**
over the break, and it keeps the same discipline: one required AI pass, one post, one
reply.

**First, the required Reproducibility Auditor pass.** Open the course group's
**HONR464 Reproducibility Auditor** and paste the cleared contents of your assigned
package, never any identifying header. It returns a **reproduction gap list** and a
**claim-to-output trace** (full role briefing:
`genai_studio/roles/reproducibility_auditor.md`; quick card in
`ai_resources/agent_role_cards.md`). The role proposes; as configured here it does
not execute anything (an AI agent connected to an execution environment can run code,
but that is not this role, and a run without a command log, environment and package
versions, the code and data version it ran, each command's exit status, the raw
output, and the files it touched is not evidence). Treat every gap it names as a
**suspect you test yourself**, and report which flags survived your own run, which
you dropped, and which you added.

**What to attack, on the board.**
- Post your single **most-threatening weakness**: the one finding most likely to
  move, widen, or break the peer's headline claim, the number behind it, and the
  **one check** that would confirm or dissolve it.
- Reply substantively to **one** classmate. Attack the ranking, not the person: is
  their top weakness really the biggest threat, or is a larger one hiding lower?
  Would their proposed check actually settle it, or just tidy the reporting?

**Mind the correlated-error trap.** If your AI and the Auditor both call a package
sound, that is not two votes; it is one flaw echoed twice, since neither ran the code.
Only your own run breaks the tie.

**What to log.** Write down the strongest hit you gave and the strongest you got back,
and the specific run that decided any disagreement. A challenge that changed nothing
still earns a line saying why it did not land. That record is part of what you submit.

> **A question that often comes up here:** *"What if the Auditor's gap list and my own
> run disagree?"* Your run wins, every time. The audit is a reading; your
> restart-and-run-all is the evidence. Say in one sentence which flag you kept, which
> you dropped, and the run that decided it. Adjudicating that gap is stronger proof
> you understand the package than agreeing with the tool would be.
