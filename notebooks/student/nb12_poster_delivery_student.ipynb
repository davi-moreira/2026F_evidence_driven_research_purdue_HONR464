## 1. The One-Rule Pitch: Compress, Never Expand

**Guiding question:** *a stranger gives you thirty seconds, then maybe three
minutes. How do you shrink your project to fit without changing what it
claims?*

> *"I stop at a poster and the presenter has about a minute before I decide
> whether to stay. If those seconds are a confident, honest tour of one claim, I
> lean in. If they are an apology or a wall of data, I nod and drift. The talking
> is the poster now. The paper on the board is just the visual aid."*
> — a **skeptical peer** you will meet many times at the Expo

Your poster is locked and labeled preliminary. Now you learn to stand beside it
and speak it. What carries a poster session is a **pitch**: a short, prepared
spoken version of your project, built to fit a fixed slice of time. The 30
seconds you give a visitor who pauses is a pitch. So is the guided tour you
give the visitor who stays.

Everything you say sits on a chain this course built deliberately: the note
fixed your claim, the poster adapted the note, and the pitch speaks the poster.
A spoken sentence with no poster element behind it, and no note sentence behind
that, is new material smuggled in at the worst possible moment: out loud, to a
stranger, with no chance to edit.

The discipline has one rule, the **one-rule pitch**: you may compress the
claim, and you may never expand it. Two terms carry the rule. A **compression**
is a shorter version of a claim that keeps its boundary, and an **inflation**
is a version that quietly makes the claim bigger. Example: "in my sample,
longer commutes went with lower attendance" compresses to "in my survey, longer
commutes tracked with skipping more class," but inflates the moment it becomes
"longer commutes make people skip class."

The failure this whole week hunts is the **excited-mouth upgrade**: the moment
your spoken verb outruns the written one and the boundary moves while nobody is
watching the words. You wrote "is associated with"; the adrenaline says
"causes." You wrote "in my sample of 812"; the mouth says "in everyone." Your
boundary sentence exists to hold that line out loud.

> **A question that often comes up here:** *"How is compressing different from
> inflating? Both change the claim."* Compression drops detail while keeping the
> boundary. "In my sample of 812 commuters, longer commutes go with lower
> attendance" compresses to "in my survey, longer commutes track with skipping
> more class." Inflation moves the boundary: "longer commutes make people skip
> class." Same length, different sin. One lost a number. The other lost the truth.

## 2. Two Layers on One Spine, and the Outsider Test

**Guiding question:** *how do the 30-second and 3-minute pitches differ in job,
and how do you find out whether a stranger received the claim you sent?*

A poster session asks for two spoken versions of the same project, sized to how
much time a visitor gives you. Both are graded components of milestone **M11**,
your presentation package. They share one skeleton, the **pitch architecture**:
the fixed set of beats every good pitch moves through. Its five beats are
**hook, claim, evidence, boundary, invitation**.

- **Hook**: one sentence a stranger outside your field cares about. Example: "every
  morning thousands of commuters trade sleep for a drive."
- **Claim**: your headline, said plainly and inside its boundary. Example: "in my
  survey, longer commutes went with lower attendance."
- **Evidence**: the finding that earns the claim, in one spoken sentence per block.
- **Boundary sentence**: one sentence naming what the claim does NOT show. Example:
  "this is an association in one sample, so it cannot say the drive caused it."
  Mark this sentence and rehearse it until it is reflex. It is the one pressure
  will try to drop.
- **Invitation**: a reason to look closer. Example: "the surprising part is on the
  left, want the walk?" The invitation can also carry your preliminary label
  honestly: "this is a preliminary edition, and I want your hardest question."

The two layers are the same architecture at two lengths:

- **The 30-second pitch**: hook, claim, boundary, invitation. Its whole job is
  to earn the next few minutes. No evidence blocks yet.
- **The 3-minute pitch**: the same spine with two or three evidence blocks
  added, moving the visitor's eye along the poster, and ending on the boundary
  and the invitation.

The longer layer **adds** blocks to the same spine. It never restates a bigger
claim. If your 3-minute pitch says something larger than your 30-second one,
the extra time inflated the claim instead of supporting it.

Two more moves keep the spoken version honest, and both are graded in M11.

- **Audience adaptation**: changing which words you use for who is listening,
  while keeping the claim identical. Example: you lead a fellow biology major
  with the mechanism, and a visitor from the art school with the stakes, but the
  headline and its boundary do not change between them.
- **Jargon**: a field-specific word a stranger cannot decode without training.
  Example: "heteroskedasticity" is jargon at a poster. Explaining your method
  *without* jargon means naming what you did in plain words, so the visitor
  follows the reasoning rather than nodding at a term they cannot climb.

And one check closes the loop. The **outsider test**: deliver your 30-second
pitch to someone outside your field, then ask them to say your claim back to
you. The test passes when the claim comes back with its boundary intact. If it
comes back bigger ("so commuting wrecks attendance?"), your pitch inflated
somewhere, and the flagged words are your revision list. Example: a listener
who repeats "longer commutes went with lower attendance in one campus survey"
just certified your compression.

> **A question that often comes up here:** *"I'm not a performer. Won't a scripted
> pitch sound fake?"* A script is not a performance. It is preparation. You are
> not memorizing lines to recite. You are deciding, calmly and in advance, exactly
> what you will say when a stranger is staring and your pulse is up. The boundary
> sentence especially: you rehearse it until it is automatic, because nerves erase
> precisely the careful wording you most need.

---

```python
# Draw the pitch architecture: the five beats, and how the two layers nest.
# (networkx is imported once in the setup cell.)
G = nx.DiGraph()
pos = {
    "Hook":        (0.0, 0.0),
    "Claim":       (1.3, 0.0),
    "Evidence 1":  (2.6, 0.6),
    "Evidence 2":  (2.6, -0.6),
    "Boundary\nsentence": (3.9, 0.0),
    "Invitation":  (5.2, 0.0),
}
edges = [
    ("Hook", "Claim"),
    ("Claim", "Evidence 1"),
    ("Claim", "Evidence 2"),
    ("Evidence 1", "Boundary\nsentence"),
    ("Evidence 2", "Boundary\nsentence"),
    ("Claim", "Boundary\nsentence"),        # the 30-second skip: straight to the bound
    ("Boundary\nsentence", "Invitation"),
]
G.add_edges_from(edges)

# The boundary node is drawn with a heavy dark outline (not a color code) so it
# reads as the load-bearing beat even in grayscale.
node_colors = ["#DCE6F1"] * len(pos)
outline = ["#4C72B0"] * len(pos)
widths = [1.5] * len(pos)
nodes = list(pos.keys())
bi = nodes.index("Boundary\nsentence")
outline[bi] = "#222222"
widths[bi] = 3.5

fig, ax = plt.subplots(figsize=(11, 4.5))
nx.draw_networkx_nodes(G, pos, node_color=node_colors, edgecolors=outline,
                       linewidths=widths, node_size=4600, ax=ax)
nx.draw_networkx_labels(G, pos, font_size=9, ax=ax)
nx.draw_networkx_edges(G, pos, arrowstyle="-|>", arrowsize=16, edge_color="#666666",
                       width=1.4, node_size=4600, ax=ax)
ax.set_title("The pitch architecture: one spine, two lengths")
ax.axis("off")
# Annotate which layer uses which beats.
ax.text(0.0, -1.4, "30-second pitch: Hook → Claim → Boundary → Invitation (skips evidence)",
        fontsize=9, color="#333333")
ax.text(0.0, -1.7, "3-minute pitch: the same spine, with evidence blocks added",
        fontsize=9, color="#333333")
plt.tight_layout()
plt.show()

print("✓ Pitch architecture drawn.")
print("  The heavy-outlined 'Boundary sentence' beat is the one you rehearse to reflex.")
print("  The 3-minute layer ADDS evidence to this spine; it never enlarges the claim.")
```

**Reading the diagram.** Every pitch walks the same path, left to right: hook,
claim, then either straight to the boundary (the 30-second version) or through the
evidence blocks first (the 3-minute version), and out on the invitation. The
heavy-outlined boundary beat is the one that keeps the claim honest, and it is the
one you cannot skip at either length.

*In plain terms, the picture says both pitches are the same skeleton. The
short one is the skeleton; the long one hangs evidence on it without growing the
head.*

```python
# Pacing calculator — how long will a script actually take to say?
RATE_WPM = 130   # a clear, unrushed pace; under pressure people speed up, so budget slow

def words_that_fit(seconds, wpm=RATE_WPM):
    # Seconds times words-per-second gives the word budget for that slice of time.
    return int(seconds * wpm / 60)

def duration_seconds(script_text, wpm=RATE_WPM):
    # Words in the script, divided by words-per-second, gives seconds aloud.
    return len(script_text.split()) / (wpm / 60.0)

budget_30 = words_that_fit(30)
budget_3min = words_that_fit(180)
print(f"At {RATE_WPM} words per minute, an honest budget is:")
print(f"  30-second pitch : about {budget_30} words")
print(f"  3-minute pitch  : about {budget_3min} words")

# A sample 30-second pitch: hook -> claim -> boundary sentence -> invitation.
my_pitch = (
    "Every morning thousands of Purdue commuters trade sleep for a long drive. "
    "In my survey of 812 of them, longer commutes went with lower reported class "
    "attendance. I want to be precise. This is an association in one campus sample, "
    "not proof the drive itself causes the absences. Come see the pattern, and "
    "tell me what you think could explain it."
)
words = len(my_pitch.split())
secs = duration_seconds(my_pitch)
print(f"\nSample pitch: {words} words -> {secs:.0f} seconds (budget 30s, so it fits with room).")
```

```python
# Self-check: the pacing math is an identity, and the sample pitch fits its 30 seconds.
assert words_that_fit(60) == 130, "one minute must be 130 words at 130 wpm"
assert budget_30 == 65 and budget_3min == 390, "word budgets drifted — check RATE_WPM"
assert secs <= 30, "sample pitch overruns 30 seconds — cut words, never speed up"
print(f"✓ Self-check passed: 30s holds {budget_30} words, 3 minutes holds {budget_3min}; "
      f"the {words}-word sample pitch lands at {secs:.0f}s, inside the 30-second budget.")
```

## 3. The Five Hard Questions and the Defense Judgment

**Guiding question:** *what will strangers actually ask at your board, and how
do you decide, on the spot, whether to defend your claim or concede a point?*

> *"I ask three kinds of question at a poster, and I learn the most from the one
> the author cannot answer. A confident bluff tells me to distrust everything. A
> clean 'here is what I know, here is where it stops' tells me to trust the rest."*
> — a **methods reviewer** working a poster session

At the Expo a stranger will ask you the hardest fair question about your
project, and how you answer decides whether they leave trusting you. The
instrument that turns that from an ambush into a strength is the **question
bank**: a short written list of the hardest fair questions a visitor could ask,
each with a prepared answer. Think of it as a study guide you write for an exam
only you will take. Yours needs five, one of each recurring type, and it is a
graded component of M11.

- **Method**: probes whether your design can bear the claim. Example:
  "why is a description enough here, didn't you need a comparison group?"
- **Alternative explanation**: names a rival cause of your pattern. Example:
  "couldn't selection into your sample produce this on its own?"
- **So-what**: asks why the finding matters. Example: "suppose you're right, what
  changes because of it?"
- **Generalization**: asks how far the claim travels. Example: "does this hold
  beyond the people in your sample?"
- **AI-use**: probes the integrity of your process. Example: "you used an AI tool
  to find sources, how did you confirm they exist?" Your answer here points at your
  ledger and your verification step, Ask then Verify then Document, performed.

It helps to know who tends to ask what, because the asker tells you the
defense. Three kinds of questioner cover almost every visitor. The **outsider**
does not share your vocabulary and usually asks the so-what and generalization
questions: their test is whether your claim survives translation. The
**specialist** knows your tools and asks the method question: their test is
whether your **specification**, the specific design and analysis choices you
made, was defensible. The **skeptic** grants your data and asks the
alternative-explanation question: their test is your claim boundary, and it is
the one you usually cannot study your way out of at the board.

Under every answer sits one judgment, **defend versus concede**. You **defend**
when the objection asks for something your evidence already licenses: you point
at the claim-evidence row and hold your ground with the number. You **concede**
when the objection asks for a crossing your design never paid for: you narrow
the claim on the spot, out loud. The test is simple. Does my evidence back this?
If yes, defend. If no, concede and state the smaller claim that survives. A
graceful concession raises your credibility; a dug-in defense of an
indefensible point sinks it, and an upgrade under pressure sinks it fastest.

And when you genuinely do not know, the **honest-don't-know formula** keeps the
concession expert: "I don't know, but here is what I DO know that bears on it,
and here is how one would find out." Three parts: the honest admission, the
adjacent thing you can stand on, and the method that would resolve it. A
prepared don't-know reads as confidence. **Bluffing**, inventing an answer you
do not have, is the one move that can destroy a presenter's credibility on the
spot.

> **A question that often comes up here:** *"Isn't saying 'I don't know' at my own
> poster a disaster?"* Only as a bare shrug. In the full form it is the opposite.
> "I don't know whether it holds for graduate peers, my frame was undergraduates,
> what I *do* know is the pattern was stable across three undergraduate years, and
> to find out you'd re-run it on a graduate sample" reads as someone who knows
> exactly where their knowledge stops. That precision is the most credible thing at
> the poster.

## 4. Uncertainty and Limitations, Spoken Like a Scientist

**Guiding question:** *can you say what your study cannot show without either
bluffing or apologizing your finding away?*

> *"When a presenter tells me, unprompted, exactly what their study can't show, I
> trust everything else they said more. When they insist it's airtight, I start
> looking for the crack myself. Knowing where your claim stops is the most expert
> thing you can say."*
> — a **policy stakeholder** who has heard a thousand pitches

Here is the counterintuitive heart of scientific communication: stating your
limits makes you more believable, not less. Two terms carry the move.

- **Uncertainty statement**: a sentence saying how much your number could wobble,
  and why. Example: "with 812 people, my estimate could easily be a few points
  higher or lower."
- **Limitation**: a true sentence about what your design cannot establish. Example:
  "a one-time survey cannot show that commuting causes the absences."

The move that delivers both out loud is the **ULN move**: uncertainty, then
limitation, then next step, three sentences in order with a straight spine.

- **U, uncertainty:** how much wiggle sits around your estimate, and why.
- **L, limitation:** what your design cannot establish, named plainly. This is the
  boundary itself.
- **N, next step:** the design that *would* settle it. This is the study you would
  run next.

The **next step** is what turns a limitation from an apology into a research
agenda. It says: I know exactly what I would do next, which is the mark of someone
in command of their evidence.

The ULN move has two failure modes, and both misstate the evidence.

- **Hedging**: vague self-protection that blurs where the claim stops. Example:
  "sort of," "probably wrong," "I guess." Hedging protects you and tells the
  listener nothing.
- **Precision**: naming the exact boundary. Example: "this is an association in one
  campus sample." Precision informs the listener.

The worst hedging failure has its own name, the **apology spiral**: burying a real,
defensible finding under self-erasure. Example: "honestly this is probably all
wrong, my sample's tiny, I'm not a statistician." The antidote is physical and
verbal. Plant your feet, drop "sorry" and "just" and "I know this is bad," and end
on the next step, not the flaw. You are reporting a boundary, not confessing a crime.

Partners score each ULN delivery on the **strength-not-apology scale**, a 1-to-5
anchored rating where 1 is the apology spiral and 5 is expertise. Aim for 4 or 5.
The figure below draws the scale so you can hear where a delivery lands.

> **A question that often comes up here:** *"If I lead with what my study can't do,
> won't people just dismiss it?"* They dismiss the drowner, not the namer. "This is
> an association in one campus sample, it can't establish cause, and a panel study
> could" is three seconds of pure competence. It tells the visitor you are the
> person in the room who most precisely knows the finding's shape.

```python
# Draw the strength-not-apology scale: what a ULN delivery sounds like at each score.
labels = ["1\nApology\nspiral", "2\nHedge-\nheavy", "3\nFlat", "4\nInformative", "5\nExpertise"]
sounds = ["'probably all\nwrong, sorry'", "finding buried\nunder qualifiers",
          "limitation stated,\nno next step", "limitation +\nnext step",
          "precise bound +\nlimitation + the\ndesign that resolves it"]
scores = [1, 2, 3, 4, 5]

fig, ax = plt.subplots(figsize=(11, 3.4))
ax.plot(scores, [0]*5, color="#888888", linewidth=2, zorder=1)
# Marker SIZE grows with the score (not a color code), so the scale reads in grayscale.
ax.scatter(scores, [0]*5, s=[140, 200, 260, 340, 440], color="#4C72B0",
           edgecolors="#222222", linewidths=1.5, zorder=3)
for x, lab, snd in zip(scores, labels, sounds):
    ax.text(x, 0.16, lab, ha="center", va="bottom", fontsize=9)
    ax.text(x, -0.16, snd, ha="center", va="top", fontsize=8, color="#333333")
ax.axvspan(3.5, 5.5, color="#DCE6F1", alpha=0.6, zorder=0)  # the target band, labeled below
ax.text(4.5, 0.42, "aim here", ha="center", fontsize=9, color="#1a1a1a")
ax.set_xlim(0.5, 5.5)
ax.set_ylim(-0.5, 0.5)
ax.axis("off")
ax.set_title("The strength-not-apology scale: a limitation can sound like weakness or expertise")
plt.tight_layout()
plt.show()

print("✓ Strength-not-apology scale drawn.")
print("  A 1 or 2 misstates the evidence as badly as HIDING the limitation would.")
print("  The growing marker is the same limitation, delivered with more spine.")
```

**Reading the diagram.** The exact same limitation can land anywhere on this
scale. At a 1 it erases the finding; at a 5 it names a precise bound, the specific
thing the design cannot establish, and the study that would resolve it, delivered
with a straight spine. Nothing about the finding changed between the two. Only the
spine did.

*In plain terms, the picture says your limitation is not the problem. The posture
you say it with decides whether a stranger trusts you more or less.*

```python
# Apology-spiral detector — counts self-erasing words and checks for a next step.
APOLOGY_MARKERS = ["sorry", "just", "probably", "i guess", "i'm no", "not sure",
                   "sort of", "kind of", "maybe", "possibly", "all wrong", "can't really"]
NEXT_STEP_MARKERS = ["next step", "the study that could", "the design that could",
                     "would settle", "how you'd find out", "how one would find out"]

def audit_delivery(text):
    low = text.lower()
    flags = sum(low.count(m) for m in APOLOGY_MARKERS)
    ends_on_next_step = any(m in low for m in NEXT_STEP_MARKERS)
    return flags, ends_on_next_step

deliveries = {
    "A (apology spiral)": (
        "Honestly this is probably all wrong, it's just 812 people from one campus, "
        "and I'm no statistician, so I guess you can't really conclude anything."),
    "B (hedge-heavy)": (
        "It sort of seems like maybe there might be a kind of weak association, "
        "possibly, but I'm not sure it really holds up."),
    "C (clean ULN move)": (
        "The estimate carries a wide interval because it is one campus sample, so I "
        "read it as a real pattern, not a precise number. Because it is an "
        "association, it cannot show the drive causes absences. The study that could "
        "settle it is a panel that follows people as their commutes change."),
}

# Paste your own delivery here to test it, then rerun.
my_delivery = ""

print(f"{'Delivery':<22}{'apology flags':>14}{'ends on next step?':>22}")
for name, text in deliveries.items():
    flags, nxt = audit_delivery(text)
    print(f"{name:<22}{flags:>14}{('yes' if nxt else 'no'):>22}")
if my_delivery.strip():
    f, n = audit_delivery(my_delivery)
    print(f"{'MINE':<22}{f:>14}{('yes' if n else 'no'):>22}")
```

```python
# Self-check: the spiral and hedge deliveries trip flags; the clean ULN trips none
# AND is the only one that ends on a next step.
fa, na = audit_delivery(deliveries["A (apology spiral)"])
fb, nb_ = audit_delivery(deliveries["B (hedge-heavy)"])
fc, nc = audit_delivery(deliveries["C (clean ULN move)"])
assert fa >= 5 and fb >= 4, "sample apology/hedge deliveries should trip several flags"
assert fc == 0, "the clean ULN delivery must trip zero apology flags"
assert nc and not na and not nb_, "only the clean ULN delivery ends on a next step"
print(f"✓ Self-check passed: A flags {fa}, B flags {fb}, C flags {fc}; only C ends on a next step.")
print("  The detector catches self-erasure. It CANNOT see whether C's boundary is the RIGHT one.")
```

### The causal question aimed at a noncausal project

Now the puzzle's payoff. A visitor asks **"so does it CAUSE it?"** about your
association-only finding. The honest reply is a compressed ULN move, two sentences,
and it does both jobs at once: it refuses the causal word, and it hands the visitor
something to chew on so they stay interested.

The shape is: **name the boundary, then offer the design that would answer their
actual question.** Example for the commute study: "I can't say it causes the
absences, because I only measured that the two go together in one sample. What
*would* answer your question is following the same people as their commutes change,
and that's the study I'd run next." The first half protects the truth (it refuses
"causes"), and the second half keeps the visitor leaning in (it treats their causal
curiosity as a real research question, not a mistake).

That reply is exactly what the colleague's transcript lacked. "We checked it a
lot of ways, so basically yes" answers the visitor's enthusiasm instead of the
design, and the upgrade rode in on the word "basically." Consistency across
checks strengthens the association; it buys no causal license at all.

> **A question that often comes up here:** *"Isn't admitting 'I can't say it causes
> it' just conceding I have nothing?"* The opposite. You are showing you know
> exactly what your design can and cannot bear, which is the most credible thing at
> the poster. A presenter who says "sure, basically it causes it" to a friendly
> stranger has just committed the excited-mouth upgrade in public, and any careful
> visitor now trusts nothing else on the board.

## 5. The Mock Symposium, the Rehearsal Record, and the Agenda

**Guiding question:** *how does a rehearsal become evidence, and what do you do
with the question you could not answer?*

Friday's studio is the **mock symposium**: a full rehearsal of the poster
session in a safe room, so you find out today whether every piece works under
live traffic. It runs as **station rounds**. A few posters go live at a time
with their presenters standing beside them, while everyone else rotates past as
conference visitors, then you swap, so you both present and visit.

When you visit, you carry the **visitor role card**: across your visits, you ask
at least one question of each of the five types (method, alternative
explanation, so-what, generalization, AI-use). You are the drill for someone
else's question bank, just as they are for yours. The stance is
critical-but-kind: a real stranger, asking real questions, without trying to
wound.

Two written artifacts turn the rehearsal into M11 evidence. The **rehearsal
record** is a table with one row per hard question fired at you: the question,
your answer or your concession, and the change it requires in your materials.
Example: "asked why my cutoff was 15 minutes; defended with the claim-evidence
row; no change required." A rehearsal that leaves no record was practice for
your nerves, not evidence of your readiness.

The **unanswered-question agenda** is the shortlist of questions you could not
answer, recorded verbatim. It is not a confession. A question you cannot answer
in rehearsal is evidence, and it sends you back to the work that owns it: some
rows send you to revise a claim, some name the follow-up study, and all of them
are the watch list you carry into next week's public test. The strongest
presenters at the Expo are the ones who already know their unanswered
questions by name.

> **A question that often comes up here:** *"What if my mock symposium goes badly?"*
> Then it did its job. A rough rehearsal is a gift: every fumble is one you now get
> to fix with time to spare instead of discovering at the Expo. The presenters who
> plan to "figure out the hard questions live" are not braver. They just move the
> fumble from the practice room to the real audience. Bring the pieces you drafted
> in this notebook, and spend the studio finding the lines that break.
