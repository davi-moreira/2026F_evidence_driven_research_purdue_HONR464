## 🎯 Definition of Done

You are done with this sprint when your submission carries all six parts of the
declaration, applied to **your own project**:

1. **The route declaration.** One of the five pathways (observational
   descriptive, observational causal, experimental descriptive, experimental
   causal, prediction), named and defended in five short moves: why your
   question's **kind** and **reach** point at this road, why your M3
   permission status supports it (a change of pathway can change that status,
   so recheck it rather than assume it), the **crossing licence** your route
   holds and the tempting one it lacks (plus the hard sentence: what this
   route can never establish, no matter how clean the execution), one real,
   retrievable **anchor study** that used this pathway on a comparable
   question, and the **hybrid-stage test** answered in writing: does your
   design run in **stages**, meaning one part of it feeds a later part, such
   as a screening survey that selects who enters an experiment? If no, say so
   in one sentence and stop there. If yes, add the stage-by-stage map, name
   the weakest link in the chain, and justify keeping or cutting that stage in
   one sentence. The required **Causal Identification Skeptic** critique rides
   inside this part, with every ranked attack answered.
2. **The mandated contrast.** The contrasting route assigned to you, engaged
   in \~one page on its merits: what it would require, what its licence would
   buy, and why it is not yours (or what adopting it would change).
3. **The jigsaw advocate summary.** The route you represented at the studio's
   route jigsaw, in five sentences a non-specialist could repeat.
4. **Research Contract v1.** The milestone's own product, fed by the three
   parts above. Name exactly which **Research Contract v0** fields this
   declaration revises, then issue the result as a dated, numbered **Research
   Contract v1** under three disciplines: a version reason a reader could use
   to reconstruct your thinking ("declared observational descriptive because
   my permission status forbids assignment", never "updated after the pathway
   week"), both **claim boundary** sentences redrawn to the licence your route
   actually holds, and the uncertainty statement updated to the sources your
   route carries (sampling error on a descriptive road, confounding on an
   observational causal road, drift between past and future cases on a
   prediction road).
5. **AI Research Ledger rows.** One row for every AI-assisted step, each with
   a named verification method. Two rows are mandatory: the required Skeptic
   run with what you did with each ranked attack, and a counter-pathway
   argument where your AI argues for a different route and you answer it in
   your own words.
6. **The dossier update line.** Where your **Research Contract** component now
   lives, named by file or section: the declared pathway, its written limits,
   the redrawn claim boundary, and the mandated contrast that priced the road
   you did not take.

> **A question that often comes up here:** *"My question is causal, but my
> data give me no leverage to identify the effect. Do I switch to a
> descriptive route?"* No. A question's kind lives in its words, not in your
> data. A causal question with weak data stays **causal, currently
> unidentified**: declare the observational causal route, name the leverage
> you lack, and keep your claims on the *associated with* side until
> identification is earned. The honest declaration is a passing answer; the
> silent relabel is the one move this milestone forbids.

```python
# Route-map scaffold — edit the EDIT ME blocks to match YOUR declaration.
# It renders the five pathways and marks three roles on the map: the route you
# DECLARE (green), the mandated CONTRAST assigned to you (red dashed), and the
# route you ADVOCATE at the jigsaw (dotted ring). The licence line under each
# node states the crossing licence that road holds. The last block runs your
# hybrid-stage test.

# EDIT ME — set each to one of the five route keys below.
my_route      = "observational descriptive"   # the pathway you declare
contrast      = "experimental causal"         # the contrast assigned to you
jigsaw_route  = "prediction"                  # the route you advocate at the jigsaw

# EDIT ME — the hybrid-stage test. Leave `stages` EMPTY if your design does not
# run in stages. Otherwise name each stage in order, then name the weakest link.
stages       = []    # e.g. ["screening survey", "randomized follow-up"]
weakest_link = ""    # e.g. "screening survey"

routes = {
    "observational descriptive": (0, 1, "sampling from a named frame\n→ population claims"),
    "observational causal":      (1, 1, "argued identification\n→ causal claims (when earned)"),
    "experimental descriptive":  (0, 0, "assignment serving measurement\n→ descriptive claims"),
    "experimental causal":       (1, 0, "assignment you control\n→ causal claims"),
    "prediction":                (2, 0.5, "honest held-out check\n→ claims about unseen cases"),
}
for name in (my_route, contrast, jigsaw_route):
    assert name in routes, f"'{name}' is not one of the five route keys"

fig, ax = plt.subplots(figsize=(11, 4.8))
for name, (x, y, licence) in routes.items():
    if name == my_route:
        face, edge, lw, style = "#d9efd9", "#1a7f37", 2.6, "solid"
    elif name == contrast:
        face, edge, lw, style = "#f8e0de", "#b3261e", 2.2, "dashed"
    else:
        face, edge, lw, style = "#e8eef6", "#33475b", 1.4, "solid"
    box = plt.Rectangle((x - 0.42, y - 0.16), 0.84, 0.32, facecolor=face,
                        edgecolor=edge, linewidth=lw, linestyle=style)
    ax.add_patch(box)
    ax.text(x, y + 0.04, name.replace(" ", "\n", 1), ha="center", va="center",
            fontsize=9, fontweight="bold")
    ax.text(x, y - 0.30, licence, ha="center", va="top", fontsize=7.5,
            color="#33475b")
    if name == jigsaw_route:
        ring = plt.Rectangle((x - 0.48, y - 0.22), 0.96, 0.44, fill=False,
                             edgecolor="#8a6d1d", linewidth=1.6,
                             linestyle="dotted")
        ax.add_patch(ring)

ax.text(0.5, 1.62, "observe the world as you find it", ha="center", fontsize=9)
ax.text(0.5, -0.62, "assign conditions yourself", ha="center", fontsize=9)
ax.text(2.0, 1.15, "score a forecast honestly", ha="center", fontsize=9)
ax.set_xlim(-0.7, 2.7); ax.set_ylim(-0.8, 1.8)
ax.set_title("Your route map — green: declared · red dashed: mandated contrast "
             "· dotted ring: jigsaw advocacy", fontsize=10.5)
ax.axis("off")
plt.tight_layout()
plt.show()

print(f"✓ Declared: {my_route} · contrast: {contrast} · jigsaw: {jigsaw_route}")
print("  Now write the licence your route holds, the one it lacks, and the")
print("  sentence naming what it can never establish.")

print("— Hybrid-stage test —")
if not stages:
    print("  No stages declared. Your written answer is one sentence saying so,")
    print("  and the hybrid chapter stays OFF your reading list this week.")
else:
    print("  Stage chain: " + " → ".join(stages))
    if weakest_link.strip():
        print(f"  Weakest link: {weakest_link}. Justify keeping or cutting that")
        print("  stage in one sentence.")
    else:
        print("  ⚠️ Weakest link not named. A stage map without one names no risk.")
    print("  Stages put the hybrid chapter ON your reading list this week.")
```

### 🗡️ Red-Team Exchange

The attack runs on your own declaration, at the studio with your AI assistant.
The adversarial voice is the **Causal Identification Skeptic**, the
**required** GenAI Studio reviewer for M4 (role card in
`ai_resources/agent_role_cards.md`; full briefing in
`genai_studio/roles/causal_identification_skeptic.md`).

**Round 1: the Skeptic attacks your licence.**
On a causal route, paste your identification leverage and let it attack. On
every other route, paste your **descriptive-boundary defense** (your argument
for why your claims stop short of cause) and let it hunt for causal language
you did not notice. Answer every ranked attack in writing: rule it out, or
carry it as a named limitation. It supplies the attack; you supply the
defense. An unanswered top attack is an open hole in the declaration.

**Round 2: the room attacks your contrast.**
At the studio's route jigsaw you advocate the route assigned to you, and the
room presses on everyone's mandated contrast. A contrast dismissed in one
sentence is a strawman, and a strawman defends nothing. The page earns its
place when a reader could finish it believing you would have taken the other
road if the facts had pointed there.

**What to log.** The Skeptic's ranked attacks and your written answers ride
inside the declaration and get a ledger row. An attack that changed nothing
still gets a line saying why it did not land.

> **A question that often comes up here:** *"What if the Skeptic and my own
> judgment disagree?"* You are the judge, not the tool. Say which you found
> more convincing and why, in one sentence. And remember the correlated-error
> warning: a second model agreeing "looks fine" is not confirmation. Two
> models can share the same blind spot on causal reasoning.
