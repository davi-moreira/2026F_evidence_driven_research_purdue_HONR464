## 1. The Stale-Run Rule

**Guiding question:** *when does a passed cold run stop being evidence about
your package?*

> *"I download your package a year from now and press run. I was not in the
> room, I do not know what you meant, and I cannot ask. The only version of
> your work that exists for me is the one in the folder, and the only
> verification that counts is one that ran on that exact folder."*
> — a **replication reviewer**, the stranger your release is really for

Your package earned its credibility at Studio 11 the hard way: a **cold run**,
a run of the packaged files from a fresh start, on a machine that is not
yours, with no author in the room. This week adds the rule that protects that
credibility until release.

A **stale run** is a cold run that predates a change to what it verified. The
test is one question: since the run finished, did anything change that a
stranger would download, run, or follow? A number, a claim, the data, the
code, the environment record, a README instruction: any of these makes the
run stale. Example: your run passed on Monday, you corrected one data value on
Wednesday, and the Monday run is now a certificate for a package that no
longer exists.

The rule deliberately ignores the author's label for the change. "Just a
wording fix" is a judgment about your own work, made by the one person who
cannot see it from outside, at the exact moment you most want to be done.
The boundary does the judging instead: package changed, run stale. And a
stale run has one honest remedy: **rerun from the packaged files**, record
the date, and let the new run testify.

> **A question that often comes up here:** *"The rerun takes an hour and the
> change was one word. Is the rule not overkill?"* Compare the two costs. The
> rerun costs you an hour tonight. A stale run costs you the one sentence you
> cannot say at the defense: "these exact files reproduce these exact
> numbers." If a rerun truly cannot finish before a deadline, the honest move
> is to release with the rerun queued and *declared*, never to release the
> silence. A changed package with an old green run is not verified. It is
> nostalgic.

### 🛠️ Run the Study: the stale-run test

**🔴 Live in class: we run this one together.**

Run the cell below. It applies the stale-run rule to all six changes: a change
makes the run stale when it lands after the cold run *and* touches something a
stranger would download, run, or follow. Read the six verdicts and the release
decision off the printout before you read the next markdown cell, and check
them against the count you committed.

```python
# The stale-run test: one package, one cold run, six changes on record.
# Day 0 is the day the Studio 11 cold run finished clean.
cold_run_day = 0

changes = pd.DataFrame([
    {"day": 2, "where": "research note (packaged)",
     "change": "headline sentence reworded ('just a wording fix')", "in_package": True},
    {"day": 3, "where": "defense slide deck (not packaged)",
     "change": "closing figure added to the slides", "in_package": False},
    {"day": 4, "where": "data file (packaged)",
     "change": "one mistyped value corrected", "in_package": True},
    {"day": 5, "where": "analysis notebook (packaged)",
     "change": "seed switched during exploration and saved", "in_package": True},
    {"day": 6, "where": "README (packaged)",
     "change": "install instructions edited for clarity", "in_package": True},
    {"day": 6, "where": "AI Research Ledger (dossier, not packaged)",
     "change": "new row logging yesterday's verification", "in_package": False},
])

# The rule: after the run + inside the package = the run is stale.
changes["makes_run_stale"] = (changes["day"] > cold_run_day) & changes["in_package"]

print("Cold run finished clean on day 0. Changes on record since:")
for _, ev in changes.iterrows():
    verdict = "STALE-MAKING" if ev["makes_run_stale"] else "run stays fresh"
    print(f"  day {ev['day']}: {ev['where']:<42} -> {verdict}")

n_stale = int(changes["makes_run_stale"].sum())
print(f"\nStale-making changes: {n_stale} of {len(changes)}")
print("Release decision:", "RERUN REQUIRED before release."
      if n_stale else "The cold run still testifies.")

# Timeline: green line = the cold run; orange = package-touching change (stale-making).
fig, ax = plt.subplots(figsize=(10, 4))
event_colors = ["#DD8452" if s else "#4C72B0" for s in changes["makes_run_stale"]]
ax.scatter(changes["day"], range(len(changes)), c=event_colors, s=150, zorder=3,
           edgecolors="#333333", linewidths=1.0)
for i, (_, ev) in enumerate(changes.iterrows()):
    ax.text(ev["day"] + 0.18, i, ev["where"], va="center", fontsize=9)
ax.axvline(cold_run_day, color="#55A868", linewidth=2.5)
ax.text(cold_run_day + 0.08, len(changes) - 0.35, "cold run\nfinished clean",
        color="#55A868", fontsize=9, va="top")
ax.set_yticks([])
ax.set_xlim(-1.0, 10.5)
ax.set_xlabel("Days since the Studio 11 cold run")
ax.set_title("Every orange change touches the package — each one alone makes the run stale")
plt.tight_layout()
plt.show()
```

```python
# Self-check: four of the six changes touch the package, so the run is stale.
assert n_stale == 4, "expected 4 package-touching changes after the cold run"
assert bool(changes.loc[0, "makes_run_stale"]), "the 'wording fix' edits the packaged note — stale-making"
assert not bool(changes.loc[1, "makes_run_stale"]), "slides are not in the package"
assert not bool(changes.loc[5, "makes_run_stale"]), "a ledger row documents; a stranger reruns the package, not the ledger"
print(f"✓ Self-check passed: {n_stale} of 6 changes make the run stale, the 'wording fix' included.\n"
      "  The rule reads the package boundary, never the author's label for the change.")
```

## 2. The Five-Part Release Audit and the Dossier

**Guiding question:** *what must be true before this work may leave your
hands, and where does a stranger find the proof?*

> *"Do not hand me a result. Hand me the folder that would let a stranger
> reconstruct the result, the permissions behind it, and every decision in
> between. If the folder cannot answer for you, then a year from now, neither
> can you."*
> — your **thesis advisor**, on what "finished" means

Release is a gate, not a date. The **release audit** is the five-part check
your work passes before it ships, and each part is one line in your M15
submission with its evidence named.

1. **Permissions.** You hold the right to ship everything in the package: the
   data or its access recipe, every figure, every quoted word. A licence you
   never checked is a claim you never verified.
2. **Disclosure.** Your AI disclosure and the ledger behind it are complete
   records, not summaries: what you delegated, what you sent to tools, what
   you did not verify.
3. **Reproducibility.** The stale-run test from Section 1 passes, and with it
   the **claim sweep**: every quantitative claim in the chapter resolves to an
   output your *latest* rerun produced. A number with no cell behind it is an
   untraceable number, and the defense will ask.
4. **Claim boundaries.** Every claim still sits inside its declared compass
   reach. The final polish is where verbs quietly upgrade, so the audit reads
   the finished sentences, not your memory of them.
5. **Promises to participants.** Every promise you made about anonymity,
   aggregation, storage, or deletion is kept in the shipped files. Release is
   the last point at which a permission problem can still be prevented,
   because after release there is no recall.

Passing the audit is what licenses the words "release-ready." Failing a part
is not a scandal; it is a finding, and the repair is the deliverable.

The audit's evidence lives in one place. The **dossier** is the folder that
lets a stranger reconstruct not just your result but your decisions: contract
versions, the evidence registry, permissions, measurement records, the
analysis and claim ledgers, the AI Research Ledger, your artifacts, the
package, and the revision history. Nine components. This week you open the
**dossier manifest**: the one-page list where each component is either present
with its location or named as missing. A missing component named is a plan; a
missing component unnamed is a surprise at grading.

## 3. Wire the Roles, Then Diagnose Their Agreement

**Guiding question:** *how do you run a team of AI reviewers on your chapter
without inheriting their shared blind spots?*

> *"Do not tell me your reviewers agreed. Tell me whether they could have agreed for
> the same wrong reason. Agreement is the easiest thing in the world to manufacture,
> and the most dangerous to trust."*
> — a **skeptical peer** at your practice defense

First, be exact about the word "agents." Your agents this week are **human-run
reviewer roles**: AI critics you open, task, and read one at a time, with your
own decision at every hand-off. You plan the order, carry each output to the
next role, and integrate. Nothing here plans or acts on its own, and that is a
sentence your defense can say with a straight face.

Three terms turn a pile of roles into a workflow you can supervise.

- **Scoped task**: a job narrow enough to state in one sentence and check in
  one pass. Example: "flag every claim in section 2 that reaches past my
  sample." A scoped task has a clear pass or fail. "Make it better" does not.
- **Worker-critic structure**: pairing a role that produces (the worker) with
  a role that attacks what it produced (the critic). Example: a worker drafts
  your limitations paragraph, then a critic hunts for the limitation it left
  out. A worker alone is confident about its own quality, which is exactly the
  blind spot it cannot see.
- **Independent review**: the critic checks the worker without sharing its
  assumptions. The checker has to be able to disagree, or it is not a check.

Two words decide the wiring. **Parallel** roles run at the same time because
neither needs the other's output: a clarity reviewer and a citation-checker
can read the same draft at once. **Sequential** roles wait, because one needs
the other's output as input: a critic cannot attack a draft that does not
exist. A required order between two tasks is a **dependency**, and the longest
chain of dependencies is the **critical path**: the rounds no wiring trick can
collapse. The cell below builds the standard wiring as a directed graph, a
picture of arrows where each arrow means "this feeds that." Before you run it,
commit two numbers in writing: the critical-path depth in arrows, and the most
roles that can run at once.

```python
# Build the worker-critic workflow as a directed graph.
# Nodes: five AI roles (worker, three critics, editor) plus YOU, the integrator.
# (networkx is imported once in the setup cell.)
G = nx.DiGraph()
parallel_edges = [
    ("Worker\ndraft", "Reviewer"),
    ("Worker\ndraft", "Methodologist"),
    ("Worker\ndraft", "Skeptic"),
    ("Reviewer", "You\nintegrate"),
    ("Methodologist", "You\nintegrate"),
    ("Skeptic", "You\nintegrate"),
    ("You\nintegrate", "Editor"),
]
G.add_edges_from(parallel_edges)

generations = list(nx.topological_generations(G))
depth = nx.dag_longest_path_length(G)          # longest dependency chain, in arrows
width = max(len(layer) for layer in generations)  # most roles that can run at once

print("Worker-critic workflow (three critics in PARALLEL):")
print(f"  roles + you (nodes)     : {G.number_of_nodes()}")
print(f"  dependencies (arrows)   : {G.number_of_edges()}")
print(f"  critical-path depth     : {depth} arrows  ({len(generations)} rounds)")
print(f"  maximum parallel width  : {width} roles at once")
print(f"  the critical path       : {' -> '.join(p.replace(chr(10), ' ') for p in nx.dag_longest_path(G))}")

# The SEQUENTIAL variant: the skeptic now waits for the reviewer's notes.
H = nx.DiGraph()
H.add_edges_from([
    ("Worker\ndraft", "Reviewer"),
    ("Worker\ndraft", "Methodologist"),
    ("Reviewer", "Skeptic"),               # skeptic now depends on the reviewer
    ("Skeptic", "You\nintegrate"),
    ("Methodologist", "You\nintegrate"),
    ("You\nintegrate", "Editor"),
])
gens_seq = list(nx.topological_generations(H))
print("\nIf the skeptic instead waits for the reviewer (one critic made SEQUENTIAL):")
print(f"  critical-path depth     : {nx.dag_longest_path_length(H)} arrows  ({len(gens_seq)} rounds)")
print(f"  maximum parallel width  : {max(len(l) for l in gens_seq)} roles at once")

# Draw the parallel workflow.
pos = {
    "Worker\ndraft":    (0.0, 1.0),
    "Reviewer":          (1.6, 1.9),
    "Methodologist":     (1.6, 1.0),
    "Skeptic":           (1.6, 0.1),
    "You\nintegrate":   (3.2, 1.0),
    "Editor":            (4.8, 1.0),
}
colors = {
    "Worker\ndraft": "#4C72B0", "Reviewer": "#8AA9D6", "Methodologist": "#8AA9D6",
    "Skeptic": "#8AA9D6", "You\nintegrate": "#55A868", "Editor": "#4C72B0",
}
fig, ax = plt.subplots(figsize=(10, 4.5))
nx.draw_networkx_nodes(G, pos, node_color=[colors[n] for n in G.nodes()],
                       edgecolors="#333333", linewidths=1.5, node_size=4600, ax=ax)
nx.draw_networkx_labels(G, pos, font_size=8.5, ax=ax)
nx.draw_networkx_edges(G, pos, arrowstyle="-|>", arrowsize=16, edge_color="#777777",
                       width=1.5, node_size=4600, ax=ax)
ax.set_title("A worker-critic workflow: three critics in parallel, then you integrate")
ax.axis("off")
plt.tight_layout()
plt.show()

print("\n✓ Workflow built. The green node is YOU: the integrator no role can replace.")
```

```python
# Self-check: the parallel workflow has depth 3 and width 3; sequencing a critic costs a round.
assert G.number_of_nodes() == 6, "expected 5 AI roles + you"
assert depth == 3, "parallel critical-path depth should be 3 arrows"
assert width == 3, "three critics should run in parallel (width 3)"
assert nx.dag_longest_path_length(H) == 4, "sequencing the skeptic should deepen the path to 4"
print(f"✓ Self-check passed: parallel depth {depth} (width {width}); the sequential "
      f"variant deepens to {nx.dag_longest_path_length(H)} arrows.\n"
      f"  Parallel buys you speed and three independent lenses at once; sequencing buys "
      f"the skeptic the reviewer's context, at the cost of one more round.")
```

*The drawing reads left to right: the worker drafts, three critics read that
draft at the same time, you integrate their notes, and an editor tightens the
result. The green node is the point of the picture: every path runs through
you, and the integration is a decision no role is allowed to settle.*

The wiring tells you who reads what, and when. It cannot tell you what their
answers are worth. That takes a diagnosis, because when your roles come back,
there are three very different things their agreement or disagreement can
mean, and they call for opposite responses.

- **Real disagreement**: the roles genuinely read the evidence differently.
  Example: your reviewer calls a sentence "clear and strong" while your
  methodologist flags that same sentence as reaching past your data. This is
  *useful*. It points you straight at the claim you need to check.
- **Correlated error**: the roles *agree*, but on the same wrong assumption,
  so the agreement proves nothing. Example: four reviewers built from the same
  base model, given the same prompt, all miss the same flaw because they share
  the blind spot. Ask: could they be wrong the *same* way? If yes, their
  consensus is one signal, not four.
- **False consensus**: the roles agree only because your prompt framed the
  task so they had to. Example: you ask "confirm my chapter is ready," and
  every role obliges. You led all the witnesses. Reframe the ask neutrally and
  run it again.

> **A question that often comes up here:** *"Conflict feels like the problem. Isn't
> agreement what I want?"* It is the reverse. Conflict announces itself and tells you
> where to look. Correlated error and false consensus feel like confirmation while
> quietly agreeing on something false. The trap to fear most is agreement, not
> conflict. When roles agree on something that matters, your job is to ask whether
> they were ever independent before you trust it.

```python
# Two four-reviewer teams on 5,000 flawed chapters. Fresh seed so this is
# reproducible on its own, regardless of what ran before (seed 464).
rng_sim = np.random.default_rng(SEED)

n_chapters = 5000
k = 4        # reviewers per team
detect = 0.5 # each reviewer catches the hidden flaw about half the time, on its own
rho = 0.40   # the correlated team shares a blind spot on ~40% of chapters

# Independent team: each reviewer catches the flaw on its own. A MISS = all four fail.
caught_indep = rng_sim.random((n_chapters, k)) < detect
miss_indep = ~caught_indep.any(axis=1)
rate_indep = miss_indep.mean()

# Correlated team: on a shared-blind-spot chapter, EVERY reviewer misses together;
# otherwise each catches on its own, just like the independent team.
blind = rng_sim.random(n_chapters) < rho
caught_corr = (rng_sim.random((n_chapters, k)) < detect) & (~blind[:, None])
miss_corr = ~caught_corr.any(axis=1)
rate_corr = miss_corr.mean()

print("A team MISS = all four reviewers say 'fine' while a real flaw is present.")
print(f"  Independent team miss rate : {rate_indep:.1%}   ({miss_indep.sum()} of {n_chapters})")
print(f"  Correlated team miss rate  : {rate_corr:.1%}   ({miss_corr.sum()} of {n_chapters})")
print(f"  The correlated team is unanimous-and-wrong about {rate_corr/rate_indep:.1f}x as often.")

fig, ax = plt.subplots(figsize=(7, 5))
# Independent gets the calm blue; correlated gets the warning orange (never the truth hue).
ax.bar(["Independent\nreviewers", "Correlated\nreviewers"],
       [rate_indep, rate_corr], color=["#4C72B0", "#DD8452"], edgecolor="white")
ax.set_ylim(0, 0.55)
ax.set_ylabel("Share of flawed chapters the team MISSES")
ax.set_title("Same four reviewers: a shared blind spot manufactures correlated error")
for i, r in enumerate([rate_indep, rate_corr]):
    ax.text(i, r + 0.012, f"{r:.1%}", ha="center", fontsize=11)
plt.tight_layout()
plt.show()
```

```python
# Self-check: the correlated team misses real flaws several times more often.
assert 0.04 < rate_indep < 0.09, "independent miss rate drifted — is the seed 464?"
assert 0.40 < rate_corr < 0.47, "correlated miss rate drifted — is the seed 464?"
assert rate_corr > 5 * rate_indep, "the shared blind spot should dominate"
print(f"✓ Self-check passed: independent misses {rate_indep:.1%} of flawed chapters, "
      f"correlated misses {rate_corr:.1%}.\n  Four reviewers either way, about "
      f"{rate_corr/rate_indep:.0f}x the false-confidence rate. Agreement is only worth "
      f"the independence behind it.")
```

The independent team is unanimous-and-wrong on about **6%** of flawed chapters. The
correlated team, the same four reviewers, is unanimous-and-wrong on about **43%**,
roughly seven times as often. Nothing changed except whether the reviewers could miss
the flaw *together*. When they share a blind spot, four "looks fine" verdicts are not
four confirmations. They are close to one. This is why asking the same model twice,
or four roles built the same way, can feel like a chorus of agreement while every
voice is repeating the same mistake.

## 4. Escalate, Stop, and Direct the Next Cycle

**Guiding question:** *at what moment do you take the decision back from the
tools, how do you know when you are done, and what does this work hand to the
next study?*

> *"Bring me a chapter, not a pile of AI output. I want to know which part each tool
> touched, what you told it to do, and how you checked it. If you cannot tell me
> that, you did not write the chapter. The tools did, and neither of us can defend
> it."*
> — your **thesis advisor**, on the draft you will defend this week

Some decisions are never the tools' to make. The habit that protects them is
**escalation to human judgment**: the moment an AI output would *settle* rather than
*inform* a decision that is yours, you stop and decide it yourself. Example: a role
proposes calling your result "causal." That would settle your claim boundary, so it
escalates to you. The five triggers that force an escalation are listed in
[`ai_resources/escalation_protocol.md`](https://github.com/davi-moreira/2026F_evidence_driven_research_purdue_HONR464/blob/main/ai_resources/escalation_protocol.md);
any one is enough.

What you do at an escalation is a **human override**: you take the pen back and make
the call in your own words, backed by your own verification, even when the roles
agreed on something else. Example: four roles agree your causal wording is fine, and
you overrule all four because your design has no identification argument. A
decision that just counted the roles is not a decision. It is a vote, and votes among
correlated models are meaningless.

You also need a **stopping rule**: a written statement of what "done" looks like,
stated as a decision about *your* verified confidence, not the roles' agreement.
Example: "I stop here because every surviving claim traces to a check I ran myself,
and the remaining uncertainty is declared; one more robustness check would not
change what the chapter may claim." A bad stopping rule is "when the roles all
agree," because your simulation just priced that. And some calls are **decisions
that should never be automated** no matter how good the tools get: which research
problem to pursue, the claim boundary, an ethical judgment, and the public defense.
These stay human by design.

> **A question that often comes up here:** *"If I keep overriding the AI, why use it
> at all?"* Because the tools are excellent at the parts you *can* delegate: drafting,
> surfacing leads, listing threats, tightening prose. Escalation does not mean
> distrusting everything. It means keeping a human at the few decisions that carry
> your name. A semester-long ledger with zero escalations is a warning sign, not a
> badge. It usually means AI was never used on anything that mattered, or a boundary
> was crossed without anyone noticing.

A research cycle does not trail off. It ends with a hand-forward. The
**next-study agenda** is the last page of your dossier: the question this work
opened, and what you would do differently the next time. Its one discipline is
the verb tense of honesty: the agenda is written as **directions, never
findings**. "The next study should test whether the gap persists when
participation is required" is a direction. "The gap likely persists" is a
finding you never showed, smuggled past your own release audit in the future
tense. Example: a descriptive project that found a turnout gap hands forward
"randomize the reminder to test whether it *causes* the gap to close," and
claims nothing about what that experiment would find.

The agenda is also where stopping becomes respectable. You are not abandoning
the open question; you are naming it, pointing the next cycle at it, and
signing the stopping rule that says why here. A research cycle ends by
declaring the next one.

The last thing this week stages is the room itself. The **Evidence Defense**
is a ten-minute defense followed by a five-minute cross-examination, in which
you defend not just *what* you claim but *why* your compass position, your
data, and your checks license it. The full format lives in
[`project/final_dossier/evidence_defense_protocol.md`](https://github.com/davi-moreira/2026F_evidence_driven_research_purdue_HONR464/blob/main/project/final_dossier/evidence_defense_protocol.md).
Your ten minutes have three movements, in order:

- **Claims (about three minutes)**: your headline claim and the exact edge it stops
  at. Example: "In my sample I find X; I do not claim it holds for the population."
- **Choices (about four minutes)**: why this compass position, this data strategy,
  this operationalization, and the road you did not take.
- **Verification (about three minutes)**: how you checked, the sensitivity that
  survived, the sources confirmed real. Verification is the spine of the defense, not
  an afterthought.

Two more terms decide whether the defense goes well. **Concede gracefully** means
that where your ledger says a claim reaches too far, you say so precisely and name the
smaller true claim, instead of digging in. Conceding the right point on time is
scored as strength, not weakness. And the **forbidden moves** are three ways to lose
the room: inventing a new claim under pressure to escape a hard question, un-conceding
a point by insistence rather than evidence, and blaming the AI tool. "The AI gave me
that" is never a defense. You verified, or you did not.

> **A question that often comes up here:** *"What if I get a question I cannot
> answer?"* Answer from your ledger if the row exists, and if it does not, concede the
> point and name the smaller claim you *can* defend. That is a far stronger move than
> inventing an answer, which is a forbidden move precisely because the room can tell.
> A defense is not a performance of certainty. It is an honest account of what your
> evidence holds and where it stops.

**Before you ask:** predict in one sentence what happens to each team's miss rate as
you add more reviewers. Commit before the tool explains it.

> 💡 **AI Prompt:** "This cell sweeps the number of reviewers from 2 up to 16 for
> both an independent team and a correlated team that shares a blind spot, and prints
> each team's miss rate at every team size: [paste the next cell]. Explain why the
> independent team's miss rate keeps shrinking toward zero while the correlated team's
> flattens out at a floor. Then tell me what that floor equals and why adding more
> correlated reviewers cannot push the miss rate below it."
>
> **After running, verify (counters *correlated errors across tools or roles*: 'more
> reviewers' is not the same as 'more independence'):**
> - [ ] Confirm the floor your AI names matches the correlated rate your cell printed
>   at the largest team size, not a rounder number it preferred.
> - [ ] Check that its independent-team numbers match the printed sweep at each team
>   size, not a smooth curve it imagined.
> - [ ] Log this use in your AI Research Ledger: task, tool, decision, verification.

```python
# Sweep the team size. Fresh seed so the sweep is reproducible on its own (seed 464).
rng_sweep = np.random.default_rng(SEED)

team_sizes = [2, 4, 8, 16]
indep_rates, corr_rates = [], []
for kk in team_sizes:
    caught_i = rng_sweep.random((n_chapters, kk)) < detect
    indep_rates.append((~caught_i.any(axis=1)).mean())
    bl = rng_sweep.random(n_chapters) < rho
    caught_c = (rng_sweep.random((n_chapters, kk)) < detect) & (~bl[:, None])
    corr_rates.append((~caught_c.any(axis=1)).mean())

print("Miss rate as the team grows:")
print(f"  {'reviewers':>10} | {'independent':>12} | {'correlated':>11}")
for kk, ri, rc in zip(team_sizes, indep_rates, corr_rates):
    print(f"  {kk:>10} | {ri:>11.1%} | {rc:>10.1%}")
print(f"\n  The correlated floor sits near the shared-blind-spot rate ({rho:.0%}): "
      f"adding roles cannot push below it.")

fig, ax = plt.subplots()
ax.plot(team_sizes, indep_rates, marker="o", color="#4C72B0", label="independent reviewers")
ax.plot(team_sizes, corr_rates, marker="s", color="#DD8452", label="correlated reviewers")
ax.axhline(rho, color="#8C8C8C", linestyle=":", linewidth=1.3, label=f"shared blind-spot rate ({rho:.0%})")
ax.set_xlabel("Number of reviewers on the team")
ax.set_ylabel("Share of flawed chapters missed")
ax.set_title("More reviewers help only if their errors are independent")
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# Self-check: independence buys you everything; correlation floors out near rho.
assert indep_rates[0] > indep_rates[-1], "independent miss rate should fall as the team grows"
assert indep_rates[-1] < 0.01, "16 independent reviewers should almost never all miss"
assert corr_rates[-1] > 0.30, "the correlated team should stay stuck near the shared blind-spot floor"
print(f"✓ Self-check passed: independent misses fall from {indep_rates[0]:.1%} to "
      f"{indep_rates[-1]:.1%} across 2 to 16 reviewers;\n  the correlated team barely "
      f"moves, from {corr_rates[0]:.1%} to {corr_rates[-1]:.1%}. Count is not independence.")
```

*The two lines say it plainly: the independent team's misses drop toward zero as it
grows, while the correlated team flattens against a floor set by its shared blind
spot. Piling on more correlated reviewers cannot buy what independence buys.*
