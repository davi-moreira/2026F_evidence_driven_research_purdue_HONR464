## 1. The Release Gate, Run Against Real Dates

**Guiding question:** *what has to be true tonight before a finished artifact
stops being yours alone?*

> *"I will open your folder a year from now. You will not be in the room, I
> cannot ask you anything, and the only version of your work that exists for me
> is the one you shipped. Every promise you made about it has to survive
> without you."*
> — a **replication reviewer**, the stranger your release is really for

You built this machinery at Studio 12 and ran it once, on a project that was
still moving. Nothing about it is new. What is new is that the artifact is
finished, the dates are real, and nothing comes after tonight.

Three words carry the whole gate, so hold them precisely.

- **Disposition**: the verdict you write beside an audit row. **Clear** means a
  record exists and you opened it today. **Pending** means the record exists but
  is incomplete, and the work can continue while you finish it. **Blocking**
  means nothing else in the audit can compensate, so release waits. Example: an
  unwritten ledger session is pending; a permission you never checked is
  blocking.
- **Package-current gate**: the rule that a release requires the package to
  match the cold run that tested it. Any change after that run to a file a
  stranger would download, run, or follow makes the run stale, and the honest
  remedy is a dated rerun. Example: a corrected data value inside the package
  retires the run; a redrawn poster figure outside the package does not.
- **Blocking gate**: a check that is not scored and cannot be averaged away.
  This week has two. The first is the permission determination, which no work
  proceeds past. The second is the package-current gate.

The seven rows the audit grades, by the names your milestone uses:
**permissions · participant promises · disclosure · package matches cold run ·
reproducibility · claim boundaries · uncertainty**. Each one gets a disposition,
and each disposition names the record behind it. A row with no named record is
pending, never clear.

> **A question that often comes up here:** *"The rerun takes an hour and the
> change was one word. Is the gate not overkill?"* Compare the two costs. The
> rerun costs you an hour tonight. A stale run costs you the one sentence you
> cannot say at your defense: "these exact files reproduce these exact numbers."
> If the rerun genuinely cannot finish before the deadline, the honest move is
> to release with the rerun queued and declared in the revision history. A
> changed package with an old green run is not verified. It is nostalgic.

### 🛠️ Run the Study: the release gate

**🔴 Live in class: we run this one together.**

Run the cell below. It takes the three edits, applies the package-current rule
to each, then gives every audit row a disposition from one rule: a row is clear
when its record is named and complete, blocking when it is not and nothing else
can compensate, and pending otherwise. Read the tally and the decision line off
the printout before you read the narration under it, and check them against the
two things you just committed.

Then swap in your own three dates on the marked lines and run it again: your
cold-run date, your last edit to a packaged file, and today.

```python
# The release gate. Day 0 is the day your cold run finished clean at the
# packaging studio. EDIT THE THREE `day` VALUES to your own record, then rerun:
# everything below is derived from them, so nothing else needs changing.
cold_run_day = 0          # <-- your cold run, as day 0

edits = pd.DataFrame([
    {"day": 8,  "file": "Expo poster figure (not packaged)", "in_package": False,
     "what": "closing figure redrawn for the poster"},
    {"day": 16, "file": "analysis notebook (packaged)", "in_package": True,
     "what": "red-team repair: the interval recomputed on the right grouping"},
    {"day": 30, "file": "research note (packaged)", "in_package": True,
     "what": "headline sentence reworded during revision week"},
])

# The package-current rule: after the cold run AND inside the package boundary.
edits["breaks_the_run"] = (edits["day"] > cold_run_day) & edits["in_package"]
n_breaking = int(edits["breaks_the_run"].sum())
package_current = n_breaking == 0

print("Edits on record since the cold run:")
for _, e in edits.iterrows():
    flag = "BREAKS THE RUN" if e["breaks_the_run"] else "run stays fresh"
    print(f"  day +{e['day']:>2}: {e['file']:<34} -> {flag}")
print(f"\nPackage-current gate: {'PASS' if package_current else 'FAIL'} "
      f"({n_breaking} packaged edit(s) after the cold run)")

# The seven rows. `named` = a record is pointed at; `complete` = you opened it
# today and it answers the row; `blocks` = nothing else can compensate for it.
rows = pd.DataFrame([
    {"row": "permissions", "named": False, "complete": False, "blocks": True,
     "record": "no terms-of-use line anywhere in the evidence registry"},
    {"row": "participant promises", "named": True, "complete": True, "blocks": True,
     "record": "deposit agreement's aggregation clause, checked against the shipped table"},
    {"row": "disclosure", "named": True, "complete": False, "blocks": False,
     "record": "AI Research Ledger, last three sessions still unwritten"},
    {"row": "package matches cold run", "named": True, "complete": package_current,
     "blocks": True, "record": f"cold run on day {cold_run_day}, plus the edit log above"},
    {"row": "reproducibility", "named": True, "complete": False, "blocks": False,
     "record": "claim sweep run, one chapter number still has no cell behind it"},
    {"row": "claim boundaries", "named": True, "complete": True, "blocks": True,
     "record": "headline read against the inquiry declaration, reach unchanged"},
    {"row": "uncertainty", "named": True, "complete": True, "blocks": True,
     "record": "headline carries its interval in the same sentence"},
])

def disposition(r):
    if r["named"] and r["complete"]:
        return "CLEAR"
    return "BLOCKING" if r["blocks"] else "PENDING"

rows["disposition"] = rows.apply(disposition, axis=1)

print("\nThe seven release-audit rows:")
for _, r in rows.iterrows():
    print(f"  {r['disposition']:<8} {r['row']:<24} <- {r['record']}")

n_clear = int((rows["disposition"] == "CLEAR").sum())
n_pend = int((rows["disposition"] == "PENDING").sum())
n_block = int((rows["disposition"] == "BLOCKING").sum())
print(f"\nTally: {n_clear} clear, {n_pend} pending, {n_block} blocking.")

repairs = "; ".join(rows.loc[rows["disposition"] == "BLOCKING", "row"])
print("Release decision:",
      f"WITHHOLD pending a named repair for: {repairs}." if n_block
      else "RELEASE.")
```

```python
# Self-check: two rows block, and the package-current gate fails on two edits.
assert n_breaking == 2, "two edits land inside the package after the cold run"
assert not package_current, "with packaged edits after the run, the gate must FAIL"
assert n_block == 2, "permissions and package-matches-cold-run both block"
assert not bool(edits.loc[0, "breaks_the_run"]), "the poster figure is outside the package"
print(f"✓ Self-check passed: {n_block} blocking, {n_pend} pending, {n_clear} clear, "
      f"and the gate FAILS on {n_breaking} packaged edits.\n"
      "  The gate reads the package boundary and the calendar, never the author's "
      "label for the change.")
```

## 2. Withhold Is a Disposition, and the Defense Standard

**Guiding question:** *what do you write when a row blocks, and how do you say
all of it out loud in five minutes?*

> *"Tell me what your work cannot show before I have to ask. When someone does
> that unprompted, I trust everything else they said more."*
> — a **policy stakeholder** who has heard a thousand presentations

A release decision has exactly three honest forms, and one dishonest one that
is not on the menu.

- **Release.** Nothing a stranger would download, run, or follow has changed
  since the run that vouches for the package, and every row carries a record you
  opened. Your defense is the file list, not your confidence.
- **Withhold pending a named repair.** A row blocks. You name the repair,
  write it as this version's reason, and the artifact waits. This is a fully
  creditable outcome. What the audit governs is what your work may claim, never
  whether you show up.
- **Release with the rerun queued and declared.** The rerun cannot finish
  before the deadline. Both verbs matter: *queued* means scheduled with a date,
  and *declared* means the revision history says so out loud.
- The fourth option, a changed package on an old green run with nothing said,
  is the one this whole gate exists to prevent.

A **version reason** is the sentence that says why this version exists and what
changed since the last one. You never edit the previous version. You write the
next one, because the sequence of changes is itself part of your research
record.

Now the room. Your **Evidence Defense** is five minutes in three movements,
followed by adversarial cross-examination.

- **Movement 1, the claim (about two minutes).** Your headline claim and the
  exact edge it stops at. Example: "in my sample I find X, with this interval;
  I do not claim it holds for the population."
- **Movement 2, the choices (about two minutes).** Why this compass position,
  this data strategy, this measurement, and the road you did not take.
- **Movement 3, verification and the AI collaboration (about one minute).** How
  you checked, what survived, how the release audit came out, and where you
  directed AI and took the pen back.

Two more terms decide how it goes. **Concede gracefully** means that where your
ledger says a claim reaches too far, you say so precisely and name the smaller
true claim on the spot. Conceding the right point on time is scored as strength.
The **forbidden moves** are three ways to lose the room: inventing a new claim
under pressure, un-conceding a point by insistence rather than evidence, and
blaming the AI tool. "The AI gave me that" is never a defense. You verified, or
you did not.

When you are not defending, you are a cross-examiner, and you fill a **verdict
card**: three yes-or-no judgments on the defender you just heard. Is the claim
inside its stated boundary? Was verification shown rather than asserted? Was the
AI collaboration owned rather than blamed? Every non-defender asks at least one
question grounded in a ledger row.

> **A question that often comes up here:** *"What if I get a question I cannot
> answer?"* Answer from your ledger if the row exists. If it does not, concede
> the point, name the smaller claim you can defend, and say how one would find
> out. A defense is not a performance of certainty. It is an honest account of
> what your evidence holds and where it stops.

## 3. The Claim Sweep: Every Number Against the Latest Rerun

**Guiding question:** *does every number in your chapter resolve to an output
your latest rerun actually produced?*

> *"I do not need your notebook to be beautiful. I need to put my finger on a
> number in your text and have you show me the cell that made it, today, from
> the files you shipped."*
> — a **journal reviewer**, on the check that takes four minutes and saves a paper

A **claim sweep** is the row-by-row check that every quantitative claim in your
artifact resolves to an output your latest rerun produced. Not to a screenshot,
not to a number you remember, and not to an earlier version of the notebook.
Example: your chapter says the gap was four points, so you find the cell that
printed it in the run you are shipping, or the sentence changes.

A swept row comes back in one of four states, and they call for different
repairs.

- **MATCHES**: the number in the text is the number the rerun printed. It is
  worth exactly that: one number, today, from these files.
- **MISMATCH**: the text and the rerun disagree. The text changes to what the
  output supports, or the claim comes out. Example: the chapter says 90 minutes
  and the current rerun says about 60.9.
- **OUTSIDE BOUNDARY**: the number is fine and the sentence built on it is not.
  Example: a mean computed from a voluntary survey is correct for the people who
  answered, and the sentence says the whole program.
- **NO OUTPUT**: nothing in the rerun produced this at all. This is the one
  that blocks, because there is no repair short of running the analysis or
  cutting the claim.

Two of the four statuses turn on **reach**, which is how far a sentence claims
to travel: the data at hand, the population it was sampled from, or cases nobody
observed. A voluntary survey licenses the data at hand, so a sentence about
everyone in the program has left its reach even when its number is exactly
right.

The demonstration below is one small project at its close, the same shape as
yours: a voluntary survey, a headline comparison, an interval, and a four-row
claim table. Then you sweep your own.

> **A question that often comes up here:** *"The number moved by half a point.
> Does that really matter?"* It matters for a reason that has nothing to do with
> the half point. The number in your text and the number in your files are
> supposed to be the same object. When they differ, something in between changed
> and nobody wrote it down, and you cannot tell from the text alone whether that
> something was a rounding habit or a repaired data value. Fixing the sentence
> costs a minute. Being unable to say where a number came from costs the
> defense.

### 🛠️ Run the Study: the claim sweep

**🔴 Live in class: we run this one together.**

Run the cell below. It rebuilds the worked example's survey from its own seed,
prints the two numbers and the interval its latest rerun produced, then resolves
each claim-table row against those outputs and prints a status. **Expect** three
of the four rows to fail, each for its own reason. Read the four statuses off the
printout before the narration under it, and check the count against what you
just committed.

```python
# One project at its close, the same shape as yours. The worked example is a
# voluntary survey about a campus recreation program: 148 people answered, and
# the chapter's headline compares the ones who used the group classes with the
# ones who did not. The data is rebuilt from the course seed so the rerun is
# identical every time; yours comes from your own packaged files instead.
rng_demo = np.random.default_rng(SEED)
n_resp = 148
used_class = rng_demo.random(n_resp) < 0.45          # answered "yes, I use them"
minutes = (rng_demo.normal(150, 55, n_resp) + 60 * used_class).clip(0, None)

mean_minutes = minutes.mean()
class_gap = minutes[used_class].mean() - minutes[~used_class].mean()

# The interval that ships beside the gap: 2,000 seeded resamples of the answers.
draws = np.empty(2000)
for b in range(2000):
    take = rng_demo.integers(0, n_resp, size=n_resp)
    mb, ub = minutes[take], used_class[take]
    draws[b] = mb[ub].mean() - mb[~ub].mean()
gap_lo, gap_hi = np.percentile(draws, [2.5, 97.5])

rerun = {"mean_minutes": mean_minutes, "class_gap": class_gap}
LICENSED_REACH = "the respondents"   # a voluntary survey reaches no further

print("Latest rerun of the worked example, from its own files:")
print(f"  answers, and class users      : {n_resp}, of whom {int(used_class.sum())}")
print(f"  mean weekly minutes           : {mean_minutes:.2f}")
print(f"  gap, class users vs the rest  : {class_gap:.2f} minutes")
print(f"  95% interval on the gap       : [{gap_lo:.2f}, {gap_hi:.2f}] minutes")

# Step two: sweep the chapter's claim table against those outputs.
# `resolves_to` names the rerun output the sentence leans on; a name the rerun
# never produced is a claim with no cell behind it. `reach` is how far the
# sentence travels, checked against what the design licenses.
claim_table = pd.DataFrame([
    {"claim": "Class users trained 90 minutes more a week.",
     "resolves_to": "class_gap", "number": 90.0, "reach": "the respondents"},
    {"claim": "Members of the program train about 174.6 minutes a week.",
     "resolves_to": "mean_minutes", "number": 174.6, "reach": "the whole program"},
    {"claim": "Among the respondents, class users trained about 60.9 minutes more.",
     "resolves_to": "class_gap", "number": 60.9, "reach": "the respondents"},
    {"claim": "The gap is larger for members who joined this year.",
     "resolves_to": "newcomer_gap", "number": None, "reach": "the respondents"},
])

TOL = 0.05  # anything larger is a different number, not a rounding habit

def sweep(row, outputs, licensed_reach):
    """Return one of the four statuses for a single claim-table row."""
    if row["resolves_to"] not in outputs:
        return "NO OUTPUT"
    if row["number"] is not None and abs(row["number"] - outputs[row["resolves_to"]]) > TOL:
        return "MISMATCH"
    if row["reach"] != licensed_reach:
        return "OUTSIDE BOUNDARY"
    return "MATCHES"

claim_table["status"] = claim_table.apply(sweep, axis=1,
                                          args=(rerun, LICENSED_REACH))

print("\nClaim sweep, four rows from the chapter's claim table:")
for _, r in claim_table.iterrows():
    print(f"  {r['status']:<17} {r['claim']}")
n_fail = int((claim_table["status"] != "MATCHES").sum())
print(f"\n{n_fail} of {len(claim_table)} rows fail the sweep.")

# One picture, and it can only show the rows that are about a number.
fig, ax = plt.subplots(figsize=(9, 3.4))
ax.hlines(1, gap_lo, gap_hi, color="#4C72B0", linewidth=3)
ax.plot(class_gap, 1, "o", color="#4C72B0", markersize=10,
        label="rerun gap, class users vs the rest")
ax.plot(90.0, 1, "X", color="#DD8452", markersize=12,
        label="the number the chapter printed")
ax.text(gap_lo, 1.04, f"{gap_lo:.2f}", color="#4C72B0", fontsize=9,
        ha="center", va="bottom")
ax.text(gap_hi, 1.04, f"{gap_hi:.2f}", color="#4C72B0", fontsize=9,
        ha="center", va="bottom")
ax.text(gap_hi, 1.19, "95% interval on the gap", color="#4C72B0",
        fontsize=9, ha="right", va="bottom")
ax.set_ylim(0.80, 1.32)
ax.set_yticks([])
ax.set_xlabel("Weekly minutes, difference between the two groups")
ax.set_title("The claim table against the latest rerun")
ax.legend(loc="lower right", fontsize=8)
plt.tight_layout()
plt.show()
```

```python
# Self-check: three of four rows fail, and each fails for its own reason.
statuses = list(claim_table["status"])
assert n_fail == 3, "three of the four claim-table rows should fail the sweep"
assert statuses[0] == "MISMATCH", "90 is not the number this rerun printed"
assert statuses[1] == "OUTSIDE BOUNDARY", "a voluntary survey does not reach the program"
assert statuses[2] == "MATCHES", "this sentence quotes the rerun, at the reach it earned"
assert statuses[3] == "NO OUTPUT", "no cell in the rerun compared newer members"
print(f"✓ Self-check passed: {n_fail} of 4 rows fail. The rerun's gap is "
      f"{class_gap:.2f} minutes,\n  the mean is {mean_minutes:.2f} minutes, and the "
      f"interval on the gap is [{gap_lo:.2f}, {gap_hi:.2f}].")
```

> *"Bring me a chapter, not a pile of AI output. I want to know which part each

You already know the three diagnoses and the escalation habit from the release
studio, so one sentence is the whole retrieval: agreement between roles is worth
only the independence behind it, and an **independence check** is what turns
agreement into evidence. An independence check is a test that an agreement is
worth more than a head count: a check you ran yourself, or a way you made two
roles genuinely independent through different tools, different framings, or
different context.

Two records close the semester's AI work. A **human override** is the moment you
took the pen back and decided in your own words, backed by a check no model
touched. Your **conflict-and-override log** collects those moments with the
non-AI check that settled each one, and it is the second part of the portfolio
you submit.

Then you stop. A **stopping rule** is a written statement of what "done" means,
stated as a decision about *your own verified confidence*. Example: "I stop here
because every surviving claim has a check I ran myself, and the one uncertainty
I could not resolve is stated beside the headline." The disqualified form is "my
reviewer roles stopped objecting," which is a statement about tools rather than
about evidence. Beside the rule sits your **never-automate list**: the decisions
you refuse to hand to any tool, however good it gets. The problem you pursue,
the claim boundary, an ethical judgment, the release decision, and the defense
itself.

A research cycle does not trail off. It ends with a hand-forward. The
**next-study agenda** names the question this work opened, tied to a limitation
you could not resolve, and its one discipline is the verb tense of honesty. The
agenda is written as **directions, never findings**. "The next study should
randomize the reminder to test whether it causes the gap to close" is a
direction. "The gap likely persists" is a finding you never showed, smuggled
past your own audit in the future tense.

> **A question that often comes up here:** *"If I stop with an open question,
> did I fail?"* The opposite. You are not abandoning the question. You are naming
> it, pointing the next cycle at it, and signing the rule that says why you stop
> here rather than running one more check. Stopping becomes respectable exactly
> when the rule is explicit, and a research cycle ends by declaring the next
> one.
