### 🤝 AI Research Partner

All semester you work with **AI as a research partner**, not an answer machine. A
**research partner** is a tool you task with real work and then check, the way you
would check a fast but unproven research assistant. And AI today is more than a
chatbot: **agentic tools** can carry out multi-step work for you, running an
analysis plan, trying variations, assembling a draft. It is fine to ask your AI to
*do* things, not only to answer things. The deal never changes: you review, you
curate, and you make the final decision.

AI is also a good partner for **brainstorming**: candidate questions, rival
explanations, directions you had not considered. Generate widely with it; commit
narrowly yourself. A good use looks like this: you commit your own answer first,
then ask your AI to stretch it, list items you will check, or explain a cell so
you can confirm it against the printed output. A bad use is letting it decide your
question, wave a claim past your evidence, or hand you a source you never opened.

**Never delegate these:** which problem and question you pursue, which claims your
evidence justifies, the ethics of your study, and how you state your uncertainty.
The full account is next week's reading,
[Research Responsibility and Intellectual Ownership](https://davi-moreira.github.io/2026F_evidence_driven_research_purdue_HONR464/book/part1-research-with-ai/04-research-responsibility-and-intellectual-ownership.html).

**Commit first, then ask.** Every AI block in this course asks you to write your
own answer before you open your tool, so the reply has something of yours to
disagree with.

> **A question that often comes up here:** *"If AI is faster and often right, why
> write my answer first?"* Because the habit protects your judgment. If you read
> the tool's answer first, it anchors you, and you lose the one thing you bring
> that it cannot: knowing what your project is really about. Writing first keeps
> you the researcher and keeps AI the assistant.

## 8. The Funnel: Four Stops

**Guiding question:** *how does a feeling become a specific unknown worth months
of work?*

Four stops, each one throwing away vagueness the stop above it still allowed.

- **Experience:** what happened, or what you noticed. Example: a menu you kept
  from 2019 lists a meal at eleven dollars; the same meal now costs sixteen.
- **Curiosity:** what you want to understand about it. Example: "has eating near
  campus really gotten more expensive faster than everything else?"
- **Topic:** the area of inquiry that contains it. Example: local restaurant
  prices and overall inflation since 2019.
- **Research problem:** the specific unknown that is not yet understood well
  enough for the decision, explanation, or audience you care about. Example: "we
  do not know how menu prices near campus moved against overall prices from 2019
  to 2025."

**What the next cell does:** draws the four stops as a narrowing funnel. Expect
four stacked bands, widest at the top, plus two ✓ lines naming the rule.

```python
# The funnel: an experience narrows, stop by stop, into a research problem.
stops = ["EXPERIENCE\n(what happened, or what you noticed)",
         "CURIOSITY\n(what you want to understand about it)",
         "TOPIC\n(the area of inquiry that contains it)",
         "RESEARCH PROBLEM\n(the specific unknown that matters)"]
widths = [1.00, 0.76, 0.52, 0.30]
fills  = ["#EAF1F8", "#D4E2F0", "#BFD3E8", "#A9C4E0"]  # light enough for black text

fig, ax = plt.subplots(figsize=(8, 5))
for i, (label, w, c) in enumerate(zip(stops, widths, fills)):
    y = 3 - i
    ax.barh(y, w, height=0.82, left=(1 - w) / 2, color=c, edgecolor="#33506E", linewidth=1.4)
    ax.text(0.5, y, label, ha="center", va="center", fontsize=9)
ax.set_xlim(0, 1)
ax.set_ylim(-0.6, 3.6)
ax.axis("off")
ax.set_title("The funnel: an experience narrows into a research problem")
plt.tight_layout()
plt.show()

print("✓ Funnel drawn. Top = widest and vaguest; bottom = one specific unknown.")
print("  Stop at the problem. Units, outcomes, and the formal question come next week.")
```

*In plain terms, the picture says the same idea gets narrower and more
concrete as you move down. The top band is something that happened to you; the
bottom band is one missing piece someone could go find.*

Now the boundary, stated once so nobody overshoots. **Stop at the problem.** Units
of analysis, outcomes, conditions, the question's kind, and how far its answer
reaches all belong to the formal declaration that closes next week's studio. A
problem is not yet a question, and trying to write the question today is how
people end up committed to a sentence they never chose.

> **A question that often comes up here:** *"My curiosity feels too big for this.
> I care about something huge, like inequality or mental health. Do I have to
> shrink it?"* You narrow the problem, not the caring. The huge thing stays as
> your motivation. The problem is the part small enough to actually chase this
> term. Chasing it is how you earn the right to say something about the huge
> thing.

## 12. The Accuracy Lock

**Guiding question:** *what has to be true before any of this enters your record?*

Four things get locked now, before anything wrong survives into a ledger.

**1. The funnel stops at the problem.** A problem is not a question. The question
gets declared next week, with its units, its outcome, and its reach.

**2. The three tests, in the book's wording.** **Importance** asks who could use
the answer and for what. **Feasibility** asks whether you can reach suitable
evidence with the time, skills, and permissions you actually have. **Possible
contribution** asks what the project might add, and stays provisional until the
evidence map.

**3. The two failures from the opening challenge, finally named.** A **silent
scope change** is when a tool swaps in a population, outcome, or comparison you
never chose, and says nothing about it. The blue-light reply did exactly that:
your curiosity was sleep and health, and it came back as screens and REM. You
catch it by reading the tool's who, what, and whom against the ones you committed.
A **confident fabrication** is a settled-sounding claim with no source behind it,
like "studies show blue light suppresses melatonin". You catch it by finding and
opening a real study yourself, or by dropping the line.

**4. Where the jump came from.** Moving from something you noticed to a possible
explanation has a name: **abduction**, proposing what might explain a pattern.
Deduction asks what must follow if an explanation is true; induction looks for
patterns across many observations; abduction invents the candidate in the first
place. Example: menu prices seem to change every third week, and "payday timing
moves menu updates" is an abductive candidate, not evidence. A Google DeepMind
position paper argues that today's tools cannot make that jump, because they hold
text about the world rather than experience of it. Treat that as an argued
position and not a settled boundary. Either way the jump is yours, because you are
the one who answers for the claim.
