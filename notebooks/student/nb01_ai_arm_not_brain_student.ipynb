## How Every Week Runs

Every week of this course has the same three-beat rhythm: two flipped lectures
(Monday and Wednesday) and a Friday project studio. Knowing the shape ahead of
time means you walk into every room already knowing your job.

**Monday and Wednesday are flipped lectures.** Both are run by the day's
**Student Research Lead** (a classmate, solo or in a pair; your instructor
leads only in Week 1), built on a fixed four-section, 50-minute architecture
whose center of gravity is a roughly 22-minute controlled AI investigation.
Your instructor does not lecture from the front: they monitor conceptual
accuracy while the lead runs the room, then formalize the concept at the end.
Monday and Wednesday use the same skeleton but a different rhythm. Monday
discovers a concept; Wednesday applies and defends it.

**Monday: concept discovery (9 / 22 / 12 / 7).**

1. **Student-led research puzzle (9 min).** The lead opens with a concrete
   puzzle or contradiction, draws out the room's prior beliefs, and everyone
   commits an answer in writing before any AI.
2. **Guided AI research-partner investigation (22 min), the core.** You work
   the notebook with your AI as a directed research partner: specify the
   decision, get an initial answer, surface its assumptions, ask for a
   competing explanation, demand evidence or a test, then verify against the
   data. The lead steers; your instructor watches for errors.
3. **Human verification + instructor formalization (12 min).** You verify the
   AI-assisted result independently (a calculation, a simulation, a
   counterexample, a DAG, a holdout check), then your instructor formalizes
   the concept, corrects misconceptions, and states the research-decision rule.
4. **Decision and defense (7 min).** You record a decision, its justification,
   a verification statement, the project implication, and an AI Research
   Ledger entry; the meeting closes on a spoken Claim Ticket.

**Wednesday: applied lab + adversarial defense (7 / 23 / 12 / 8).**

1. **Student-led retrieval and challenge (7 min).** The lead presents a common
   error, an ambiguous claim, a flawed AI output, or a result needing
   interpretation, and the room decides what is wrong before touching your AI.
2. **Intensive applied AI laboratory (23 min), the core.** You use the
   concept: repair or generate code, build a simulation, diagnose a design,
   audit a claim, hunt for data leakage, stress-test a model, always under the
   required iterative control (initial prompt → diagnose → revise → new output
   → independent test → decision), never copy-and-accept.
3. **Peer defense and adversarial questioning (12 min).** You compare
   solutions and defend what you delegated, why, what the AI proposed, what
   you rejected, how you verified it, and what is still uncertain.
4. **Transfer to the final project (8 min).** You apply the concept to your
   own research project and update the cumulative Research Project Dossier and
   ledger.

**Friday is the project studio**, with no new topic content. It is the
fixed weekly beat where the concepts from Monday and Wednesday get turned into
that week's milestone. Unlike Monday and Wednesday, Friday is not led by a
Student Research Lead: your instructor frames it and runs rotating consults
while you work. Same 50-minute discipline, four sections (10 / 5 / 30 / 5):

1. **Weekly topic quiz (10 min).** You answer a short set of multiple-choice
   questions, solo and on paper, about the topic covered during the week. It
   is quick, it counts toward your grade, and it tells you immediately what
   stuck.
2. **Research stand-up (5 min).** You report, in a sentence or two: what
   changed since last time, the current blocker, the research decision you are

3. **Milestone kickoff + AI-supported sprint (30 min), the core.** Your

### 🤝 AI Research Partner

All semester you work with **AI as a research partner**, not an answer machine.
A **research partner** is a tool you task with real work and then check, the way
you would check a fast but unproven research assistant. And AI today is more
than a chatbot: **agentic tools** can carry out multi-step work for you, running
an analysis plan, trying variations, assembling a draft. It is fine to ask your
AI to *do* things, not just answer things. The deal never changes: you review,
you curate, and you make the final decision.

AI is also a good partner for **brainstorming**: candidate questions, rival
explanations, designs you had not considered. Generate widely with it; commit
narrowly yourself. A good use looks like this: you commit your own answer first,
then ask your AI to locate candidate sources you will retrieve, list items you
will check against your data, or explain a cell so you can confirm it against
the printed output. A bad use is letting it decide your question, wave a claim
past your evidence, or hand you a source you never opened.

**Never delegate these:** which problem and question you pursue, which claims your
evidence justifies, the ethics of your study, and how you state your uncertainty.
Those are yours to declare and defend. The full list of never-delegate decisions
lives in the book:
[Chapter 4 — Research Responsibility and Intellectual Ownership](https://davi-moreira.github.io/2026F_evidence_driven_research_purdue_HONR464/book/part1-research-with-ai/04-research-responsibility-and-intellectual-ownership.html).

**Commit first, then ask.** Every AI block below asks you to write your own answer
before you open your AI tool, so the reply has something of yours to disagree with.

> **A question that often comes up here:** *"If AI is faster and often right,
> why write my answer first?"* Because the habit protects your judgment. If you read
> the tool's answer first, it anchors you, and you lose the one thing you bring that
> it cannot: knowing what your project is really about. Writing first keeps you the
> researcher and keeps AI the assistant.

## 3. Why Smart People Trust a Wrong Machine

**Guiding question:** *what makes a fluent, confident answer so hard to doubt, and
what discipline keeps you doubting anyway?*

> *"The most dangerous output isn't the one that's obviously wrong. It's the one
> that's confident, detailed, well-organized, and wrong. That one gets used."*
> — a **skeptical peer**, reviewing a draft that leaned too hard on a tool

Three named habits explain the trap. Learn the names, because you cannot guard against
a danger you cannot name.

- **Automation bias** is the tendency to over-trust an automated system and stop
  checking, precisely because it is automated. Example: drivers who follow a GPS
  straight into a closed road because "the machine said turn."
- **False confidence** is when fluent, detailed output makes you feel sure of
  something you never actually verified. Example: a polished paragraph makes a wrong
  statistic feel solid, so you quote it.
- **Illusion of understanding** is mistaking a smooth explanation for real
  comprehension. Example: you nod along to an AI's line-by-line walkthrough of a cell,
  then find you could not defend it to a skeptic.

None of this is new to AI. Long before these tools, Tversky and Kahneman (1974)
showed people lean on confident mental shortcuts that misfire in predictable ways. And
Open Science Collaboration (2015) re-ran 100 published psychology studies and found
that many did not hold up, a reminder that even peer-reviewed, confident-sounding
results need checking. A fluent AI answer simply makes the old trap faster and
smoother.

> **A question that often comes up here:** *"If these tools invent things and mislead,
> why use them at all?"* Because they are genuinely powerful, and you will use them
> for the rest of your career no matter what any course says. A search that once took
> an afternoon now takes minutes. The skill worth building is not avoidance. It is
> using the tool ambitiously and then verifying like a professional, every time.

---

The discipline that beats the trap is a loop with a human at both ends. You met its
shorthand, Ask → Verify → Document. Here is the full version the course trains,
**SDIIVDD**: **Specify, Delegate, Interrogate, Inspect, Verify, Document, Defend**.

You **specify** the task and write down your own expected answer first. You
**delegate** the well-specified work to the tool. You **interrogate** the draft for
assumptions and errors. You **inspect** what it actually produced, not its summary of
itself. You **verify** with a real check before trusting anything. You **document** the
use in your Ledger. You **defend** the result in your own words, with the tool out of
the room.

The two ends are human on purpose. That connects to one more term. **Never-delegate
decisions** are the calls that stay yours no matter how good the tool gets: the
question, the design and population, the measures, the ethics, the claim boundary, the
uncertainty, and the public defense. Example: you may ask your AI to list candidate
measures for your concept, but deciding which one truly captures it is your call.

**What the next cell does:** draws the full SDIIVDD chain as one figure. Expect
seven boxes and a ✓ line naming the everyday shorthand.

```python
# The full loop, drawn as a chain: Specify -> Delegate -> Interrogate -> Inspect ->
# Verify -> Document -> Defend. Specify and Defend are the human bookends.
steps = ["Specify", "Delegate", "Interrogate", "Inspect", "Verify", "Document", "Defend"]
G2 = nx.DiGraph()
G2.add_edges_from([(steps[i], steps[i + 1]) for i in range(len(steps) - 1)])
pos2 = {s: (i, 0) for i, s in enumerate(steps)}

# Human-owned bookends get a darker outline AND are named in the caption (not color
# alone). Verify, where trust is earned, is filled a distinct shade.
fill = ["#B7C9E2", "#DCE6F1", "#DCE6F1", "#DCE6F1", "#55A868", "#DCE6F1", "#B7C9E2"]
edgecol = ["#1F3B6E", "#4C72B0", "#4C72B0", "#4C72B0", "#2F6B45", "#4C72B0", "#1F3B6E"]

fig, ax = plt.subplots(figsize=(12, 3))
nx.draw_networkx_nodes(G2, pos2, node_color=fill, edgecolors=edgecol,
                       linewidths=2.0, node_size=3400, ax=ax)
nx.draw_networkx_labels(G2, pos2, font_size=8.5, ax=ax)
nx.draw_networkx_edges(G2, pos2, arrowstyle="-|>", arrowsize=16, edge_color="#555555",
                       width=1.5, node_size=3400, ax=ax)
ax.set_title("SDIIVDD: the AI-collaboration loop, human at both ends (Specify, Defend)")
ax.set_xlim(-0.6, 6.6)
ax.set_ylim(-0.6, 0.6)
ax.axis("off")
plt.tight_layout()
plt.show()
print("✓ SDIIVDD loop drawn. Shorthand: Ask (Specify + Delegate) → Verify (Interrogate")
print("  + Inspect + Verify) → Document, and Defend closes it. You own the first and last step.")
```

**Reading the output.** *In plain terms, the chain says: you set the task, hand out the work, argue with the
draft, look at what it really did, confirm it with a real check, log it, and defend it
yourself. The two darker end circles are the steps you never hand to a tool.*

**What the next cell does:** builds a tiny randomized study: 200 people, half
assigned to a study-skills workshop by coin flip, and a written summary that
claims a **+9 point** effect. It prints both group means, the gap verified from
the data, and the claimed gap, then draws the two score distributions. Watch
whether the printout backs the paragraph.

```python
# A small randomized study: does a study-skills workshop raise a quiz score?
# Re-seed a fresh generator so this result is reproducible on its own (seed 464).
rng_study = np.random.default_rng(SEED)

n = 120
TRUE_EFFECT = 3.0                                        # the real effect in this toy world
workshop = rng_study.permutation(np.repeat([1, 0], n // 2))  # random assignment, 60 vs 60
score = rng_study.normal(72, 8, n) + TRUE_EFFECT * workshop
study = pd.DataFrame({"workshop": workshop, "score": score.round(1)})

means = study.groupby("workshop")["score"].mean()
observed_gap = means[1] - means[0]
CLAIMED_GAP = 9.0                                        # the confident one-line summary

print(f"Mean score, no workshop : {means[0]:.2f}")
print(f"Mean score, workshop    : {means[1]:.2f}")
print(f"Verified gap (from data): {observed_gap:+.2f} points")
print(f"Claimed gap (summary)   : {CLAIMED_GAP:+.2f} points")
print(f"The summary overstated the gap by {CLAIMED_GAP - observed_gap:.2f} points.")

# Plot the claimed number next to the verified one. Hue AND the value label both carry
# the meaning, so the picture never depends on color alone. The verified bar is blue;
# the unverified claim gets its own orange so it never reads as the trusted value.
fig, ax = plt.subplots(figsize=(7, 5))
bars = ax.bar(["Verified gap\n(from the data)", "Claimed gap\n(confident summary)"],
              [observed_gap, CLAIMED_GAP],
              color=["#4C72B0", "#DD8452"], edgecolor="white")
for b, v in zip(bars, [observed_gap, CLAIMED_GAP]):
    ax.text(b.get_x() + b.get_width() / 2, v + 0.15, f"{v:+.1f}", ha="center", fontsize=11)
ax.set_ylim(0, 10)
ax.set_ylabel("Effect on quiz score (points)")
ax.set_title("Verify the number: the confident summary is not what the data show")
plt.tight_layout()
plt.show()
```

**Reading the output.** The verified gap is **+1.68 points** (75.01 vs 76.69)
against the summary's claimed **+9.00**. The printout, not the paragraph, is
the evidence; the overstatement is 7.32 points.

**What the next cell does:** a self-check that restates that comparison as one
✓ line.

```python
# Self-check: random assignment licenses a causal gap here, but it is far below 9.
assert observed_gap > 0, "the built-in effect is positive; the observed gap should be too"
assert observed_gap < CLAIMED_GAP - 2, "the claim should clearly overstate the data"
print(f"✓ Self-check passed: the verified gap is {observed_gap:+.2f} points, well below the "
      f"confident '{CLAIMED_GAP:+.0f} points'.\n  Random assignment lets this gap estimate "
      f"the workshop's effect in the toy world, but the claimed number was never in the data.")
```
