## 🎯 Definition of Done

You are done with this sprint when your submission carries every part below,
built from **your own** afternoon at the Expo.

**The live half (graded in the room, and terminal):**

1. **The Expo presentation and peer evaluation.** You stand your poster for
   strangers and deliver a **bounded claim** out loud: what your evidence shows
   and, in the same breath, the boundary it does not cross (which kind, which
   reach). You also evaluate at least **three** other posters, taking notes at
   each one on its central claim, the evidence shown for it, and the single place
   the two do not line up. While you present, capture every question on your
   coding sheet as one line plus a one-word tag, and star anything you hear twice.

**The written half (submitted at the reflection studio):**

2. **The coded question tally.** Every question and reaction, in a real table,
   coded by **type** (method, alternative-explanation, so-what, generalization,
   AI-use, other), **target** (claims or communication, tagged where the *fix*
   lives), and **recurrence** (how many times you got a version of it).
3. **The pattern you found.** One or two sentences read straight off the counts:
   which type carried the most recurrence-weighted questions, whether the heavier
   target was claims or communication, and the one claims question that recurred
   enough to be a finding about your project. Cite the rows it comes from.
4. **The honest reflection paragraph.** What the public test changed in how you
   see your project, stated with its uncertainty and its limits, naming whether
   what the day exposed lives in your **claims** or your **communication**.
5. **The defense revision record.** Every place the questioning exposed something,
   as a row with three cells: what it exposed, **the revision** written out in the
   words you will actually use, and the reason the evidence licenses that much and
   no more. Each row says which of three things moved: the **claim**, the
   **boundary**, or the **defense**. "Nothing changed" is legal only with a
   defense: name your hardest question and show the boundary that already answered
   it.
6. **The poster-criticism portfolio.** Three structured, kind-but-critical
   readings of posters you evaluated. Each names an **axis** (claims or
   communication), points at evidence really on that poster, and attacks the
   claim, never the person.
7. **The post-release note.** Half a page closing the loop your M17 release audit
   opened: whether anything moved between the audit and the Expo (which would make
   that audit stale), which audit rows the public test touched, and whether the
   release decision still stands.
8. **The self-evaluation.** Two lines, measured against your counts and not the
   mood of the room: one thing your presentation did well, one thing you would do
   differently.
9. **AI Research Ledger rows.** One row per AI-assisted step, each with a named
   verification method. A missing ledger scores Craft 0 and returns the submission.
10. **The version line and the dossier update line.** Open the artifact with
    **Book Milestone 10, version 3**, dated, with a reason a reader could use to
    reconstruct your thinking, and close with one line naming the file or section
    where each finalized piece now lives.

> **A question that often comes up here:** *"Most of my feedback was positive.
> Doesn't that mean my project is solid?"* It means you communicated well, which
> is worth a great deal and worth almost nothing to the truth of your claim. A
> warm room is a fact about your communication. Only the coded, counted questions
> tell you what your evidence still owes.

### 🤝 AI Research Partner

AI is your arm and your research assistant in this sprint, not your brain. Hand it
the mechanical work: re-tallying, spotting an inconsistent code, drafting candidate
wording for a revision, or listing where a peer poster's evidence might fall short.
Keep every judgment for yourself.

**Good to delegate here:**
- Re-counting your coded rows and flagging a **target** that looks mis-coded, so
  you can double-check it against your coding sheet.
- Drafting candidate wordings for a revision you have already decided to make.
- Listing candidate gaps between a peer poster's claim and its shown evidence, so
  you have a longer list to check against the actual poster.

**Never delegate (these are yours to decide):**
- Whether a question doubted your **claims** or your **communication**.
- Whether a revision is **licensed** at all. A tool may propose wording; the
  judgment that reads your counts, your evidence, and your claim boundary together
  is the one you will defend at M16.
- Your honest **self-evaluation** and the final wording of any criticism you put
  in your portfolio. A gap you did not confirm on the poster is one you cannot
  defend.

Whatever you delegate, verify what comes back and log it in your **AI Research
Ledger** below. See `ai_resources/human_responsibility_checklist.md` for the full
never-delegate list.

> 💡 **AI Prompt:** "Here are the questions I was asked at my poster, with
> how I coded each one for type and target (claims vs communication): [paste your
> rows]. First, flag any row where my TARGET looks mis-coded (a legibility
> complaint I filed under 'claims', or an evidence doubt I filed under
> 'communication'), and say why. Then tell me which single question TYPE recurred
> the most across my rows. Do not rewrite my reflection or decide any target for
> me."
>
> **After running, verify (counters *confident fabrication*: a fluent, confident
> "your most-recurring type is X" can name a type your counts do not support):**
> - [ ] Run the scaffold cell first. Check the type your AI calls "most-recurring"
>   against the printed `by_type` counts. If it names a different type than your
>   top row, trust the printout, not the sentence.
> - [ ] For every re-code it proposes, decide yourself whether the question doubted
>   your *evidence* (claims) or your *telling* (communication). That call is a
>   never-delegate one: the tool can flag, only you decide.
> - [ ] Log this use in your AI Research Ledger: task, tool, decision, verification.

```python
# Coded-tally scaffold — replace the EDIT ME rows with YOUR OWN Expo questions.
# The DataFrame `tally` is the single source of truth: every summary below is
# recomputed from it, so there is no separate count to fall out of sync. Keep the
# four columns; use only the type codes in TYPE_CODES and the two target values.

# The valid vocabulary, defined once. Editing a row to an off-list code trips the
# check below instead of silently mis-tallying.
TYPE_CODES   = ["method", "alternative-explanation", "so-what",
                "generalization", "AI-use", "other"]
TARGET_CODES = ["claims", "communication"]

# EDIT ME — replace these placeholder rows with the questions you actually logged.
tally = pd.DataFrame([
    {"question": "Would this hold on a different campus or sample?",      "type": "generalization",          "target": "claims",        "count": 3},
    {"question": "Could something you didn't measure explain the gap?",   "type": "alternative-explanation", "target": "claims",        "count": 2},
    {"question": "I couldn't tell which line was your comparison group.", "type": "other",                   "target": "communication", "count": 4},
    {"question": "How exactly did you estimate that number?",             "type": "method",                  "target": "claims",        "count": 2},
    {"question": "Did you use AI in the analysis, and did you check it?", "type": "AI-use",                  "target": "claims",        "count": 1},
])

# Edit-safe checks: catch a typo'd code before it corrupts the counts.
assert set(tally["type"]).issubset(TYPE_CODES),     f"unknown type code — use only {TYPE_CODES}"
assert set(tally["target"]).issubset(TARGET_CODES), f"unknown target — use only {TARGET_CODES}"

# Everything below is DERIVED from `tally` (the single source of truth).
total     = int(tally["count"].sum())
by_type   = tally.groupby("type")["count"].sum().sort_values(ascending=False)
by_target = tally.groupby("target")["count"].sum().sort_values(ascending=False)
loudest   = tally.loc[tally["count"].idxmax()]
recurring_claims = tally[(tally["target"] == "claims") & (tally["count"] >= 2)]

print(f"✓ Coded {len(tally)} distinct questions, {total} asked in total")
print()
print("By TYPE (weighted by recurrence):")
print(by_type.to_string())
print()
print("By TARGET (claims vs communication):")
print(by_target.to_string())
print()
print(f"Most-repeated single question ({int(loudest['count'])}x, target={loudest['target']}):")
print(f'  "{loudest["question"]}"')
print()
print("CLAIMS questions that recurred (candidates for your defense revision record):")
for _, row in recurring_claims.sort_values("count", ascending=False).iterrows():
    print(f'  {int(row["count"])}x [{row["type"]}]  "{row["question"]}"')
print("  Each one needs a row: what it exposed, the revision, and the reason.")

# Why capture in the moment: memory resamples your afternoon. Expand to one row per
# asked question, then keep a random subset (seeded) to imitate partial recall.
instances = tally.loc[tally.index.repeat(tally["count"])].reset_index(drop=True)
true_claims_share = (instances["target"] == "claims").mean()
print()
print(f"Full sheet: claims share = {true_claims_share:.2f} "
      f"({int((instances['target']=='claims').sum())} of {len(instances)})")
n_recalled = max(3, round(len(instances) * 0.4))  # you recall only part of the afternoon
for i in range(3):
    idx = rng.choice(len(instances), size=n_recalled, replace=False)
    share = (instances.iloc[idx]["target"] == "claims").mean()
    print(f"  Memory-only recall #{i + 1} ({n_recalled} of {len(instances)} "
          f"questions): claims share = {share:.2f}")
print("  A partial recall wobbles from draw to draw; the full sheet does not.")
```

### 🗡️ Red-Team Exchange

The reflection studio's red-team block. Pair with one partner and run two rounds of
attack. Bring in an AI reviewer as a third voice using the **Poster Critic** role
card (`ai_resources/agent_role_cards.md`; full briefing in
`genai_studio/roles/poster_critic.md`). It flags untraceable numbers and
claim-boundary overreach in a pasted poster claim, which is the read you need in both
rounds.

**Round 1: attack your pattern and your revision record (about 6 minutes).**
Your partner reads your pattern sentence, your reflection, and one row of your
defense revision record. They check whether the pattern is read off your **counts**
or off your **mood**, then press the revision row: "show me the count that licenses
this much change, and no more." A revision you promised but never wrote out is not a
revision. You defend or concede, then rewrite so every sentence points at a row in
your tally and every revision appears in the words you will actually use.

**Round 2: attack one portfolio entry (about 6 minutes).**
Take one criticism from your poster-criticism portfolio. Your partner, and the Poster
Critic reading the peer's pasted claim and evidence, test two things: is the gap you
named **really on that poster**, and did you tag the right **axis** (claims or
communication)? A gap neither of you can point to is one you imagined, so cut it. A
criticism that names no axis is a vibe, so fix it.

**Then check the post-release note.** Ask your partner one question out loud: did
anything move between your M17 release audit and the Expo? If something did, that
audit is **stale**, and your note has to say so, along with whether the release
decision still stands.

**What to log.** Write down the single strongest hit from each round and how you
changed the package in response. An attack that changed nothing still gets a line
saying why it did not land. That record is part of what you submit.

> **A question that often comes up here:** *"What if my partner and the Poster Critic
> disagree about a peer's poster?"* Good, because you are the judge, not either of
> them. The Poster Critic never saw the rendered poster; it reads only what you paste,
> so it can invent a flaw or miss a real one. Say which read you found more
> convincing and why, in one sentence, checked against the actual poster in front of
> you.
