## 🎯 Definition of Done

You are done with this sprint when your submission carries every part below,
built from **your own** project.

1. **The repaired reusable package.** A **README** a stranger enters by, whose
   first three lines say what the project asks, what the headline found, and how
   to run everything; the **seeded notebook** (`SEED = 464`) that passes
   **restart-and-run-all**; your **data or its access recipe** with version and
   provenance; the **environment record**; the **licence and permissions note**;
   the **decision log**; and the **dated, frozen manifest**, with your current
   **claim-to-evidence table** frozen beside it. Audited against the **five
   package sins** (hard-coded path, missing seed, by-hand edit, undocumented
   exclusion, stale or unversioned data), with the one you found and fixed named.
2. **Your own cold run, first.** The run you made before the exchange: a fresh
   session, your README only, after a real break, every place you stalled and the
   documentation line you added for it, labeled honestly **"solo proxy; external
   cold run pending."**
3. **The report you wrote, and your response to the one you received.** Attach
   the cold-run report you wrote on your colleague's package (the `ms13` workbench
   helps you write it). Then answer every item on the report you received with one
   of two outcomes: **fixed**, naming the change and where it lives in the
   package, or **logged**, with the reason it stays. Nothing on the report is
   answered with silence.
4. **Gate 1: the author self-reproduction log (blocking).** A dated,
   fresh-runtime, restart-and-run-all rerun of the repaired package in which every
   number and every bound on your poster came back: each poster number beside the
   value the rerun produced, including its interval, marked matched or not
   matched. Your poster is printed, so a mismatch is not hidden: name it, give its
   corrected value, and record the correction in the README and the
   claim-to-evidence table.
5. **Gate 2: the release preflight (blocking).** Taken immediately before you
   freeze the package, each row marked clear, pending, or blocking:
   **permissions** (everything inside your permission determination, nothing
   identifiable that should not be there), **disclosure** (the README carries your
   AI-use disclosure, drafted from the full ledger, naming what you sent to tools
   and anything you did not verify), **reproducibility** (the package complete,
   its numbers matching Gate 1), and **claim boundary** (the README's one-line
   finding inside what your design licenses, with its uncertainty beside it).
6. **The second clean run.** Any change to the package makes the previous run's
   record stale. After your **last** edit, run it once more in a fresh environment,
   following only your written instructions, restart-and-run-all, and record the
   date, the environment, and the result. Reproduce the **uncertainty statement**,
   not only the point estimate. Label it honestly as an author rerun, and keep your
   colleague's run beside it as the independent evidence.

Plus your **Instructor Request Record** and your **action plan's commitments**,
which open the record, your **AI Research Ledger** rows, and the **dossier update
line**. At the studio a partner reads your response record and reruns your
package cold.

> **A question that often comes up here:** *"My colleague's run came back clean.
> Is there anything left to do?"* Yes. Gate 1 matches every number and bound on
> your poster, which your colleague's run never checked. Gate 2 checks what
> arithmetic cannot see. And if you changed anything after their run, even one
> line of the README, their run vouches for a package that no longer exists, so
> your second clean run has to come after your last change.

### 🤝 AI Research Partner

AI is your arm and your research assistant in this sprint, not your brain. This week
it works as a **documentation critic**, never the author of your decisions. A good
use looks like this: you draft a README section, then ask a tool what a stranger
would still need, or you ask it to sort the items on the report you received so
you can check the sort against the report itself. A bad use is letting it write
your responses, decide whether a gate row is clear, or tell you the package
reproduces.

**Good to delegate here:**
- Sorting and de-duplicating the items on the cold-run report you received, so
  repeated items collapse into one row for you to confirm.
- Red-teaming the README's one-line finding, which **you** already wrote, for any
  word that claims more than your evidence supports.
- Drafting README sections and the environment record, then checking each against
  a fresh run.

**Never delegate (these are yours to decide):**
- The response to each report item: fixed or logged, each with a reason you own.
- Whether a gate row is clear, pending, or blocking, and whether a number
  reproduced. Only your run settles that.
- Which sources are real, and the words you put your name on.

**Commit first, then ask.** Draft your own sentence before you open a tool, so its
output has something of yours to disagree with. A cleaner sentence is not a bigger
finding: keep only edits that leave the claim its size, and reject every edit that
quietly upgrades a verb or drops an "in this sample". Verify what comes back, and log
it in the **AI Research Ledger** below. See
`ai_resources/human_responsibility_checklist.md` for the full never-delegate list.

### 🛠️ Run It Live: check the four floors your package must clear

Run the scaffold below on its default inputs first to see what it does. Then edit the
**four inputs** at the top with the pieces of *your own* project, and rerun. Everything
downstream is derived from those four edits, so there is one source of truth for every
number: change an input once and the whole readout stays consistent. These are floors,
not ceilings. Clearing them is necessary, never sufficient, and the rest of the
milestone is where the grade lives.

```python
# --- Package self-check — edit ONLY the four inputs below, then rerun.
# Everything after them is DERIVED, so there is one source of truth per number.

# EDIT ME (1) — the cold-run report you RECEIVED: one row per item your colleague
# raised, with what it touches, your response, and where the fix lives or why the
# item stays. Repeated items collapse into one row. An item you have not answered
# yet keeps response = None, which the check below will surface.
report_items = [
    {"item": "README never named the run order",            "touches": "instructions", "response": "fixed",  "note": "README, 'How to run', the step order"},
    {"item": "load cell pointed at the author's own drive", "touches": "package",      "response": "fixed",  "note": "cell 2 now loads through load_course_data()"},
    {"item": "environment record left out scipy",           "touches": "package",      "response": "logged", "note": "scipy is never imported on the headline path"},
    {"item": "the interval did not come back on the rerun", "touches": "evidence",     "response": None,     "note": ""},
]

# EDIT ME (2) — the per-unit numbers BEHIND your headline finding: one value per
# independent unit (a respondent, a site, a trial). The defaults let the cell run.
headline_values = [4.2, 3.1, 5.0, 2.8, 4.6, 3.7, 5.3, 2.2, 4.0, 3.9,
                   4.8, 2.5, 3.3, 5.1, 4.4, 3.0, 4.7, 2.9, 3.6, 4.1]

# EDIT ME (3) — a few key lines from your package, exactly as they appear in it.
package_lines = [
    "SEED = 464",
    "rng = np.random.default_rng(SEED)",
    'df = load_course_data("your_data.csv")   # provenance + license in the ledger',
    'df = df[df["outcome"].notna()]   # reason: pre-registered complete cases; in the decision log',
]

# EDIT ME (4) — the ORDER of your last two events, as step numbers in your own work
# log. No dates needed: what matters is which came last.
last_package_edit_step = 7   # the step at which you last changed ANYTHING in the package
last_clean_run_step    = 6   # the step at which you last ran restart-and-run-all

# ---- everything below is DERIVED from the four edits above ----

RESPONSES = ["fixed", "logged"]

# FLOOR 1 — the response record: report items in, responses out, and they must match.
unanswered = [r for r in report_items if r["response"] not in RESPONSES]
answered = len(report_items) - len(unanswered)
print(f"FLOOR 1  response record: {len(report_items)} report items, {answered} answered")
for v in RESPONSES:
    rows = [r for r in report_items if r["response"] == v]
    print(f"         {v:7s}: {len(rows)}  ({', '.join(r['touches'] for r in rows) or 'none'})")
bare = [r for r in report_items if r["response"] in RESPONSES and not r["note"].strip()]
if unanswered:
    print("         -> UNANSWERED items (nothing on the report is answered with silence):")
    for r in unanswered:
        print(f"            {r['item']}")
else:
    print("         -> every item answered. Now check that each one names where the")
    print("            fix lives, or a reason a stranger would accept for keeping it.")
if bare:
    print("         -> answered with no location or reason (a bare verdict is a dodge):")
    for r in bare:
        print(f"            {r['item']}")

# FLOOR 2 — the headline's uncertainty: a seeded 95% bootstrap interval on the mean.
# This is the statement your second clean run has to reproduce, not just the estimate.
vals = np.asarray(headline_values, dtype=float)
n = len(vals)
boot = np.array([rng.choice(vals, size=n, replace=True).mean() for _ in range(2000)])
lo, hi = np.percentile(boot, [2.5, 97.5])
print(f"\nFLOOR 2  headline (n = {n}): mean = {vals.mean():.2f}   "
      f"95% interval [{lo:.2f}, {hi:.2f}]   width {hi - lo:.2f}")
print("         Report the interval in the SAME sentence as the estimate, and")
print("         reproduce BOTH in your second clean run. A wider interval is a")
print("         smaller claim, not a failed one.")

# FLOOR 3 — the package floor: scan the pasted lines for the five package sins.
low = [ln.lower() for ln in package_lines]
joined = "\n".join(low)
sins = []
if any(("/users/" in ln) or ("/home/" in ln) for ln in low):
    sins.append("hard-coded path")
if not any(k in joined for k in ("seed", "random_state", "default_rng")):
    sins.append("missing seed")
if any((".loc[" in ln and "=" in ln) or ("by hand" in ln) for ln in low):
    sins.append("by-hand edit")
filtered = [ln for ln in low if (".drop(" in ln) or ("df[df[" in ln.replace(" ", ""))]
if filtered and not any("reason" in ln for ln in low):
    sins.append("undocumented exclusion")
loads_csv = any("read_csv" in ln for ln in low)
shareable = any(("http" in ln) or ("load_course_data" in ln) for ln in low)
if loads_csv and not shareable and not any(("version" in ln) or ("license" in ln) for ln in low):
    sins.append("stale/unversioned data")
print(f"\nFLOOR 3  package sins flagged: {sins if sins else 'none (runnable, not yet proven correct)'}")
print("         A zero-flag package is RUNNABLE, never proven CORRECT. Name the one")
print("         sin you found and fixed in YOUR package; the scan is a prompt, not the audit.")

# FLOOR 4 — the stale-run check: does the run that vouches for the package come AFTER
# the last change to it? If not, the record vouches for a package that no longer exists.
stale = last_clean_run_step < last_package_edit_step
print(f"\nFLOOR 4  stale-run check: last edit at step {last_package_edit_step}, "
      f"last clean run at step {last_clean_run_step}")
if stale:
    print("         -> STALE: the record vouches for a package that no longer exists.")
    print("            Rerun restart-and-run-all in a fresh environment AFTER your last")
    print("            change, then record the date, the environment, and the result.")
else:
    print("         -> current: the run record comes after the last change.")
    print("            Any further edit stales it again, so rerun before you submit.")
```

### 🗡️ Red-Team Exchange

The studio's red-team block. Pair with one partner and run two rounds, with an AI
reviewer as a third voice. The role for this milestone is the **Reproducibility
Auditor**, which reads your package the way a cold replicator would and returns a
**reproduction gap list** and a **claim-to-output trace** (quick card in
`ai_resources/agent_role_cards.md`; full briefing in
`genai_studio/roles/reproducibility_auditor.md`). It proposes; as configured here it
never runs your package.

**First, the Reproducibility Auditor pass (before the rounds).** Paste the cleared
contents of your repaired package: the README, the run order, and each key number
with one line on where it comes from. Every gap it names is a **suspect you test
yourself** against your own run and the M17 brief, not a verdict, and every missing
instruction it names is one **you** must write.

**Round 1: the response-record read (about 15 minutes).** Your partner reads the
cold-run report you received beside your response record and does two things. They
count: items in, responses out, and the numbers must match. Then they pick your
weakest *logged* item and press its reason: "why does this stay, and would a stranger
accept that reason?" You defend or concede, then rewrite the reason, or fix the item.

**Round 2: rerun the package cold (about 15 minutes).** Your partner opens your package
on a fresh kernel and runs **restart-and-run-all**, from your written instructions only
and with no help from you. They log whether your headline number **and its uncertainty
statement** come back, and where, if anywhere, it broke. A package that runs for you and
not for them is a finding you fix before you submit, not a failure to hide. Remember the
closing rule: any fix you make afterward stales this run, so your **second clean run**
must come after your last change.

**Mind the correlated-error trap.** If your AI polished your README and the
Reproducibility Auditor calls the package sound, that agreement is not confirmation
that it reproduces. Neither tool ran your package. Only a rerun settles whether the
number comes back.

**What to log.** Write down the strongest hit you gave and the strongest you got back,
the response you rewrote, and the one rerun that decided any disagreement. A challenge
that changed nothing still earns a line saying why it did not land. That record is part
of what you submit.

> **A question that often comes up here:** *"What if the Reproducibility Auditor and my
> partner's rerun disagree?"* Your run settles it, not either voice. The Auditor read
> only your text and can miss a broken step or invent a missing file; your partner ran
> the package but cannot see your ledger. Say in one sentence which reading your own
> printout and claim-to-evidence table support, and why. That adjudication is stronger
> proof you understand your package than agreeing with a tool would be.
