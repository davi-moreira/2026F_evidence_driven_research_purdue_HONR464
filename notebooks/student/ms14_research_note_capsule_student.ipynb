## 🎯 Definition of Done

You are done with this sprint when your submission carries all six parts below,
built from **your own** project.

1. **The two streams, gathered into one list.** One numbered table holding every
   item that arrived from outside, each row tagged with its **source** (Expo
   criticism or peer cold run), the **item** in one sentence, and the part of the
   project it touches: the claim, the evidence, the writing, or the package. A
   question asked five times is one item with its recurrence noted, not five items.
2. **The adjudication record.** Dated, and it opens the note. Every row on the
   inbound list gets exactly one of three verdicts, each with a reason you own and
   could defend aloud: **accept** (the criticism was right, and here is what changed
   and where a reader sees it), **rebut** (it does not hold, and here is the evidence
   why), or **bound** (it holds in part, and the honest answer is a narrower claim,
   stated inside the claim sentence itself). A reader should be able to count your
   rows, count your verdicts, and see the two numbers match.
3. **Research note v1.** The same five-move spine as your first draft: the question,
   the method in one paragraph, the results with uncertainty in the same eye-span,
   the bounded headline claim carrying its compass **kind** and **reach**, then
   limitations and retrievable references. Revised wherever an accepted or bounded
   verdict demands it, grown only where public questioning showed a reader needed
   more, findings ordered by how much evidence backs each, the exploratory labeled
   last and never leading.
4. **The repaired reusable package.** Five parts: a **README** a stranger enters by,
   the **seeded notebook** (`SEED = 464`) that passes **restart-and-run-all**, your
   **data or its access recipe** with version and provenance, the **environment
   record**, and the **licence and permissions note** with your decision log. Audited
   against the **five package sins** (hard-coded path, missing seed, by-hand edit,
   undocumented exclusion, stale or unversioned data), with the one you found and
   fixed named. Every question your cold-runner had to ask is now answered in writing
   or logged with the reason it stays.
5. **The second clean run.** Any change to the package makes the previous run's
   record stale. After your **last** edit, run it once more in a fresh environment,
   following only your written instructions, restart-and-run-all, and record the
   date, the environment, and the result. Reproduce the **uncertainty statement**,
   not only the point estimate. Label it honestly as a clean rerun by the author, and
   keep your peer's M16 run beside it as the independent evidence.
6. **The updated claim–evidence table and AI disclosure.** Every sentence still
   traces to a row (claim · evidence · verification · boundary), and when
   adjudication moved a claim its row moved with it. Redraft your disclosure from the
   full ledger, never from memory: it now covers the poster weeks and the peer
   exchange, includes what you *sent* to tools, and names anything you did not verify.

Plus your **AI Research Ledger** rows and the **dossier update line**. At the studio
a partner table-reads your adjudication record and reruns your package cold.

> **A question that often comes up here:** *"I already drafted this note at M10. Isn't

### 🤝 AI Research Partner

AI is your arm and your research assistant in this sprint, not your brain. This week
it works as a **writing critic**, never a ghostwriter. A good use looks like this: you
draft a paragraph, then ask a tool to red-team it, or you ask it to sort your inbound
list so you can check the sort against the two reports it came from. A bad use is
letting it write your discussion, decide which finding leads, or hand you a verdict.

**Good to delegate here:**
- Sorting and de-duplicating the inbound list, so recurring items collapse into one
  row with a recurrence count for you to confirm.
- Red-teaming a discussion paragraph **you** already wrote, for any sentence that
  claims more than your evidence supports.
- Drafting README sections, then ranking your limitations so you can check the order
  against your own printed interval.

**Never delegate (these are yours to decide):**
- The verdict. Accept, rebut, and bound are yours, each with a reason you own.
- Which claim is your headline and how big it is allowed to be.
- Which sources are real, and the words you put your name on.

**Commit first, then ask.** Draft your own sentence before you open a tool, so its
output has something of yours to disagree with. A cleaner sentence is not a bigger
finding: keep only edits that leave the claim its size, and reject every edit that
quietly upgrades a verb or drops an "in this sample". Verify what comes back, and log
it in the **AI Research Ledger** below. See
`ai_resources/human_responsibility_checklist.md` for the full never-delegate list.

### 🛠️ Run It Live: check the four floors your note v1 and package must clear

Run the scaffold below on its default inputs first to see what it does. Then edit the
**four inputs** at the top with the pieces of *your own* project, and rerun. Everything
downstream is derived from those four edits, so there is one source of truth for every
number: change an input once and the whole readout stays consistent. These are floors,
not ceilings. Clearing them is necessary, never sufficient, and the rest of the
milestone is where the grade lives.

```python
# --- Note & package self-check — edit ONLY the four inputs below, then rerun.
# Everything after them is DERIVED, so there is one source of truth per number.

# EDIT ME (1) — your inbound list: every item that arrived from OUTSIDE, one row
# each, with its source, what it touches, and the verdict YOU decided. Recurring
# items collapse into one row; put the count in the item text. A row still under
# consideration keeps verdict = None, which the check below will surface.
inbound = [
    {"source": "expo",    "item": "generalization doubt, asked 3x",        "touches": "claim",    "verdict": "bound"},
    {"source": "expo",    "item": "comparison line unreadable on figure",  "touches": "writing",  "verdict": "accept"},
    {"source": "expo",    "item": "reading a warm reception as support",   "touches": "claim",    "verdict": "rebut"},
    {"source": "coldrun", "item": "README never named the run order",      "touches": "package",  "verdict": "accept"},
    {"source": "coldrun", "item": "clustering ignored in the interval",    "touches": "evidence", "verdict": None},
]

# EDIT ME (2) — the per-unit numbers BEHIND your headline finding: one value per
# independent unit (a respondent, a site, a trial). The defaults let the cell run.
headline_values = [4.2, 3.1, 5.0, 2.8, 4.6, 3.7, 5.3, 2.2, 4.0, 3.9,
                   4.8, 2.5, 3.3, 5.1, 4.4, 3.0, 4.7, 2.9, 3.6, 4.1]

# EDIT ME (3) — a few key lines from your package, exactly as they appear in it.
package_lines = [
    "SEED = 464",
    "rng = np.random.default_rng(SEED)",
    'df = load_course_data("your_data.csv")   # provenance + license in the ledger',
    'df = df[df["outcome"].notna()]   # reason: pre-registered complete cases; in the decision log',
]

# EDIT ME (4) — the ORDER of your last two events, as step numbers in your own work
# log. No dates needed: what matters is which came last.
last_package_edit_step = 7   # the step at which you last changed ANYTHING in the package
last_clean_run_step    = 6   # the step at which you last ran restart-and-run-all

# ---- everything below is DERIVED from the four edits above ----

VERDICTS = ["accept", "rebut", "bound"]

# FLOOR 1 — adjudication completeness: rows in, verdicts out, and they must match.
undecided = [r for r in inbound if r["verdict"] not in VERDICTS]
decided = len(inbound) - len(undecided)
print(f"FLOOR 1  adjudication: {len(inbound)} inbound items, {decided} verdicts recorded")
for v in VERDICTS:
    rows = [r for r in inbound if r["verdict"] == v]
    print(f"         {v:7s}: {len(rows)}  ({', '.join(r['touches'] for r in rows) or 'none'})")
if undecided:
    print("         -> UNDECIDED items (each one caps Evidence integrity at Developing):")
    for r in undecided:
        print(f"            [{r['source']}] {r['item']}")
else:
    print("         -> every item decided. Now check each verdict carries a REASON")
    print("            you own; a verdict with no reason is a dodge, not a decision.")

# FLOOR 2 — the headline's uncertainty: a seeded 95% bootstrap interval on the mean.
# This is the statement your second clean run has to reproduce, not just the estimate.
vals = np.asarray(headline_values, dtype=float)
n = len(vals)
boot = np.array([rng.choice(vals, size=n, replace=True).mean() for _ in range(2000)])
lo, hi = np.percentile(boot, [2.5, 97.5])
print(f"\nFLOOR 2  headline (n = {n}): mean = {vals.mean():.2f}   "
      f"95% interval [{lo:.2f}, {hi:.2f}]   width {hi - lo:.2f}")
print("         Report the interval in the SAME sentence as the estimate, and")
print("         reproduce BOTH in your second clean run. A wider interval is a")
print("         smaller claim, not a failed one.")

# FLOOR 3 — the package floor: scan the pasted lines for the five package sins.
low = [ln.lower() for ln in package_lines]
joined = "\n".join(low)
sins = []
if any(("/users/" in ln) or ("/home/" in ln) for ln in low):
    sins.append("hard-coded path")
if not any(k in joined for k in ("seed", "random_state", "default_rng")):
    sins.append("missing seed")
if any((".loc[" in ln and "=" in ln) or ("by hand" in ln) for ln in low):
    sins.append("by-hand edit")
filtered = [ln for ln in low if (".drop(" in ln) or ("df[df[" in ln.replace(" ", ""))]
if filtered and not any("reason" in ln for ln in low):
    sins.append("undocumented exclusion")
loads_csv = any("read_csv" in ln for ln in low)
shareable = any(("http" in ln) or ("load_course_data" in ln) for ln in low)
if loads_csv and not shareable and not any(("version" in ln) or ("license" in ln) for ln in low):
    sins.append("stale/unversioned data")
print(f"\nFLOOR 3  package sins flagged: {sins if sins else 'none (runnable, not yet proven correct)'}")
print("         A zero-flag package is RUNNABLE, never proven CORRECT. Name the one")
print("         sin you found and fixed in YOUR package; the scan is a prompt, not the audit.")

# FLOOR 4 — the stale-run check: does the run that vouches for the package come AFTER
# the last change to it? If not, the record vouches for a package that no longer exists.
stale = last_clean_run_step < last_package_edit_step
print(f"\nFLOOR 4  stale-run check: last edit at step {last_package_edit_step}, "
      f"last clean run at step {last_clean_run_step}")
if stale:
    print("         -> STALE: the record vouches for a package that no longer exists.")
    print("            Rerun restart-and-run-all in a fresh environment AFTER your last")
    print("            change, then record the date, the environment, and the result.")
else:
    print("         -> current: the run record comes after the last change.")
    print("            Any further edit stales it again, so rerun before you submit.")
```

### 🗡️ Red-Team Exchange

The studio's red-team block. Pair with one partner and run two rounds, with an AI
reviewer as a third voice. The role for this milestone is the **Research Note
Reviewer**, which reads the *written genre*: whether every claim carries its
uncertainty and traces to your evidence (quick card in
`ai_resources/agent_role_cards.md`; full briefing in
`genai_studio/roles/research_note_reviewer.md`). It reviews the draft; it never writes
it, and it cannot run your package.

**First, the Research Note Reviewer pass (before the rounds).** Paste your full draft
with your compass classification and, for each key number, one line on where it comes
from. It returns a structure check, an uncertainty-and-limitations check, and a
traceability check. Every flag is a **suspect you test yourself** against the M16
brief, not a verdict, and every missing claim or limitation it names is one **you**
must write.

**Round 1: the adjudication table read (about 15 minutes).** Your partner reads your
inbound list beside your adjudication record and does two things. They count: rows in,
verdicts out, and the numbers must match. Then they pick your weakest verdict and press
its reason: "why does the evidence license *bound* here rather than *accept*, and where
does a reader see the change?" You defend or concede, then rewrite the reason so it is
one you could defend under questioning.

**Round 2: rerun the package cold (about 15 minutes).** Your partner opens your package
on a fresh kernel and runs **restart-and-run-all**, from your written instructions only
and with no help from you. They log whether your headline number **and its uncertainty
statement** come back, and where, if anywhere, it broke. A package that runs for you and
not for them is a finding you fix before you submit, not a failure to hide. Remember the
closing rule: any fix you make afterward stales this run, so your **second clean run**
must come after your last change.

**Mind the correlated-error trap.** If your AI polished your prose and the Research Note
Reviewer calls it sound, that agreement is not confirmation your claims are bounded.
Neither tool ran your package. Only the partner rerun settles whether the number
reproduces.

**What to log.** Write down the strongest hit you gave and the strongest you got back,
the verdict you rewrote, and the one rerun that decided any disagreement. A challenge
that changed nothing still earns a line saying why it did not land. That record is part
of what you submit.

> **A question that often comes up here:** *"What if the Research Note Reviewer and my
> partner disagree about whether a claim is bounded?"* Your evidence settles it, not
> either of them. The Reviewer read only your text and can miss a naked claim or invent
> a structural worry; your partner read the page but not your ledger. Say in one
> sentence which read your own printout and claim ledger support, and why. That
> adjudication is stronger proof you understand your note than agreeing with a tool
> would be.
