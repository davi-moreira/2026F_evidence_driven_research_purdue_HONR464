costs than it did in 2019, relative to overall prices.
**Units.** Menu items at restaurants within walking distance.
**Outcome.** The listed price. **Conditions.** 2019 versus 2025.
**Kind.** Descriptive: the words ask "how much more," not "what

retrievable for both years, and only those. **Claim boundary.**

the overall index. Cannot defend: that nearby restaurants as a
group did (that claim needs a sampling story this design does
not have), and not that the wage increase caused anything.
