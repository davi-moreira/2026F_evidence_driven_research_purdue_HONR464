## A worked example

**Working rules.** Delegate freely: locating candidate series,
drafting summaries. Delegate then verify: any number, any
citation, any code. Never delegate: the measure choice, the base
year, the claim and its uncertainty.

basket · responsible researcher: the reader's own name.

not people; defensible to attempt.
