## Choosing your format

The goal of this studio is an artifact ready for publication or
presentation: a manuscript a venue could receive, a talk a room
could hear, a poster a conference could hang. Start with a **venue
contract**: your written record of what the audience will receive
and what the venue allows. Name the audience, the medium, the
length or space limit, the required elements, the accessibility
needs, the submission rule, and the question channel. One line each
is enough. Example: a five-minute recorded talk, six slides,
captions required, no live questions.

Every format carries the same research core: the bounded claim, the
evidence that earns it, uncertainty and limitations, traceable
numbers and citations, and whatever disclosure or permission
information the audience needs. The format only decides where those
pieces live. Choose the row that fits your venue, work the lessons it
names, and when you borrow a technique from another row, name the
technique rather than claiming the whole lesson transfers.

| If your venue asks for | build it with | then test and rehearse |
|---|---|---|
| a paper or research note | your Studio 9 note, developed to the venue contract: sections, length, references, and the figure rules for any figure | the audit on the manuscript; Difficult Questions in writing |
| a seminar or conference talk | Research Pitches' spine, expanded section by section to the venue's length; slides follow the poster lesson's figure rules | the audit on the deck; Difficult Questions, spoken |
| a short pitch | Research Pitches | Difficult Questions, spoken |
| a poster | Research Posters | Poster Criticism, then Difficult Questions |
| another format | the venue contract plus the nearest row, naming what transfers | Difficult Questions, in the venue's own mode |

The poster lessons are this book's fully worked visual example: one
surface, no narrator, every number exposed. The paper route runs on
your Studio 9 note; the seminar route runs on the pitches lesson's
spine. Difficult Questions is for everyone. If your venue has no
live questions, the rehearsal still happens: have a peer or an AI
reviewer send you hard questions in writing, and record defend,
concede, or do-not-know for each.
