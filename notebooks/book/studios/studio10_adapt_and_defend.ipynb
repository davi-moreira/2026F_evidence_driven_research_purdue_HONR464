## Choosing your format

Start with a **venue contract**: your written record of what the
audience will receive and what the venue allows. Name the audience,
the medium, the length or space limit, the required elements, the
accessibility needs, the submission rule, and the question channel.
One line each is enough. Example: a five-minute recorded talk, six
slides, captions required, no live questions.

Every format carries the same research core: the bounded claim, the
evidence that earns it, uncertainty and limitations, traceable
numbers and citations, and whatever disclosure or permission
information the audience needs. The format only decides where those
pieces live. Choose the row that fits your venue, work the lessons it
names, and when you borrow a technique from another row, name the
technique rather than claiming the whole lesson transfers.

everyone. If your venue has no live questions, the rehearsal still
happens: have a peer or an AI reviewer send you hard questions in
writing, and record defend, concede, or do-not-know for each.
