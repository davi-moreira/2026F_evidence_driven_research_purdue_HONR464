## A worked example

Research Contract v0 for the restaurant-prices project, compact.

**Objective.** Describe local menu-price growth against overall
prices, 2019 to 2025. **Target estimand.** The ratio of the two
growth rates. **Population.** Restaurants near campus whose
menus are retrievable for both years. **Data strategy.** Archived
menus plus the agency's published index. **Provisional
operationalization (marked for revision).** The listed price of
a standard dinner entree. **Warrant.** A descriptive comparison;
no causal claim. **Answer strategy.** Mean growth ratio across
the menu panel. **Uncertainty.** A coverage caveat, not an
interval: archived menus survive selectively. **Permission
status.** Cleared: prices, not people.

**Diagnosis note.** On simulated menus, price-dependent archiving
biased the growth estimate upward: pricier restaurants archive
more, and no restriction inside these data removes that
selection. **Redesign.** Narrow the estimand to the retrievable
panel and say so in the claim; the coverage caveat stays.
**The honest call.** Run it on the panel, bounded, and name what
more data of the same kind could never fix.
