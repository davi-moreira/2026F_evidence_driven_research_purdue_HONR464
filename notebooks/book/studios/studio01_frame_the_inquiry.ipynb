index."

**The stretch.** The brainstorm returned five directions; two
introduced NEW elements (delivery fees; a second town). The
reader kept the delivery-fee direction as a fallback, rejected
the second town (no access), and wrote both reasons down.

**The choice.** Chosen problem: how menu prices near campus
moved against overall prices, 2019 to 2025. Importance: every
budget feels it. Feasibility: archived menus plus a public
index. Contribution: nobody has looked at this campus. Expected
answer: menus outran the index; a flat result would surprise
the reader most.
