## Prompts de IA recomendados

Comprométete primero con tu propia versión y después delega. Cada prompt es una tarea
verificable, no una petición de veredicto.

Espera dos o tres rondas en cada uno. La primera compresión suele ser demasiado lisa, así que
señalas el matiz que dejó caer y preguntas de nuevo. Las herramientas agénticas iterarán hacia
lo que suene mejor según sus propios criterios, que es exactamente la presión que ensancha las
afirmaciones, así que tu frase de límite tiene que ser algo que sostienes y no algo que
delegas.

**Localiza la jerga y tradúcela.**

```text
Here is one sentence from my poster: "[paste it]". My headline claim and its
boundary are: "[paste them]". List every word a curious stranger with no training
in my field could not decode. Then give two plain rewrites a first-year in any
major would follow, each keeping my claim exactly as narrow as it is now.
```

Después de ejecutarlo, verifica: lee cada reescritura junto a tu límite y rechaza cualquiera
que haya dejado caer un matiz como "reservado para prueba" o que haya convertido "señaló" en
"predice". Contrarresta el *cambio silencioso de alcance* (una reescritura "sencilla" que
ensancha la afirmación en silencio mientras la simplifica).

✍️ **Tu registro.** Haz doble clic en esta celda y anota: qué devolvió la IA (una o dos líneas), qué verificaste y cómo, y tu fila del ledger.

**Enumera las palabras que inflan mi afirmación.**

```text
Here is my 30-second hook: "[paste it]". Here is my poster's boundary sentence:
"[paste it]". Play a hurried visitor. List every word in the hook that says
something bigger than the boundary allows. Rewrite nothing; just name the words.
```

Después de ejecutarlo, verifica: si no señala nada y llama "fuerte" al gancho, insiste y exige
el peor infractor. Contrarresta la *adulación complaciente* (el elogio revisa tu ego, no tus
palabras).

✍️ **Tu registro.** Haz doble clic en esta celda y anota: qué devolvió la IA (una o dos líneas), qué verificaste y cómo, y tu fila del ledger.

**Haz red team (revisión adversaria) al pitch como un desconocido apurado.**

```text
Play a skeptical conference visitor with thirty seconds of patience, from outside
my field. Here is my full pitch: "[paste it]". Name the exact sentence where you
lose interest or stop believing me, and the single place I promise more than I show.
```

Después de ejecutarlo, verifica: contrasta su respuesta con un ensayo real con alguien de tu
confianza, y pregúntale "¿qué es lo único que dejaste fuera?". Contrarresta la *ilusión de
exhaustividad* (un pulido "está claro" que se salta el momento en que un desconocido de verdad
se distrae).

✍️ **Tu registro.** Haz doble clic en esta celda y anota: qué devolvió la IA (una o dos líneas), qué verificaste y cómo, y tu fila del ledger.

> **No delegues.**
>
> Tres decisiones se quedan contigo. Tú decides **qué afirmación respalda realmente tu
> evidencia**, **dónde tiene que detenerse la afirmación** (la frase de límite exacta) y **si una
> frase comprimida sigue diciendo lo que dice el póster**. Una herramienta puede recortar palabras
> y señalar jerga. No puede decidir que la frase más corta sigue siendo verdadera, y no va a estar
> parada frente al tablero cuando un desconocido pregunte. Tú dices la frase. Tú respondes por
> ella.
