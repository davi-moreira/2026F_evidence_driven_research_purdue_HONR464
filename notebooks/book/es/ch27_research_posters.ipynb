**Enumera las preguntas del escéptico y después asígnalas a auditorías.**

```text
Act as a hostile poster reviewer. My headline figure is a bar chart of mean
checkout completion rate for two page versions, y-axis 70 to 76, red versus green
bars, no interval shown, titled "Version B BOOSTS completion." Generate the six
hardest questions you would ask, then name the one this figure cannot answer at all.
```
Después de ejecutarlo, verifica: asigna cada pregunta a una auditoría (límite de la
afirmación, honestidad de la figura, incertidumbre, accesibilidad); una auditoría sin
ninguna pregunta es el hueco, y más vale que aparezca el eje truncado. Contrarresta la
*ilusión de exhaustividad* (una lista larga y ordenada que se salta en silencio la falla que
más importa).

✍️ **Tu registro.** Haz doble clic en esta celda y anota: qué devolvió la IA (una o dos líneas), qué verificaste y cómo, y tu fila del ledger.

**Haz red team (revisión adversaria) al verbo del titular.**

```text

associational / causal]". Act as a skeptical reviewer and find every word that
claims more than that position licenses. List the exact words. Do not soften them.
```

Después de ejecutarlo, verifica: si elogia el titular o no encuentra nada, insiste y exige la
única peor palabra; después contrasta cada palabra señalada con tu intervalo. Contrarresta la
*adulación complaciente* (elogios que revisan tu valentía, no tu evidencia).
