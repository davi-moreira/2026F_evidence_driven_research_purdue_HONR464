## Prompts de IA recomendados

Comprométete primero con tu propia respuesta y después delega. Escribir con una
herramienta de IA es un **bucle de IA**, no una sola petición: el ciclo prompt →
respuesta → interrogación → refinamiento → nueva ejecución. Envías el prompt, lees la
salida, empujas contra la parte que está mal y la corres de nuevo. Mantén el bucle
apuntado a tu propio texto. En el momento en que empieza a producir oraciones que tú no
pensaste y no puedes comprobar, ya no estás editando: te están editando. Cada prompt de
abajo es una tarea verificable, no una petición de veredicto.

**Localiza fuentes reales para el vacío.**

```text
Act as a literature assistant in service operations. List peer-reviewed or
official sources that measure customer wait times or the effect of pre-ordering
on queues, in a table with title, authors, venue, year, and one-line finding.
Only include work you are confident exists; mark anything uncertain. Do not
write my gap paragraph.
```

Después de ejecutarlo, verifica: recupera por tu cuenta cada fuente antes de que entre
en la nota; una fuente que no puedes abrir en segundos está fabricada hasta que se
demuestre lo contrario. Contrarresta la *fabricación confiada* (una cita inventada llega
con la misma seguridad que una real).

✍️ **Tu registro.** Haz doble clic en esta celda y anota: qué devolvió la IA (una o dos líneas), qué verificaste y cómo, y tu fila del ledger.

**Enumera cada exceso para verificarlo.**

```text
Here is my discussion paragraph and the cell that printed my median wait-time
drop with its interval: [paste both]. List every sentence that claims more than
the printout supports: a bigger number, a wider setting, or "faster for everyone"
where I measured one simulated demand pattern. List them; do not rewrite them.
```

Después de ejecutarlo, verifica: contrasta cada oración señalada con tu impresión de
resultados y quédate solo con las ediciones que encogen una afirmación hasta que quepa
en la evidencia. Contrarresta el *cambio silencioso de alcance* (una oración más afilada
que asciende "sobre este patrón de llegadas" a "para todo el mundo" ya respondió, en
silencio, una pregunta distinta).

✍️ **Tu registro.** Haz doble clic en esta celda y anota: qué devolvió la IA (una o dos líneas), qué verificaste y cómo, y tu fila del ledger.

**Hazle red team (revisión adversaria) a la cápsula.**

```text
Here are the key lines of my reproducibility capsule: [paste them]. Playing a cold
replicator who has only these lines, list every input, version, or by-hand step you
would need to rerun my number and might not find here. Then name one problem this
line-scan can never catch that only a person running it cold would hit.
```

Después de ejecutarlo, verifica: contrasta cada hueco con lo que de verdad puedes ver
que falta, y confirma que el problema "que no se puede detectar" es real releyendo tus
propias líneas. Contrarresta la *ilusión de exhaustividad* (una cápsula puede verse
completa y aun así omitir el único insumo que una persona desconocida necesita).

✍️ **Tu registro.** Haz doble clic en esta celda y anota: qué devolvió la IA (una o dos líneas), qué verificaste y cómo, y tu fila del ledger.
