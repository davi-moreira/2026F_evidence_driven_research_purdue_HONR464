```text

defensible specifications I did NOT list, and for each say which direction you expect
```

Después de ejecutarlo, verifica: contrasta cada sugerencia con las cuatro palancas y agrega

Después de ejecutarlo, verifica: corre contra tus propios datos la verificación que nombre;
