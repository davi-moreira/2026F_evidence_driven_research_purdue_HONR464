## Prompts de IA recomendados

Comprométete primero con tu propia respuesta y después delega. Empaquetar es un **bucle
de IA** que correrás más de una vez: el ciclo prompt → respuesta → interrogación →
refinamiento → nueva ejecución. Preguntas, arreglas lo que vuelve, vuelves a ejecutar la
cápsula en frío, y esa nueva ejecución saca a la superficie el siguiente hueco. Algunas
herramientas ya corren ese bucle solas, editando archivos y reejecutando hasta que nada
dé error. Un bucle que termina sin errores demostró que el código se ejecuta. No
demostró que los números estén bien, y solo tú puedes notar la diferencia. Cada prompt
es una tarea verificable, no un veredicto.

**Localiza la herramienta estándar.**

```text
Act as a reproducibility assistant. Name the standard file and format for recording
the exact package versions a notebook needs in order to rerun, and cite the official
documentation. Only name tools you are confident exist.
```

Después de ejecutarlo, verifica: abre la documentación oficial y confirma que el archivo
y su sintaxis existen. Contrarresta la *fabricación confiada* (un nombre de herramienta
inventado llega con la misma seguridad que uno real).

✍️ **Tu registro.** Haz doble clic en esta celda y anota: qué devolvió la IA (una o dos líneas), qué verificaste y cómo, y tu fila del ledger.

**Enumera para que puedas verificar (la persona que replica en frío).**

```text
Here are the key lines of my capsule: [paste]. Playing a replicator who has only
these lines and none of my memory, list every input, file, or by-hand step you would
need to rerun my headline number and might not find here.
```

Después de ejecutarlo, verifica: contrasta cada hueco nombrado con lo que tu propia
ejecución de `audit_capsule` de verdad marcó, y descarta cualquier hueco que no
corresponda a ninguna línea realmente faltante. Contrarresta la *ilusión de
exhaustividad* (una cápsula ordenada que se ve entera mientras falta un insumo).

✍️ **Tu registro.** Haz doble clic en esta celda y anota: qué devolvió la IA (una o dos líneas), qué verificaste y cómo, y tu fila del ledger.

**Hazle red team (revisión adversaria) a la afirmación.**

```text
My claim is: "a stranger can rerun this capsule and get my number." Act as a hostile
replication reviewer and name every way it could fail on a machine that is not mine.
Do not fix it for me.
```

Después de ejecutarlo, verifica: si solo te tranquiliza, empuja de vuelta y exige la
peor falla individual. Contrarresta el *acuerdo complaciente* (elogio que revisa tu ego,
no tu paquete).

✍️ **Tu registro.** Haz doble clic en esta celda y anota: qué devolvió la IA (una o dos líneas), qué verificaste y cómo, y tu fila del ledger.

> **No delegues.**
