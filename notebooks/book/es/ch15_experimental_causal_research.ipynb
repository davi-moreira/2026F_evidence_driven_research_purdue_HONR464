## Código del capítulo

Las celdas de abajo vienen del capítulo. Córrelas, luego cambia algo y corre de nuevo — los números deben moverse como dice el capítulo.

*De la sección “Una simulación con semilla”.*

````python
import numpy as np
import matplotlib.pyplot as plt

SEED = 464
rng = np.random.default_rng(SEED)

n, tau = 200, 5.0                            # efecto verdadero: +5 puntos
y0 = np.clip(rng.normal(70, 12, n), 20, 99)  # tasa de renovación sin recordatorio
y1 = np.clip(y0 + tau, 20, 100)              # tasa de renovación con recordatorio

estimaciones = []
for _ in range(2000):
    tratados = rng.permutation(n) < n // 2   # una nueva asignación al azar
    estimaciones.append(y1[tratados].mean() - y0[~tratados].mean())
estimaciones = np.array(estimaciones)

fig, ax = plt.subplots(figsize=(7.6, 3.2))
ax.hist(estimaciones, bins=40, color="#2a78d6", edgecolor="white", lw=.4)
ax.axvline(tau, color="#333333", ls="--", lw=1.2)
ax.text(.02, .92, f"efecto verdadero = {tau:.1f} pp", color="#333333",
        fontsize=9, transform=ax.transAxes)
ax.text(.02, .82, f"media de las estimaciones = {estimaciones.mean():.1f} pp",
        color="#2a78d6", fontsize=9, transform=ax.transAxes)
ax.set_xlabel("Efecto estimado del recordatorio (puntos porcentuales)")
ax.set_ylabel("Número de reasignaciones")
plt.show()
````
