```

myself if someone asked how I know it is true. Do not rewrite anything. List the weak
spots.
```

Después de ejecutarlo, verifica: para cada oración señalada, o produces tu propia
revisión o cortas la afirmación. Contrarresta el **acuerdo complaciente**, elogio que
revisa tu ego en lugar de tu evidencia.

> **No delegues.**
>
> La herramienta puede buscar, redactar y calcular. Nunca puede decidir en qué

> ética de tu muestreo, cómo enuncias tu incertidumbre, ni qué dice tu declaración de
> uso de IA. La autoría es una responsabilidad, no una tarea, y no se transfiere.
