````python
import numpy as np, pandas as pd
SEED = 464
rng = np.random.default_rng(SEED)

# Constructed transcripts: eight owners, one line each, for the example only.
transcripts = {
    "O1": "We reprint the menu once a year. Reprinting costs real money, "
          "so everything goes up at once.",
    "O2": "The breakfast burrito stays at eight dollars. People notice that one, "
          "so the increase goes on the sides.",
    "O3": "Eggs doubled last spring. I ate it for two months because I figured "
          "it would pass.",
    "O4": "Our prices live on a screen now. I change them most weeks when the "
          "supplier invoice lands.",
    "O5": "New menus cost me four hundred dollars, so I wait and do all the "
          "prices in one go.",
    "O6": "Nobody tracks what a side salad costs. They track the burger.",
    "O7": "If beef stays high past the summer, then I move the price. Not before.",
    "O8": "I raised the burger and lost two regulars in a week. Now I raise the "
          "fries instead.",
}
# Two of the contrasts you chose the owners for: menu medium and service.
cases = pd.DataFrame({
    "menu":     ["printed", "printed", "printed", "screen",
                 "printed", "printed", "screen", "printed"],
    "service":  ["sit-down", "counter", "sit-down", "counter",
                 "sit-down", "sit-down", "sit-down", "counter"],
}, index=[f"O{i}" for i in range(1, 9)])

ONE, SHIELD, WAIT = ("reprice in one round", "shield the watched dishes",
                     "wait out a passing shock")
rows = [  # excerpt, owner, code, theme, quote
    ("E01", "O1", "reprint cost",         ONE,    "Reprinting costs real money"),
    ("E02", "O5", "reprint cost",         ONE,    "New menus cost me four hundred dollars"),
    ("E03", "O5", "batch the changes",    ONE,    "do all the prices in one go"),
    ("E04", "O4", "prices change weekly", ONE,    "I change them most weeks"),  # CONTRARY
    ("E05", "O2", "anchor dish",          SHIELD, "The breakfast burrito stays at eight dollars"),
    ("E06", "O6", "anchor dish",          SHIELD, "They track the burger"),
    ("E07", "O8", "shift to sides",       SHIELD, "Now I raise the fries instead"),
    ("E08", "O2", "shift to sides",       SHIELD, "the increase goes on the sides"),
    ("E09", "O3", "absorb a short spike", WAIT,   "I figured it would pass"),
    ("E10", "O7", "judge if it lasts",    WAIT,   "If beef stays high past the summer"),
]
trail = pd.DataFrame(rows, columns=["excerpt", "owner", "code", "theme", "quote"])
trail["contrary"] = trail.excerpt.eq("E04")

# Every quote must appear WORD FOR WORD in its owner's transcript.
trail["verbatim"] = [q in transcripts[o] for o, q in zip(trail.owner, trail.quote)]
print(f"quotes found verbatim : {trail.verbatim.sum()} of {len(trail)}")

# Case coverage: WHICH of your owners, and which kinds, each theme draws on.
# It shows where a theme has and has not been heard. It is not a score.
def kinds(owners, col):
    return "/".join(sorted(cases.loc[sorted(set(owners)), col].unique()))
coverage = pd.DataFrame({
    theme: {
        "owners": ",".join(sorted(set(g.owner[~g.contrary]))),
        "against": ",".join(sorted(set(g.owner[g.contrary]))) or "-",
        "menu": kinds(g.owner[~g.contrary], "menu"),
        "service": kinds(g.owner[~g.contrary], "service"),
    }
    for theme, g in trail.groupby("theme")
}).T
print(coverage.to_string())

ai_quote = "I never pass cost increases on to my customers"
print(f"\nAI quote found in O3's transcript : {ai_quote in transcripts['O3']}")
audit = sorted(str(e) for e in rng.choice(trail.excerpt.to_numpy(), 3, replace=False))
print(f"excerpts for a critical reader     : {audit}")
print("\ncoverage shows where each theme was heard among THESE eight owners;")
print("it is not a prevalence estimate and not a measure of a theme's importance")
````
