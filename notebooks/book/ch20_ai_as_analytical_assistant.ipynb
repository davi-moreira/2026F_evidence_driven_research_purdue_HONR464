## Recommended AI prompts

Commit your own expectation first, then delegate. Each prompt is a checkable job, not a
request for a verdict.

**Propose the checks you might have missed.**

```text
Here is my robustness grid for one estimate: I vary the sample (all enrollees /
drop those who enrolled with an offer in hand), the measurement (mean / median
weeks to first offer), and the specification (raw gap / adjusted for prior work
experience). Act as a hostile reviewer. Name up to three additional, equally
defensible checks I did NOT list, and for each say which direction you expect it
to move the estimate.
```
After running, verify: check each proposal against your four handles and confirm
you can actually compute it in your data. Counters *illusion of completeness* (a
tidy grid that still omits the one check that would break the claim).

**Delegate the computation, keep the numbers checkable.**

```text
Recompute my time-to-offer gap under each row of this grid and return a table:
row label, the gap, and the exact columns and filter used for that row.
Do not summarize; show one number and its code path per row.
```
After running, verify: recompute two rows by hand with a second expression and
confirm they match the table. Counters *plausible-but-wrong-method* (a row whose
label and code quietly disagree).

✍️ **Your run.** Double-click this cell and record: what the AI returned (one or two lines), what you verified and how, and your ledger row.

**Make the loop confess.**

```text
Before I accept this table: list every version of this analysis you ran while
producing it, in order, with what changed each time and why. If you tried a
variant and set it aside, it goes on the list too.
```
After running, verify: compare its list against your own cycle log and re-run any
variant it mentions that you never saw. Counters an *unlogged specification search*
(the loop quietly trying options until one looks good).

✍️ **Your run.** Double-click this cell and record: what the AI returned (one or two lines), what you verified and how, and your ledger row.

**Red-team the surviving claim.**

```text
Here is my claim: "the program gets people to a first offer about 0.7 SD sooner."
Act as a hostile reviewer of labor-market program evaluations. Name the single
worst way this result could mislead, and the exact data check that would confirm
or kill it. Do not rewrite my claim.
```
After running, verify: run the check it names, and treat the flaw as real only if
your own output confirms it. Counters *sycophantic agreement* (praise that reviews
your ego, not your evidence).

✍️ **Your run.** Double-click this cell and record: what the AI returned (one or two lines), what you verified and how, and your ledger row.
