## Prompts de IA recomendados

Comprometa-se com o seu próprio rascunho primeiro, depois delegue. Cada prompt é uma
tarefa conferível, não um veredito para acreditar. Rode-os como um loop: a primeira
resposta nomeia os instrumentos óbvios, e a passagem útil é a segunda, em que você diz
qual faceta a ferramenta pulou e pede que ela tente de novo. Ferramentas agênticas vão
alegremente dar três ou quatro voltas e devolver um plano de mensuração polido. Polido
não é válido, e a frase que nomeia o que o seu número deixa de fora continua tendo que
ser escrita por você.

**Localize os indicadores padrão.**

```text
Act as a survey-methods assistant. I want to measure the construct "willingness to
recommend one's workplace" in an organizational study. Name the standard, published
instruments researchers use for this construct, with the instrument name, how it is
scored, and a real source I can open. Only list instruments you are confident
exist; mark anything uncertain.
```
Depois de rodar, verifique: abra uma das fontes nomeadas e confirme que o instrumento
existe e mede o que a ferramenta diz que ele mede. Enfrenta a *fabricação confiante*
(o nome de uma escala inventada chega tão fluente quanto o de uma real).

✍️ **Seu registro.** Clique duas vezes nesta célula e anote: o que a IA devolveu (uma ou duas linhas), o que você verificou e como, e a sua linha do ledger.

**Liste os limites, para você poder verificá-los.**

```text
Here is my operationalization: concept = employee engagement, construct =
willingness to recommend the workplace, indicator = mean of a single 1-7 agreement
item. Return a table of every facet of engagement this indicator does NOT capture,
one row each, with why it matters and who it fails to hear from.
```
Depois de rodar, verifique: confronte a tabela com um modelo publicado de engajamento
que você mesmo recupere. Enfrenta a *ilusão de completude* (uma lista bem arrumada que
mesmo assim omite a faceta que mais ameaça a sua alegação).

✍️ **Seu registro.** Clique duas vezes nesta célula e anote: o que a IA devolveu (uma ou duas linhas), o que você verificou e como, e a sua linha do ledger.

**Faça red team na troca.**

```text
Act as a hostile reviewer. Here is my claim: "the pilot sites are more engaged,
because more people there would recommend the company." Name every place my
indicator fails to support the word "engaged." Do not rewrite the claim for me.
```
Depois de rodar, verifique: se a ferramenta só elogiar o desenho, insista e exija a
única pior lacuna. Enfrenta a *concordância bajuladora* (elogio que revisa o seu ego,
não a sua mensuração).

✍️ **Seu registro.** Clique duas vezes nesta célula e anote: o que a IA devolveu (uma ou duas linhas), o que você verificou e como, e a sua linha do ledger.
