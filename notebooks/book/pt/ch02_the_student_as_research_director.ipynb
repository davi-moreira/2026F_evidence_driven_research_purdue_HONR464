## Prompts de IA recomendados

Cada prompt abaixo é do tipo **comprometa-se primeiro**: escreva a sua própria
resposta antes de abrir a ferramenta, para que a resposta dela tenha algo seu com
que discordar. Espere rodar cada um mais de uma vez. Você lê a resposta, encontra o
ponto fraco, afia o prompt e vai de novo. O loop é normal; parar o loop antes de
ter conferido qualquer coisa é que é o erro.

**Localize um recurso de bancada (delegável).**

```text
Act as a lab-methods assistant in microbiology. I am measuring bacterial growth
by optical density (OD600). List 3 published, retrievable protocols or textbook
methods for reading a growth curve. For each, give authors or source, year, and
where it appears. Only include work you are confident exists; mark anything
uncertain. Then tell me which one to verify first.
```

Depois de rodar, verifique (combate a *fabricação confiante*): abra você mesmo
cada protocolo antes de citá-lo; uma fonte que você não consegue recuperar é
inventada até que ela abra.

✍️ **Seu registro.** Clique duas vezes nesta célula e anote: o que a IA devolveu (uma ou duas linhas), o que você verificou e como, e a sua linha do ledger.

**Transforme a sua triagem em uma lista checável.**

```text
Here are eight tasks in my project: [paste your list]. For each, label it "safe to
delegate," "delegate then verify," or "never delegate," and give a one-line reason.
Return it as a table so I can audit each row on its own.
```

Depois de rodar, verifique (combate a *ilusão de completude*): compare a tabela com
a triagem que você escreveu antes e com a sua lista do que nunca se delega; uma
tabela organizada pode arquivar no lugar errado, sem alarde, justamente a decisão
que importava.

✍️ **Seu registro.** Clique duas vezes nesta célula e anote: o que a IA devolveu (uma ou duas linhas), o que você verificou e como, e a sua linha do ledger.

**Faça o red team (revisão adversária) do seu próprio limite.**

```text
I marked this task "never delegate": [paste one task]. Act as a hostile reviewer
and argue that it is actually safe to hand to an AI. Give your strongest case. Do
not agree with me.
```

Depois de rodar, verifique (combate a *concordância bajuladora*): se a revisão
ceder e elogiar a sua escolha, conteste, e depois leve a objeção mais forte dela a
um colega ou a um mentor, em vez de confiar na aprovação da ferramenta.

✍️ **Seu registro.** Clique duas vezes nesta célula e anote: o que a IA devolveu (uma ou duas linhas), o que você verificou e como, e a sua linha do ledger.

> **Não delegue.**
>
> Os próprios grupos são seus para preencher, e três escolhas nunca saem do grupo
> "nunca delegar": **qual pergunta o seu projeto persegue**, **se a sua evidência
> sustenta a alegação que você quer fazer** e **quanta incerteza você reporta**. Você
> pode pedir a uma ferramenta que exponha as opções, mas escolher entre elas, e
> responder pela escolha, é o trabalho que faz de você quem dirige.
