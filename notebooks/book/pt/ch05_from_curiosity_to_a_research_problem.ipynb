## Prompts de IA recomendados

Comprometa-se primeiro com a sua própria resposta. Só então abra a ferramenta, para
que o resultado dela tenha algo seu com que discordar. E conte com dar mais de uma
volta. A primeira lista de problemas candidatos quase nunca é a útil; você estreita,
pergunta de novo e estreita mais. A única coisa que não pode se deslocar ao longo
desses ciclos é de quem é a curiosidade que está sendo delimitada.

**Localize o que já se sabe** (antes de alegar que o seu problema é novo):

```
Act as a research-scoping assistant. My topic is [your topic]. List what is
already reasonably well established about it and where a genuine open question
might remain. For every established fact, give one real, retrievable source I
can open myself. Mark anything you are unsure exists.
```

*Depois de rodar, verifique:* abra cada fonte antes de confiar nela. Isso combate a
**fabricação confiante**, em que uma resposta fluente cita um estudo com cara de
real que não existe.

✍️ **Seu registro.** Clique duas vezes nesta célula e anote: o que a IA devolveu (uma ou duas linhas), o que você verificou e como, e a sua linha do ledger.

**Alargue uma curiosidade em candidatos que você confere:**

```
Here is my curiosity: [one sentence]. Produce five candidate research problems
inside it. For each, name its units of analysis, its outcome, and its target
population, in a table. Do not add any topic I did not mention.
```

*Depois de rodar, verifique:* confira cada linha contra a curiosidade que você de
fato escreveu. Isso combate a **mudança silenciosa de escopo**, em que a ferramenta
responde discretamente a uma pergunta vizinha e a devolve como se fosse a sua.

✍️ **Seu registro.** Clique duas vezes nesta célula e anote: o que a IA devolveu (uma ou duas linhas), o que você verificou e como, e a sua linha do ledger.

**Faça o red team (revisão adversária) do problema que você escolheu:**

```
Act as a hostile grant reviewer. Here is the research problem I want to commit
months of work to: [paste it]. Argue the strongest case that it is not worth
doing. Attack it on importance, on feasibility, and on whether it adds anything
new.
```

*Depois de rodar, verifique:* fique só com as objeções que sobrevivem ao seu próprio
raciocínio. Isso combate a **concordância bajuladora**, em que a ferramenta elogia a
sua ideia porque foi você quem a propôs.

✍️ **Seu registro.** Clique duas vezes nesta célula e anote: o que a IA devolveu (uma ou duas linhas), o que você verificou e como, e a sua linha do ledger.
