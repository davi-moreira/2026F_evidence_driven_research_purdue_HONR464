## Prompts de IA recomendados

Comprometa-se primeiro com a sua própria versão e só depois delegue. Cada prompt é uma
tarefa verificável, não um pedido de veredito.

Espere duas ou três voltas em cada um. A primeira compressão costuma ser lisa demais,
então você aponta o qualificador que ela descartou e pergunta de novo. Ferramentas
agênticas vão iterar rumo ao que soa melhor pelos próprios critérios delas, que é
exatamente a pressão que alarga alegações, então a sua frase de fronteira precisa ser
algo que você segura, e não algo que você delega.

**Localize o jargão e traduza.**

```text
Here is one sentence from my poster: "[paste it]". My headline claim and its
boundary are: "[paste them]". List every word a curious stranger with no training
in my field could not decode. Then give two plain rewrites a first-year in any
major would follow, each keeping my claim exactly as narrow as it is now.
```

Depois de rodar, verifique: leia cada reescrita ao lado da sua fronteira e rejeite
qualquer uma que tenha descartado um qualificador como "separado para teste" ou
transformado "sinalizou" em "prevê". Neutraliza a *mudança silenciosa de escopo* (uma
reescrita "simples" que alarga a alegação enquanto simplifica).

✍️ **Seu registro.** Clique duas vezes nesta célula e anote: o que a IA devolveu (uma ou duas linhas), o que você verificou e como, e a sua linha do ledger.

**Liste as palavras que inflam a minha alegação.**

```text
Here is my 30-second hook: "[paste it]". Here is my poster's boundary sentence:
"[paste it]". Play a hurried visitor. List every word in the hook that says
something bigger than the boundary allows. Rewrite nothing; just name the words.
```

Depois de rodar, verifique: se ela não sinalizar nada e chamar o gancho de "forte",
insista e exija o pior transgressor, um só. Neutraliza a *concordância bajuladora*
(elogio revisa o seu ego, não as suas palavras).

✍️ **Seu registro.** Clique duas vezes nesta célula e anote: o que a IA devolveu (uma ou duas linhas), o que você verificou e como, e a sua linha do ledger.

**Faça red team no pitch como um estranho apressado.**

```text
Play a skeptical conference visitor with thirty seconds of patience, from outside
my field. Here is my full pitch: "[paste it]". Name the exact sentence where you
lose interest or stop believing me, and the single place I promise more than I show.
```

Depois de rodar, verifique: confronte a resposta dela com um ensaio de verdade com um
amigo, e pergunte a ela "o que foi a única coisa que você deixou de fora?". Neutraliza
a *ilusão de completude* (um "está claro" polido que pula o momento em que um estranho
de fato se dispersa).

✍️ **Seu registro.** Clique duas vezes nesta célula e anote: o que a IA devolveu (uma ou duas linhas), o que você verificou e como, e a sua linha do ledger.

> **Não delegue.**
>
> Três decisões continuam suas. Você decide **qual alegação a sua evidência de fato
> sustenta**, **onde a alegação precisa parar** (a frase de fronteira exata) e **se uma
> frase comprimida ainda diz o que o pôster diz**. Uma ferramenta pode enxugar palavras e
> sinalizar jargão. Ela não pode decidir que a frase mais curta continua verdadeira, e
> não vai estar diante do painel quando o estranho perguntar. Você diz a frase. Ela é
> sua.
