## Prompts de IA recomendados

Comprometa-se primeiro com o seu próprio rascunho de tabela e só depois delegue. Cada
prompt é uma tarefa verificável, não um pedido de veredito.

Nenhum deles é de tiro único. Você envia o prompt, lê a resposta, insiste na linha que
parece lisa demais e envia de novo. Ferramentas agênticas vão iterar por você e devolver
uma tabela pronta, e é exatamente aí que se deve desacelerar. Uma tabela montada em
cinco voltas invisíveis ainda precisa ser aberta célula por célula.

**Localize a fonte de um fato citado.**

```text
Act as a reference assistant in political science. I want to cite the officially
reported turnout for this county in this election. Name the figure and one real,
retrievable source that reports it, such as a state or county election office
release, and give me enough detail to find it myself. If you are not confident a
source exists, say so instead of guessing.
```
Depois de rodar, verifique: abra você mesmo a fonte indicada e confirme que ela relata
aquele valor antes de ele encostar no seu pôster. Neutraliza a *fabricação confiante*
(uma citação inventada chega com a mesma confiança de uma real).

✍️ **Seu registro.** Clique duas vezes nesta célula e anote: o que a IA devolveu (uma ou duas linhas), o que você verificou e como, e a sua linha do ledger.

**Liste todas as alegações, para que nenhuma fique sem respaldo.**

```text
Here is my poster text: [paste it]. List every distinct claim it makes, one per
row. For each, name what evidence a reader would need to see to believe it. Do not
judge whether my evidence is good; just surface every claim, including implicit
ones.
```
Depois de rodar, verifique: confronte a lista com a sua própria tabela e acrescente
qualquer alegação para a qual você não tinha escrito uma linha. Neutraliza a *ilusão de
completude* (uma lista arrumadinha que pula em silêncio justamente a alegação sem
evidência por trás).

✍️ **Seu registro.** Clique duas vezes nesta célula e anote: o que a IA devolveu (uma ou duas linhas), o que você verificou e como, e a sua linha do ledger.

**Faça red team no encaixe alegação–evidência.**

```text
Here is one row of my claim-evidence table: claim "[paste]", evidence "[paste]".
Act as a hostile reviewer. Tell me every way this evidence fails to license this
claim, or claims more than it shows. Do not soften the claim for me; name the words
that overreach.
```
Depois de rodar, verifique: leia cada objeção contra a sua coluna de fronteira e
rejeite qualquer uma que peça para você conceder algo que a sua evidência de fato
autoriza. Neutraliza a *mudança silenciosa de escopo* (evidência para uma alegação mais
fácil, disfarçada de respaldo para a difícil).

✍️ **Seu registro.** Clique duas vezes nesta célula e anote: o que a IA devolveu (uma ou duas linhas), o que você verificou e como, e a sua linha do ledger.

> **Não delegue.**
>
> A **redação de cada alegação**, o **julgamento de que uma peça de evidência realmente a
> sustenta** e a **fronteira que cada alegação carrega** continuam suas. Uma IA pode
> listar alegações e rascunhar linhas, mas não pode abrir a sua célula do notebook para
> confirmar que um número é real, e não pode decidir que a evidência é honesta o bastante
> para ser impressa. A decisão de que a tabela está completa e o trabalho está travado é
> sua para declarar e defender.
