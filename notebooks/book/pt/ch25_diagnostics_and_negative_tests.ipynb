## Prompts de IA recomendados

Comprometa-se com a sua própria expectativa primeiro, depois delegue uma tarefa
conferível. Espere trabalhar em ciclos: propor, interrogar, refinar, perguntar de novo.
Um julgamento, porém, nunca melhora com um novo prompt: se o seu controle é
genuinamente nulo. Isso você resolve uma vez só, a partir do mecanismo, e nenhuma
quantidade de voltas resolve por você.

**Localize o controle padrão.**

```text
Act as a lab-methods assistant. I ran an antibiotic disk-diffusion assay with a
plant extract dissolved in 70% ethanol. Before any advice, name the standard
negative (vehicle) control for this assay and cite a protocol or textbook where it
appears. Only name controls you are confident are standard practice.
```
Depois de rodar, verifique: abra o protocolo citado e confirme que o controle é real e
padrão antes de usá-lo. Contra-ataca a *fabricação confiante* (um controle "padrão"
inventado chega com a mesma confiança de um real).

✍️ **Seu registro.** Clique duas vezes nesta célula e anote: o que a IA devolveu (uma ou duas linhas), o que você verificou e como, e a sua linha do ledger.

**Liste testes negativos candidatos e depois verifique cada um.**

```text
Here is my analysis: [one-sentence design, your cause, your outcome]. Propose three
negative tests: a placebo label, a falsification outcome measured where the effect
cannot yet exist, and a negative-control outcome my cause could not plausibly affect.
Para cada um, nomeie a quantidade que é verdadeiramente zero, diga que leitura contaria
como comum no meu tamanho de amostra e qual seria grande o bastante para preocupar, e
diga se a checagem conseguiria detectar um artefato grande o bastante para importar.
```
Depois de rodar, verifique: para cada teste, confirme que a sua causa genuinamente
*não pode* tocar o controle negativo. Um controle que a sua causa consegue afetar não é
nulo, e ele falha em silêncio. Contra-ataca o *método plausível mas errado* (um
"controle negativo" que não é de fato zero).

✍️ **Seu registro.** Clique duas vezes nesta célula e anote: o que a IA devolveu (uma ou duas linhas), o que você verificou e como, e a sua linha do ledger.

**Faça o red team de uma passagem limpa.**

```text
Here is a negative test I ran and its result: [describe the control and its
near-zero outcome]. Act as a hostile reviewer. Name every reason this test could
read as zero for the WRONG reason: too few samples to detect anything, a control too
weak to carry the artifact, a zero I would have gotten no matter what. Do not reassure me.
```
Depois de rodar, verifique: se ela apenas elogiar o resultado limpo, insista e exija a
única maneira pela qual o teste é não informativo. Contra-ataca a *concordância
bajuladora* (elogio que revisa o seu alívio, não a sua evidência).

✍️ **Seu registro.** Clique duas vezes nesta célula e anote: o que a IA devolveu (uma ou duas linhas), o que você verificou e como, e a sua linha do ledger.

> **Não delegue.**
>
> Você decide **de qual teste negativo o seu desenho de pesquisa realmente precisa**, **se
> o controle é genuinamente nulo** (um lugar que a sua causa de fato não alcança) e **o
> que uma passagem limpa autoriza você a alegar**. Uma ferramenta pode listar controles
> candidatos, mas só você conhece o mecanismo bem o bastante para atestar que a sua causa
> não toca o que você escolheu. O julgamento de que um teste negativo aprovado descarta um
> artefato *sem* provar que o seu efeito é real continua sendo seu.
