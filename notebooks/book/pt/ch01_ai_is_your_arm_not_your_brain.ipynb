## Prompts de IA recomendados

Comprometa-se primeiro com a sua própria resposta, e só então delegue. E espere um
loop, não um tiro único: você pergunta, lê o que volta, aperta na parte que parece
frágil e pergunta de novo. Esse vaivém é o **loop de IA**, o ciclo prompt →
resposta → interrogação → refinamento → nova execução, e é assim que o trabalho
com IA funciona de verdade. O que importa é onde o loop para. Ele para quando você
mesmo rodou a checagem, não quando a ferramenta finalmente soa segura.

Cada prompt abaixo entrega uma tarefa que você consegue checar, com uma nota de
verificação que nomeia a falha contra a qual ela defende.

**Localize fontes candidatas (você verifica cada uma).**

```text
Act as an economic-research assistant. I need annual consumer price index values
for 2019 and 2025. List 3 candidate published sources, each with the exact table
or series name and the agency that publishes it. Only include documents you are
confident exist; mark anything uncertain.
```

Depois de rodar, verifique (combate a *fabricação confiante*): abra cada documento
você mesmo e leia o número. Uma tabela que você não consegue recuperar é inventada
até você encontrá-la.

✍️ **Seu registro.** Clique duas vezes nesta célula e anote: o que a IA devolveu (uma ou duas linhas), o que você verificou e como, e a sua linha do ledger.

**Transforme uma decisão em uma lista de checagem (você audita a lista).**

```text
For converting a 2019 restaurant price into 2025 dollars, list every input my
calculation needs. For each, say why it belongs and which single choice most
governs the result.
```

Depois de rodar, verifique (combate a *ilusão de completude*): uma lista longa e
organizada ainda pode omitir a escolha que governa a sua resposta. Compare-a com
uma lista de checagem que você escreveu antes e confirme que a escolha
determinante é mesmo aquela em que a comparação se apoia.

✍️ **Seu registro.** Clique duas vezes nesta célula e anote: o que a IA devolveu (uma ou duas linhas), o que você verificou e como, e a sua linha do ledger.

**Faça o red team (revisão adversária) da sua própria escolha (a decisão continua
sua).**

```text
Here is my price index and base year for comparing 2019 and 2025 menu prices:
[paste it]. Act as a hostile referee. Name every place the choice reaches past my
evidence. Do not rewrite it for me; list the weaknesses.
```

Depois de rodar, verifique (combate a *concordância bajuladora*): se toda objeção
for branda, ou se a ferramenta chamar a sua escolha de "bem equilibrada", conteste
e peça o pior defeito, um só.

✍️ **Seu registro.** Clique duas vezes nesta célula e anote: o que a IA devolveu (uma ou duas linhas), o que você verificou e como, e a sua linha do ledger.

> **Não delegue.**
>
> Estas continuam sendo suas, por mais fluente que a ferramenta soe. Decidir qual
> medida se ajusta à sua pergunta e qual ano-base a comparação exige. Julgar se uma
> fonte é confiável o bastante para servir de base. Fixar a alegação que você vai
> defender e enunciar a sua incerteza. A ferramenta reúne opções e rascunha texto;
> escolher entre elas, e responder pela escolha, é o trabalho de quem pesquisa.
