## Prompts de IA recomendados

Cada prompt é **comprometa-se primeiro**: anote o valor ou a fonte em que você já
acredita, e só então deixe a ferramenta discordar de algo seu. Trate cada um deles
como o começo de um loop, e não como um pedido único. A primeira resposta lhe dá
pistas; a sua segunda mensagem nomeia a pista que não sobreviveu à recuperação e
pergunta de novo. Ferramentas que rodam o loop de busca sozinhas vão lhe entregar uma
lista de fontes mais longa e mais arrumada. Mais longa e mais arrumada não quer dizer
aberta, e abrir cada uma continua sendo trabalho seu.

```text
Act as a research-data assistant. I have this figure: [paste the value, the place,

anything uncertain, and say for each whether it is the original record or a
re-report.
```

**Transforme números emprestados em uma tabela de proveniência (delegável).**

```text
Here are five numbers my project borrows: [paste them]. For each, return a row with
the value, its exact definition (which population, which time window, which
denominator), the primary source, and whether the value was counted, estimated, or
self-reported. Return it as a table so I can audit each row on its own.
```

Depois de rodar, verifique (enfrenta a *mudança silenciosa de escopo*): leia a
definição de cada linha contra a definição de que a sua pergunta realmente precisa;
um número construído sobre outra população pode lhe ser entregue como se fosse o seu.

✍️ **Seu registro.** Clique duas vezes nesta célula e anote: o que a IA devolveu (uma ou duas linhas), o que você verificou e como, e a sua linha do ledger.

**Faça red team no número de que você gostou (delegável).**

```text
I plan to build on this figure: [paste value and source]. Act as a hostile reviewer
and argue it is unreliable for a claim about [your population and time window]:
wrong denominator, self-reported rather than counted, or copied through a chain of
summaries. Do not reassure me.
```

Depois de rodar, verifique (enfrenta a *concordância bajuladora*): se a revisão só
elogiar o seu número, insista e depois leve a objeção mais forte dela à fonte
primária ou a alguém que conheça os dados melhor que você, em vez de confiar em uma
aprovação fácil.
