## Prompts de IA recomendados

Comprometa-se primeiro com a sua própria resposta e só depois delegue. Cada prompt é uma
tarefa verificável, não um pedido de veredito.

Rode o loop, e registre. Cada volta é um prompt, uma saída, uma objeção e uma nova
rodada, e a linha do registro anota o que você finalmente aceitou e por quê. Isso importa
aqui mais do que em qualquer outra parte do livro. Se uma ferramenta agêntica rodou dez
voltas invisíveis para produzir o seu parágrafo, a sua declaração ainda tem que
descrever apenas aquilo por que você pode pessoalmente responder.

**Localize o padrão.**

```text
Act as a research-integrity assistant. Name the elements a complete AI-use
disclosure should contain for a research poster, and cite one real, retrievable
source such as a university research-integrity office or a major publisher's AI
policy. Only name a source you are confident exists.
```
Depois de rodar, verifique: abra a política citada e confirme que cada elemento nomeado
está de fato nela. Neutraliza a *fabricação confiante* (um elemento de política inventado
chega com a mesma confiança de um real).

✍️ **Seu registro.** Clique duas vezes nesta célula e anote: o que a IA devolveu (uma ou duas linhas), o que você verificou e como, e a sua linha do ledger.

**Transforme anotações numa tabela de declaração a verificar.**

```text
Here are my raw notes on every time I used an AI tool on this project: [paste].
Build a table with columns tool, task, and verification method, one row per use.
Do not invent a verification I did not mention. If a use has no check, leave that
cell blank and flag the row.
```
Depois de rodar, verifique: confronte cada linha com o seu registro; uma verificação
preenchida onde as suas anotações não tinham nenhuma é fabricada, então corte.
Neutraliza a *ilusão de completude* (uma tabela arrumadinha que relata em silêncio
checagens que você nunca fez).

✍️ **Seu registro.** Clique duas vezes nesta célula e anote: o que a IA devolveu (uma ou duas linhas), o que você verificou e como, e a sua linha do ledger.

**Faça red team na sua declaração.**

```text
Here is my drafted AI-use disclosure: [paste]. Act as a hostile reviewer who
suspects I am hiding a use or overstating my checks. Name every place I claim a
verification I have not shown, or omit a use a reviewer would expect. Do not
rewrite it for me.
```
Depois de rodar, verifique: para cada linha sinalizada, apresente a linha do registro que
a respalda ou apague a alegação. Neutraliza a *concordância bajuladora* (um revisor que
elogia a sua honestidade em vez de testá-la).

✍️ **Seu registro.** Clique duas vezes nesta célula e anote: o que a IA devolveu (uma ou duas linhas), o que você verificou e como, e a sua linha do ledger.

> **Não delegue.**
>
> A honestidade da declaração é só sua. Você decide **quais usos declarar** (inclusive
> aquele que você preferiria não mencionar), **se uma verificação que você afirma de fato
> aconteceu** e **se o enunciado final é verdadeiro**. Uma ferramenta pode formatar o seu
